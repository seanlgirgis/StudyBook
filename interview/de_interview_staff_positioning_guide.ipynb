## 1. What Is Staff-Level Impact?

```
SCOPE OF IMPACT BY LEVEL
──────────────────────────────────────────────────────────────────

  L4 SWE:        ──────▶  Feature / task
                          "Implement the ETL job"

  L5 Senior:     ──────▶  Component / service
                          "Design and own the ETL pipeline"

  L6 Staff:      ──────▶  System / team boundary
                          "Define the data platform architecture
                           that 4 teams build on"

  L7 Principal:  ──────▶  Org / company
                          "Set the data strategy for the division"

THE STAFF SHIFT — THREE CHANGES
────────────────────────────────
  1. From IC output  →  Multiplied team output
     You write less code; the team ships better code because of you.

  2. From manager tells you  →  You set the agenda
     You identify the problem worth solving, not just solve assigned problems.

  3. From one team  →  Cross-team / cross-org
     Your work unblocks or shapes what other teams do.

ARCHETYPES (Will Larson — Staff Engineer)
──────────────────────────────────────────
  Tech Lead:    Embedded in a team, guides technical direction
  Solver:       Rotates to the hardest problems, fixes, moves on
  Architect:    Owns cross-cutting technical strategy
  Right Hand:   Extends a VP/CTO's reach into engineering decisions
```

## 3. Core Competencies

```
STAFF ENGINEER TOOLS
──────────────────────────────────────────────────────────────────
TOOL                    WHEN TO USE                 OUTPUT
────────────────────────────────────────────────────────────────
RFC / Design Doc        Before building anything    Written alignment
                        cross-team or risky

Architecture Review     New system or major change  Go/no-go + conditions

Tech Roadmap            Quarterly planning          Prioritized problem list

1:1 Mentoring           Individual growth gaps      Engineer leveled up

Office Hours / Review   Scale your expertise        Standards adopted

Post-mortem / RCA       After incidents             System never fails same way

Standards & Templates   Recurring decisions         Decisions automated away

Coalition Building      Cross-team decision needed  Aligned stakeholders

THINGS YOU DO NOT DO AT STAFF LEVEL
─────────────────────────────────────
❌  Own every line of code in the system
❌  Say yes to every review request (you become a bottleneck)
❌  Make every decision unilaterally (creates dependence, not ownership)
❌  Optimize for personal output over team output
❌  Avoid ambiguous, hard-to-measure work
```

````python
# Core competency tracker — maps signals to behaviors

def assess_staff_signal(behavior):
    """
    Given a behavior description, classify it as IC, Senior, or Staff signal.
    Used in interview prep to calibrate story-telling level.
    """
    # Keywords that signal Staff-level contribution
    staff_keywords = [
        "multiple teams", "cross-team", "org-wide", "adopted by",
        "roadmap", "RFC", "design review", "standard", "mentored",
        "leveled up", "coalition", "without authority", "unblocked",
        "shaped", "changed how", "framework", "template",
    ]
    senior_keywords = [
        "designed", "led", "owned", "delivered", "improved", "reduced",
        "built", "shipped", "diagnosed", "optimized",
    ]

    behavior_lower = behavior.lower()
    staff_matches = [k for k in staff_keywords if k in behavior_lower]
    senior_matches = [k for k in senior_keywords if k in behavior_lower]

    if staff_matches:
        level = "STAFF"
        signals = staff_matches
    elif senior_matches:
        level = "SENIOR"
        signals = senior_matches
    else:
        level = "IC"
        signals = []

    return level, signals


# Test with example story fragments
examples = [
    "I built the ETL pipeline for the marketing team.",
    "I designed and owned the data quality framework end-to-end.",
    "I wrote an RFC that was adopted by three teams, creating a shared standard for event schemas.",
    "I mentored two engineers; both were promoted to senior within 18 months.",
    "I built a coalition across the data platform, ML, and analytics teams to align on a lakehouse migration roadmap.",
]

print("STORY LEVEL CALIBRATION")
print("=" * 60)
for ex in examples:
    level, signals = assess_staff_signal(ex)
    print(f"\n[{level}] {ex[:70]}..." if len(ex) > 70 else f"\n[{level}] {ex}")
    if signals:
        print(f"  Staff signals found: {signals}")
````

## 4. Decision Map — When to Use Each Mode

```
SITUATION → STAFF RESPONSE
──────────────────────────────────────────────────────────────────
Team is making a bad technical decision
  → Write a short 1-pager with data; propose alternative; let them choose
  → Do NOT override; DO make your case with evidence

Two teams have conflicting priorities / approaches
  → Facilitate a structured working session; document the decision
  → Write RFC; get written sign-off; remove ambiguity

A junior engineer is stuck / making slow progress
  → 1:1 coaching, not fix-it-for-them
  → Ask: "What's your current plan? What are the options?"
  → Transfer understanding, not just the answer

System has recurring class of bugs / incidents
  → Write a post-mortem + systemic fix (standard, template, lint rule)
  → The fix should make that class of bug impossible, not just fix this one

Leadership asks for the technical strategy for Q3
  → Translate business priorities into a sequenced technical roadmap
  → Frame in impact, dependencies, and risk — not just tech topics

A senior engineer on another team is doing something dangerous
  → Have a direct 1:1 conversation first — assume good intent
  → If unresolved, escalate to their manager with data, not judgment

You're being asked to own a project across 4 teams
  → Clarify: am I the DRI or a consultant?
  → Get explicit executive sponsorship for cross-team decisions
  → Write a project charter with scope, owners, and decision rights
```

## 5. Pattern 1: Influence Without Authority

---

```
WHAT IT IS:
  Driving a significant technical decision or change when you have no
  reporting relationship with the people you need to convince.

THE TRAP:
  Relying on "I'm right" as the only argument. Technical correctness
  is necessary but not sufficient. People follow people they trust,
  whose ideas they feel ownership of, and who make their lives easier.

INFLUENCE LADDER (bottom = least effective, top = most):
  5. Mandate (you have authority)         ← NOT available at staff level
  4. Logical argument with data           ← necessary but often not enough
  3. Shared story / shared pain           ← creates emotional buy-in
  2. Demonstrated prototype / proof       ← show don't tell
  1. Let them co-author the solution      ← their idea = their commitment

SLOW MOTION TRACE — Data snapshot tool story:
  Step 1: Identified the pain (training data not reproducible).
  Step 2: Built the tool for MY team first — proved value locally.
  Step 3: Shared a demo with the analytics team; asked for feedback.
  Step 4: Incorporated their feedback — now it was partly their idea.
  Step 5: They adopted it. Shared it with team 3 organically.
  Key: never pushed; pulled them toward adoption by making it valuable.

INTERVIEW SIGNAL:
  Strong:  "I built a shared tool, got their input, they adopted it"
  Weak:    "I told them they should use my approach"
```

````python
# Influence without authority — story framework and calibration tool

def print_story(title, situation, task, actions, result, timing_s=120):
    print(f"\n{'='*60}")
    print(f"STORY: {title}  (~{timing_s}s)")
    print(f"{'='*60}")
    print("[S]", situation)
    print("[T]", task)
    print("[A]")
    for i, a in enumerate(actions, 1):
        print(f"  {i}. {a}")
    print("[R]", result)


# Influence without authority — canonical story
print_story(
    title="Influence Without Authority: Data Snapshot Standard",
    situation=(
        "At Citi, three data science teams (mine + analytics + risk) all trained"
        " models on shared Snowflake data with no versioning. When we hit a regression"
        " we couldn't reproduce, I realized the root cause was reproducibility — not"
        " the model. I had no authority over the other teams."
    ),
    task=(
        "Drive adoption of a shared data snapshot standard across three teams"
        " without any management mandate."
    ),
    actions=[
        "Built the tool for my own team first: logged training set manifests to S3"
        " before every run. Used it for 2 weeks to validate it solved our problem.",

        "Scheduled a 30-min demo with the analytics team lead. Framed it as:"
        " 'we solved a problem you might have too.' Asked for their feedback at the end.",

        "Incorporated their feedback (they wanted manifest to include Snowflake query hash)."
        " Now they were co-authors. Shared the updated version back.",

        "They adopted it. Three months later, the risk team asked the analytics lead"
        " for their reproducibility process — and got pointed to my tool.",
    ],
    result=(
        "All three teams adopted within 5 months — zero mandate."
        " The analytics team credited it in a post-mortem when it caught a training data issue."
        " I was cited in a quarterly review as a cross-team force multiplier."
    ),
    timing_s=120,
)

# Anti-patterns in influence
print("\nCOMMON INFLUENCE MISTAKES:")
mistakes = [
    ("WRONG", "Sending a long Slack message explaining why your approach is better"),
    ("WRONG", "Going to their manager before trying 1:1"),
    ("WRONG", "Insisting on credit when others adopt your idea"),
    ("RIGHT",  "Build a working prototype they can touch and experiment with"),
    ("RIGHT",  "Ask what problems they have BEFORE proposing your solution"),
    ("RIGHT",  "Incorporate their feedback — shared ownership = shared commitment"),
]
for tag, m in mistakes:
    icon = "✅" if tag == "RIGHT" else "❌"
    print(f"  {icon}  {m}")
````

## 6. Pattern 2: RFC / Design Review Process

---

```
WHAT IT IS:
  A written document that proposes a technical decision, presents
  alternatives, and explicitly asks for review and sign-off before
  implementation begins.

WHY IT MATTERS AT STAFF LEVEL:
  - Async alignment: gets 6 people aligned without a 2-hour meeting
  - Written record: prevents "but we decided X" disputes months later
  - Forces clear thinking: if you can't write it, you don't understand it
  - Scales your expertise: others can learn from your reasoning

RFC STRUCTURE (the one that works):
  Title:         One sentence — what decision is this?
  Status:        Draft | In Review | Approved | Superseded
  Authors:       Who wrote it
  Reviewers:     Who must approve
  Context:       Why does this problem exist? (2–3 paragraphs)
  Decision:      What are we doing? (clear, unambiguous)
  Alternatives:  What else was considered? Why rejected?
  Trade-offs:    What are we giving up?
  Implementation:High-level plan, timeline, owners
  Open Questions:What is still uncertain?

SLOW MOTION TRACE — Tiered retrain RFC:
  Day 1:  PM asks for one-click full retrain button
  Day 2:  I write draft RFC: propose tiered system instead
  Day 3:  Share with PM and 2 engineers — async review
  Day 5:  Address feedback (add cost estimates, failure modes)
  Day 7:  PM signs off — decision recorded, not in someone's head
  Outcome: No ambiguity; new team members can read the RFC to understand why

KEY INSIGHT:
  The RFC is not bureaucracy — it's the artifact that lets you
  move faster because everyone is aligned before you start building.
```

````python
# RFC template generator — produces a structured RFC outline

def generate_rfc_outline(title, decision, alternatives, trade_offs, reviewers):
    """
    Generate a structured RFC outline.
    Staff engineers write RFCs to drive async alignment at scale.
    """
    sections = [
        ("TITLE",        title),
        ("STATUS",       "Draft"),
        ("REVIEWERS",    ", ".join(reviewers)),
        ("CONTEXT",      "[Why does this problem exist? What breaks without this decision?]"),
        ("DECISION",     decision),
        ("ALTERNATIVES", "\n".join(f"  - {a}" for a in alternatives)),
        ("TRADE-OFFS",   "\n".join(f"  - {t}" for t in trade_offs)),
        ("PLAN",         "[High-level implementation steps + owner per step + timeline]"),
        ("OPEN QUESTIONS", "[What is still uncertain? What do we need to learn?]"),
    ]
    print("=" * 60)
    print("RFC OUTLINE")
    print("=" * 60)
    for section, content in sections:
        print(f"\n[{section}]")
        print(content)


# Example: Tiered model retrain RFC
generate_rfc_outline(
    title="Tiered Model Retrain System — HorizonScale ML Platform",
    decision=(
        "Implement three retrain tiers: Tier-1 (free, re-score existing model),"
        " Tier-2 ($200, subset retrain), Tier-3 ($2.4k, full retrain with approval)."
        " Remove the proposed one-click full retrain button."
    ),
    alternatives=[
        "One-click full retrain (rejected: $2.4k/run, no safeguards, reliability risk)",
        "No self-serve retrain (rejected: ops team bottleneck, PM goal not met)",
        "Scheduled-only retrain (rejected: too slow to respond to data drift)",
    ],
    trade_offs=[
        "Tier-2 adds engineering complexity: need subset config UI (+2 sprint)",
        "Tier-3 requires approval flow: adds ~1 day latency for full retrains",
        "PSI drift detection requires feature monitoring infra (already planned for Q3)",
    ],
    reviewers=["PM Lead", "ML Platform Lead", "SRE Lead", "Data Science Lead"],
)

# RFC anti-patterns
print("\n\nRFC ANTI-PATTERNS:")
antipatterns = [
    ("❌", "Writing the RFC after the decision is made (rubber stamp, not alignment)"),
    ("❌", "RFC with no alternatives section (looks like you didn't think it through)"),
    ("❌", "Never getting a real sign-off (decisions stay in people's heads)"),
    ("❌", "10-page RFC for a 1-week change (overhead > benefit)"),
    ("✅", "1-pager for small decisions; full RFC for cross-team or risky ones"),
    ("✅", "Include 'open questions' — shows intellectual honesty"),
    ("✅", "Share draft early; iterate before final review — avoid surprise rejections"),
]
for icon, ap in antipatterns:
    print(f"  {icon}  {ap}")
````

## 7. Pattern 3: Mentoring and Leveling Up Others

---

```
WHAT IT IS:
  Investing deliberately in the growth of other engineers so that your
  expertise scales beyond what you can build yourself.

STAFF MENTORING vs SENIOR MENTORING:
  Senior: helps when asked, reviews PRs, answers questions
  Staff:  proactively identifies growth gaps, creates systems that
          scale mentoring beyond 1:1 (standards, templates, workshops)

THE TRANSFER-OF-UNDERSTANDING PRINCIPLE:
  The goal is not to give someone the answer.
  The goal is to transfer the reasoning process.
  Wrong: "Here's the solution to your problem."
  Right: "What are the options you've considered? What's your current best guess?"

SLOW MOTION TRACE — Engineer leveled up:
  Month 1: Noticed engineer always asked me before making decisions.
  Month 2: Changed approach: instead of answering, asked "what would you do?"
  Month 3: Started doing weekly 30-min "design 1:1" — they presented,
            I asked questions, rarely gave answers.
  Month 6: Engineer started leading design reviews independently.
  Month 9: Engineer was promoted to senior.
  Month 12: Engineer was mentoring others.

SCALABLE MENTORING ARTIFACTS:
  Code review checklist     → encodes your standard; reviewers apply it
  RFC template              → anyone can write a well-structured RFC
  Architecture decision log → history is searchable; new hires learn from it
  "How to oncall" runbook   → your incident response expertise, written down
  Tech radar / principles   → documented opinionated guidance, applied by all
```

````python
# Mentoring effectiveness framework

class MentoringSession:
    """
    Tracks a structured mentoring session.
    Staff engineers use a consistent framework to maximize transfer.
    """

    def __init__(self, mentee, topic):
        self.mentee = mentee
        self.topic = topic
        self.questions_asked = []
        self.insights_from_mentee = []
        self.next_action = None

    def run_session(self):
        print(f"MENTORING SESSION: {self.mentee} — {self.topic}")
        print("-" * 50)

        # Phase 1: Understand where they are
        phase1 = [
            f"What's your current understanding of {self.topic}?",
            "What have you already tried?",
            "Where exactly are you stuck?",
        ]
        print("\nPhase 1 — Diagnose (listen, don't solve)")
        for q in phase1:
            print(f"  Mentor: '{q}'")
            self.questions_asked.append(q)

        # Phase 2: Expand their thinking
        phase2 = [
            "What are all the options you see? Even bad ones?",
            "If you had to pick one approach right now, which would it be?",
            "What would make you more confident?",
        ]
        print("\nPhase 2 — Expand (ask, don't answer)")
        for q in phase2:
            print(f"  Mentor: '{q}'")
            self.questions_asked.append(q)

        # Phase 3: Transfer one key insight (only if needed)
        print("\nPhase 3 — Transfer (one insight, not the answer)")
        print(f"  Mentor: 'The thing I've found useful here is: [share mechanism, not answer]'")
        print(f"  Mentor: 'Does that change how you're thinking about it?'")

        # Phase 4: Commit to next action
        print("\nPhase 4 — Commit (specific next action with date)")
        print(f"  Mentor: 'What will you do by Thursday?'")
        print(f"  Mentor: 'Send me a 3-bullet summary of your decision after you decide.'")


session = MentoringSession("Dave (L4 → L5 track)", "system design for the new ingestion pipeline")
session.run_session()

# Scalable mentoring: code review checklist as a mentoring artifact
print("\n\nSCALABLE ARTIFACT: Code Review Checklist (encodes expertise)")
code_review_checklist = [
    "[ ] Is this change idempotent? What happens if it runs twice?",
    "[ ] Are all error paths handled? What does the DLQ contain on failure?",
    "[ ] Does this add a new external dependency? Has the team discussed the SLA?",
    "[ ] Is this change observable? Are there metrics/logs to confirm it worked?",
    "[ ] Could this fail silently? How would we know 24 hours later?",
    "[ ] Is there a rollback plan if this causes a production issue?",
]
for item in code_review_checklist:
    print(f"  {item}")
````

## 8. Pattern 4: Technical Roadmap & Prioritization

---

```
WHAT IT IS:
  Translating business priorities into a sequenced technical investment
  plan that leadership can make resource decisions from.

THE STAFF ENGINEER'S ROLE:
  - Identify the right problems, not just solve the assigned ones
  - Frame technical work in terms of business risk and opportunity
  - Sequence work so each item enables the next (dependency chain)
  - Make the implicit explicit ("we've been ignoring observability debt")

PRIORITIZATION FRAMEWORKS:

  RICE score:
    Reach × Impact × Confidence / Effort
    Good for product features; less good for infrastructure

  Cost-of-delay:
    What happens if we don't do this for 6 more months?
    Good for reliability/security work

  Now/Next/Later:
    Now:    Must-do this quarter (blocking other work)
    Next:   High value, no current blocker
    Later:  Good idea, not yet justified

SLOW MOTION TRACE — Q3 roadmap conversation:
  Step 1: Engineering manager asks for "tech debt" list for Q3.
  Step 2: I don't send a list of tickets. I ask:
           - What is the business trying to do in Q3?
           - What would slow that down or make it fail?
  Step 3: Map each tech investment to a business risk it mitigates
           or an opportunity it enables.
  Step 4: Present as: "Here are 4 investments. Skipping #1 creates $X risk."
  Outcome: Leadership can now make informed trade-off decisions.
```

````python
# Technical roadmap prioritization tool

class RoadmapItem:
    def __init__(self, title, business_value, effort_weeks, risk_if_skipped, dependencies=None):
        self.title = title
        self.business_value = business_value  # 1-10
        self.effort_weeks = effort_weeks
        self.risk_if_skipped = risk_if_skipped  # text description
        self.dependencies = dependencies or []

    @property
    def priority_score(self):
        # Simple score: value per unit effort
        return round(self.business_value / max(self.effort_weeks, 1), 2)


def build_roadmap(items):
    """Sort by priority score, respecting dependency order."""
    # Simple topological sort by dependency
    completed = set()
    ordered = []

    remaining = list(items)
    iterations = 0
    while remaining and iterations < 100:
        for item in remaining[:]:
            if all(dep in completed for dep in item.dependencies):
                ordered.append(item)
                completed.add(item.title)
                remaining.remove(item)
        iterations += 1

    return ordered


# Example Q3 roadmap for ML platform
items = [
    RoadmapItem(
        "Observability: model drift detection",
        business_value=9, effort_weeks=4,
        risk_if_skipped="Silent model degradation → forecast errors → inventory mismatch",
        dependencies=[]
    ),
    RoadmapItem(
        "Feature store: online serving (Redis)",
        business_value=8, effort_weeks=6,
        risk_if_skipped="Real-time inference blocked → ML use cases limited to batch",
        dependencies=[]
    ),
    RoadmapItem(
        "A/B test framework",
        business_value=7, effort_weeks=3,
        risk_if_skipped="Cannot validate new models in production without full rollout risk",
        dependencies=["Feature store: online serving (Redis)"]
    ),
    RoadmapItem(
        "Model registry migration to MLflow",
        business_value=5, effort_weeks=8,
        risk_if_skipped="Continued spreadsheet tracking — brittle, not scalable",
        dependencies=[]
    ),
]

ordered = build_roadmap(items)

print("Q3 TECHNICAL ROADMAP — PRIORITIZED")
print("=" * 70)
print(f"{'#':<3} {'Item':<40} {'Value':<7} {'Effort':<8} {'Score'}")
print("-" * 70)
for i, item in enumerate(ordered, 1):
    dep_str = f" [needs: {item.dependencies[0][:20]}...]" if item.dependencies else ""
    print(f"{i:<3} {item.title:<40} {item.business_value:<7} {item.effort_weeks}w{'':<5} {item.priority_score}")

print("\nRISK IF SKIPPED:")
for item in ordered:
    print(f"  [{item.title[:35]}] {item.risk_if_skipped}")
````

## 9. Pattern 5: Cross-Team Alignment

---

```
WHAT IT IS:
  Getting multiple teams with competing priorities and different
  managers to agree on a shared technical direction.

WHY IT'S HARD:
  - Each team has its own roadmap and resource constraints
  - Technical arguments don't land without relationship capital
  - "Aligned" in a meeting ≠ "committed" in execution

ALIGNMENT PLAYBOOK:
  1. Map the stakeholders: who has a stake, who has a veto, who is neutral?
  2. Understand each team's priority: what are they trying to do this quarter?
  3. Find the shared pain: "we all suffer from the same problem"
  4. Propose a solution that makes each stakeholder's life easier
  5. Get written commitment (not just verbal agreement)
  6. Make the commitment visible to leadership (executive sponsorship)

SLOW MOTION TRACE — Lakehouse migration alignment:
  Week 1: Three teams (data platform, ML, analytics) all hit S3 partition limits.
           Each team planning their own local fix.
  Week 2: I run 3 separate 1:1s. Identify: all have same root cause.
  Week 3: Write a 1-pager: "shared problem, shared solution, shared cost"
  Week 4: Facilitate a 45-min working session with all three leads.
           Let them agree on the solution — I facilitate, not drive.
  Week 5: Document decision with owner per team. Share with eng managers.
  Week 6: Executive sponsor named → decision can't be reversed locally.

KEY INSIGHT:
  Alignment dies when only one team delivers. Always make sure
  each team has a clear, bounded commitment — not "help when you can."
```

````python
# Cross-team alignment tracker

class AlignmentTracker:
    """
    Tracks stakeholder positions and commitments for a cross-team initiative.
    Staff engineers use structured tracking to prevent drift post-alignment.
    """

    POSITIONS = ["champion", "supportive", "neutral", "skeptical", "blocker"]

    def __init__(self, initiative):
        self.initiative = initiative
        self.stakeholders = {}
        self.commitments = {}
        self.open_concerns = {}

    def add_stakeholder(self, name, team, position, concern=None):
        self.stakeholders[name] = {"team": team, "position": position}
        if concern:
            self.open_concerns[name] = concern

    def record_commitment(self, name, commitment, deadline):
        self.commitments[name] = {"commitment": commitment, "deadline": deadline}

    def report(self):
        print(f"ALIGNMENT TRACKER: {self.initiative}")
        print("=" * 60)

        print("\nSTAKEHOLDERS:")
        for name, info in self.stakeholders.items():
            concern = self.open_concerns.get(name, "none")
            print(f"  {name:<20} [{info['team']:<15}] {info['position']:<12} concern: {concern}")

        blockers = [n for n, i in self.stakeholders.items() if i["position"] == "blocker"]
        skeptics = [n for n, i in self.stakeholders.items() if i["position"] == "skeptical"]
        if blockers:
            print(f"\n⚠️  BLOCKERS to resolve: {blockers}")
        if skeptics:
            print(f"⚠️  SKEPTICS to address: {skeptics}")

        print("\nCOMMITMENTS:")
        if self.commitments:
            for name, c in self.commitments.items():
                print(f"  {name:<20} by {c['deadline']}: {c['commitment']}")
        else:
            print("  ⚠️  No written commitments yet — high risk of drift!")


# Lakehouse migration alignment
tracker = AlignmentTracker("Shared Delta Lake migration — Q4")

tracker.add_stakeholder("Alice",  "Data Platform", "champion")
tracker.add_stakeholder("Bob",    "ML Platform",   "supportive")
tracker.add_stakeholder("Carol",  "Analytics",     "skeptical",
                         concern="Worried about Athena query latency regression")
tracker.add_stakeholder("Dave",   "Data Eng Mgr",  "neutral")
tracker.add_stakeholder("Eve",    "Security",       "blocker",
                         concern="Delta Lake encryption at rest not yet validated")

tracker.record_commitment("Alice", "Deliver Delta Lake POC for data platform tables", "2024-10-15")
tracker.record_commitment("Bob",   "Validate ML feature store on Delta Lake",         "2024-10-22")

tracker.report()

print("\nNEXT ACTIONS:")
print("  1. Schedule 1:1 with Eve to understand encryption concern specifics")
print("  2. Run Athena latency benchmark on Delta Lake vs Parquet → address Carol's concern")
print("  3. Get Dave to name exec sponsor before week 3")
````

## 11. Interview Cheat Sheet

### When to use each Staff pattern:

| Signal in the question | Pattern to demonstrate |
|------------------------|------------------------|
| "No formal authority" | Influence without authority — show pull, not push |
| "Cross-team" | Alignment tracker + RFC + executive sponsorship |
| "Technical direction" | Roadmap prioritization tied to business risk |
| "Grew / developed" | Structured 1:1 + transfer of reasoning, not answers |
| "System / standard" | Scalable artifact — checklist, template, RFC process |
| "Difficult stakeholder" | Coalition building + written commitment |
| "Most impactful" | Multiplied team output — what others built because of you |

### Staff-level framing — memorize these:

```python
# Instead of:  "I built X"
# Say:         "I enabled 3 teams to build X by creating the shared standard"

# Instead of:  "I fixed the bug"
# Say:         "I fixed the bug and wrote the standard that prevents this class of bug"

# Instead of:  "I designed the system"
# Say:         "I wrote the RFC, ran the review, got alignment across 4 teams,
#               and tracked implementation to completion"

# Instead of:  "I helped the junior engineer"
# Say:         "I ran structured 1:1s using the Socratic method — asked questions
#               instead of giving answers. They were promoted within 18 months."
```

### Common templates:

```
# INFLUENCE TEMPLATE
"I couldn't mandate the change, so I:
  1. Built a working prototype for my own team first (proved value)
  2. Shared the demo and asked for their feedback (co-authorship)
  3. Incorporated their input (they owned part of it)
  4. [Result: they adopted it / shared it / it became the standard]"

# RFC TEMPLATE
"For cross-team decisions, I always write an RFC: context, decision,
  alternatives, trade-offs, open questions. I get written sign-off.
  This prevents the 'but we decided X' conflict 6 months later."

# MENTORING TEMPLATE
"My goal in mentoring is to transfer the reasoning process, not the answer.
  I ask: 'What are your options? Which would you pick? Why?'
  Then I share one insight — not the solution.
  The engineer who can reason through it once can do it again alone."

# ROADMAP TEMPLATE
"When asked for a tech roadmap, I start with business priorities.
  For each investment, I frame it as: 'here's the business risk
  we're mitigating or opportunity we're enabling, here's the cost,
  here's what happens if we skip it for 6 months.'"
```

### Gotchas:

```
❌  Saying "we" for everything — interviewers want your specific contribution
❌  Only citing technical work — Staff requires organizational work too
❌  Vague outcomes — "improved things" vs "2 engineers promoted, 3 teams adopted"
❌  Skipping the "why" — Staff interviewers want to see your judgment, not your deeds
✅  "I" for your specific contributions; "we" for the team's
✅  Always include: what you identified, what you built, what others did because of it
✅  Include one failure or trade-off — shows intellectual honesty
✅  Quantify organizational impact: teams aligned, engineers leveled up, months saved
```
