````python
"""
They want you to calculate the **average score for each student**.

The tricky part: the input is organized by **subject**, not by student.

## What the problem wants

You are given:

```text
N = number of students
X = number of subjects
```

Then you are given `X` lines.

Each line contains the scores for all `N` students in one subject.

So if there are:

```text
5 students
3 subjects
```

The input gives you 3 rows of marks.

## Sample input

```text
5 3
89 90 78 93 80
90 91 85 88 86
91 92 83 89 90.5
```

Meaning:

```text
N = 5 students
X = 3 subjects
```

Subject 1 scores:

```text
student 1 = 89
student 2 = 90
student 3 = 78
student 4 = 93
student 5 = 80
```

Subject 2 scores:

```text
student 1 = 90
student 2 = 91
student 3 = 85
student 4 = 88
student 5 = 86
```

Subject 3 scores:

```text
student 1 = 91
student 2 = 92
student 3 = 83
student 4 = 89
student 5 = 90.5
```

Now you need to group by student.

## What `zip()` helps with

You start with subject rows:

```text
Subject 1: 89   90   78   93   80
Subject 2: 90   91   85   88   86
Subject 3: 91   92   83   89   90.5
```

But you need student columns:

```text
Student 1: 89, 90, 91
Student 2: 90, 91, 92
Student 3: 78, 85, 83
Student 4: 93, 88, 89
Student 5: 80, 86, 90.5
```

That is exactly what `zip()` does. It flips rows into columns.

## Output needed

Print each student’s average on a separate line.

For the sample:

```text
Student 1 average = (89 + 90 + 91) / 3 = 90.0
Student 2 average = (90 + 91 + 92) / 3 = 91.0
Student 3 average = (78 + 85 + 83) / 3 = 82.0
Student 4 average = (93 + 88 + 89) / 3 = 90.0
Student 5 average = (80 + 86 + 90.5) / 3 = 85.5
```

Expected output:

```text
90.0
91.0
82.0
90.0
85.5
```

## Example 2

Input:

```text
3 2
10 20 30
40 50 60
```

Meaning:

```text
3 students
2 subjects
```

Student scores:

```text
Student 1: 10, 40 -> average 25.0
Student 2: 20, 50 -> average 35.0
Student 3: 30, 60 -> average 45.0
```

Expected output:

```text
25.0
35.0
45.0
```

## Key idea

You read the subject rows into a list like:

```python
marks = [
    [89, 90, 78, 93, 80],
    [90, 91, 85, 88, 86],
    [91, 92, 83, 89, 90.5]
]
```

Then `zip(*marks)` gives student groups:

```python
(89, 90, 91)
(90, 91, 92)
(78, 85, 83)
(93, 88, 89)
(80, 86, 90.5)
```

Then for each group, calculate:

```python
sum(student_scores) / X
```

Print with one decimal place.

"""
# N = number of students
# X = number of subjects
N, X = map(int , input().strip().split())
lsts = []
for _ in range(X):
    lsts.append(list( map(float , input().strip().split())))
m = zip(*lsts)    
for row in m:
    print (sum(row)/X)

    
````
