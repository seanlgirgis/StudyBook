````python
"""
They want you to try **every possible choice of one number from each list**, 
then find the best score.

No full solution yet.

## What the problem wants

You are given:

```text
K = number of lists
M = modulo number
```

Then you are given `K` lists.

From each list, pick **exactly one number**.

For every possible pick, calculate:

```text
sum of squares of selected numbers, then % M
```

In math:

```text
(x1² + x2² + x3² + ... + xK²) % M
```

You need to print the **maximum possible value**.

## Input needed

First line:

```text
K M
```

Then `K` lines follow.

Each list line starts with the list length, then the actual values:

```text
N value1 value2 value3 ...
```

## Sample input

```text
3 1000
2 5 4
3 7 8 9
5 5 7 8 9 10
```

Meaning:

```text
K = 3
M = 1000
```

There are 3 lists:

```text
List 1: [5, 4]
List 2: [7, 8, 9]
List 3: [5, 7, 8, 9, 10]
```

The first number on each list line is just the count, so you ignore it after reading.

## Output needed

One integer:

```text
maximum possible S value
```

## Sample output

```text
206
```

Why?

Pick:

```text
5 from list 1
9 from list 2
10 from list 3
```

Then calculate:

```text
5² + 9² + 10²
= 25 + 81 + 100
= 206
```

Then:

```text
206 % 1000 = 206
```

So the answer is:

```text
206
```

## Another example

Input:

```text
2 10
2 3 4
2 5 6
```

Possible choices:

```text
3 and 5 -> (9 + 25) % 10 = 34 % 10 = 4
3 and 6 -> (9 + 36) % 10 = 45 % 10 = 5
4 and 5 -> (16 + 25) % 10 = 41 % 10 = 1
4 and 6 -> (16 + 36) % 10 = 52 % 10 = 2
```

Maximum is:

```text
5
```

Expected output:

```text
5
```

## Key Python idea

This is a Cartesian product problem.

You need all possible combinations like:

```text
one from list 1
one from list 2
one from list 3
```

The useful tool is:

```python
itertools.product
```

It creates combinations across multiple lists.

Think:

```text
product([5, 4], [7, 8, 9], [5, 7, 8, 9, 10])
```

creates choices like:

```text
(5, 7, 5)
(5, 7, 7)
(5, 7, 8)
...
(4, 9, 10)
```

For each tuple, calculate the score and keep the maximum.
"""
from itertools import product
k, m = map(int, input().split())
lists = []
best = -float('inf')
for _ in range(k):
    _, *lst = map(int, input().split())
    lists.append(lst)
prds = list(product(*lists))
for prd in prds:
    s = 0
    for i in range(k):
        s += prd[i] ** 2
    best = max(best, s % m)
print(best)
    

````
