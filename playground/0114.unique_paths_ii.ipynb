## Unique Paths II

Same as Unique Paths but the grid has **obstacles** (`1`). You cannot pass through an obstacle. Start top-left, end bottom-right.

### Example

```
obstacleGrid = [[0,0,0],
                [0,1,0],
                [0,0,0]]  → 2
```
- `[[0,0],[0,1]]` → `0` (blocked)
- `[[0,1],[0,0]]` → `1`

### Hint
Same 2D DP as unique_paths — `dp[i][j] = dp[i-1][j] + dp[i][j-1]` — but set `dp[i][j] = 0` if obstacle.

### Input/Output
- **[input]** `array.array.integer` `obstacleGrid`
- **[output]** `integer`

````python
%run _timer.py
````

````python
def solution(obstacleGrid):
    if not obstacleGrid or obstacleGrid[0][0] == 1:
        return 0
    
    rows = len(obstacleGrid)
    cols = len(obstacleGrid[0])
    
    # dp table 2D to calculate path
    dp = [[0] * cols for _ in range(rows)]
    
    # Starting point
    dp[0][0] = 1
    
    for r in range(rows):
        for c in range(cols):

            if obstacleGrid[r][c] == 1:  #obstacle
                dp[r][c] = 0
                continue
            

            if r > 0:
                dp[r][c] += dp[r-1][c]
            

            if c > 0:
                dp[r][c] += dp[r][c-1]
                
    return dp[rows-1][cols-1]


assert solution([[0,0,0],[0,1,0],[0,0,0]]) == 2
assert solution([[0,0],[0,1]])             == 0
assert solution([[0,1],[0,0]])             == 1
assert solution([[0]])                     == 1
assert solution([[1]])                     == 0
print('All Pass!')
````
