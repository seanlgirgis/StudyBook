## Hand of Straights (LC 846)

Can you split `hand` into groups of `groupSize` consecutive cards?

**Pattern:** Greedy + Counter — always start each group from the smallest remaining card.
- If total cards not divisible by `groupSize` → impossible
- Sort unique card values; for each starting card `c`, try to consume `groupSize` consecutive values
- If any `c+i` has fewer cards than needed → impossible

```python
%run _timer.py
```

```python
from collections import Counter

def solution(hand, groupSize):
    if len(hand) % groupSize != 0:
        return False

    count = Counter(hand)

    for card in sorted(count):          # always start from the smallest available card
        n = count[card]                 # how many groups start here
        if n > 0:
            for i in range(groupSize):  # need n copies of card, card+1, ..., card+groupSize-1
                if count[card + i] < n:
                    return False
                count[card + i] -= n

    return True


assert solution([1,2,3,6,2,3,4,7,8], 3) == True
assert solution([1,2,3,4,5],          4) == False
assert solution([1,2,3],              3) == True
assert solution([1,2,3,4],            3) == False   # 4 not divisible by 3
assert solution([1,1,2,2,3,3],        2) == False   # groups of 2 consec: need [1,2],[1,2],[3,?]
assert solution([1,1,2,2,3,3],        3) == True    # [1,2,3] x2
print('All Pass!')
```
