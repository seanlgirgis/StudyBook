## Token Ring with Node Cooldown (Custom Simulation)

`n` nodes in a ring. Each node has `work[i]` (time units contributed) and `cooldown[i]` (unavailable after use). Pass a token for `t` total time units.

- Scan nodes in circular deque order. Skip if not ready.
- Count only full uses. Partial final use → return count without counting.
- No node available at current time → return `-1`.

**Pattern:** Same as phone battery but using a deque rotate instead of modular pointer.

```python
from collections import deque

def solution(t, work, cooldown):
    n = len(work)
    ready_at = [0] * n
    order    = deque(range(n))   # circular order of node indices
    clock    = 0
    used     = 0

    while clock < t:
        found = False
        for _ in range(n):           # one full circle scan
            i = order[0]
            order.rotate(-1)         # advance pointer
            if ready_at[i] <= clock:
                found = True
                rem = t - clock
                if work[i] <= rem:           # full use
                    clock       += work[i]
                    used        += 1
                    ready_at[i]  = clock + cooldown[i]
                else:                        # partial — done
                    return used
                break
        if not found:
            return -1

    return used


assert solution(8,  [2,5],  [10,4]) == -1
assert solution(10, [7,5],  [12,4]) == 1
assert solution(6,  [3,3],  [1,1])  == 2
assert solution(1,  [5],    [3])    == 0
assert solution(0,  [5,5],  [0,0])  == 0
print('All Pass!')
```
