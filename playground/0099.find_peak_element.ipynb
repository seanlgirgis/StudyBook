## Find Peak Element

A peak element is one that is strictly greater than its neighbors. Given an integer array `nums`, find a peak element and return its **index**. If there are multiple peaks, return any.

You may assume `nums[-1]` and `nums[n]` are `-∞` (so the first/last element can be a peak).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- `nums = [1,2,3,1]` → `2` (nums[2]=3 is a peak)
- `nums = [1,2,1,3,5,6,4]` → `1` or `5` (either valid)

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums`
- **[output]** `integer`

```python
def solution(nums):
    # The global maximum is always a peak
    max_val = max(nums)
    return nums.index(max_val)

        
        
    


# [1,2,3,1]       → index 2  (value 3, only peak)
# [1,2,1,3,5,6,4] → index 1 or 5  (value 2 or 6, both valid peaks)
# [1]             → index 0  (single element, always a peak)
# [3,1]           → index 0  (value 3 > right neighbor 1)
# [1,3]           → index 1  (value 3 > left neighbor 1)

def check(result, nums):
    i = result
    left  = nums[i-1] if i > 0           else float('-inf')
    right = nums[i+1] if i < len(nums)-1 else float('-inf')
    return nums[i] > left and nums[i] > right

assert check(solution([1,2,3,1]),        [1,2,3,1])
assert check(solution([1,2,1,3,5,6,4]), [1,2,1,3,5,6,4])
assert check(solution([1]),             [1])
assert check(solution([3,1]),           [3,1])
assert check(solution([1,3]),           [1,3])
print('All Pass!')
```

```python
def solution(nums):
    # The global maximum is always a peak
    for i in range(len(nums)):
        left  = nums[i-1] if i > 0           else float('-inf')
        right = nums[i+1] if i < len(nums)-1 else float('-inf')
        if nums[i] > left and nums[i] > right:
            return i

        
        
    


# [1,2,3,1]       → index 2  (value 3, only peak)
# [1,2,1,3,5,6,4] → index 1 or 5  (value 2 or 6, both valid peaks)
# [1]             → index 0  (single element, always a peak)
# [3,1]           → index 0  (value 3 > right neighbor 1)
# [1,3]           → index 1  (value 3 > left neighbor 1)

def check(result, nums):
    i = result
    left  = nums[i-1] if i > 0           else float('-inf')
    right = nums[i+1] if i < len(nums)-1 else float('-inf')
    return nums[i] > left and nums[i] > right

assert check(solution([1,2,3,1]),        [1,2,3,1])
assert check(solution([1,2,1,3,5,6,4]), [1,2,1,3,5,6,4])
assert check(solution([1]),             [1])
assert check(solution([3,1]),           [3,1])
assert check(solution([1,3]),           [1,3])
print('All Pass!')
```
