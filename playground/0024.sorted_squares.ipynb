## Sorted Squares

Given an array of integers `nums` sorted in **non-decreasing order**, return an array of the **squares of each number**, also sorted in non-decreasing order.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.

### Example

- For `nums = [-4,-1,0,3,10]`, the output should be `solution(nums) = [0,1,9,16,100]`.

  **Explanation:** Squares are `[16,1,0,9,100]` — sorted gives `[0,1,9,16,100]`.

- For `nums = [-7,-3,2,3,11]`, the output should be `solution(nums) = [4,9,9,49,121]`.

  **Explanation:** Squares are `[49,9,4,9,121]` — sorted gives `[4,9,9,49,121]`.

- For `nums = [-3,-2,-1]`, the output should be `solution(nums) = [1,4,9]`.

  **Explanation:** All negative — squares reverse the order.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array sorted in non-decreasing order.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^4`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[output]** `array.integer` — Squares sorted in non-decreasing order.

```python
import heapq
def solution(nums):
    ret = []
    heap = []
    for num in nums:
        heapq.heappush(heap, num **2)
    while heap:
        ret.append(heapq.heappop(heap))
    return ret


assert solution([-4,-1,0,3,10])  == [0,1,9,16,100]   # given example 1
assert solution([-7,-3,2,3,11])  == [4,9,9,49,121]   # given example 2 — duplicate square
assert solution([-3,-2,-1])      == [1,4,9]           # given example 3 — all negative
assert solution([0,1,2])         == [0,1,4]           # all non-negative
assert solution([1])             == [1]               # single element
assert solution([-1,0,1])        == [0,1,1]           # symmetric — two equal squares
assert solution([-5,-3,-2,-1])   == [1,4,9,25]        # all negative — reversed order
assert solution([0])             == [0]               # single zero
print('All Pass!')
```

```python
def solution(nums):
    n = len(nums)
    ret = [0] * n
    l, r = 0, n - 1
    for i in range(n - 1, -1, -1):   # fill from back — largest first
        if abs(nums[l]) > abs(nums[r]):
            ret[i] = nums[l] ** 2
            l += 1
        else:
            ret[i] = nums[r] ** 2
            r -= 1
    return ret
assert solution([-4,-1,0,3,10])  == [0,1,9,16,100]   # given example 1
assert solution([-7,-3,2,3,11])  == [4,9,9,49,121]   # given example 2 — duplicate square
assert solution([-3,-2,-1])      == [1,4,9]           # given example 3 — all negative
assert solution([0,1,2])         == [0,1,4]           # all non-negative
assert solution([1])             == [1]               # single element
assert solution([-1,0,1])        == [0,1,1]           # symmetric — two equal squares
assert solution([-5,-3,-2,-1])   == [1,4,9,25]        # all negative — reversed order
assert solution([0])             == [0]               # single zero
print('All Pass!')
```
