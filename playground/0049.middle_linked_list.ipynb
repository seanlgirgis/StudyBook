## Middle of the Linked List

Given the `head` of a singly linked list, return the **middle node**.

If there are **two middle nodes**, return the **second** one.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `head = [1,2,3,4,5]`, the output should be `solution(head) = [3,4,5]`.

  **Explanation:** The middle node has value `3`. Return the node and everything after it.

- For `head = [1,2,3,4,5,6]`, the output should be `solution(head) = [4,5,6]`.

  **Explanation:** Two middle nodes are `3` and `4` — return the second one.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `head`  
  Guaranteed constraints: `1 ≤ n ≤ 100`, `1 ≤ val ≤ 100`
- **[output]** `ListNode` — The middle node (return the node itself, not just its value).

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

def to_ll(lst):
    dummy = ListNode()
    cur = dummy
    for v in lst:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def solution(head):
    n = 0
    curr = head
    while curr:
        n += 1
        curr = curr.next
    hlf = n // 2
    curr = head
    for _ in range(hlf):
        curr = curr.next
    return curr


assert to_list(solution(to_ll([1,2,3,4,5])))       == [3,4,5]    # given example 1 — odd length
assert to_list(solution(to_ll([1,2,3,4,5,6])))     == [4,5,6]    # given example 2 — even, second middle
assert to_list(solution(to_ll([1])))                == [1]        # single node
assert to_list(solution(to_ll([1,2])))              == [2]        # two nodes — second middle
assert to_list(solution(to_ll([1,2,3])))            == [2,3]      # three nodes
assert to_list(solution(to_ll([1,2,3,4])))          == [3,4]      # four nodes — second middle
assert to_list(solution(to_ll([1,2,3,4,5,6,7])))   == [4,5,6,7]  # seven nodes
assert to_list(solution(to_ll([1,2,3,4,5,6,7,8]))) == [5,6,7,8]  # eight nodes
print('All Pass!')
```
