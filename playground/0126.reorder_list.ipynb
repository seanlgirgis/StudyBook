## Reorder List

Given list `L0→L1→…→Ln`, reorder to `L0→Ln→L1→Ln-1→L2→Ln-2→…` in-place.

- `[1,2,3,4]` → `[1,4,2,3]`
- `[1,2,3,4,5]` → `[1,5,2,4,3]`

**Steps:** 1) Find middle (slow/fast pointers). 2) Reverse second half. 3) Merge two halves alternately.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def make_list(arr):
    dummy = ListNode()
    cur = dummy
    for v in arr: cur.next = ListNode(v); cur = cur.next
    return dummy.next

def to_list(head):
    out = []
    while head: out.append(head.val); head = head.next
    return out


def solution(head):
    if not head or not head.next:
        return head

    # Step 1: find middle
    slow, fast = head, head
    while fast.next and fast.next.next:
        slow = slow.next
        fast = fast.next.next

    # Step 2: reverse second half
    prev, curr = None, slow.next
    slow.next = None          # cut first half
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    second = prev             # head of reversed second half

    # Step 3: merge alternately
    first = head
    while second:
        tmp1, tmp2 = first.next, second.next
        first.next  = second
        second.next = tmp1
        first  = tmp1
        second = tmp2

    return head


assert to_list(solution(make_list([1,2,3,4])))   == [1,4,2,3]
assert to_list(solution(make_list([1,2,3,4,5]))) == [1,5,2,4,3]
assert to_list(solution(make_list([1])))          == [1]
assert to_list(solution(make_list([1,2])))        == [1,2]
print('All Pass!')
```
