## Next Greater Element — Circular List

Same as single list but the array **wraps around** — an element at the end can look forward into the beginning.

### Example

- `[1,2,1]` → `[2,-1,2]`  (last 1 wraps and finds 2 at index 0)
- `[3,1,2,4]` → `[4,2,4,-1]`  (3 is the max — no greater even after wrap)
- `[1,2,3,4,3]` → `[2,3,4,-1,4]`

### Trick

Iterate `2*n` times using `i % n`. Only push indices for the first `n` iterations.

```python
%run _timer.py
```

```python
def solution(nums):
    n = len(nums)
    ans = [-1] * n
    stack = []
    for i in range(2 * n):
        val = nums[i % n]
        while stack and val > nums[stack[-1]]:
            ans[stack.pop()] = val
        if i < n:
            stack.append(i)
    return ans


assert solution([1,2,1])     == [2,-1,2]
assert solution([3,1,2,4])   == [4,2,4,-1]
assert solution([1,2,3,4,3]) == [2,3,4,-1,4]
assert solution([1])         == [-1]
print('All Pass!')
```
