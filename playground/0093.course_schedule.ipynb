## Course Schedule

There are `numCourses` courses labeled `0` to `numCourses-1`. You are given a list of `prerequisites` where `prerequisites[i] = [a, b]` means you must take course `b` before course `a`.

Return `True` if you can finish all courses, `False` if there is a cycle (impossible).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n + e)** will fit within the execution time limit.

### Example

- `numCourses=2`, `prerequisites=[[1,0]]` → `True` (take 0 then 1)
- `numCourses=2`, `prerequisites=[[1,0],[0,1]]` → `False` (cycle: 0→1→0)

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `integer` `numCourses`, `array.array.integer` `prerequisites`
- **[output]** `boolean`

```python
from collections import defaultdict

def solution(numCourses, prerequisites):
    adj = defaultdict(list)
    for course, pre in prerequisites:
        adj[pre].append(course)

    state = [0] * numCourses   # 0=unvisited, 1=visiting, 2=done

    def has_cycle(i):
        if state[i] == 1: return True    # back edge — cycle
        if state[i] == 2: return False   # already cleared
        state[i] = 1
        for nei in adj[i]:
            if has_cycle(nei):
                return True
        state[i] = 2
        return False

    for i in range(numCourses):
        if state[i] == 0:
            if has_cycle(i):
                return False
    return True


assert solution(2, [[1,0]])                      == True
assert solution(2, [[1,0],[0,1]])                == False
assert solution(1, [])                           == True
assert solution(3, [[1,0],[2,1]])                == True
assert solution(3, [[1,0],[2,1],[0,2]])          == False
assert solution(4, [[1,0],[2,0],[3,1],[3,2]])    == True
print('All Pass!')
```
