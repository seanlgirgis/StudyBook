## Trapping Rain Water

Given `height[]`, compute how much water is trapped between bars after rain.

**Pattern:** Two pointers from both ends.  
- Water at position `i` = `min(max_left, max_right) - height[i]`  
- Move the pointer on the **shorter side** — that side is the bottleneck  
- The taller side is guaranteed ≥ the min, so it's safe to compute water now

No extra space needed — O(1) beyond the two running maxes.

```python
"""
id: lc_0042
title: Trapping Rain Water
source: leetcode
difficulty: hard
primary: two pointers
tags: [array, two-pointers, dynamic-programming, stack]
leetcode_url: https://leetcode.com/problems/trapping-rain-water/
status: draft
last_updated: 2026-04-23
notes:
- key idea: use two pointers to track max_left and max_right; move the pointer with the smaller max.
- time: O(n)
- space: O(1)
"""

# ============================================================================
# File: 042_lc_0042_trapping_rain_water_empty.py
# LeetCode 42: Trapping Rain Water
# Difficulty: Hard
#
# PROBLEM STATEMENT:
# Given n non-negative integers representing an elevation map where the width 
# of each bar is 1, compute how much water it can trap after raining.
#
# RULES:
# - Water at index i is min(max_left, max_right) - height[i].
# - Use O(1) space by moving pointers from the bottleneck side.
#
# EXAMPLES:
# Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
# Output: 6
#
# Input: height = [4,2,0,3,2,5]
# Output: 9
# ============================================================================

from typing import List

def trap(height: List[int]) -> int:
    n = len(height)
    l , r = 0, n-1
    max_l, max_r = 0, 0 
    water = 0 
    while l< r :
        if height[l] <= height[r]:                       # process left height
            if height[l] > max_l:
                max_l  = height[l]
            else:
                water += max_l -  height[l]
            l += 1
        else:                                          # process right height
            if height[r] > max_r:
                max_r = height[r]
            else:
                water += max_r - height[r]
            r -= 1
    return water
    

print(trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]))  #6
print(trap([4, 2, 0, 3, 2, 5])) #9

def test():
    assert trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]) == 6  # given example 1
    assert trap([4, 2, 0, 3, 2, 5]) == 9                   # given example 2
    assert trap([1, 0, 1]) == 1                            # simple pit
    assert trap([3, 0, 2, 0, 4]) == 7                      # complex valley
    assert trap([1, 2, 3, 4, 5]) == 0                      # strictly increasing
    assert trap([5, 4, 3, 2, 1]) == 0                      # strictly decreasing
    assert trap([5, 5, 5]) == 0                            # flat surface
    assert trap([2, 0, 2]) == 2                            # symmetric hole
    assert trap([]) == 0                                   # empty list
    assert trap([5, 2, 5]) == 3                            # single deep hole
    print('All Pass!')

test()
```

```python
def solution(height):
    l, r  = 0, len(height) - 1
    max_l = max_r = 0
    water = 0

    while l < r:
        if height[l] <= height[r]:          # left side is the bottleneck
            if height[l] >= max_l:
                max_l = height[l]           # new left wall — no water here
            else:
                water += max_l - height[l]  # trapped between max_l and current bar
            l += 1
        else:                               # right side is the bottleneck
            if height[r] >= max_r:
                max_r = height[r]           # new right wall — no water here
            else:
                water += max_r - height[r]  # trapped between max_r and current bar
            r -= 1

    return water


assert solution([0,1,0,2,1,0,1,3,2,1,2,1]) == 6
assert solution([4,2,0,3,2,5])             == 9
assert solution([1,0,1])                   == 1
assert solution([3,0,2,0,4])               == 7
assert solution([1,1])                     == 0
assert solution([3])                       == 0
print('All Pass!')
```
