## Circular Workers with Rest Time (Custom Simulation)

Workers in a ring. Worker `i` handles `capacity[i]` tasks then rests `rest[i]` minutes. Complete `t` tasks total.

**Same pattern as 0139–0141** — the core circular simulation template. Mastering one means mastering all four.

```python
from collections import deque

def solution(t, capacity, rest):
    n        = len(capacity)
    ready_at = [0] * n
    order    = deque(range(n))
    clock    = 0
    used     = 0

    while clock < t:
        found = False
        for _ in range(n):
            i = order[0]
            order.rotate(-1)
            if ready_at[i] <= clock:
                found = True
                rem = t - clock
                if capacity[i] <= rem:
                    clock      += capacity[i]
                    used       += 1
                    ready_at[i] = clock + rest[i]
                else:
                    return used
                break
        if not found:
            return -1

    return used


assert solution(5,  [2,2,100], [100,1,100]) == 2
assert solution(12, [2,10],   [100,1])     == 2
assert solution(13, [2,10],   [100,1])     == -1
assert solution(20, [3,3],    [1,1])       == 6
print('All Pass!')
```
