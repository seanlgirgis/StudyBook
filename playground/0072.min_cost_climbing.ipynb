## Min Cost Climbing Stairs

You are given an integer array `cost` where `cost[i]` is the cost of the `i`th step. Once you pay the cost, you can climb **1 or 2 steps**.

You can start from index `0` or index `1`. Return the **minimum cost** to reach the top of the floor (one step beyond the last index).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `cost = [10,15,20]`, the output should be `15`.

  **Explanation:** Start at index 1, pay 15, jump 2 steps to the top.

- For `cost = [1,100,1,1,1,100,1,1,100,1]`, the output should be `6`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `cost`  
  Guaranteed constraints: `2 ≤ n ≤ 1000`, `0 ≤ cost[i] ≤ 999`
- **[output]** `integer`

```python
def solution(cost):
    #guranteed least length of cost is 2
    prevprev = cost[0]
    prev = cost[1]
    n = len(cost)
    for i in range(2,n):
        prevprev, prev = prev, min(prev, prevprev) +cost[i]
    return min(prev, prevprev)

assert solution([10,15,20])                       ==  15  # given example 1
assert solution([1,100,1,1,1,100,1,1,100,1])      ==   6  # given example 2
assert solution([0,0])                            ==   0  # all zeros
assert solution([1,2])                            ==   1  # pick cheaper start
assert solution([10,15])                          ==  10  # pick cheaper start
assert solution([1,1,1,1])                        ==   2  # every other step
assert solution([5,5,5,5,5])                      ==  10  # uniform cost
assert solution([0,1,0,1,0])                      ==   0  # skip all cost
print('All Pass!')
```
