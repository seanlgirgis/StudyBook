## Decode String

Given an encoded string, return its decoded version.

The encoding rule is: `k[encoded_string]` means `encoded_string` is repeated exactly `k` times.

You may assume the input is always valid — no extra spaces, brackets are well-formed, and `k` is a positive integer.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "3[a]2[bc]"`, the output should be `"aaabcbc"`.
- For `s = "3[a2[c]]"`, the output should be `"accaccacc"` (nested).
- For `s = "2[abc]3[cd]ef"`, the output should be `"abcabccdcdcdef"`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `s`
- **[output]** `string`

```python
def process_stack_sofar(stack):
    alpha = ""
    num = ""
    while stack and stack[-1].isalpha():
        alpha += stack.pop()
    alpha = alpha[::-1]
    stack.pop()                          # guaranteed '[' after all alphas
    while stack and stack[-1].isdigit():
        num += stack.pop()
    s = ""
    for _ in range(int(num[::-1])):
        s += alpha
    stack.append(s[::-1])


def solution(s):
    stack = []
    for ch in s:
        if ch == ']':
            process_stack_sofar(stack)
        else:
            stack.append(ch)
    out = []
    while stack:
        out.append(stack.pop()[::-1])
    return ''.join(out[::-1])


assert solution("3[a]2[bc]")     == "aaabcbc"
assert solution("3[a2[c]]")      == "accaccacc"
assert solution("2[abc]3[cd]ef") == "abcabccdcdcdef"
assert solution("abc")           == "abc"
assert solution("1[a]")          == "a"
assert solution("2[2[a]]")       == "aaaa"
assert solution("10[a]")         == "aaaaaaaaaa"
assert solution("2[a3[b]]")      == "abbbabbb"
print('All Pass!')
```
