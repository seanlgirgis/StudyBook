## Decode String

Given an encoded string, return its decoded version.

The encoding rule is: `k[encoded_string]` means `encoded_string` is repeated exactly `k` times.

You may assume the input is always valid — no extra spaces, brackets are well-formed, and `k` is a positive integer.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "3[a]2[bc]"`, the output should be `"aaabcbc"`.
- For `s = "3[a2[c]]"`, the output should be `"accaccacc"` (nested).
- For `s = "2[abc]3[cd]ef"`, the output should be `"abcabccdcdcdef"`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `s`
- **[output]** `string`

```python
"""
id: lc_0394
title: Decode String
source: leetcode
difficulty: medium
primary: stack
tags: [string, stack, recursion]
leetcode_url: https://leetcode.com/problems/decode-string/
status: draft
last_updated: 2026-04-23
notes:
- key idea: use stack to track multipliers and current strings
- time: O(n) where n is the length of the decoded string
- space: O(n)
"""

# ============================================================================
# File: 394_lc_0394_decode_string_empty.py
# LeetCode 394: Decode String
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Given an encoded string, return its decoded version. 
# The encoding rule is: k[encoded_string] means the encoded_string is 
# repeated exactly k times.
#
# RULES:
# - k is guaranteed to be a positive integer.
# - Input is always valid (no extra spaces, well-formed brackets).
#
# EXAMPLES:
# s = "3[a]2[bc]" -> "aaabcbc"
# s = "3[a2[c]]" -> "accaccacc"
# s = "2[abc]3[cd]ef" -> "abcabccdcdcdef"
# ============================================================================

from typing import List

def decodeString(s: str) -> str:

    stack = []
    for c in s:
        if c != "]":
            stack.append(c)
        else:
            word = ""
            words = ""
            n = 0
            nord = 0
            while stack and stack[-1] != "[":
                word += stack.pop()
            stack.pop()
            while stack and stack[-1].isdigit():
                n += int(stack.pop()) * (10 ** nord)
                nord+= 1
            words = word * n
            stack.append(words[::-1])
    out = ""            
    while stack:
        out = stack.pop() + out
        
            
    return out
            
            


print(decodeString("10[a]"))
print(decodeString("3[a2[c]]"))
print(decodeString("3[a]2[bc]"))
def test():
    assert decodeString("3[a]2[bc]") == "aaabcbc"  # basic sequence
    assert decodeString("3[a2[c]]") == "accaccacc"  # nested brackets
    assert decodeString("2[abc]3[cd]ef") == "abcabccdcdcdef"  # mixed content
    assert decodeString("abc3[cd]xyz") == "abccdcdcdxyz"  # outer text
    assert decodeString("10[a]") == "aaaaaaaaaa"  # multi-digit k
    assert decodeString("2[2[b]]") == "bbbb"  # nested repeats
    assert decodeString("z") == "z"  # no brackets
    assert decodeString("2[a]3[b]4[c]") == "aabbbcccc"  # multiple blocks
    assert decodeString("3[a]2[b4[c]]") == "aaabccccbcccc"  # complex nesting
    
    print('All Pass!')

test()
```

```python
# Canonical tuple-stack approach (clean interview version)
from typing import List

def decodeString_tuple_stack(s: str) -> str:
    stack = []  # (previous_string, repeat_count)
    curr_num = 0
    curr_str = ''

    for ch in s:
        if ch.isdigit():
            curr_num = curr_num * 10 + int(ch)
        elif ch == '[':
            stack.append((curr_str, curr_num))
            curr_str = ''
            curr_num = 0
            print(stack)
        elif ch == ']':
            prev_str, repeat = stack.pop()
            curr_str = prev_str + curr_str * repeat
        else:
            curr_str += ch

    return curr_str


print(decodeString_tuple_stack('3[a2[c]]'))
#print(decodeString_tuple_stack('10[a]'))
def test_tuple_stack():
    assert decodeString_tuple_stack('3[a]2[bc]') == 'aaabcbc'
    assert decodeString_tuple_stack('3[a2[c]]') == 'accaccacc'
    assert decodeString_tuple_stack('2[abc]3[cd]ef') == 'abcabccdcdcdef'
    assert decodeString_tuple_stack('abc3[cd]xyz') == 'abccdcdcdxyz'
    assert decodeString_tuple_stack('10[a]') == 'aaaaaaaaaa'
    assert decodeString_tuple_stack('2[2[b]]') == 'bbbb'
    assert decodeString_tuple_stack('z') == 'z'
    assert decodeString_tuple_stack('2[a]3[b]4[c]') == 'aabbbcccc'
    assert decodeString_tuple_stack('3[a]2[b4[c]]') == 'aaabccccbcccc'
    print('Tuple-stack: All Pass!')

test_tuple_stack()
```
