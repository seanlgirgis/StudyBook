## Reverse Linked List

Given the `head` of a singly linked list, **reverse the list** and return the new head.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

The node class is already defined:

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

### Example

- For `head = [1,2,3,4,5]`, the output should be `solution(head) = [5,4,3,2,1]`.
- For `head = [1,2]`, the output should be `solution(head) = [2,1]`.
- For `head = [1]`, the output should be `solution(head) = [1]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `head` — Head of a singly linked list.  
  Guaranteed constraints: `0 ≤ n ≤ 5000`, `−5000 ≤ val ≤ 5000`
- **[output]** `ListNode` — Head of the reversed list.

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val      # the data stored in this node
        self.next = next    # pointer to the next node (None = end of list)

def to_list(head):          # helper: linked list → Python list (for assertions)
    result = []
    while head:             # walk until we fall off the end (None)
        result.append(head.val)
        head = head.next    # move to next node
    return result

def to_ll(lst):             # helper: Python list → linked list
    dummy = ListNode()      # throwaway node so we always have a .next to write to
    cur = dummy
    for v in lst:
        cur.next = ListNode(v)  # attach new node
        cur = cur.next          # advance tail
    return dummy.next       # skip the throwaway, return real head


def solution(head):
    curr = head
    prev = None             # the reversed chain starts empty

    while curr:
        nxt = curr.next     # 1. save next before we overwrite it
        curr.next = prev    # 2. flip this node's arrow to point backward
        prev = curr         # 3. prev advances — now the front of the reversed chain
        curr = nxt          # 4. curr advances — move to the node saved in step 1

    return prev             # curr is None (fell off end); prev is the new head


assert to_list(solution(to_ll([1,2,3,4,5]))) == [5,4,3,2,1]  # given example 1
assert to_list(solution(to_ll([1,2])))        == [2,1]        # given example 2
assert to_list(solution(to_ll([1])))          == [1]          # single node
assert to_list(solution(to_ll([])))           == []           # empty list
assert to_list(solution(to_ll([1,2,3])))      == [3,2,1]      # three nodes
assert to_list(solution(to_ll([5,4,3,2,1]))) == [1,2,3,4,5]  # already reversed
assert to_list(solution(to_ll([1,1,1])))      == [1,1,1]      # all same values
assert to_list(solution(to_ll([-1,0,1])))     == [1,0,-1]     # negative values
print('All Pass!')
````
