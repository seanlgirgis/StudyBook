## Maximum Subarray

Given an array of integers `nums`, find the **contiguous subarray** with the largest sum and return that sum.

A subarray is a contiguous non-empty part of the array.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [-2,1,-3,4,-1,2,1,-5,4]`, the output should be `solution(nums) = 6`.

  **Explanation:** The subarray `[4,-1,2,1]` has the largest sum `4 + (-1) + 2 + 1 = 6`.

- For `nums = [5,4,-1,7,8]`, the output should be `solution(nums) = 23`.

  **Explanation:** The entire array sums to `23` — no negative drag anywhere.

- For `nums = [-1,-2,-3]`, the output should be `solution(nums) = -1`.

  **Explanation:** All negative — the best subarray is the least negative single element.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^5`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[output]** `integer` — The largest subarray sum.

```python
def solution(nums):
    best = nums[0]
    for i in range(len(nums)):
        total = 0
        for j in range(i, len(nums)):
            total += nums[j]
            best = max(best, total)
    return best


assert solution([-2,1,-3,4,-1,2,1,-5,4]) == 6   # given example 1 — [4,-1,2,1]
assert solution([5,4,-1,7,8])            == 23   # given example 2 — entire array
assert solution([-1,-2,-3])              == -1   # given example 3 — all negative, least bad wins
assert solution([1])                     == 1    # single element
assert solution([1,2,3,4,5])            == 15   # all positive — entire array
assert solution([-2,-1])                 == -1   # two negatives — pick larger
assert solution([2,-1,2,3,-4,3])        == 6    # [2,-1,2,3] = 6
assert solution([0,0,0])                 == 0    # all zeros
print('All Pass!')
```

```python
def solution(nums):
    current = best = nums[0]
    for num in nums[1:]:
        current = max(num, current + num)
        best = max(best, current)
    return best


assert solution([-2,1,-3,4,-1,2,1,-5,4]) == 6   # given example 1 — [4,-1,2,1]
assert solution([5,4,-1,7,8])            == 23   # given example 2 — entire array
assert solution([-1,-2,-3])              == -1   # given example 3 — all negative, least bad wins
assert solution([1])                     == 1    # single element
assert solution([1,2,3,4,5])            == 15   # all positive — entire array
assert solution([-2,-1])                 == -1   # two negatives — pick larger
assert solution([2,-1,2,3,-4,3])        == 6    # [2,-1,2,3] = 6
assert solution([0,0,0])                 == 0    # all zeros
print('All Pass!')
```
