## Channel Surfing

You are given an array of integers `channels` representing the sequence of TV channels a viewer switched through. The viewer's remote control logs every channel tuned to, including repeated ones.

Your task is to count the total number of times the viewer **changed** to a different channel — that is, the number of times a channel in the log differs from the one immediately before it.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(channels.length)** will fit within the execution time limit.

### Example

- For `channels = [5, 5, 3, 3, 7, 5]`, the output should be `solution(channels) = 3`.

  **Explanation:**
  - `5 → 5` same channel, no change
  - `5 → 3` different channel, **change**
  - `3 → 3` same channel, no change
  - `3 → 7` different channel, **change**
  - `7 → 5` different channel, **change**
  - Total changes: 3

- For `channels = [1, 2, 3, 2, 1]`, the output should be `solution(channels) = 4`.

  **Explanation:**
  Every consecutive pair is different, so there are 4 changes.

- For `channels = [7, 7, 7]`, the output should be `solution(channels) = 0`.

  **Explanation:**
  The viewer never changed the channel.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `channels` — An array of integers representing the channel log. It is guaranteed the array is non-empty.
  Guaranteed constraints: `1 ≤ channels.length ≤ 1000`, `1 ≤ channels[i] ≤ 1000`
- **[output]** `integer` — The number of channel changes.

```python
def solution(channels):
    changes = -1
    prev = -float('INF')
    for ch in channels:
        if prev != ch:
            changes += 1
            prev = ch
    return changes
        


assert solution([5, 5, 3, 3, 7, 5])    == 3  # given example 1
assert solution([1, 2, 3, 2, 1])        == 4  # given example 2 — every pair differs
assert solution([7, 7, 7])              == 0  # given example 3 — never changed
assert solution([1])                    == 0  # single channel — no changes possible
assert solution([1, 2])                 == 1  # minimal change
assert solution([1, 1])                 == 0  # two same — no change
assert solution([1, 2, 1, 2, 1])        == 4  # alternating — every step changes
assert solution([5, 5, 5, 3, 3, 3])    == 1  # one transition in the middle
print('All Pass!')
```
