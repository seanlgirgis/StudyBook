## Car Fleet

`n` cars drive toward a target. Each car has a `position` and `speed`. A faster car that catches up to a slower one forms a **fleet** and travels at the slower car's speed. Return the number of fleets that arrive at the target.

### Example

- `target=12`, `position=[10,8,0,5,3]`, `speed=[2,4,1,1,3]`  → `3`
  - Car at 10 → arrives in (12-10)/2 = 1.0
  - Car at 8  → arrives in (12-8)/4  = 1.0  → same time as car 10 → **fleet**
  - Car at 5  → arrives in (12-5)/1  = 7.0  → alone
  - Car at 3  → arrives in (12-3)/3  = 3.0  → alone
  - Car at 0  → arrives in (12-0)/1  = 12.0 → alone
  - Result: 3 fleets

### Hint
Sort by position descending (closest to target first). Use a stack of arrival times — if the next car arrives **at or before** the car ahead, they merge (same fleet, don't push).

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `integer` `target`, `array.integer` `position`, `array.integer` `speed`
- **[output]** `integer`

```python
def solution(target, position, speed):
    n = len(position)
    if n < 2: return n
    cars = sorted(zip([target - x for x in position], speed))
    times = [x / s for x, s in cars]
    stack = []
    for time in times:
        if not stack or stack[-1] < time:   # slower than leader → new fleet
            stack.append(time)
        # else: catches fleet ahead → skip
    return len(stack)


assert solution(12, [10,8,0,5,3], [2,4,1,1,3]) == 3
assert solution(10, [3],          [3])          == 1
assert solution(100,[0,2,4],      [4,2,1])      == 1
assert solution(10, [6,8],        [3,2])        == 2
print('All Pass!')
```

```python
def solution(target, position, speed):
    n = len(position)
    if n < 2: return n
    cars = [target - x for x in position]
    cars = [(x, s) for x, s in zip(cars, speed)]
    cars.sort()
    cars = [x/s for (x,s) in cars]
    fleets = 1
    fleet_leader = cars[0]
    for i in range(1, n):
        if cars[i] > fleet_leader:
            fleets +=1
            fleet_leader = cars[i]
    return fleets


assert solution(12, [10,8,0,5,3], [2,4,1,1,3]) == 3
assert solution(10, [3],          [3])          == 1
assert solution(100,[0,2,4],      [4,2,1])      == 1
assert solution(10, [6,8],        [3,2])        == 2
print('All Pass!')
```
