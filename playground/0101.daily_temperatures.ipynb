## Daily Temperatures

Given an array of integers `temperatures`, return an array `answer` where `answer[i]` is the number of days you have to wait after day `i` to get a warmer temperature. If there is no future warmer day, `answer[i] = 0`.

**Note:** O(n) solution expected.

### Example

- `temperatures = [73,74,75,71,69,72,76,73]` → `[1,1,4,2,1,1,0,0]`
- `temperatures = [30,40,50,60]` → `[1,1,1,0]`
- `temperatures = [30,60,90]` → `[1,1,0]`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `temperatures`
- **[output]** `array.integer`

```python
%run _timer.py
```

```python
def solution(temps):
    stack = []
    n = len(temps)
    ret = [0] * n
    
    for i, temp in enumerate(temps):
        while stack and temp > temps[stack[-1]]:
            popped_ind = stack.pop()
            ret[popped_ind] = i - popped_ind
        stack.append(i)
    return ret
    


assert solution([73,74,75,71,69,72,76,73]) == [1,1,4,2,1,1,0,0]
assert solution([30,40,50,60])             == [1,1,1,0]
assert solution([30,60,90])               == [1,1,0]
assert solution([90,60,30])               == [0,0,0]
assert solution([70,70,70])               == [0,0,0]
print('All Pass!')
```
