## Find All Numbers Disappeared in an Array

Given an array `nums` of `n` integers where each integer is in the range `[1, n]`, return all integers in that range that **do not appear** in `nums`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [4,3,2,7,8,2,3,1]`, the output should be `solution(nums) = [5,6]`.

  **Explanation:** The array has 8 elements, so numbers 1–8 are expected. `5` and `6` are missing.

- For `nums = [1,1]`, the output should be `solution(nums) = [2]`.

  **Explanation:** `1` appears twice, `2` is missing.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — Length `n`, values in `[1, n]`.  
  Guaranteed constraints: `1 ≤ n ≤ 10^5`
- **[output]** `array.integer` — Missing numbers in ascending order.

```python
%run _timer.py
```

```python
def solution(nums):
    m = len(nums)
    nums = set(nums)
    ret =[]
    for i in range(1, m+1):
        if i not in nums:
            ret.append(i)
    return ret


assert solution([4,3,2,7,8,2,3,1]) == [5,6]     # given example 1
assert solution([1,1])             == [2]        # given example 2 — duplicate, one missing
assert solution([1])               == []         # nothing missing
assert solution([2])               == [1]        # missing first
assert solution([2,2])             == [1]        # duplicate at end
assert solution([1,2,3,4,5])      == []         # all present
assert solution([5,4,3,2,1])      == []         # reverse order, none missing
assert solution([1,1,1,1])        == [2,3,4]    # only 1s — everything else missing
print('All Pass!')
```
