## Remove Nth Node From End of List

Given the `head` of a linked list, remove the **nth node from the end** and return the head.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `head = [1,2,3,4,5]` and `n = 2`, the output should be `[1,2,3,5]`.

  **Explanation:** The 2nd node from the end is `4`. Remove it.

- For `head = [1]` and `n = 1`, the output should be `[]`.

- For `head = [1,2]` and `n = 1`, the output should be `[1]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `head`, `integer` `n`  
  Guaranteed constraints: `1 ≤ n ≤ sz`, `1 ≤ sz ≤ 30`
- **[output]** `ListNode` — Head of the list after removal.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

def to_ll(lst):
    dummy = ListNode()
    cur = dummy
    for v in lst:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def solution(head, n):
    def length_of_ll(head):
        count = 0
        curr = head
        while curr:
            count += 1
            curr = curr.next
        return count    
    def delete_at_index(head, n):
        # Case 1: List is empty
        if not head:
            return None
    
        # Case 2: Deleting the head (index 0)
        if n == 0:
            return head.next
    
        curr = head
        # Navigate to the node at index (n-1)
        for i in range(n - 1):
            if curr.next is None:  # Index is out of bounds
                return head
            curr = curr.next
    
        # Case 3: Delete the node by skipping it
        # curr is now at (n-1), so curr.next is the node to delete
        if curr.next:
            curr.next = curr.next.next
    
        return head 
    dummy = delete_at_index(head, length_of_ll(head) - n)
    return dummy


assert to_list(solution(to_ll([1,2,3,4,5]), 2)) == [1,2,3,5]  # given example 1 — remove 4
assert to_list(solution(to_ll([1]), 1))          == []         # given example 2 — remove only node
assert to_list(solution(to_ll([1,2]), 1))         == [1]       # given example 3 — remove last
assert to_list(solution(to_ll([1,2]), 2))         == [2]       # remove head
assert to_list(solution(to_ll([1,2,3]), 3))       == [2,3]     # remove head of 3
assert to_list(solution(to_ll([1,2,3]), 1))       == [1,2]     # remove tail of 3
assert to_list(solution(to_ll([1,2,3]), 2))       == [1,3]     # remove middle
assert to_list(solution(to_ll([1,2,3,4,5]), 5))  == [2,3,4,5] # remove head of 5
print('All Pass!')
```
