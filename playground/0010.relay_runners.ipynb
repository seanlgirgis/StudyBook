## Relay Runners

A relay race team has `n` runners arranged in a rotation. Runner `i` can run for `stamina[i]` minutes before becoming exhausted. After exhausting, they need `recovery[i]` minutes of rest before they can run again.

Runners take turns **in cyclic order** — after the last runner, the rotation goes back to runner `0`. If the next runner in rotation is still recovering, skip them and try the next one. If **all** runners are recovering at the same time, the race cannot continue.

Your task is to calculate how many runners were **fully exhausted** (ran their entire stamina) over the course of `t` total race minutes. The last runner active when `t` is reached does **not** count if they weren't fully exhausted. Return `-1` if the race cannot be completed.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(t · n)** will fit within the execution time limit.

### Example

- For `t = 16`, `stamina = [2, 5, 6]`, and `recovery = [12, 1, 4]`, the output should be `solution(t, stamina, recovery) = 3`.

  **Explanation:**
  - `clock = 0`: Runner 0 runs for 2 min. `clock = 2`. Runner 0 recovers until `t = 14`.
  - `clock = 2`: Runner 1 runs for 5 min. `clock = 7`. Runner 1 recovers until `t = 8`.
  - `clock = 7`: Runner 2 runs for 6 min. `clock = 13`. Runner 2 recovers until `t = 17`.
  - `clock = 13`: Runner 0 still recovering. Runner 1 is ready — runs for 5 min. `clock = 18 > 16`.
  - Last runner was not fully exhausted — don't count. Total fully exhausted: **3**.

- For `t = 16`, `stamina = [2, 5, 6]`, and `recovery = [12, 8, 4]`, the output should be `solution(t, stamina, recovery) = -1`.

  **Explanation:**
  - At `clock = 13` all runners are still recovering — race cannot continue.

- **[output]** `integer` — Number of fully exhausted runners, or `-1` if impossible.

```python
import heapq
def solution(t, stamina, recovery):
    ticket_counter = 0                 # is being used to keep the Order of the runners. Gets incremented  When gets into recovery

    # ready_line: (ticket, battery) — lowest ticket gets picked next
    ready = []
    for runner in range(len(stamina)):
        heapq.heappush(ready, (ticket_counter, runner))
        ticket_counter += 1

    # charger: (done_at, ticket, battery) — earliest done floats to the top
    # ticket is stamped when the battery is SENT away — so it knows its place in line when it returns
    recovering = []                    # tuple of 3 when recover , ticket and runner id
    clock         = 0
    fully_exhausted = 0

    def move_to_ready(recovering, ready, clock):
        while recovering and recovering[0][0] <= clock:
            _, ticket, runner = heapq.heappop(recovering)
            heapq.heappush(ready, (ticket, runner))     
    
    while clock < t:
        move_to_ready(recovering, ready, clock)
        if not ready:
            return -1
    
        _, runner = heapq.heappop(ready)
        fully_exhausted += 1
    
        clock += stamina[runner]                        # runner ran — clock at exhaustion point
        ready_again = clock + recovery[runner]          # recovery starts exactly here
        heapq.heappush(recovering, (ready_again, ticket_counter, runner))
        ticket_counter += 1
    
    if clock > t:
        fully_exhausted -= 1
    return fully_exhausted

        
    
assert solution(16, [2, 5, 6], [12, 1, 4])  == 3   # given example 1
assert solution(16, [2, 5, 6], [12, 8, 4])  == -1  # given example 2 — all recovering
assert solution(1,  [1],       [99])         == 1   # single runner, exact finish
assert solution(2,  [1],       [99])         == -1  # single runner, can't recover in time
assert solution(5,  [10],      [1])          == 0   # runner finishes race before exhausting
assert solution(10, [5, 3],    [1, 2])       == 2   # two runners, two full shifts
assert solution(4,  [2, 2],    [1, 1])       == 2   # clean two-runner finish
assert solution(3,  [1, 1],    [5, 5])       == -1  # both recovering simultaneously
print('All Pass!')
```
