# 001 - Multisort Tupe List

```python
from operator import itemgetter

def multisort_items(xs, specs):
    # specs: (index, reverse)  -> reverse=True means descending
    for idx, reverse in reversed(specs):
        xs.sort(key=itemgetter(idx), reverse=reverse)
    return xs

rows = [('dave', 'B', 10), ('jane', 'B', 12), ('john', 'A', 15)]

# 0 -> desc, 1 -> asc, 2 -> desc
sorted_rows = multisort_items(rows[:], ((0, True), (1, False), (2, True)))

print(sorted_rows)
# [('john', 'A', 15), ('jane', 'B', 12), ('dave', 'B', 10)]
```

```python
from collections import Counter
def solution(nums):
    counts = Counter(nums)
    return counts.most_common()[-1][0]
solution([3,2,3])   
```

```python
def localPeak(nums):
    if len(nums) == 1:
        return 0 
    if len(nums) == 2:
        return 0 if nums[0] > nums[1] else (1)
    for i in range(1, len(nums) - 1):
        if nums[i] > nums[i-1] and nums[i] > nums[i+1]:
            return i 
            
localPeak([1,2,1,3,5,6,4])
```

```python
#0067
def jumpGame(nums):
    max_reach = 0
    for i, jump in enumerate(nums):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + jump)
    return True
        
jumpGame([3,2,2,0,4])  
```
