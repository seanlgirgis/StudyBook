## Majority Element

Given an array `nums` of size `n`, return the **majority element** — the element that appears **more than `n/2` times**. You may assume the majority element always exists.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [3,2,3]`, the output should be `solution(nums) = 3`.
- For `nums = [2,2,1,1,1,2,2]`, the output should be `solution(nums) = 2`.

  **Explanation:** `2` appears 4 times out of 7 — more than 7/2 = 3.5.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `1 ≤ n ≤ 5·10^4`, `-10^9 ≤ nums[i] ≤ 10^9`
- **[output]** `integer` — The majority element.

```python
from collections import Counter
def solution(nums):
    counts = Counter(nums)
    return counts.most_common(1)[0][0]


assert solution([3,2,3])              == 3  # given example 1
assert solution([2,2,1,1,1,2,2])     == 2  # given example 2
assert solution([1])                  == 1  # single element
assert solution([1,1,2])             == 1  # majority at start
assert solution([2,1,1])             == 1  # majority at end
assert solution([6,6,6,7,7])         == 6  # clear majority
assert solution([1,2,1,2,1])         == 1  # alternating, 1 wins
assert solution([3,3,4,3,4,3,4,3,4]) == 3  # 3 appears 5/9 times
print('All Pass!')
```

```python
#Boyer-Moore for your toolkit —
from collections import Counter
def solution(nums):
    candidate, count = nums[0], 0
    for num in nums:
        if count == 0:
            candidate = num
        count += 1 if num == candidate else -1
    return candidate


assert solution([3,2,3])              == 3  # given example 1
assert solution([2,2,1,1,1,2,2])     == 2  # given example 2
assert solution([1])                  == 1  # single element
assert solution([1,1,2])             == 1  # majority at start
assert solution([2,1,1])             == 1  # majority at end
assert solution([6,6,6,7,7])         == 6  # clear majority
assert solution([1,2,1,2,1])         == 1  # alternating, 1 wins
assert solution([3,3,4,3,4,3,4,3,4]) == 3  # 3 appears 5/9 times
print('All Pass!')
```
