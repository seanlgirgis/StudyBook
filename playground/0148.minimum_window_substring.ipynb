## Minimum Window Substring (LC 76)

Find the smallest window in `s` that contains all characters of `t` (including duplicates).

**Pattern:** Sliding window with frequency matching.
- `need` = Counter(t), `required` = number of unique chars to satisfy
- Expand `r` until all satisfied (`have == required`)
- Shrink `l` to minimise window, record best, then break the condition and keep expanding
- `have` tracks how many chars meet their exact count — not just presence

```python
from collections import Counter

def solution(s, t):
    if not t or not s:
        return ""

    need     = Counter(t)        # how many of each char we must have
    window   = {}                # current window counts
    have     = 0                 # chars whose count is satisfied
    required = len(need)         # unique chars we need to satisfy

    l       = 0
    best    = (float('inf'), 0, 0)   # (length, l, r)

    for r, c in enumerate(s):
        window[c] = window.get(c, 0) + 1
        if c in need and window[c] == need[c]:   # exactly met — not exceeded
            have += 1

        while have == required:          # valid window — try to shrink
            if r - l + 1 < best[0]:
                best = (r - l + 1, l, r)

            window[s[l]] -= 1
            if s[l] in need and window[s[l]] < need[s[l]]:
                have -= 1                # dropped below required count
            l += 1

    return s[best[1]:best[2] + 1] if best[0] != float('inf') else ""


assert solution("ADOBECODEBANC", "ABC")  == "BANC"
assert solution("a",             "a")   == "a"
assert solution("a",             "aa")  == ""
assert solution("aa",            "aa")  == "aa"
assert solution("bba",           "ab")  == "ba"
assert solution("cabwefgewcwaefgcf", "cae") == "cwae"
print('All Pass!')
```
