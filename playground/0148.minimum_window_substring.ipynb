## Minimum Window Substring (LC 76)

Find the smallest window in `s` that contains all characters of `t` (including duplicates).

**Pattern:** Sliding window with frequency matching.
- `need` = Counter(t), `required` = number of unique chars to satisfy
- Expand `r` until all satisfied (`have == required`)
- Shrink `l` to minimise window, record best, then break the condition and keep expanding
- `have` tracks how many chars meet their exact count — not just presence

```python
"""
id: lc_0076
title: Minimum Window Substring
source: leetcode
difficulty: hard
primary: sliding-window
tags: [string, hash-table, sliding-window]
leetcode_url: https://leetcode.com/problems/minimum-window-substring/
status: draft
last_updated: 2026-04-22
notes:
- key idea: sliding window with two frequency maps (target and current)
- time: O(m + n)
- space: O(m + n)
"""

# ============================================================================
# File: 076_lc_0076_minimum_window_substring_empty.py
# LeetCode 76: Minimum Window Substring
# Difficulty: Hard
#
# PROBLEM STATEMENT:
# Given two strings s and t of lengths m and n respectively, return the minimum 
# window substring of s such that every character in t (including duplicates) 
# is included in the window. If there is no such substring, return the empty 
# string "".
#
# RULES:
# - The testcases will be generated such that the answer is unique.
# - s and t consist of uppercase and lowercase English letters.
#
# EXAMPLES:
# s = "ADOBECODEBANC", t = "ABC" -> "BANC"
# s = "a", t = "a" -> "a"
# s = "a", t = "aa" -> ""
# ============================================================================

from typing import List

def minWindow(s: str, t: str) -> str:
    if t == "": return ""
        
    countT, window = {}, {}
    for c in t:
        countT[c]  = countT.get(c, 0) + 1

    have, need = 0 , len(countT)
    res, resLen = [-1, -1], float("INF")
    l = 0

    for r, c in enumerate(s):
        window[c] = window.get(c, 0) + 1

        if c in countT and window[c] == countT[c]:
            have += 1
            
        while have  == need:          #shrink 
            if (r -l +1) < resLen:
                res = [l, r]
                resLen = (r - l + 1)
            window[s[l]] -= 1
            if s[l] in countT and window[s[l]] < countT[s[l]]:
                have -= 1
            l += 1
    l, r = res
    return s[l:r+1] if resLen !=  float('INF') else ""
        

print(minWindow("aaabdabcefaecbef", "abc") )
print( minWindow("bba", "ab") )                        
print(minWindow("ADOBECODEBANC", "ABC"))
print(minWindow("a", "aa"))

def test():
    assert minWindow("aaabdabcefaecbef", "abc") == "abc" # 
    assert minWindow("bba", "ab") == "ba"  # breaks ret-tracking/shrink logic
    assert minWindow("ADOBECODEBANC", "ABC") == "BANC"  # example 1
    assert minWindow("a", "a") == "a"  # example 2
    assert minWindow("a", "aa") == ""  # example 3
    assert minWindow("ab", "b") == "b"  # single char match
    assert minWindow("aa", "aa") == "aa"  # identical strings
    assert minWindow("ABCDEFG", "GD") == "DEFG"  # start and end chars
    assert minWindow("DONOTPANIC", "NOO") == "ONO"  # duplicate target chars
    assert minWindow("a", "b") == ""  # no match possible
    assert minWindow("ADOBECODEBANC", "ABCZ") == ""  # missing one char
    assert minWindow("cabbacc", "aba") == "abba"  # overlap/repeats
    print('All Pass!')

test()
```

```python
from collections import Counter

def solution(s, t):
    if not t or not s:
        return ""

    need     = Counter(t)        # how many of each char we must have
    window   = {}                # current window counts
    have     = 0                 # chars whose count is satisfied
    required = len(need)         # unique chars we need to satisfy

    l       = 0
    best    = (float('inf'), 0, 0)   # (length, l, r)

    for r, c in enumerate(s):
        window[c] = window.get(c, 0) + 1
        if c in need and window[c] == need[c]:   # exactly met — not exceeded
            have += 1

        while have == required:          # valid window — try to shrink
            if r - l + 1 < best[0]:
                best = (r - l + 1, l, r)

            window[s[l]] -= 1
            if s[l] in need and window[s[l]] < need[s[l]]:
                have -= 1                # dropped below required count
            l += 1

    return s[best[1]:best[2] + 1] if best[0] != float('inf') else ""


assert solution("ADOBECODEBANC", "ABC")  == "BANC"
assert solution("a",             "a")   == "a"
assert solution("a",             "aa")  == ""
assert solution("aa",            "aa")  == "aa"
assert solution("bba",           "ab")  == "ba"
assert solution("cabwefgewcwaefgcf", "cae") == "cwae"
print('All Pass!')
```
