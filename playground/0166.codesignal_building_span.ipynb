## CodeSignal — Building Span

Given an array of building heights, for each building `i` compute its **span**:
the number of consecutive buildings ending at `i` (including `i` itself)
where every building in that range has height `<= heights[i]`.

Return an array of spans, one per building.

### Examples

```
[1, 2, 3]      → [1, 2, 3]
  h=1: span 1 (just itself)
  h=2: absorbs h=1 (1<=2) → span 2
  h=3: absorbs h=2 span=2 (2<=3) → span 3

[1, 1, 2, 2]   → [1, 2, 3, 4]
[3, 1, 2]      → [1, 1, 2]
  h=3: span 1
  h=1: 3 > 1, stop → span 1
  h=2: absorbs h=1 span=1 (1<=2); 3>2, stop → span 2

[1000000]      → [1]
[5, 5, 5]      → [1, 2, 3]   (equal height counts as <=)
```

### Rules
- Minimum span is always 1 (each building counts itself).
- Equal-height buildings count toward the span (`<=`, not `<`).
- **Must use a monotonic stack** — nested loops will TLE on `[1000000]`.

### Hint
Stack stores `(height, accumulated_span)` tuples.
When a taller-or-equal building arrives, pop and absorb the span of everything it towers over.
