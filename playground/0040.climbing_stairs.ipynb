## Climbing Stairs

You are climbing a staircase with `n` steps. Each time you can climb either **1 or 2 steps**. Return the number of **distinct ways** you can reach the top.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `n = 2`, the output should be `solution(n) = 2`.

  **Explanation:** Two ways: `(1,1)` or `(2)`.

- For `n = 3`, the output should be `solution(n) = 3`.

  **Explanation:** Three ways: `(1,1,1)`, `(1,2)`, `(2,1)`.

- For `n = 5`, the output should be `solution(n) = 8`.

  **Explanation:** `(1,1,1,1,1)`, `(1,1,1,2)`, `(1,1,2,1)`, `(1,2,1,1)`, `(2,1,1,1)`, `(1,2,2)`, `(2,1,2)`, `(2,2,1)` — 8 ways.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `integer` `n`  
  Guaranteed constraints: `1 ≤ n ≤ 45`
- **[output]** `integer` — Number of distinct ways to climb to the top.

```python
def solution(n):
    dp = [1] * n 
    if n <= 2 : return n
    dp[1] = 2
    for i in range(2, n):
        dp[i] = dp[i-2] + dp[i-1]
    return dp[-1]
        


assert solution(1)  == 1   # one step — only one way
assert solution(2)  == 2   # given example 1
assert solution(3)  == 3   # given example 2
assert solution(5)  == 8   # given example 3
assert solution(4)  == 5   # 5 ways
assert solution(6)  == 13  # fibonacci continues
assert solution(10) == 89
assert solution(45) == 1836311903
print('All Pass!')
```

```python
def solution(n):
    prevprev , prev = 0, 1
    for i in range(n):
        prevprev , prev = prev, prev + prevprev
    return prev
        


assert solution(1)  == 1   # one step — only one way
assert solution(2)  == 2   # given example 1
assert solution(3)  == 3   # given example 2
assert solution(5)  == 8   # given example 3
assert solution(4)  == 5   # 5 ways
assert solution(6)  == 13  # fibonacci continues
assert solution(10) == 89
assert solution(45) == 1836311903
print('All Pass!')
```
