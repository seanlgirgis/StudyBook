```python
def longestConsecutive(nums: List[int]) -> int:
    """
    Given an unsorted array of integers nums, return the length of the longest
    consecutive elements sequence.

    You must write an algorithm that runs in O(n) time.
    """
    my_nums = set(nums)
    max_seq = 0
    def seq_length (theSet, num):
        ret = 0 
        while num in theSet:
            ret+=1
            num+=1
        return ret
        
    for num in my_nums:
        if (num - 1) not in my_nums:    #start of a seq
            max_seq = max(max_seq, seq_length(my_nums, num))
    return max_seq
            
            

print(longestConsecutive([-3, -2, -1, 0, 1, 2, 50]))
```

```python
def productExceptSelf(nums: List[int]) -> List[int]:
    """
    Given an integer array nums, return an array answer such that answer[i]
    is equal to the product of all the elements of nums except nums[i].

    Constraints (LeetCode style):
    - Solve without using division
    - Aim for O(n) time
    """
    ret = [1] * len(nums)
    prefix , postfix = 1,1
    for i in range(1, len(nums)):
        prefix *= nums[i-1]
        ret[i] = prefix 
    for i in range(len(nums) -1 , -1, -1):
        ret[i] *= postfix
        postfix *= nums[i]
    print(ret)

productExceptSelf([1,2,3,4])
```

```python
def containsDuplicate(nums: List[int]) -> bool:
    seen = set()
    for n in nums:
        if n in seen:
            return True
        seen.add(n)
    return False

print(containsDuplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2]))
```

```python
# LeetCode 155: Min Stack

from typing import Any, List, Tuple

# --- TEST CASES ---
# Format: (operations_list, arguments_list, expected_returns)
min_stack_tests: List[Tuple[List[str], List[List[int]], List[Any]]] = [
    (
        ["MinStack", "push", "push", "push", "getMin", "pop", "top", "getMin"],
        [[], [-2], [0], [-3], [], [], [], []],
        [None, None, None, None, -3, None, 0, -2]
    ), # 1. Standard LC Example 1
    (
        ["MinStack", "push", "push", "getMin", "getMin", "push", "getMin", "getMin", "top", "getMin", "pop", "push", "push", "getMin", "pop", "getMin", "pop", "getMin"],
        [[], [-10], [14], [], [], [-20], [], [], [], [], [], [10], [-7], [], [], [], [], []],
        [None, None, None, -10, -10, None, -20, -20, -20, -20, None, None, None, -10, None, -10, None, -10]
    ), # 2. Extended operations mix
    (
        ["MinStack", "push", "getMin"],
        [[], [5], []],
        [None, None, 5]
    ), # 3. Single element stack
    (
        ["MinStack", "push", "push", "push", "getMin", "pop", "getMin", "pop", "getMin"],
        [[], [5], [4], [3], [], [], [], [], []],
        [None, None, None, None, 3, None, 4, None, 5]
    ), # 4. Strictly decreasing inserts
    (
        ["MinStack", "push", "push", "push", "getMin", "pop", "getMin", "pop", "getMin"],
        [[], [3], [4], [5], [], [], [], [], []],
        [None, None, None, None, 3, None, 3, None, 3]
    ), # 5. Strictly increasing inserts (min never changes until empty)
    (
        ["MinStack", "push", "push", "push", "getMin", "pop", "getMin"],
        [[], [2], [2], [2], [], [], []],
        [None, None, None, None, 2, None, 2]
    )  # 6. Duplicate minimums (testing proper duplicate tracking on pop)
]

# --- TEST HARNESS ---
def test_harness(stack_class: type, test_cases: List[Tuple[List[str], List[List[int]], List[Any]]]) -> None:
    """
    Test harness for LeetCode #155: Min Stack.
    Validates O(1) class methods using dynamic dispatch.
    """
    print(f"--- Running Tests for: {stack_class.__name__} ---")
    passed: int = 0
    for i, (operations, arguments, expected) in enumerate(test_cases):
        try:
            results: List[Any] = []
            obj = None
            
            for op, arg in zip(operations, arguments):
                if op == "MinStack":
                    obj = stack_class()
                    results.append(None)
                else:
                    # Dynamically call the method based on the string name
                    method = getattr(obj, op)
                    # Unpack arguments if they exist, otherwise call empty
                    result = method(*arg) if arg else method()
                    results.append(result)
            
            if results == expected:
                print(f"Test {i+1}: PASSED ({len(operations)} operations)")
                passed += 1
            else:
                print(f"Test {i+1}: FAILED | {len(operations)} operations")
                print(f"    Expected: {expected}")
                print(f"    Got:      {results}")
        except Exception as e:
            print(f"Test {i+1}: ERROR  | {type(e).__name__}: {e}")

    print(f"\nSummary: {passed}/{len(test_cases)} tests passed.")


# --- USER TO IMPLEMENT SOLUTION BELOW ---
class MinStack:
    """
    LC 155 — Min Stack
    
    PROBLEM:
    Design a stack that supports push, pop, top, and retrieving the minimum element 
    in constant time O(1).
    
    HINT / APPROACH:
    1. Pattern: Two Stacks OR Tuple Stack.
    2. The constraint: You CANNOT scan the stack to find the minimum when `getMin()` 
       is called, because that would be O(n). It must be O(1).
    3. Approach A (Tuple Stack - Recommended for Python): 
       Instead of just appending the raw value to your list, append a tuple: 
       `(value, current_minimum)`. 
       When you push a new number, the `current_minimum` is simply 
       `min(new_value, previous_minimum)`.
    4. Approach B (Two Stacks):
       Maintain `self.stack` for the actual values, and `self.min_stack` to track 
       the minimums. Every time you push to `self.stack`, you also push the current 
       minimum to `self.min_stack`.
    5. Because you track the minimum *state* alongside every single element, popping 
       an element naturally rolls back the minimum state to whatever it was previously!
    """

    def __init__(self) -> None: 
        self.stack = []
        self.min_stack =  []              #keep adding if smaller than top 
    def push(self, val: int) -> None:
        self.stack.append(val)
        if self.min_stack and self.min_stack[-1] >= val:
            self.min_stack.append(val)
        elif not self.min_stack:
            self.min_stack.append(val)
            
            
    def pop(self) -> None: 
        if self.stack:
            val = self.stack.pop()
            if self.min_stack and val == self.min_stack[-1]:
                self.min_stack.pop()
        
    def top(self) -> int: 
        return self.stack[-1] if self.stack else None
    def getMin(self) -> int: 
        return self.min_stack[-1] if self.min_stack else None


# Execute harness without __main__ block
test_harness(MinStack, min_stack_tests)
mystack = MinStack()
mystack.push(-2)
print(mystack.top())
print(mystack.getMin())
```

```python
def lengthOfLongestSubstring(s: str) -> int:
    if len(s) <= 1: return len(s)
    ret = 1
    seen = set(s[0])
    j = 0
    for i in range(1, len(s)):
        while s[i] in seen and j <= i:
            seen.remove(s[j])
            j+=1
        seen.add(s[i])
        ret = max(ret, i-j+1)
    return ret
            

        
        

print(lengthOfLongestSubstring("abcabcbb"))
```
