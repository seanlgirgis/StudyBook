## Number of Recent Calls

You have a `RecentCounter` class that counts the number of recent requests within a certain time frame.

Implement the class:
- `RecentCounter()` — initializes the counter
- `ping(t)` — adds a new request at time `t` (in milliseconds) and returns the number of requests that have happened in the past `3000` milliseconds (inclusive of `t - 3000` and `t`). All calls to `ping` are made with strictly increasing values of `t`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(1)** per call will fit within the execution time limit.

### Example

```
rc = RecentCounter()
rc.ping(1)    → 1   # requests in [max(0,-2999), 1] = [1] → 1
rc.ping(100)  → 2   # requests in [-2900, 100] = [1,100] → 2
rc.ping(3001) → 3   # requests in [1, 3001] = [1,100,3001] → 3
rc.ping(3002) → 3   # requests in [2, 3002] = [100,3001,3002] → 3 (1 dropped)
```

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** Series of `ping(t)` calls with strictly increasing `t`
- **[output]** `integer` per call — count of pings in `[t - 3000, t]`

````python
from collections import deque

class RecentCounter:
    def __init__(self):
        self.q = deque()

    def ping(self, t):
        while self.q and t - self.q[0] > 3000:
            self.q.popleft()       # drop pings outside the 3000ms window
        self.q.append(t)
        return len(self.q)


rc = RecentCounter()
assert rc.ping(1)    == 1
assert rc.ping(100)  == 2
assert rc.ping(3001) == 3
assert rc.ping(3002) == 3

rc2 = RecentCounter()
assert rc2.ping(5000) == 1
assert rc2.ping(5001) == 2
assert rc2.ping(8001) == 2
assert rc2.ping(8002) == 2
print('All Pass!')
````
