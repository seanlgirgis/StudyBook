## Top K Frequent Elements

Given an array of integers `nums` and an integer `k`, return the `k` most frequent elements in any order.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.

### Example

- For `nums = [1,1,1,2,2,3]`, `k = 2`, the output should be `solution(nums, k) = [1, 2]` (any order).

  **Explanation:** `1` appears 3 times, `2` appears 2 times, `3` appears once. Top 2 are `1` and `2`.

- For `nums = [1]`, `k = 1`, the output should be `solution(nums, k) = [1]`.

- For `nums = [1,2]`, `k = 2`, the output should be `solution(nums, k) = [1, 2]` (any order).

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^5`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[input]** `integer` `k` — Guaranteed `1 ≤ k ≤` number of unique elements
- **[output]** `array.integer` — The `k` most frequent elements (order does not matter).

```python
import heapq
from collections import Counter
def solution(nums, k):
    counts = Counter(nums)
    heap = []
    for n, freq in counts.items():
        heapq.heappush(heap, (freq, n))
        if len(heap) > k:
             heapq.heappop(heap)
    return [n for _,n in heap]


def check(result, expected):
    return sorted(result) == sorted(expected)

assert check(solution([1,1,1,2,2,3], 2),       [1,2])      # given example 1
assert check(solution([1], 1),                  [1])        # given example 2 — single element
assert check(solution([1,2], 2),                [1,2])      # given example 3 — k = all unique
assert check(solution([3,3,3,1,1,2], 1),        [3])        # clear winner
assert check(solution([4,4,4,1,1,2], 2),        [4,1])      # clear top 2
assert check(solution([5,5,5,5], 1),            [5])        # single unique element
assert check(solution([1,1,2,2,2,3,3,3,3], 2), [2,3])      # top 2 clear
assert check(solution([1,1,2,2,2,3], 3),        [1,2,3])    # k = all unique
print('All Pass!')
```
