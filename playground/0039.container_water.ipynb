## Container With Most Water

You are given an array `height` where `height[i]` represents the height of a vertical line at position `i`. Two lines together with the x-axis form a container. Find the two lines that together with the x-axis form a container that holds the **most water**.

The amount of water a container holds is: `min(height[l], height[r]) × (r - l)`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `height = [1,8,6,2,5,4,8,3,7]`, the output should be `solution(height) = 49`.

  **Explanation:** Lines at index `1` (height 8) and index `8` (height 7): `min(8,7) × (8-1) = 7 × 7 = 49`.

- For `height = [1,1]`, the output should be `solution(height) = 1`.

  **Explanation:** Only two lines: `min(1,1) × 1 = 1`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `height`  
  Guaranteed constraints: `2 ≤ height.length ≤ 10^5`, `0 ≤ height[i] ≤ 10^4`
- **[output]** `integer` — Maximum water the container can hold.

```python
def solution(height):
    l, r = 0, len(height) -1
    max_water = 0 
    while l < r:
        max_water = max(max_water,( (r-l) * min(height[l], height[r])) )
        if height[l] <= height[r]:
            l+=1
        else:
            r-=1
    return max_water


assert solution([1,8,6,2,5,4,8,3,7]) == 49  # given example 1
assert solution([1,1])               == 1   # given example 2 — two lines
assert solution([4,3,2,1,4])         == 16  # same height at both ends — 4×4
assert solution([1,2,1])             == 2   # middle taller — l and r walls win
assert solution([1,2,4,3])           == 4   # 2×2 or 1×3 — best is 4 (index 1,3: min(2,3)×2=4)
assert solution([2,3,4,5,18,17,6])   == 17  # tall walls near center
assert solution([1,0,0,0,1])         == 4   # zeros in middle — width matters
assert solution([5,5])               == 5   # two equal tall walls
print('All Pass!')
```
