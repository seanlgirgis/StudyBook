## Move Zeroes

Given an integer array `nums`, move all `0`s to the end while maintaining the relative order of the non-zero elements. Do this **in-place**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [0,1,0,3,12]`, the output should be `[1,3,12,0,0]`.
- For `nums = [0]`, the output should be `[0]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums`
- **[output]** `array.integer`

```python
def solution(nums):
    l = 0
    for r in range(len(nums)):
        if nums[r] != 0:
            nums[l], nums[r] = nums[r], nums[l]
            l += 1
    return nums


assert solution([0,1,0,3,12])  == [1,3,12,0,0]
assert solution([0])           == [0]
assert solution([1,0,0,0,3])   == [1,3,0,0,0]
assert solution([1,2,3])       == [1,2,3]
assert solution([0,0,0,1])     == [1,0,0,0]
assert solution([0,0])         == [0,0]
print('All Pass!')
```
