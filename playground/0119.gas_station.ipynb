## Gas Station

There are `n` gas stations in a circle. `gas[i]` = gas at station `i`. `cost[i]` = gas to travel from `i` to `i+1`. Find the starting station index to complete the circuit, or return `-1` if impossible.

### Example

- `gas=[1,2,3,4,5]`, `cost=[3,4,5,1,2]` → `3`
- `gas=[2,3,4]`, `cost=[3,4,3]` → `-1`

### Hint
If `sum(gas) < sum(cost)` → impossible, return `-1`. Otherwise a solution always exists. Greedy: track running tank; when it goes negative, reset start to `i+1`.

### Input/Output
- **[input]** `array.integer` `gas`, `array.integer` `cost`
- **[output]** `integer`

```python
%run _timer.py
```

```python
def solution(gas, cost):
    # If total gas < total cost, no solution exists
    if sum(gas) < sum(cost):
        return -1

    tank  = 0
    start = 0

    for i in range(len(gas)):
        tank += gas[i] - cost[i]
        # Tank went negative — can't start at or before i
        # Reset: try starting from the next station
        if tank < 0:
            start = i + 1
            tank  = 0

    return start


assert solution([1,2,3,4,5], [3,4,5,1,2]) == 3
assert solution([2,3,4],     [3,4,3])     == -1
assert solution([5,1,2,3,4], [4,4,1,5,1]) == 4
assert solution([2],         [2])         == 0
print('All Pass!')
```
