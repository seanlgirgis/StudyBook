## First Unique Character

Given a string `s`, find the **index** of the first character that appears exactly once. If every character repeats, return `-1`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(s.length²)** will fit within the execution time limit.

### Example

- For `s = "abacd"`, the output should be `solution(s) = 1`.

  **Explanation:**
  - `'a'` appears at indices 0 and 2 — repeats.
  - `'b'` appears only at index 1 — **first unique**.
  - Answer: `1`

- For `s = "abacabad"`, the output should be `solution(s) = 3`.

  **Explanation:**
  - `'a'` appears at 0, 2, 4, 6 — repeats.
  - `'b'` appears at 1, 5 — repeats.
  - `'c'` appears only at index 3 — **first unique**.
  - Answer: `3`

- For `s = "aabbcc"`, the output should be `solution(s) = -1`.

  **Explanation:**
  Every character appears more than once — no unique character exists.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `string` `s` — A string of lowercase English letters.
  Guaranteed constraints: `1 ≤ s.length ≤ 1000`
- **[output]** `integer` — Index of the first unique character, or `-1` if none exists.

```python
from collections import Counter
def solution(s):
    cnts = Counter(s)
    for i, c in enumerate(s):
        if cnts[c] == 1:
            return i
    return -1


assert solution('abacd')    == 1   # given example 1 — 'b' at index 1
assert solution('abacabad') == 3   # given example 2 — 'c' at index 3
assert solution('aabbcc')   == -1  # given example 3 — all repeat
assert solution('a')        == 0   # single character — always unique
assert solution('abcd')     == 0   # all unique — first one wins
assert solution('aab')      == 2   # 'b' at end is first unique
assert solution('aabbccdd') == -1  # all paired — no unique
assert solution('aabbccd')  == 6   # 'd' at end is only unique
print('All Pass!')
```

```python
from collections import Counter
def solution(s):
    counts = Counter(s)
    ones = {x for x, v in counts.items() if v == 1}
    for i, ch in enumerate(s):
        if ch in ones:
            return i
    return -1

assert solution('abacd')    == 1   # given example 1 — 'b' at index 1
assert solution('abacabad') == 3   # given example 2 — 'c' at index 3
assert solution('aabbcc')   == -1  # given example 3 — all repeat
assert solution('a')        == 0   # single character — always unique
assert solution('abcd')     == 0   # all unique — first one wins
assert solution('aab')      == 2   # 'b' at end is first unique
assert solution('aabbccdd') == -1  # all paired — no unique
assert solution('aabbccd')  == 6   # 'd' at end is only unique
print('All Pass!')
```
