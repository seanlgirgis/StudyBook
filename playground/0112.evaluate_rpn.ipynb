## Evaluate Reverse Polish Notation

Evaluate an expression in **Reverse Polish Notation** (postfix). Tokens are either integers or operators `+`, `-`, `*`, `/`. Division truncates toward zero.

### Example

- `["2","1","+","3","*"]` → `9`   → `(2+1)*3 = 9`
- `["4","13","5","/","+"]` → `6`  → `4+(13/5) = 4+2 = 6`
- `["10","6","9","3","+","-11","*","/","17","+","5","+"]` → `22`

### Hint
Push numbers onto a stack. When you see an operator, pop two numbers, apply the operator, push the result.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.string` `tokens`
- **[output]** `integer`

```python
"""
id: lc_0150
title: Evaluate Reverse Polish Notation
source: leetcode
difficulty: medium
primary: stack
tags: [stack, math, array]
leetcode_url: https://leetcode.com/problems/evaluate-reverse-polish-notation/
status: draft
last_updated: 2026-04-23
notes:
- key idea: Use a stack to store operands and apply operators to the top two elements.
- time: O(n)
- space: O(n)
"""

# ============================================================================
# File: 150_lc_0150_evaluate_reverse_polish_notation_empty.py
# LeetCode 150: Evaluate Reverse Polish Notation
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Evaluate the value of an arithmetic expression in Reverse Polish Notation.
# Valid operators are +, -, *, and /. Each operand may be an integer or 
# another expression.
#
# RULES:
# 1. Division between two integers should truncate toward zero.
# 2. The input tokens are always valid RPN.
# 3. There will be no division by zero.
#
# EXAMPLES:
# ["2","1","+","3","*"] -> ((2 + 1) * 3) -> 9
# ["4","13","5","/","+"] -> (4 + (13 / 5)) -> 6
# ["10","6","9","3","+","-11","*","/","17","+","5","+"] -> 22
# ============================================================================

from typing import List

def evalRPN(tokens: List[str]) -> int:
    s = []
    for tk in tokens:
        if tk not in {"+", "-", "*", "/"}:
            s.append(int(tk))
        elif tk == "+":
            val = s.pop() + s.pop()
            s.append(val)
        elif tk == "*":
            val = s.pop() * s.pop()
            s.append(val)
        elif tk == "-":
            b = s.pop()
            a = s.pop()
            val = a - b
            s.append(val)
        elif tk == "/":
            b = s.pop()
            a = s.pop()
            val = int(a / b)
            s.append(val)            
    return s[-1]

print(evalRPN(["2", "1", "+", "3", "*"]))
print(evalRPN(["4", "13", "5", "/", "+"]))

def test():
    assert evalRPN(["2", "1", "+", "3", "*"]) == 9          # simple add and multiply
    assert evalRPN(["4", "13", "5", "/", "+"]) == 6        # division truncation
    assert evalRPN(["10","6","9","3","+","-11","*","/","17","+","5","+"]) == 22 # complex
    assert evalRPN(["3", "-4", "+"]) == -1                  # negative result
    assert evalRPN(["-128", "128", "/"]) == -1              # division of negatives
    assert evalRPN(["10", "3", "/"]) == 3                   # positive truncation
    assert evalRPN(["-10", "3", "/"]) == -3                 # negative truncation (toward zero)
    assert evalRPN(["4", "3", "-"]) == 1                    # subtraction order
    assert evalRPN(["18"]) == 18                            # single token
    assert evalRPN(["4", "13", "5", "*", "+"]) == 69        # order of operations check
    print('All Pass!')

test()
```

```python
def solution(tokens):
    ops = set(['+', '-', '*', '/'])
    stack = []
    
    for tk in tokens:
        if not tk in ops:
            stack.append(int(tk))
        else:
            b = stack.pop()
            a = stack.pop()
            if tk == "+":
                stack.append(a+b)
            elif tk == "-":
                stack.append(a-b)
            elif tk == "*":
                stack.append(a*b)    
            #The division between two integers always truncates toward zero.
            elif tk == "/":
                stack.append(int(a/b))    

    return stack[-1]


assert solution(["2","1","+","3","*"])                                          == 9
assert solution(["4","13","5","/","+"])                                         == 6
assert solution(["10","6","9","3","+","-11","*","/","17","+","5","+"])         == 22
assert solution(["3","4","-"])                                                  == -1
assert solution(["6","2","/"])                                                  == 3
print('All Pass!')
```
