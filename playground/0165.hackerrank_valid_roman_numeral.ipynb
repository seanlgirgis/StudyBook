# 0165 HackerRank - Valid Roman Numeral

Given a string, validate whether it is a valid Roman numeral.
Print `True` if valid, otherwise print `False`.


## Regex Pattern

```python
regex_pattern = r"^(?=.+)M{0,3}(CM|CD|D?C{0,3})(XC|XL|L?X{0,3})(IX|IV|V?I{0,3})$"
```


````python
import re

regex_pattern = r"^(?=.+)M{0,3}(CM|CD|D?C{0,3})(XC|XL|L?X{0,3})(IX|IV|V?I{0,3})$"

roman = input("Enter Roman numeral: ").strip().upper()

is_valid = bool(re.match(regex_pattern, roman))
print("Valid Roman numeral:", is_valid)

if is_valid:
    values = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000,
    }

    total = 0

    for i in range(len(roman)):
        current_value = values[roman[i]]

        if i + 1 < len(roman):
            next_value = values[roman[i + 1]]
        else:
            next_value = 0

        if current_value < next_value:
            total -= current_value
        else:
            total += current_value

    print("Decimal value:", total)
else:
    print("Cannot convert because the Roman numeral is invalid.")
# MMCM => 2900
````
