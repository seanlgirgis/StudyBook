## Best Time to Buy and Sell Stock with Cooldown

After selling, you must wait one day before buying again. Maximize profit.

- `[1,2,3,0,2]` → `3` (buy@1, sell@3, cooldown, buy@0, sell@2)
- `[1]` → `0`

**DP states per day:** `held` (holding stock), `sold` (just sold today), `idle` (not holding, not just sold).

```python
%run _timer.py
```

```python
def solution(prices):
    # held  = max profit when holding stock
    # sold  = max profit on day we just sold
    # idle  = max profit when resting (not holding, didn't just sell)
    held  = float('-inf')
    sold  = 0
    idle  = 0

    for price in prices:
        prev_held, prev_sold, prev_idle = held, sold, idle
        held  = max(prev_held, prev_idle - price)  # hold: keep holding OR buy (only from idle)
        sold  = prev_held + price                  # sell: was holding, sell today
        idle  = max(prev_idle, prev_sold)          # idle: was idle OR came off cooldown

    return max(sold, idle)


assert solution([1,2,3,0,2]) == 3
assert solution([1])         == 0
assert solution([1,2])       == 1
assert solution([2,1])       == 0
print('All Pass!')
```
