## Terrain Analysis

You are analyzing elevation data collected by a drone flying over a landscape. The readings are stored as an array of integers `terrain`, where each value represents the elevation at that point.

A point is called a **peak** if it is strictly higher than both its immediate neighbors. A point is called a **valley** if it is strictly lower than both its immediate neighbors. All other points are considered **flat**.

Your task is to count the total number of peaks and valleys in the terrain.

**Note:** Only interior points (not the first or last) can be peaks or valleys.

### Example

- For `terrain = [1, 3, 2, 4, 1]`, the output should be `solution(terrain) = 3`.

  **Explanation:**
  - `(1, 3, 2)` → `1 < 3 > 2` → **peak**
  - `(3, 2, 4)` → `3 > 2 < 4` → **valley**
  - `(2, 4, 1)` → `2 < 4 > 1` → **peak**
  - Total: 3

- For `terrain = [1, 2, 3, 4]`, the output should be `solution(terrain) = 0`.

  **Explanation:**
  All points are increasing — no peaks or valleys.

- For `terrain = [1, 2, 1, 2, 1]`, the output should be `solution(terrain) = 3`.

  **Explanation:**
  - `(1, 2, 1)` → **peak**
  - `(2, 1, 2)` → **valley**
  - `(1, 2, 1)` → **peak**
  - Total: 3

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `terrain` — An array of integers representing elevation readings.
  Guaranteed constraints: `3 ≤ terrain.length ≤ 100`, `1 ≤ terrain[i] ≤ 10^9`
- **[output]** `integer` — The total number of peaks and valleys.

```python
def solution(terrain):
    t = terrain
    n = len(t)
    def peakOrVally(end):
        if t[end] < t[end-1] > t[end-2] or t[end] > t[end-1] < t[end-2] :
            return 1
        return 0
    survey = []
    for i in range(2, n):
        survey.append(peakOrVally(i))
    return sum(survey)
    


assert solution([1, 3, 2, 4, 1])          == 3  # given example 1 — peak, valley, peak
assert solution([1, 2, 3, 4])             == 0  # given example 2 — all increasing
assert solution([1, 2, 1, 2, 1])          == 3  # given example 3 — alternating
assert solution([1, 3, 1])                == 1  # single peak
assert solution([3, 1, 3])                == 1  # single valley
assert solution([5, 5, 5])                == 0  # all equal — no peaks or valleys
assert solution([4, 3, 2, 1])             == 0  # all decreasing
assert solution([1, 2, 1, 3, 4])          == 2  # peak at 2, valley at 1
print('All Pass!')
```
