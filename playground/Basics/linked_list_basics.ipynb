# Linked List Basics
A reference notebook — 10 fundamental singly-linked-list operations.

```python
# 1 — Node class
class Node:
    def __init__(self, val):
        self.val  = val
        self.next = None

# helpers used throughout
def to_ll(lst):
    dummy = Node(0)
    cur = dummy
    for v in lst:
        cur.next = Node(v)
        cur = cur.next
    return dummy.next

def to_list(head):
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out
```

```python
# 2 — Traverse (O(n))
def traverse(head):
    cur = head
    while cur:
        print(cur.val, end=' -> ')
        cur = cur.next
    print('None')

traverse(to_ll([1,2,3,4,5]))
```

```python
# 3 — Insert at head (O(1))
def insert_head(head, val):
    node = Node(val)
    node.next = head
    return node

assert to_list(insert_head(to_ll([2,3]), 1)) == [1,2,3]
```

```python
# 4 — Insert at tail (O(n))
def insert_tail(head, val):
    node = Node(val)
    if not head:
        return node
    cur = head
    while cur.next:
        cur = cur.next
    cur.next = node
    return head

assert to_list(insert_tail(to_ll([1,2]), 3)) == [1,2,3]
```

```python
# 5 — Insert after a target value (O(n))
def insert_after(head, target, val):
    cur = head
    while cur:
        if cur.val == target:
            node = Node(val)
            node.next = cur.next
            cur.next = node
            return head
        cur = cur.next
    return head   # target not found — unchanged

assert to_list(insert_after(to_ll([1,2,4]), 2, 3)) == [1,2,3,4]
```

```python
# 6 — Delete a node by value (O(n))
#
# We use a dummy node whose .next points to the real head.
# This means even if we delete the head node, the code path
# is identical — no special 'if target is head' branch needed.
#
# We walk with 'cur' and look one step ahead at cur.next.
# When cur.next.val == target, we skip that node by rewiring:
#   cur.next = cur.next.next
# We return dummy.next (not dummy) — the real first node.

def delete_val(head, val):
    dummy = Node(0)
    dummy.next = head
    cur = dummy
    while cur.next:
        if cur.next.val == val:
            cur.next = cur.next.next   # skip the target node
            return dummy.next
        cur = cur.next
    return dummy.next

assert to_list(delete_val(to_ll([1,2,3]),   2)) == [1,3]     # middle
assert to_list(delete_val(to_ll([1,2,3]),   1)) == [2,3]     # head
assert to_list(delete_val(to_ll([1,2,3]),   3)) == [1,2]     # tail
assert to_list(delete_val(to_ll([1,2,3]),  99)) == [1,2,3]   # not found
```

```python
# 7 — Reverse (O(n))
#
# Three pointers: prev, curr, nxt.
# Each step:
#   1. Save nxt = curr.next  (before we overwrite curr.next)
#   2. Flip  curr.next = prev
#   3. Advance: prev = curr, curr = nxt
#
# Loop ends when curr is None (walked off the end).
# At that point prev is the new head — return it.
#
# Common bug: 'while curr.next' stops one node early.
# Common bug: 'return curr' returns None at the end.

def reverse(head):
    prev = None
    curr = head
    while curr:
        nxt       = curr.next
        curr.next = prev
        prev      = curr
        curr      = nxt
    return prev

assert to_list(reverse(to_ll([1,2,3,4,5]))) == [5,4,3,2,1]
assert to_list(reverse(to_ll([1])))          == [1]
assert reverse(None)                         is None
```

```python
# 8 — Length (O(n))
def length(head):
    count = 0
    while head:
        count += 1
        head = head.next
    return count

assert length(to_ll([1,2,3])) == 3
assert length(None)           == 0
```

```python
# 9 — Find: return node with target value, or None (O(n))
def find(head, val):
    while head:
        if head.val == val:
            return head
        head = head.next
    return None

ll = to_ll([1,2,3])
assert find(ll, 2).val == 2
assert find(ll, 9)     is None
```

```python
# 10 — Delete at index (O(n))  — 0-based
def delete_at_index(head, idx):
    dummy = Node(0)
    dummy.next = head
    cur = dummy
    for _ in range(idx):
        if not cur.next:
            return dummy.next   # index out of range — unchanged
        cur = cur.next
    if cur.next:
        cur.next = cur.next.next
    return dummy.next

assert to_list(delete_at_index(to_ll([1,2,3]), 0)) == [2,3]   # head
assert to_list(delete_at_index(to_ll([1,2,3]), 1)) == [1,3]   # middle
assert to_list(delete_at_index(to_ll([1,2,3]), 2)) == [1,2]   # tail
print('All operations verified.')
```
