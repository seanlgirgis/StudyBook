# Monotonic Stack — Cheat Sheet

A **monotonic stack** is a stack that stays sorted (increasing or decreasing) as you push elements.

The key idea: when a new element **breaks** the order, pop until the stack is valid again. Each pop is an **answer event** — the popped element has found its "next greater" or "next smaller".

---

## When to use

| Pattern | Use |
|---------|-----|
| Next Greater Element | Monotonic decreasing stack (pop when current > top) |
| Next Smaller Element | Monotonic increasing stack (pop when current < top) |
| Previous Greater | Same stack, answer is what's left below when you push |
| Daily Temperatures | Next greater — store indices, answer = i - stack_top |
| Largest Rectangle | Next smaller on both sides |

---

## The Template

```python
stack = []  # stores indices
for i, val in enumerate(nums):
    while stack and val > nums[stack[-1]]:   # adjust condition per problem
        idx = stack.pop()
        # idx has found its answer — record it here
    stack.append(i)
```

**Decreasing stack** (pop when `val > top`) → finds **next greater**  
**Increasing stack** (pop when `val < top`) → finds **next smaller**

````python
# ── NEXT GREATER ELEMENT ──────────────────────────────────────────────────────
# For each element, find the first element to its RIGHT that is greater.
# If none exists, answer is -1.
#
# nums = [2, 1, 2, 4, 3]  →  [4, 2, 4, -1, -1]

def next_greater(nums):
    n = len(nums)
    ans = [-1] * n
    stack = []              # stores indices, values decrease top→bottom

    for i, val in enumerate(nums):
        # current val is greater than top → top has found its answer
        while stack and val > nums[stack[-1]]:
            idx = stack.pop()
            ans[idx] = val
        stack.append(i)
    # anything left in stack has no greater element → stays -1
    return ans

print(next_greater([2, 1, 2, 4, 3]))   # [4, 2, 4, -1, -1]
print(next_greater([1, 3, 2, 4]))      # [3, 4, 4, -1]
````

````python
# ── DAILY TEMPERATURES ────────────────────────────────────────────────────────
# Same pattern — answer is the DISTANCE (index gap), not the value.
#
# temps = [73,74,75,71,69,72,76,73]  →  [1,1,4,2,1,1,0,0]

def daily_temperatures(temps):
    n = len(temps)
    ans = [0] * n
    stack = []              # stores indices

    for i, val in enumerate(temps):
        while stack and val > temps[stack[-1]]:
            idx = stack.pop()
            ans[idx] = i - idx      # distance, not value
        stack.append(i)
    return ans

print(daily_temperatures([73,74,75,71,69,72,76,73]))  # [1,1,4,2,1,1,0,0]
print(daily_temperatures([30,60,90]))                  # [1,1,0]
````

````python
# ── PREVIOUS GREATER ELEMENT ──────────────────────────────────────────────────
# For each element, find the first element to its LEFT that is greater.
# Answer lives in what's already in the stack when you push.
#
# nums = [4, 5, 2, 10, 8]  →  [-1, -1, 5, -1, 10]

def previous_greater(nums):
    ans = []
    stack = []              # values (not indices needed here)

    for val in nums:
        while stack and val >= stack[-1]:   # pop smaller-or-equal
            stack.pop()
        ans.append(stack[-1] if stack else -1)   # top is the previous greater
        stack.append(val)
    return ans

print(previous_greater([4, 5, 2, 10, 8]))   # [-1, -1, 5, -1, 10]
print(previous_greater([1, 2, 3, 4]))        # [-1, -1, -1, -1]
````

````python
# ── VISUAL TRACE: next_greater([2, 1, 2, 4, 3]) ───────────────────────────────
#
# i=0  val=2   stack=[]         push 0        stack=[0]       ans=[-1,-1,-1,-1,-1]
# i=1  val=1   1 < nums[0]=2    push 1        stack=[0,1]     ans=[-1,-1,-1,-1,-1]
# i=2  val=2   2 > nums[1]=1 → pop 1, ans[1]=2               stack=[0]
#              2 = nums[0]=2    stop, push 2  stack=[0,2]     ans=[-1, 2,-1,-1,-1]
# i=3  val=4   4 > nums[2]=2 → pop 2, ans[2]=4               stack=[0]
#              4 > nums[0]=2 → pop 0, ans[0]=4               stack=[]
#              push 3            stack=[3]                    ans=[ 4, 2, 4,-1,-1]
# i=4  val=3   3 < nums[3]=4    push 4        stack=[3,4]     ans=[ 4, 2, 4,-1,-1]
# end: stack=[3,4] → no answer → stay -1
# final: [4, 2, 4, -1, -1]  ✓

print('Trace confirms:', next_greater([2, 1, 2, 4, 3]))
````
