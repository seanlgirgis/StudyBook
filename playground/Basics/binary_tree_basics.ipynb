# Binary Tree Basics
A reference notebook — node class, builders, and the core traversal patterns every tree problem is built from.

```python
# 1 — TreeNode class + helpers
#
# make_tree builds a tree from a level-order list (same format LeetCode uses).
# None in the list means no node at that position.
# Example: [3,9,20,None,None,15,7] builds:
#       3
#      / \
#     9  20
#        / \
#       15   7

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val   = val
        self.left  = left
        self.right = right

def make_tree(vals):
    if not vals or vals[0] is None:
        return None
    root  = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root
```

```python
# 2 — Inorder traversal: LEFT → ROOT → RIGHT  (O(n))
#
# For a BST this yields nodes in sorted ascending order.
# Pattern: recurse left, visit, recurse right.

def inorder(root):
    if not root:
        return []
    return inorder(root.left) + [root.val] + inorder(root.right)

t = make_tree([4,2,6,1,3,5,7])
assert inorder(t) == [1,2,3,4,5,6,7]   # sorted — it's a BST
print('inorder OK')
```

```python
# 3 — Preorder traversal: ROOT → LEFT → RIGHT  (O(n))
#
# Root is visited first — useful for copying or serializing a tree.

def preorder(root):
    if not root:
        return []
    return [root.val] + preorder(root.left) + preorder(root.right)

t = make_tree([1,2,3,4,5])
assert preorder(t) == [1,2,4,5,3]
print('preorder OK')
```

```python
# 4 — Postorder traversal: LEFT → RIGHT → ROOT  (O(n))
#
# Root is visited last — useful when children must be processed before parent
# (e.g. deleting a tree, computing subtree sizes).

def postorder(root):
    if not root:
        return []
    return postorder(root.left) + postorder(root.right) + [root.val]

t = make_tree([1,2,3,4,5])
assert postorder(t) == [4,5,2,3,1]
print('postorder OK')
```

```python
# 5 — Level-order BFS: returns list of lists, one per level  (O(n))
#
# Use a deque. Each iteration processes all nodes at the current level
# by snapshotting len(queue) before the inner loop.
# This is the foundation for any problem that asks "per level" questions.

def level_order(root):
    if not root:
        return []
    result = []
    queue  = deque([root])
    while queue:
        level = []
        for _ in range(len(queue)):   # snapshot size = nodes at this level
            node = queue.popleft()
            level.append(node.val)
            if node.left:  queue.append(node.left)
            if node.right: queue.append(node.right)
        result.append(level)
    return result

t = make_tree([3,9,20,None,None,15,7])
assert level_order(t) == [[3],[9,20],[15,7]]
print('level_order OK')
```

```python
# 6 — Height / max depth  (O(n))
#
# Base case: None → 0.
# Recursive case: 1 + max(height(left), height(right)).

def height(root):
    if not root:
        return 0
    return 1 + max(height(root.left), height(root.right))

assert height(make_tree([3,9,20,None,None,15,7])) == 3
assert height(make_tree([1]))                     == 1
assert height(None)                               == 0
print('height OK')
```

```python
# 7 — Count nodes  (O(n))

def count_nodes(root):
    if not root:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

assert count_nodes(make_tree([3,9,20,None,None,15,7])) == 5
assert count_nodes(None)                               == 0
print('count_nodes OK')
```

```python
# 8 — Search: find a node by value  (O(n) general tree)
#
# Returns the node if found, None otherwise.
# For a BST you can prune left or right — but this works for any tree.

def search(root, val):
    if not root:
        return None
    if root.val == val:
        return root
    return search(root.left, val) or search(root.right, val)

t = make_tree([3,9,20,None,None,15,7])
assert search(t, 15).val == 15
assert search(t, 99)     is None
print('search OK')
```

```python
# 9 — BST insert  (O(h), h = height)
#
# In a BST: smaller values go left, larger go right.
# Recurse until you hit None, then attach the new node.
# Returns the (possibly new) root.

def bst_insert(root, val):
    if not root:
        return TreeNode(val)
    if val < root.val:
        root.left  = bst_insert(root.left,  val)
    else:
        root.right = bst_insert(root.right, val)
    return root

t = make_tree([4,2,6])
bst_insert(t, 3)
assert inorder(t) == [2,3,4,6]   # still sorted after insert
print('bst_insert OK')
```

```python
# 10 — BFS with (node, depth) tuple
#
# Store depth alongside each node in the queue.
# Same pattern as rotting_oranges — carry extra state in the tuple.
# Use this when you need per-node depth without recursion.

def max_depth_bfs(root):
    if not root:
        return 0
    queue     = deque([(root, 1)])
    max_depth = 0
    while queue:
        node, depth = queue.popleft()
        max_depth   = max(max_depth, depth)
        if node.left:  queue.append((node.left,  depth + 1))
        if node.right: queue.append((node.right, depth + 1))
    return max_depth

assert max_depth_bfs(make_tree([3,9,20,None,None,15,7])) == 3
assert max_depth_bfs(make_tree([1]))                     == 1
assert max_depth_bfs(None)                               == 0
print('max_depth_bfs OK')
print('All operations verified.')
```
