## Core Identity

If `pref[i]` is the sum of `nums[0]..nums[i]`, then:

- `sum(nums[l..r]) = pref[r] - pref[l-1]` when `l > 0`
- `sum(nums[0..r]) = pref[r]` when `l == 0`

So we pay O(n) once to build `pref`, then each range query is O(1).

```python
# 1 — Build prefix sum array (O(n))
def build_prefix(nums):
    pref = []
    run = 0
    for x in nums:
        run += x
        pref.append(run)
    return pref

assert build_prefix([1,2,3,4]) == [1,3,6,10]
assert build_prefix([5])       == [5]
assert build_prefix([-1,2,-3]) == [-1,1,-2]
```

```python
# 2 — Range sum query using prefix array (O(1) per query)
def range_sum(pref, l, r):
    if l == 0:
        return pref[r]
    return pref[r] - pref[l - 1]

nums = [3, 1, 4, 1, 5]
pref = build_prefix(nums)

assert range_sum(pref, 0, 2) == 8   # 3+1+4
assert range_sum(pref, 2, 4) == 10  # 4+1+5
assert range_sum(pref, 1, 3) == 6   # 1+4+1
```

```python
# 3 — Validate prefix query matches naive slicing
def naive_sum(nums, l, r):
    return sum(nums[l:r+1])

nums = [2, 7, 1, 8, 2, 8]
pref = build_prefix(nums)

for l in range(len(nums)):
    for r in range(l, len(nums)):
        assert naive_sum(nums, l, r) == range_sum(pref, l, r)

print('naive_sum == prefix_sum for all checked ranges')
```

```python
# 4 — Quick visual for manual reasoning
nums = [1, 2, 1, 2, 1]
pref = build_prefix(nums)

print('nums:', nums)
print('pref:', pref)

l, r = 1, 3
window_sum = range_sum(pref, l, r)
print(f'sum(nums[{l}:{r+1}]) = pref[{r}] - pref[{l-1}] = {window_sum}')
```

```python
# 5 — Mini practice checks
nums = [10, -3, 4, -2, 1]
pref = build_prefix(nums)

assert pref == [10, 7, 11, 9, 10]
assert range_sum(pref, 0, 4) == 10
assert range_sum(pref, 1, 3) == -1
assert range_sum(pref, 2, 2) == 4

print('All Prefix Basics Checks Passed')
```
