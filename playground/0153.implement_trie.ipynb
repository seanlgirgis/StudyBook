## Implement Trie (LC 208)

Build a prefix tree supporting `insert`, `search`, and `startsWith`.

**Pattern:** Each node = dict of children + `is_end` flag.
- Walk down character by character; create nodes as needed
- `search` → walk to end, check `is_end`
- `startsWith` → just walk to end (don't check `is_end`)

```python
class TrieNode:
    def __init__(self):
        self.children = {}   # char → TrieNode
        self.is_end   = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for c in word:
            if c not in node.children:
                node.children[c] = TrieNode()   # create missing node
            node = node.children[c]
        node.is_end = True

    def search(self, word):
        node = self.root
        for c in word:
            if c not in node.children:
                return False
            node = node.children[c]
        return node.is_end              # must be a complete inserted word

    def startsWith(self, prefix):
        node = self.root
        for c in prefix:
            if c not in node.children:
                return False
            node = node.children[c]
        return True                     # just needs the path to exist


t = Trie()
t.insert("apple")
assert t.search("apple")   == True
assert t.search("app")     == False    # not inserted as a word
assert t.startsWith("app") == True
t.insert("app")
assert t.search("app")     == True
assert t.startsWith("xyz") == False
print('All Pass!')
```
