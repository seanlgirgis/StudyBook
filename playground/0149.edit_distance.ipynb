## Edit Distance (LC 72)

Min operations (insert / delete / replace) to convert `word1` → `word2`.

**Pattern:** 2D DP — same family as Unique Paths and LCS.

```
dp[i][j] = edit distance for word1[:i] → word2[:j]

if word1[i-1] == word2[j-1]:  dp[i][j] = dp[i-1][j-1]        (free — chars match)
else:                          dp[i][j] = 1 + min(
                                   dp[i-1][j-1],   # replace
                                   dp[i-1][j],     # delete from word1
                                   dp[i][j-1]      # insert into word1
                               )
```

Base cases: `dp[i][0] = i` (delete all), `dp[0][j] = j` (insert all).

````python
def solution(word1, word2):
    m, n = len(word1), len(word2)
    # dp[i][j] = edit dist between word1[:i] and word2[:j]
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1): dp[i][0] = i   # delete all of word1
    for j in range(n + 1): dp[0][j] = j   # insert all of word2

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i-1] == word2[j-1]:
                dp[i][j] = dp[i-1][j-1]                           # chars match — free
            else:
                dp[i][j] = 1 + min(dp[i-1][j-1],   # replace
                                   dp[i-1][j],      # delete
                                   dp[i][j-1])      # insert

    return dp[m][n]


assert solution("horse",     "ros")       == 3
assert solution("intention", "execution") == 5
assert solution("",          "")          == 0
assert solution("",          "abc")       == 3
assert solution("abc",       "")          == 3
assert solution("abc",       "abc")       == 0
assert solution("a",         "b")         == 1
print('All Pass!')
````
