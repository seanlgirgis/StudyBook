## Next Greater Element

You are given two arrays `nums1` and `nums2` where `nums1` is a subset of `nums2`. For each element in `nums1`, find its **next greater element** in `nums2` (the first element to its right in `nums2` that is greater). If none exists, return `-1`.

### Example

- `nums1=[4,1,2]`, `nums2=[1,3,4,2]` → `[-1,3,-1]`
  - 4 → no greater element to its right in nums2 → -1
  - 1 → next greater in nums2 is 3 → 3
  - 2 → no greater element to its right in nums2 → -1
- `nums1=[2,4]`, `nums2=[1,2,3,4]` → `[3,-1]`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums1`, `array.integer` `nums2`
- **[output]** `array.integer`

```python
def solution(nums1, nums2):
    lookup = {val: i for i, val in enumerate(nums1)}
    out = [-1] * len(nums1)
    stack = []

    for val in nums2:
        while stack and stack[-1] < val:
            popped_val = stack.pop()
            if popped_val in lookup:
                out[lookup[popped_val]] = val
        stack.append(val)
    return out

assert solution([4,1,2], [1,3,4,2])   == [-1,3,-1]
assert solution([2,4],   [1,2,3,4])   == [3,-1]
assert solution([1],     [1,2])       == [2]
assert solution([3],     [1,2,3])     == [-1]
print('solution pass')
```

```python
def solution(nums1, nums2):
    # Mapping of value to its index in nums1 for O(1) updates
    lookup = {val: i for i, val in enumerate(nums1)}  
    out = [-1] * len(nums1)
    stack = []  

    for val in nums2:
        while stack and stack[-1] < val:
            popped_val = stack.pop()
            # If the popped value is one we care about (in nums1)
            if popped_val in lookup:
                out[lookup[popped_val]] = val
        
        stack.append(val)
        
    return out
assert solution([4,1,2], [1,3,4,2])   == [-1,3,-1]
assert solution([2,4],   [1,2,3,4])   == [3,-1]
assert solution([1],     [1,2])       == [2]
assert solution([3],     [1,2,3])     == [-1]
print('All Pass!')
```
