## Most Frequent Element

Given an array of integers `nums`, return the element that appears **most frequently**. If there is a tie, return the **smallest** of the tied elements.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.

### Example

- For `nums = [1,3,2,1,3,1]`, the output should be `solution(nums) = 1`.

  **Explanation:** `1` appears 3 times, `3` appears 2 times, `2` appears 1 time. `1` wins.

- For `nums = [1,2,2,3,3]`, the output should be `solution(nums) = 2`.

  **Explanation:** `2` and `3` both appear twice — tie broken by smallest value, so `2`.

- For `nums = [5]`, the output should be `solution(nums) = 5`.

  **Explanation:** Single element — trivially the most frequent.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^5`, `-10^9 ≤ nums[i] ≤ 10^9`
- **[output]** `integer` — The most frequent element, smallest on tie.

```python
import heapq
from collections import Counter
def solution(nums):
    counts = Counter(nums)
    heap = []
    for num, freq in counts.items():
        heapq.heappush(heap, (-freq, num))
    return(heap[0][1])

assert solution([1,3,2,1,3,1])     == 1   # given example 1 — 1 appears 3x
assert solution([1,2,2,3,3])       == 2   # given example 2 — tie, smallest wins
assert solution([5])               == 5   # given example 3 — single element
assert solution([1,2,3])           == 1   # all frequency 1 — smallest wins
assert solution([3,3,2,2,1])       == 2   # tie between 2 and 3 — smallest wins
assert solution([-1,-1,2,2,3])     == -1  # tie, negative is smallest
assert solution([4,4,4,1,1])       == 4   # clear winner
assert solution([1,1,2,2,2,3,3])   == 2   # 2 appears 3x, clear winner
print('All Pass!')
```

```python
def solution(nums):
    counts = Counter(nums)
    return min(counts, key=lambda x: (-counts[x], x))

assert solution([1,3,2,1,3,1])     == 1   # given example 1 — 1 appears 3x
assert solution([1,2,2,3,3])       == 2   # given example 2 — tie, smallest wins
assert solution([5])               == 5   # given example 3 — single element
assert solution([1,2,3])           == 1   # all frequency 1 — smallest wins
assert solution([3,3,2,2,1])       == 2   # tie between 2 and 3 — smallest wins
assert solution([-1,-1,2,2,3])     == -1  # tie, negative is smallest
assert solution([4,4,4,1,1])       == 4   # clear winner
assert solution([1,1,2,2,2,3,3])   == 2   # 2 appears 3x, clear winner
print('All Pass!')
```
