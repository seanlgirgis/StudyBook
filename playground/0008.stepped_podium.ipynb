## Stepped Podium

You are building a ceremonial podium made of stone blocks. The podium has `n` sections, and for it to look correct, each section must be exactly **2 units** taller than the section before it (ascending), or exactly **2 units** shorter (descending).

You are given an array `structures` representing the current height of each section. You can only **add** blocks — you cannot remove them. Each block added counts as one operation.

Your goal is to find the **minimum number of operations** needed to make the podium either fully ascending (each section +2 from the previous) or fully descending (each section -2 from the previous) — whichever is cheaper.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(structures.length)** will fit within the execution time limit.

### Example

- For `structures = [1, 2, 5]`, the output should be `solution(structures) = 1`.

  **Explanation:**
  - Target ascending pattern: `[1, 3, 5]` — add 1 block to the middle section.
  - Target descending pattern: `[7, 5, 3]` — costs 6+3+0 = 9 operations.
  - Minimum is **1**.

- For `structures = [5, 4, 3]`, the output should be `solution(structures) = 3`.

  **Explanation:**
  - Target ascending pattern: `[5, 7, 9]` — costs 0+3+6 = 9.
  - Target descending pattern: `[7, 5, 3]` — costs 2+1+0 = 3.
  - Minimum is **3**.

- For `structures = [1, 3, 5]`, the output should be `solution(structures) = 0`.

  **Explanation:**
  Already a perfect ascending podium — no operations needed.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `structures` — Current heights of the podium sections.
  Guaranteed constraints: `2 ≤ structures.length ≤ 10^5`, `1 ≤ structures[i] ≤ 10^9`
- **[output]** `integer64` — Minimum number of block additions needed.

```python
def solution(structures):
    A = structures
    n = len(A)
    M1 = [ i * 2 for i in range(n)]
    M2 = [ i * 2 for i in range(n-1, -1, -1)]
    def cost(A, M):
        offset  = max(a-m  for a, m in zip(A,M))
        cost  = sum(offset + m - a for a, m in zip(A,M))
        return cost
    return min(cost(A, M1), cost(A, M2))
    
assert solution([1, 3, 5])    == 0  # already ascending by 2
assert solution([5, 3, 1])    == 0  # already descending by 2
assert solution([1, 2, 5])    == 1  # add 1 to middle → [1, 3, 5]
assert solution([5, 4, 3])    == 3  # add 2+1+0 → [7, 5, 3]
assert solution([3, 5])       == 0  # two sections, already ascending by 2
assert solution([5, 3])       == 0  # two sections, already descending by 2
assert solution([1, 2])       == 1  # add 1 to second → [1, 3]
assert solution([1, 1])       == 2  # add 2 to second → [1, 3] or add 2 to first → [3, 1]
print('All Pass!') 

```
