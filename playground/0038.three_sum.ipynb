## Three Sum

Given an array of integers `nums`, return all **unique triplets** `[a, b, c]` such that `a + b + c = 0`.

The solution set must not contain duplicate triplets.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [-1,0,1,2,-1,-4]`, the output should be `solution(nums) = [[-1,-1,2],[-1,0,1]]` (any order).

  **Explanation:**
  - `-1 + (-1) + 2 = 0`
  - `-1 + 0 + 1 = 0`

- For `nums = [0,1,1]`, the output should be `solution(nums) = []`.

  **Explanation:** No three numbers sum to zero.

- For `nums = [0,0,0]`, the output should be `solution(nums) = [[0,0,0]]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `3 ≤ nums.length ≤ 3000`, `-10^5 ≤ nums[i] ≤ 10^5`
- **[output]** `array.array.integer` — Unique triplets summing to zero.

```python
def solution(nums):
    nums.sort()
    n = len(nums)
    ret = set()
    for i, x in enumerate(nums):
        seen = set()
        target = -x
        for num in nums[i+1:]:
            if (target - num) in seen:
                ret.add((x,(target - num), num ))
            seen.add(num)
    return [list(tpl) for tpl in ret]

        


def check(result, expected):
    return sorted([sorted(t) for t in result]) == sorted([sorted(t) for t in expected])

assert check(solution([-1,0,1,2,-1,-4]),  [[-1,-1,2],[-1,0,1]])  # given example 1
assert check(solution([0,1,1]),            [])                    # given example 2 — no triplets
assert check(solution([0,0,0]),            [[0,0,0]])             # given example 3 — all zeros
assert check(solution([-2,0,2]),           [[-2,0,2]])            # single triplet
assert check(solution([-4,-1,-1,0,1,2]),   [[-1,-1,2],[-1,0,1]]) # same as ex1 with extra
assert check(solution([1,2,3]),            [])                    # all positive — no solution
assert check(solution([-1,-1,-1,2,2]),     [[-1,-1,2]])           # duplicates in input
assert check(solution([-2,-1,0,1,2,3]),    [[-2,-1,3],[-2,0,2],[-1,0,1]]) # multiple triplets
print('All Pass!')
```
