# 0046 Majority Element most_common 

```python
from collections import Counter
def solution(nums):
    counts = Counter(nums)
    return counts.most_common()[-1][0]
solution([3,2,3])   
```

## 0099 find peak element

```python
def localPeak(nums):
    if len(nums) == 1:
        return 0 
    if len(nums) == 2:
        return 0 if nums[0] > nums[1] else (1)
    for i in range(1, len(nums) - 1):
        if nums[i] > nums[i-1] and nums[i] > nums[i+1]:
            return i 
            
localPeak([1,2,1,3,5,6,4])
```

# 0067 Jump Game

```python
#0067
def jumpGame(nums):
    max_reach = 0
    for i, jump in enumerate(nums):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + jump)
    return True
        
jumpGame([3,2,2,0,4]) 
```

# 0017 Kadane's Algorithm — maximum subarray sum

## Mental model:

### At each index, decide:
###       extend previous subarray
###       start fresh at current number
This method used when there are negatives in the list

```python
def max_subarray(nums):
    cursum = best = nums[0]
    
    for num in nums[1:]:
        # Either extend the current subarray or start fresh here
        cursum = max(num, cursum + num)
        
        # Track the best sum seen so far
        best = max(best, cursum)
    
    return best


assert max_subarray([-2,1,-3,4,-1,2,1,-5,4]) == 6   # [4,-1,2,1]
assert max_subarray([1])                       == 1
assert max_subarray([5,4,-1,7,8])              == 23  # whole array
assert max_subarray([-1,-2,-3])                == -1  # least negative
print('All Pass!')
```

## maximum subarray sum  with indexes

```python
def max_subarray_with_indices(nums):
    cur_sum = best_sum = nums[0]
    cur_l = best_l = best_r = 0

    for i in range(1, len(nums)):
        num = nums[i]

        # Start new window at i, or extend existing one
        if num > cur_sum + num:
            cur_sum = num
            cur_l = i
        else:
            cur_sum += num

        # Update best window
        if cur_sum > best_sum:
            best_sum = cur_sum
            best_l = cur_l
            best_r = i

    return best_sum, best_l, best_r


assert max_subarray_with_indices([-2,1,-3,4,-1,2,1,-5,4]) == (6, 3, 6)   # [4,-1,2,1]
assert max_subarray_with_indices([1])                      == (1, 0, 0)
assert max_subarray_with_indices([5,4,-1,7,8])             == (23, 0, 4)
assert max_subarray_with_indices([-1,-2,-3])               == (-1, 0, 0)
print("All Pass!")
```

# 0067	Jump Game . Max_Reach Keep max_reach = farthest index you can reach so far. if max_reach < index dead

```python
def jumpGame(nums):
    max_reach = 0

    for i in range(len(nums)):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + nums[i])
    return True
    
        

assert jumpGame([2,3,1,1,4])  == True   # given example 1
assert jumpGame([3,2,1,0,4])  == False  # given example 2 — stuck at index 3
assert jumpGame([0])          == True   # single element — already at end
assert jumpGame([1,0])        == True   # jump from 0 to end
assert jumpGame([0,1])        == False  # can't move from index 0
assert jumpGame([2,0,0])      == True   # jump over zeros
assert jumpGame([1,1,1,1,0])  == True   # reach last index exactly
assert jumpGame([1,0,1,0])    == False  # stuck at index 1
print('All Pass!')
```

## 0047  Best time to buy and sell later . no new transactions the same day

This approach keeps a running **minimum price** as the best day to buy so far.  
For each new day, it treats that day as a potential sell point and updates the best profit with `price - min_price`.

In one left-to-right pass, it captures the maximum possible profit from exactly one buy followed by one sell.  
If prices only decline, the profit correctly remains `0`.


```python
def solution(prices):
    profit = 0
    buy_price = prices[0]


    for price in prices[1:]:
        if price < buy_price:
            buy_price = price
            continue                     #no buy and sell the same day
        profit = max(profit, price - buy_price)
    return profit


assert solution([7,1,5,3,6,4])  == 5   # given example 1 — buy 1 sell 6
assert solution([7,6,4,3,1])    == 0   # given example 2 — declining, no profit
assert solution([1,2])          == 1   # two elements, profit exists
assert solution([2,1])          == 0   # two elements, no profit
assert solution([1])            == 0   # single element
assert solution([3,3,3,3])      == 0   # all same price
assert solution([1,2,3,4,5])    == 4   # steadily rising — buy first sell last
assert solution([2,4,1,7])      == 6   # dip then spike — buy at 1 sell at 7
print('All Pass!')    
```

# 001 - Multisort Tupe List

```python
from operator import itemgetter

def multisort_items(xs, specs):
    # specs: (index, reverse)  -> reverse=True means descending
    for idx, reverse in reversed(specs):
        xs.sort(key=itemgetter(idx), reverse=reverse)
    return xs

rows = [('dave', 'B', 10), ('jane', 'B', 12), ('john', 'A', 15)]

# 0 -> desc, 1 -> asc, 2 -> desc
sorted_rows = multisort_items(rows[:], ((0, True), (1, False), (2, True)))

print(sorted_rows)
# [('john', 'A', 15), ('jane', 'B', 12), ('dave', 'B', 10)]
```
