 # 0027 Binary Search
 [↑ Back to TOC](#binsearch)


```python
"""
id: lc_0704
title: Binary Search
source: leetcode
difficulty: easy
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/binary-search/
status: draft
last_updated: 2026-04-23
notes:
- key idea: classic two-pointer iterative binary search
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 704_lc_0704_binary_search_empty.py
# LeetCode 704: Binary Search
# Difficulty: Easy
#
# PROBLEM STATEMENT:
# Given a sorted array of integers nums and a target integer target, return 
# the index of target in the array. If target does not exist, return -1.
# You must write an algorithm with O(log n) runtime complexity.
#
# RULES:
# - nums is sorted in ascending order.
# - All integers in nums are unique.
#
# EXAMPLES:
# Input: nums = [-1,0,3,5,9,12], target = 9 -> Output: 4
# Input: nums = [-1,0,3,5,9,12], target = 2 -> Output: -1
# ============================================================================

from typing import List

def solution(nums, target):
    l , r  = 0 , len(nums) -1 

    while l <= r:
        mid = l+  (r -l )//2
        if nums[mid] == target:
            return mid
        if target > nums[mid]:
            l = mid + 1
        else:
            r = mid -1
    return -1





print(solution([-1,0,3,5,9,12], 9))
print(solution([-1,0,3,5,9,12], 2))

def test():
    assert solution([-1,0,3,5,9,12], 9) == 4   # given example 1 - target exists
    assert solution([-1,0,3,5,9,12], 2) == -1  # given example 2 - target missing
    assert solution([5], 5) == 0               # single element - match
    assert solution([5], 2) == -1              # single element - no match
    assert solution([2, 5], 2) == 0            # two elements - match left
    assert solution([2, 5], 5) == 1            # two elements - match right
    assert solution([2, 5], 3) == -1           # two elements - missing between
    assert solution([1, 3, 5, 7, 9], 1) == 0   # boundary - start
    assert solution([1, 3, 5, 7, 9], 9) == 4   # boundary - end
    assert solution([1, 3, 5, 7, 9], 5) == 2   # exact middle
    print('All Pass!')

test()
```

 # 0028	 Search Insert Position
 [↑ Back to TOC](#binsearch)


```python
"""
id: lc_0035
title: Search Insert Position
source: leetcode
difficulty: easy
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/search-insert-position/
status: draft
last_updated: 2026-04-23
notes:
- key idea: classic binary search (lower bound) to find insertion point
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 035_lc_0035_search_insert_position_empty.py
# LeetCode 35: Search Insert Position
# Difficulty: Easy
#
# PROBLEM STATEMENT:
# Given a sorted array of distinct integers nums and a target integer target, 
# return the index of target if it exists. Otherwise, return the index where 
# it would be inserted to keep the array sorted.
#
# RULES:
# - Must run in O(log n) time.
# - Array is sorted and contains distinct integers.
#
# EXAMPLES:
# Input: nums = [1,3,5,6], target = 5 -> Output: 2
# Input: nums = [1,3,5,6], target = 2 -> Output: 1
# Input: nums = [1,3,5,6], target = 7 -> Output: 4
# ============================================================================

from typing import List

def searchInsert(nums: List[int], target: int) -> int:

    l , r, n = 0 , len(nums) -1, len(nums)
    if target > max(nums):
        return n
    while l <= r:
        mid = l+  (r -l )//2
        if nums[mid] == target:
            return mid
        if target > nums[mid]:
            l = mid + 1
        else:
            r = mid -1
    return l


print(searchInsert([1, 3, 5, 6], 5))
print(searchInsert([1, 3, 5, 6], 2))

def test():
    assert searchInsert([1, 3, 5, 6], 5) == 2  # target exists at index 2
    assert searchInsert([1, 3, 5, 6], 2) == 1  # insert between 1 and 3
    
    assert searchInsert([1, 3, 5, 6], 0) == 0  # insert at the beginning
    assert searchInsert([1], 0) == 0           # single element smaller
    assert searchInsert([1], 2) == 1           # single element larger
    assert searchInsert([1, 3], 2) == 1        # even length insertion
    assert searchInsert([1, 3, 5], 4) == 2     # odd length insertion
    assert searchInsert([1, 3, 5, 10], 10) == 3 # target at the last index
    assert searchInsert([1, 3, 5, 10], 1) == 0  # target at the first index
    assert searchInsert([1, 3, 5, 6], 7) == 4  # insert at the end
    print('All Pass!')

test()
```

 # 0099 find peak

```python
"""
id: lc_0162
title: Find Peak Element
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/find-peak-element/
status: draft
last_updated: 2026-04-23
notes:
- key idea: binary search to find a local maximum
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 162_lc_0162_find_peak_element_empty.py
# LeetCode 162: Find Peak Element
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# A peak element is an element that is strictly greater than its neighbors.
# Given an integer array nums, find a peak element, and return its index. 
# If the array contains multiple peaks, return the index to any of the peaks.
#
# RULES:
# - You may imagine that nums[-1] = nums[n] = -∞.
# - You must write an algorithm that runs in O(log n) time.
#
# EXAMPLES:
# Input: nums = [1,2,3,1]
# Output: 2 (Index of 3)
#
# Input: nums = [1,2,1,3,5,6,4]
# Output: 1 or 5
# ============================================================================

from typing import List

def findPeakElement(nums: List[int]) -> int:
    l , r, n = 0 , len(nums)-1, len(nums)
#    if n == 1 : return 0
    while l < r:
        mid = l + (r - l) // 2
        if nums[mid] < nums[mid+1]:    #uphiull
            l = mid +1
        else:
            r = mid
    return l


print(findPeakElement([1, 2, 3, 1]))
print(findPeakElement([1, 2, 1, 3, 5, 6, 4]))

def test():
    assert findPeakElement([1, 2, 3, 1]) == 2  # given example 1
    assert findPeakElement([1, 2, 1, 3, 5, 6, 4]) in {1, 5}  # given example 2
    assert findPeakElement([1]) == 0  # single element
    assert findPeakElement([1, 2]) == 1  # strictly increasing
    assert findPeakElement([2, 1]) == 0  # strictly decreasing
    assert findPeakElement([1, 2, 3]) == 2  # peak at end
    assert findPeakElement([3, 2, 1]) == 0  # peak at start
    assert findPeakElement([1, 3, 2]) == 1  # peak in middle
    assert findPeakElement([1, 2, 3, 4, 5]) == 4  # linear increasing
    assert findPeakElement([5, 4, 3, 2, 1]) == 0  # linear decreasing
    print('All Pass!')

test()
```

 # 0075  Binary search a 2D Matrix
 [↑ Back to TOC](#binsearch)


```python
"""
id: lc_0074
title: Search a 2D Matrix
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search, matrix]
leetcode_url: https://leetcode.com/problems/search-a-2d-matrix/
status: draft
last_updated: 2026-04-23
notes:
- key idea: treat the 2D matrix as a flattened 1D sorted array
- time: O(log(m*n))
- space: O(1)
"""

# ============================================================================
# File: 074_lc_0074_search_2d_matrix_empty.py
# LeetCode 74: Search a 2D Matrix
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# You are given an m x n integer matrix with the following properties:
# 1. Each row is sorted in ascending order.
# 2. The first integer of each row is greater than the last integer of the previous row.
# Given an integer target, return True if it exists in the matrix, False otherwise.
#
# RULES:
# - Your solution must run in O(log(m * n)) time complexity.
#
# EXAMPLES:
# matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 3 -> True
# matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 13 -> False
# ============================================================================

from typing import List

def searchMatrix(matrix: List[List[int]], target: int) -> bool:
    if not matrix or not matrix[0]:
        return False
    #bianry search with a little twist
    rows, cols = len(matrix), len(matrix[0])
    n = rows * cols
    l, r = 0, n-1
    def ItemAt(pos):
        r = pos // cols
        c = pos % cols
        return matrix[r][c]
    while l <= r:
        mid = l + (r-l)//2
        if ItemAt(mid) == target:
            return True
        if target > ItemAt(mid):
            l = mid +1 
        else:
            r = mid -1
    return False
    
    

print(searchMatrix([[1,3,5,7],[10,11,16,20],[23,30,34,60]], 3))
print(searchMatrix([[1,3,5,7],[10,11,16,20],[23,30,34,60]], 13))

def test():
    matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
    assert searchMatrix(matrix, 3) == True    # given example 1
    assert searchMatrix(matrix, 13) == False  # given example 2
    assert searchMatrix(matrix, 1) == True    # start of matrix
    assert searchMatrix(matrix, 60) == True   # end of matrix
    assert searchMatrix(matrix, 0) == False   # smaller than min
    assert searchMatrix(matrix, 61) == False  # larger than max
    assert searchMatrix([[1]], 1) == True     # 1x1 matrix hit
    assert searchMatrix([[1]], 2) == False    # 1x1 matrix miss
    assert searchMatrix([[1, 3]], 3) == True  # 1-row matrix
    assert searchMatrix([[1], [3]], 3) == True # 1-column matrix
    print('All Pass!')

test()
```

 # 0105 Find Minimum in Rotated Sorted Array
 [↑ Back to TOC](#binsearch)


```python
"""
id: lc_0153
title: Find Minimum in Rotated Sorted Array
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/
status: draft
last_updated: 2026-04-23
notes:
- key idea: binary search for the inflection point
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 153_lc_0153_find_minimum_in_rotated_sorted_array_empty.py
# LeetCode 153: Find Minimum in Rotated Sorted Array
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Suppose an array of length n sorted in ascending order is rotated between 
# 1 and n times. Find the minimum element of this array.
#
# RULES:
# - You must write an algorithm that runs in O(log n) time.
# - All the integers in nums are unique.
#
# EXAMPLES:
# Input: nums = [3,4,5,1,2] -> Output: 1
# Input: nums = [4,5,6,7,0,1,2] -> Output: 0
# Input: nums = [11,13,15,17] -> Output: 11
# ============================================================================

from typing import List

def findMin(nums: List[int]) -> int:
    n = len(nums)
    l, r = 0 , n-1
    while l < r:
        mid = l + (r-l)//2

        if nums[mid] > nums[r]:
            l = mid + 1
        else:
            r = mid
    return nums[l]
        
    

print(findMin([3,4,5,1,2]))
print(findMin([4,5,6,7,0,1,2]))

def test():
    assert findMin([3,4,5,1,2]) == 1  # example 1: rotated at index 3
    assert findMin([4,5,6,7,0,1,2]) == 0  # example 2: rotated at index 4
    assert findMin([11,13,15,17]) == 11  # example 3: no rotation
    assert findMin([1]) == 1  # single element
    assert findMin([2,1]) == 1  # two elements rotated
    assert findMin([1,2]) == 1  # two elements sorted
    assert findMin([5,1,2,3,4]) == 1  # min at index 1
    assert findMin([2,3,4,5,1]) == 1  # min at the end
    assert findMin([10,20,30,5,7]) == 5  # arbitrary rotation
    assert findMin([4,5,1,2,3]) == 1  # mid point is inflection
    print('All Pass!')

test()
```

 # 0106 Find pivot (minimum) in a rotated array
 [↑ Back to TOC](#binsearch)


```python
"""
id: lc_0153_idx
title: Find Minimum in Rotated Sorted Array (Index)
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/
status: draft
last_updated: 2026-04-23
notes:
- key idea: binary search for the inflection point where nums[mid] > nums[right]
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 153_lc_0153_find_pivot_rotated_empty.py
# LeetCode 153: Find Minimum in Rotated Sorted Array (Pivot Index)
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Suppose an array of length n sorted in ascending order is rotated between 1 
# and n times. Find the pivot index (the index of the smallest element).
#
# RULES:
# - All elements are unique.
# - Must run in O(log n) time.
#
# EXAMPLES:
# [3,4,5,1,2] -> 3
# [4,5,6,7,0,1,2] -> 4
# [11,13,15,17] -> 0
# ============================================================================

from typing import List

def findPivot(nums: List[int]) -> int:
    n = len(nums)
    l , r  = 0, n-1
    while l < r:
        mid = l +  (r -l) //2
        if nums[mid] > nums[r]:
            l = mid +1
        else:
            r = mid
    return l



print(findPivot([3, 4, 5, 1, 2]))
print(findPivot([4, 5, 6, 7, 0, 1, 2]))

def test():
    assert findPivot([3, 4, 5, 1, 2]) == 3          # given example 1
    assert findPivot([4, 5, 6, 7, 0, 1, 2]) == 4    # given example 2
    assert findPivot([11, 13, 15, 17]) == 0         # given example 3 (no rotation)
    assert findPivot([2, 1]) == 1                   # given example 4 (two elements)
    assert findPivot([1]) == 0                      # single element
    assert findPivot([5, 1, 2, 3, 4]) == 1          # pivot at index 1
    assert findPivot([2, 3, 4, 5, 1]) == 4          # pivot at last index
    assert findPivot([1, 2, 3]) == 0                # sorted small array
    assert findPivot([3, 1, 2]) == 1                # rotated small array
    assert findPivot([5, 6, 7, 8, 9, 1, 2]) == 5    # larger array pivot
    print('All Pass!')

test()
```

 # 0121   search in a rotatged sorted array

```python
"""
id: lc_0033
title: Search in Rotated Sorted Array
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/search-in-rotated-sorted-array/
status: draft
last_updated: 2026-04-23
notes:
- key idea: check which half is sorted and determine if target lies within that range
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 033_lc_0033_search_in_rotated_sorted_array_empty.py
# LeetCode 33: Search in Rotated Sorted Array
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# There is an integer array nums sorted in ascending order (with distinct values).
# Prior to being passed to your function, nums is possibly rotated at an unknown 
# pivot index. Given the array nums after the possible rotation and an integer 
# target, return the index of target if it is in nums, or -1 if it is not in nums.
#
# RULES:
# - You must design an algorithm with O(log n) runtime complexity.
#
# EXAMPLES:
# Input: nums = [4,5,6,7,0,1,2], target = 0 -> Output: 4
# Input: nums = [4,5,6,7,0,1,2], target = 3 -> Output: -1
# Input: nums = [1], target = 0 -> Output: -1
# ============================================================================

from typing import List

def search(nums: List[int], target: int) -> int:
    n = len(nums)
    l , r  = 0, n-1
    while l <= r:
        mid = l +  (r -l) //2
        if target == nums[mid]:
            return mid
        # left side sorted 
        if nums[l] <= nums[mid]:
            if nums[l] <= target < nums[mid]:   # left side sorted and the target is on left side .. normal search  
                r = mid -1
            else:
                l = mid + 1
        # Sorted right side
        else:
            if nums[mid] < target <= nums[r]:  # right side is sorted and the target is on the right side
                l = mid + 1
            else:
                r = mid -1
    return -1
            
        
        

print(search([4, 5, 6, 7, 0, 1, 2], 0))
print(search([4, 5, 6, 7, 0, 1, 2], 3))

def test():
    assert search([4, 5, 6, 7, 0, 1, 2], 0) == 4    # example 1: target in right half
    assert search([4, 5, 6, 7, 0, 1, 2], 3) == -1   # example 2: target not found
    assert search([1], 0) == -1                     # example 3: single element miss
    assert search([1], 1) == 0                      # single element hit
    assert search([4, 5, 6, 7, 0, 1, 2], 4) == 0    # target is first element
    assert search([4, 5, 6, 7, 0, 1, 2], 2) == 6    # target is last element
    assert search([6, 7, 1, 2, 3, 4, 5], 6) == 0    # target in left half
    assert search([3, 1], 1) == 1                   # two elements rotated
    assert search([3, 5, 1], 3) == 0                # target at start of left
    assert search([5, 1, 3], 3) == 2                # target at end of right
    print('All Pass!')

test()
```
