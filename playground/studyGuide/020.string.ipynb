# 063 Randsom

```python
from collections import Counter
def solution(ransomNote, magazine):
    ran = Counter(ransomNote)
    mag = Counter(magazine)
    ret = True
    for key, val in ran.items():
        if not key in mag or mag[key] < val:
            return False

    return True

def test():
    
    assert solution("a", "b")        == False  # given example 1
    assert solution("aa", "ab")      == False  # given example 2
    assert solution("aa", "aab")     == True   # given example 3
    assert solution("", "abc")       == True   # empty note
    assert solution("abc", "abc")    == True   # exact match
    assert solution("abc", "ab")     == False  # magazine too short
    assert solution("aab", "aab")    == True   # duplicates exact
    assert solution("abcd", "dcba")  == True   # same letters different order
    assert solution("a", "aa")       == True   # magazine has more than needed — fine
    print('All Pass!')    
test()
```

# 68 String Compression

```python
def solution(s):
    cur = ''
    cnt = 0
    ret = ''

    for i, c in enumerate(s):
        if c != cur :    #switching
            if cnt < 2:
                ret += cur
            else:
                ret += cur + str(cnt)
            cur = c
            cnt = 1
        else:
            cnt += 1
            
    if cnt < 2:
        ret += cur
    else:
        ret +=   cur + str(cnt)
    return ret
            



print(solution("abcd"))
def test():
    assert solution("aabcccdddd") == "a2bc3d4"
    assert solution("abcd")       == "abcd"
    assert solution("aaa")        == "a3"
    assert solution("a")          == "a"
    assert solution("aa")         == "a2"
    assert solution("aabb")       == "a2b2"
    assert solution("aaabba")     == "a3b2a"
    assert solution("aaaaaaaaaa") == "a10"
    print('All Pass!')    

test()
```

 # 0073 Return all indexes of anagram of p in string s
 [↑ Back to TOC](#string)


```python
def anaMe(s):
    cnts = [0] * 26
    for c in s:
        cnts[ord(c) - ord('a')] += 1
    return tuple(cnts)
    
def solution(s, p):
    if len(s) < len(p): 
        return []
    ret = []
    p_cnt = anaMe(p)
    wl = len(p)
    for i in range(len(s) -wl+1):
        w_cnt = anaMe(s[i :i + wl])
        if p_cnt == w_cnt:
            ret.append(i)
    return ret
    
print(solution("cbaebabacd", "abc"))

def test():
    assert solution("cbaebabacd", "abc") == [0, 6]
    
    assert solution("aa",         "bb")  == []
    assert solution("a",          "a")   == [0]
    assert solution("ab",         "abc") == []          # p longer than s
    assert solution("baa",        "aa")  == [1]
    assert solution("aaaa",       "aa")  == [0, 1, 2]
    assert solution("abcdefg",    "gfedcba") == [0]    # full string is anagram
    assert solution("abab",       "ab")  == [0, 1, 2]
    print('All Pass!')
test()
```

 # 0074  for Palindrome.. You need to care for the even and odd 
 [↑ Back to TOC](#string)


```python
def maxEvenPal(s, i):
    l = i 
    r = i + 1
    ret = ''
    while r < len(s) and l >= 0:
        if s[r] == s[l]:
            ret = s[l:r+1]
            r += 1
            l -= 1
        else:
            break
    return ret
 
def maxOddPal(s, i):
    l = i-1
    r = i+1
    ret = s[i]
    while r < len(s) and l >= 0:
        if s[r] == s[l]:
            ret = s[l:r+1]
            r += 1
            l -= 1
        else:
            break
    return ret
    
def solution(s):
    ret = ''
    
    for i in range(len(s)):
        paleven = maxEvenPal(s, i)
        ret = paleven if len(paleven) > len(ret) else ret
        palodd = maxOddPal(s,i)
        ret = palodd if len(palodd) > len(ret) else ret

    return ret

def test():
    assert solution("cbbd")    == "bb"              # given example 2
    assert solution("babad")   in ("bab", "aba")   # given example 1 — two valid answers
    assert solution("cbbd")    == "bb"              # given example 2
    assert solution("racecar") == "racecar"         # whole string is palindrome
    assert solution("aacabdkacaa") == "aca"         # multiple candidates, pick longest
    assert solution("aaaa")    == "aaaa"            # all same char
    assert solution("abcba")   == "abcba"           # odd-length full palindrome
    assert solution("a")       == "a"               # single char
    assert solution("ac")      in ("a", "c")        # no palindrome longer than 1
    print('All Pass!')    
test()
```

 # 0122 Longest Repeating Character Replacement  == need to redo many times
 [↑ Back to TOC](#string)


```python
from collections import defaultdict
def solution(s, k):
    freq = defaultdict(int)
    l = 0
    best = 0
    max_freq = 0
    for r, c in enumerate(s):
        freq[c] += 1
        max_freq = max(max_freq, freq[c])
        
        while (r -l + 1) - max_freq > k:
            freq[s[l]] -= 1
            l += 1     
            
        best = max(best, r -l + 1)
        
    return best

solution("ABAB",    2)

def test():
    assert solution("ABAB",    2) == 4
    assert solution("AABABBA", 1) == 4
    assert solution("AAAA",    0) == 4
    assert solution("ABCD",    0) == 1
    assert solution("ABCD",    4) == 4
    print('All Pass!')
test()
```
