 ## 0046 Majority element .. 
 [↑ Back to TOC](#hashmap)


```python
from collections import Counter
def solution(nums):
    counts = Counter(nums)
    return counts.most_common(1)[0][0]


assert solution([3,2,3])              == 3  # given example 1
assert solution([2,2,1,1,1,2,2])     == 2  # given example 2
assert solution([1])                  == 1  # single element
assert solution([1,1,2])             == 1  # majority at start
assert solution([2,1,1])             == 1  # majority at end
assert solution([6,6,6,7,7])         == 6  # clear majority
assert solution([1,2,1,2,1])         == 1  # alternating, 1 wins
assert solution([3,3,4,3,4,3,4,3,4]) == 3  # 3 appears 5/9 times
print('All Pass!')
```

 ## 0022 Most Frequent Element

Given an array of integers `nums`, return the element that appears **most frequently**. If there is a tie, return the **smallest** of the tied elements.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.
 [↑ Back to TOC](#hashmap)


```python
import heapq
from collections import Counter
def solution(nums):
    cnts = Counter(nums)
    heap = [(-freq, n) for n, freq in cnts.items()]
    heapq.heapify(heap)
    _, n = heapq.heappop(heap)
    return n
    
assert solution([1,3,2,1,3,1])     == 1   # given example 1 — 1 appears 3x
assert solution([1,2,2,3,3])       == 2   # given example 2 — tie, smallest wins
assert solution([5])               == 5   # given example 3 — single element
assert solution([1,2,3])           == 1   # all frequency 1 — smallest wins
assert solution([3,3,2,2,1])       == 2   # tie between 2 and 3 — smallest wins
assert solution([-1,-1,2,2,3])     == -1  # tie, negative is smallest
assert solution([4,4,4,1,1])       == 4   # clear winner
assert solution([1,1,2,2,2,3,3])   == 2   # 2 appears 3x, clear winner
print('All Pass!')
```

 ## 0029      Group anagrams  setdefault
 [↑ Back to TOC](#hashmap)


```python
def makeAna(s):
    lookup = [0] * 26
    for c in s:
        lookup[ord(c) - ord('a')] += 1
    return tuple(lookup)
    
def solution(words):
    grps = {}
    for s in words:
        grps.setdefault(makeAna(s), []).append(s)
    return grps.values()
    



def check(result, expected):
    return sorted([sorted(g) for g in result]) == sorted([sorted(g) for g in expected])

assert check(solution(["eat","tea","tan","ate","nat","bat"]),
             [["eat","tea","ate"],["tan","nat"],["bat"]])   # given example 1
assert check(solution([""]),          [[""]])               # given example 2 — empty string
assert check(solution(["a"]),         [["a"]])              # given example 3 — single word
assert check(solution(["a","b"]),     [["a"],["b"]])        # no anagrams
assert check(solution(["ab","ba"]),   [["ab","ba"]])        # one group
assert check(solution(["abc","bca","cab","xyz"]),
             [["abc","bca","cab"],["xyz"]])                  # three anagrams + loner
assert check(solution(["a","a"]),     [["a","a"]])          # duplicate words
assert check(solution(["cat","act","dog","god","tac"]),
             [["cat","act","tac"],["dog","god"]])           # two groups
print('All Pass!')
```

You’ve got the core idea nailed:
- map stores counts of previous running totals
- at current running total `prefix`, matches are previous `(prefix - k)`
- add that occurrence count to total subarrays found

```python
# 1 — Build prefix sum array (O(n))
def build_prefix(nums):
    pref = []
    run = 0
    for x in nums:
        run += x
        pref.append(run)
    return pref

assert build_prefix([1,2,3,4]) == [1,3,6,10]
assert build_prefix([5])       == [5]
assert build_prefix([-1,2,-3]) == [-1,1,-2]
```

 ## 0036 Top K Frequent Elements
 [↑ Back to TOC](#hashmap)


```python
import heapq
from collections import Counter
# to get the maximum fequent k items.. maintain a minheap of length k.. 
# Because when you pop .. you pop from the top which is the smallest. meaining you are keeping the largest
def solution(nums, k):
    cnts = Counter(nums)
    heap = [(freq, n) for n, freq in cnts.items()]
    heapq.heapify(heap)
    while len(heap) > k :
        heapq.heappop(heap)
    return [n for (_, n) in heap]
    


def check(result, expected):
    return sorted(result) == sorted(expected)

assert check(solution([1,1,1,2,2,3], 2),       [1,2])      # given example 1
assert check(solution([1], 1),                  [1])        # given example 2 — single element
assert check(solution([1,2], 2),                [1,2])      # given example 3 — k = all unique
assert check(solution([3,3,3,1,1,2], 1),        [3])        # clear winner
assert check(solution([4,4,4,1,1,2], 2),        [4,1])      # clear top 2
assert check(solution([5,5,5,5], 1),            [5])        # single unique element
assert check(solution([1,1,2,2,2,3,3,3,3], 2), [2,3])      # top 2 clear
assert check(solution([1,1,2,2,2,3], 3),        [1,2,3])    # k = all unique
print('All Pass!')
```

```python
def validlist(l):
    seen = set()
    for s in l:
        if s == ".":
            continue
        if s in seen:
            return False
        seen.add(s)
    return True
    
def validSquare(board,r,c):
    seen = set()
    for i in range(r, r+3):
        for j in range(c, c+3):
            s = board[i][j]
            if s == ".":
                continue
            if s in seen:
                return False
            seen.add(s)
    return True  
        
def solution(board):
    inner_vertics = [(0,0), (0,3), (0,6),(3,0), (3,3), (3,6),(6,0), (6,3), (6,6),]
    for r,c in inner_vertics:
        if not validSquare(board,r,c):
            return False
    for row in board:
        if not  validlist(row):
            return False

    for col in zip(*board):
        if not validlist(col):
            return False
    return True



valid = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
invalid_row = [
    ["5","3",".",".","7",".",".",".","5"],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
invalid_col = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    ["5",".",".",".","8",".",".","7","9"]
]
invalid_box = [
    ["5","3",".",".","7",".",".",".","."],
    ["6","5",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]

assert solution(valid)       == True
assert solution(invalid_row) == False
assert solution(invalid_col) == False
assert solution(invalid_box) == False
print('All Pass!')
```
