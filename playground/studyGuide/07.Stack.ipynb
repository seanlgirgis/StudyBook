 # 0031 Valid Parentheses
 [↑ Back to TOC](#stack)


```python
"""
id: lc_0020
title: Valid Parentheses
source: leetcode
difficulty: easy
primary: stack
tags: [string, stack]
leetcode_url: https://leetcode.com/problems/valid-parentheses/
status: draft
last_updated: 2026-04-23
notes:
- key idea: Use a stack to track opening brackets and match them with closing ones.
- time: O(n)
- space: O(n)
"""

# ============================================================================
# File: 020_lc_0020_valid_parentheses_empty.py
# LeetCode 20: Valid Parentheses
# Difficulty: Easy
#
# PROBLEM STATEMENT:
# Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', 
# determine if the input string is valid.
#
# RULES:
# 1. Open brackets must be closed by the same type of brackets.
# 2. Open brackets must be closed in the correct order.
# 3. Every close bracket has a corresponding open bracket of the same type.
#
# EXAMPLES:
# s = "()" -> True
# s = "()[]{}" -> True
# s = "(]" -> False
# s = "([)]" -> False
# ============================================================================

from typing import List

def isValid(s: str) -> bool:
    lookup = { ')':'(',
                '}':'{',
                ']':'[',}
    sta = []
    for c in s:
        if c in lookup:
            if sta and sta[-1] == lookup[c]:
                sta.pop()
                continue
            else :
                return False
        sta.append(c)
    return not sta

print(isValid("()"))
print(isValid("([)]"))

def test():
    assert isValid("()") == True          # example 1: simple pair
    assert isValid("()[]{}") == True      # example 2: multiple types
    assert isValid("(]") == False         # example 3: mismatch
    assert isValid("([)]") == False       # example 4: interleaved
    assert isValid("{[]}") == True        # nested valid
    assert isValid("[") == False          # single opening
    assert isValid("]") == False          # single closing
    assert isValid("((") == False         # multiple opening
    assert isValid("){") == False         # wrong order
    assert isValid("({[]})") == True      # deep nesting
    print('All Pass!')

test()
```

 # 0032 Min Stack
 [↑ Back to TOC](#stack)


```python
"""
id: lc_0155
title: Min Stack
source: leetcode
difficulty: medium
primary: stack
tags: [stack, design]
leetcode_url: https://leetcode.com/problems/min-stack/
status: draft
last_updated: 2026-04-23
notes:
- key idea: Maintain an auxiliary stack to store the minimum value at each state.
- time: O(1) for all operations
- space: O(n)
"""

# ============================================================================
# File: 155_lc_0155_min_stack_empty.py
# LeetCode 155: Min Stack
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Design a stack that supports push, pop, top, and retrieving the minimum 
# element in constant time.
#
# RULES:
# - push(val) -- Push element val onto stack.
# - pop() -- Removes the element on the top of the stack.
# - top() -- Get the top element.
# - getMin() -- Retrieve the minimum element in the stack.
# - Methods pop, top and getMin operations will always be called on non-empty stacks.
#
# EXAMPLES:
# MinStack minStack = new MinStack();
# minStack.push(-2);
# minStack.push(0);
# minStack.push(-3);
# minStack.getMin(); // return -3
# minStack.pop();
# minStack.top();    // return 0
# minStack.getMin(); // return -2
# ============================================================================

from typing import List

class MinStack:
    def __init__(self):
        self.data = []
        pass

    def push(self, val: int) -> None:
        if not self.data:
            itm = (val, val)
        else:
            m = min(val, self.data[-1][1])
            itm = (val, m)
        self.data.append(itm)

    def pop(self) -> None:
        self.data.pop()

    def top(self) -> int:
        return self.data[-1][0]

    def getMin(self) -> int:
        return self.data[-1][1]

# Demo
obj = MinStack()
print(obj.push(-2), obj.push(0), obj.push(-3), obj.getMin())
print(obj.pop(), obj.top(), obj.getMin())

def test():
    ms = MinStack()
    ms.push(-2)
    ms.push(0)
    ms.push(-3)
    assert ms.getMin() == -3  # initial min after pushes
    ms.pop()
    assert ms.top() == 0      # top after popping -3
    assert ms.getMin() == -2  # min reverts to -2
    ms.push(-1)
    assert ms.getMin() == -2  # push larger than current min
    ms.push(-5)
    assert ms.getMin() == -5  # new absolute min
    ms.pop()
    assert ms.getMin() == -2  # min reverts after popping -5
    ms.push(10)
    assert ms.top() == 10     # check top identity
    assert ms.getMin() == -2  # min stays same after pushing large
    print('All Pass!')

test()
```

 #  0112  Evaluate Reverse Polish Notation
 [↑ Back to TOC](#stack)


```python
"""
id: lc_0150
title: Evaluate Reverse Polish Notation
source: leetcode
difficulty: medium
primary: stack
tags: [stack, math, array]
leetcode_url: https://leetcode.com/problems/evaluate-reverse-polish-notation/
status: draft
last_updated: 2026-04-23
notes:
- key idea: Use a stack to store operands and apply operators to the top two elements.
- time: O(n)
- space: O(n)
"""

# ============================================================================
# File: 150_lc_0150_evaluate_reverse_polish_notation_empty.py
# LeetCode 150: Evaluate Reverse Polish Notation
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Evaluate the value of an arithmetic expression in Reverse Polish Notation.
# Valid operators are +, -, *, and /. Each operand may be an integer or 
# another expression.
#
# RULES:
# 1. Division between two integers should truncate toward zero.
# 2. The input tokens are always valid RPN.
# 3. There will be no division by zero.
#
# EXAMPLES:
# ["2","1","+","3","*"] -> ((2 + 1) * 3) -> 9
# ["4","13","5","/","+"] -> (4 + (13 / 5)) -> 6
# ["10","6","9","3","+","-11","*","/","17","+","5","+"] -> 22
# ============================================================================

from typing import List

def evalRPN(tokens: List[str]) -> int:
    s = []
    for tk in tokens:
        if tk not in {"+", "-", "*", "/"}:
            s.append(int(tk))
        elif tk == "+":
            val = s.pop() + s.pop()
            s.append(val)
        elif tk == "*":
            val = s.pop() * s.pop()
            s.append(val)
        elif tk == "-":
            b = s.pop()
            a = s.pop()
            val = a - b
            s.append(val)
        elif tk == "/":
            b = s.pop()
            a = s.pop()
            val = int(a / b)
            s.append(val)            
    return s[-1]

print(evalRPN(["2", "1", "+", "3", "*"]))
print(evalRPN(["4", "13", "5", "/", "+"]))

def test():
    assert evalRPN(["2", "1", "+", "3", "*"]) == 9          # simple add and multiply
    assert evalRPN(["4", "13", "5", "/", "+"]) == 6        # division truncation
    assert evalRPN(["10","6","9","3","+","-11","*","/","17","+","5","+"]) == 22 # complex
    assert evalRPN(["3", "-4", "+"]) == -1                  # negative result
    assert evalRPN(["-128", "128", "/"]) == -1              # division of negatives
    assert evalRPN(["10", "3", "/"]) == 3                   # positive truncation
    assert evalRPN(["-10", "3", "/"]) == -3                 # negative truncation (toward zero)
    assert evalRPN(["4", "3", "-"]) == 1                    # subtraction order
    assert evalRPN(["18"]) == 18                            # single token
    assert evalRPN(["4", "13", "5", "*", "+"]) == 69        # order of operations check
    print('All Pass!')

test()
```

```python
"""
id: lc_0394
title: Decode String
source: leetcode
difficulty: medium
primary: stack
tags: [string, stack, recursion]
leetcode_url: https://leetcode.com/problems/decode-string/
status: draft
last_updated: 2026-04-23
notes:
- key idea: use stack to track multipliers and current strings
- time: O(n) where n is the length of the decoded string
- space: O(n)
"""

# ============================================================================
# File: 394_lc_0394_decode_string_empty.py
# LeetCode 394: Decode String
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Given an encoded string, return its decoded version. 
# The encoding rule is: k[encoded_string] means the encoded_string is 
# repeated exactly k times.
#
# RULES:
# - k is guaranteed to be a positive integer.
# - Input is always valid (no extra spaces, well-formed brackets).
#
# EXAMPLES:
# s = "3[a]2[bc]" -> "aaabcbc"
# s = "3[a2[c]]" -> "accaccacc"
# s = "2[abc]3[cd]ef" -> "abcabccdcdcdef"
# ============================================================================

from typing import List

def decodeString(s: str) -> str:

    stack = []
    for c in s:
        if c != "]":
            stack.append(c)
        else:
            word = ""
            words = ""
            n = 0
            nord = 0
            while stack and stack[-1] != "[":
                word += stack.pop()
            stack.pop()
            while stack and stack[-1].isdigit():
                n += int(stack.pop()) * (10 ** nord)
                nord+= 1
            words = word * n
            stack.append(words[::-1])
    out = ""            
    while stack:
        out = stack.pop() + out
        
            
    return out
            
            


print(decodeString("10[a]"))
print(decodeString("3[a2[c]]"))
print(decodeString("3[a]2[bc]"))
def test():
    assert decodeString("3[a]2[bc]") == "aaabcbc"  # basic sequence
    assert decodeString("3[a2[c]]") == "accaccacc"  # nested brackets
    assert decodeString("2[abc]3[cd]ef") == "abcabccdcdcdef"  # mixed content
    assert decodeString("abc3[cd]xyz") == "abccdcdcdxyz"  # outer text
    assert decodeString("10[a]") == "aaaaaaaaaa"  # multi-digit k
    assert decodeString("2[2[b]]") == "bbbb"  # nested repeats
    assert decodeString("z") == "z"  # no brackets
    assert decodeString("2[a]3[b]4[c]") == "aabbbcccc"  # multiple blocks
    assert decodeString("3[a]2[b4[c]]") == "aaabccccbcccc"  # complex nesting
    
    print('All Pass!')

test()
```
