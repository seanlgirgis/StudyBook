## Product Except Self

Given an array of integers `nums`, return an array `result` where `result[i]` is the **product of all elements except `nums[i]`**.

You must not use division.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [1,2,3,4]`, the output should be `solution(nums) = [24,12,8,6]`.

  **Explanation:**
  - `result[0] = 2×3×4 = 24`
  - `result[1] = 1×3×4 = 12`
  - `result[2] = 1×2×4 = 8`
  - `result[3] = 1×2×3 = 6`

- For `nums = [-1,1,0,-3,3]`, the output should be `solution(nums) = [0,0,9,0,0]`.

  **Explanation:** Any position that doesn't include the `0` at index 2 will have `0` in the product. Position 2 itself gets `(-1)×1×(-3)×3 = 9`.

- For `nums = [2,3]`, the output should be `solution(nums) = [3,2]`.

  **Explanation:** Each element is just the other one.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `2 ≤ nums.length ≤ 10^5`, `-30 ≤ nums[i] ≤ 30`
- **[output]** `array.integer` — Product of all elements except self, for each position.

```python
def solution(nums):
    n = len(nums)
    prefix = postfix = 1
    ret = [1] * n
    for i, num in enumerate(nums):
        ret[i] = prefix
        prefix *= num

    for i in range( n-1, -1, -1):
        ret[i] *= postfix
        postfix *= nums[i]

    return ret
        


assert solution([1,2,3,4])       == [24,12,8,6]   # given example 1
assert solution([-1,1,0,-3,3])   == [0,0,9,0,0]   # given example 2 — zero in array
assert solution([2,3])           == [3,2]          # given example 3 — two elements
assert solution([1,1,1,1])       == [1,1,1,1]      # all ones
assert solution([-1,-1,-1])      == [1,1,1]        # all negative-ones — product of two negatives = 1
assert solution([2,2,2,2])       == [8,8,8,8]      # all same
assert solution([0,0])           == [0,0]          # two zeros
assert solution([1,2,3])         == [6,3,2]        # three elements
print('All Pass!')
```
