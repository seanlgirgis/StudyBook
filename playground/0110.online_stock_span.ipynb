## Online Stock Span

Design a class `StockSpanner` with a method `next(price)` that returns the **span** of the stock's price for the current day — the number of consecutive days (including today) where the price was less than or equal to today's price.

### Example

```
Spanner = StockSpanner()
Spanner.next(100) → 1   (just today)
Spanner.next(80)  → 1   (80 < 100, streak breaks)
Spanner.next(60)  → 1
Spanner.next(70)  → 2   (70 >= 60, counts today + yesterday)
Spanner.next(60)  → 1
Spanner.next(75)  → 4   (75 >= 60,70,60 — counts 4 days)
Spanner.next(85)  → 6   (85 >= all previous)
```

### Hint
Store `(price, span)` pairs on the stack. When today's price >= top price, absorb its span.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** sequence of `next(price)` calls
- **[output]** `integer` per call

````python
%run _timer.py
````

````python
class StockSpanner:
    def __init__(self):
        self.stack = []    # tuples of (price, span)

    def next(self, price: int) -> int:
        span = 1
        while self.stack and self.stack[-1][0] <= price:
            prev_price, prev_span = self.stack.pop()
            span += prev_span
        self.stack.append((price, span))
        return span


s = StockSpanner()
assert s.next(100) == 1
assert s.next(80)  == 1
assert s.next(60)  == 1
assert s.next(70)  == 2
assert s.next(60)  == 1
assert s.next(75)  == 4
assert s.next(85)  == 6
print('All Pass!')
````
