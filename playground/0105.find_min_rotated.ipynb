## Find Minimum in Rotated Sorted Array

A sorted array was rotated at some pivot. Find the minimum element. You must write an O(log n) solution.

### Example

- `nums = [3,4,5,1,2]` → `1` (rotated at index 3)
- `nums = [4,5,6,7,0,1,2]` → `0`
- `nums = [11,13,15,17]` → `11` (not rotated)

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums`
- **[output]** `integer`

```python
"""
id: lc_0153
title: Find Minimum in Rotated Sorted Array
source: leetcode
difficulty: medium
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/
status: draft
last_updated: 2026-04-23
notes:
- key idea: binary search for the inflection point
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 153_lc_0153_find_minimum_in_rotated_sorted_array_empty.py
# LeetCode 153: Find Minimum in Rotated Sorted Array
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Suppose an array of length n sorted in ascending order is rotated between 
# 1 and n times. Find the minimum element of this array.
#
# RULES:
# - You must write an algorithm that runs in O(log n) time.
# - All the integers in nums are unique.
#
# EXAMPLES:
# Input: nums = [3,4,5,1,2] -> Output: 1
# Input: nums = [4,5,6,7,0,1,2] -> Output: 0
# Input: nums = [11,13,15,17] -> Output: 11
# ============================================================================

from typing import List

def findMin(nums: List[int]) -> int:
    n = len(nums)
    l, r = 0 , n-1
    while l < r:
        mid = l + (r-l)//2

        if nums[mid] > nums[r]:
            l = mid + 1
        else:
            r = mid
    return nums[l]
        
    

print(findMin([3,4,5,1,2]))
print(findMin([4,5,6,7,0,1,2]))

def test():
    assert findMin([3,4,5,1,2]) == 1  # example 1: rotated at index 3
    assert findMin([4,5,6,7,0,1,2]) == 0  # example 2: rotated at index 4
    assert findMin([11,13,15,17]) == 11  # example 3: no rotation
    assert findMin([1]) == 1  # single element
    assert findMin([2,1]) == 1  # two elements rotated
    assert findMin([1,2]) == 1  # two elements sorted
    assert findMin([5,1,2,3,4]) == 1  # min at index 1
    assert findMin([2,3,4,5,1]) == 1  # min at the end
    assert findMin([10,20,30,5,7]) == 5  # arbitrary rotation
    assert findMin([4,5,1,2,3]) == 1  # mid point is inflection
    print('All Pass!')

test()
```
