## House Robber

You are a robber planning to rob houses along a street. Each house has a certain amount of money. You **cannot rob two adjacent houses** (the alarm will trigger). Given an array `nums` where `nums[i]` is the money in house `i`, return the **maximum amount you can rob**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [1,2,3,1]`, the output should be `solution(nums) = 4`.

  **Explanation:** Rob house 0 (`1`) and house 2 (`3`) → total `4`.

- For `nums = [2,7,9,3,1]`, the output should be `solution(nums) = 12`.

  **Explanation:** Rob house 0 (`2`), house 2 (`9`), house 4 (`1`) → total `12`.

- For `nums = [5]`, the output should be `solution(nums) = 5`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `1 ≤ nums.length ≤ 100`, `0 ≤ nums[i] ≤ 400`
- **[output]** `integer` — Maximum money you can rob.

```python
%run _timer.py
```

```python
def solution(nums):
    n = len(nums)
    prevprev, prev = 0, nums[0]
    for i in range(1, n):
        prevprev, prev = prev, max(prev, prevprev + nums[i])



    return prev


assert solution([1,2,3,1])     == 4   # given example 1 — rob 0,2
assert solution([2,7,9,3,1])   == 12  # given example 2 — rob 0,2,4
assert solution([5])           == 5   # given example 3 — single house
assert solution([1,2])         == 2   # two houses — take the larger
assert solution([2,1,1,2])     == 4   # rob first and last
assert solution([0,0,0])       == 0   # all zero
assert solution([4,1,1,4])     == 8   # skip middle two
assert solution([1,3,1,3,1])   == 6   # alternating — rob index 1,3
print('All Pass!')
```
