## Is Palindrome

Given a string `s`, return `True` if it reads the same forwards and backwards, and `False` otherwise.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "racecar"`, the output should be `solution(s) = True`.

  **Explanation:** `r-a-c-e-c-a-r` — mirrors perfectly.

- For `s = "hello"`, the output should be `solution(s) = False`.

  **Explanation:** `h` ≠ `o`.

- For `s = "a"`, the output should be `solution(s) = True`.

  **Explanation:** A single character is always a palindrome.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `string` `s` — A non-empty string of lowercase English letters.  
  Guaranteed constraints: `1 ≤ s.length ≤ 10^5`
- **[output]** `boolean` — `True` if palindrome, `False` otherwise.

```python
def solution(s):
    l, r = 0 , len(s)-1
    while r > l:
        if s[r] != s[l]:
            return False
        l += 1
        r -= 1
    return True


assert solution('racecar')  == True   # given example 1 — odd length
assert solution('hello')    == False  # given example 2 — h ≠ o
assert solution('a')        == True   # given example 3 — single char
assert solution('aa')       == True   # even length, match
assert solution('ab')       == False  # even length, no match
assert solution('abcba')    == True   # odd length, classic
assert solution('abcca')    == False  # almost — center differs
assert solution('abba')     == True   # even length palindrome
print('All Pass!')
```

```python
def solution(s):
    return s == s[::-1]

assert solution('racecar')  == True   # given example 1 — odd length
assert solution('hello')    == False  # given example 2 — h ≠ o
assert solution('a')        == True   # given example 3 — single char
assert solution('aa')       == True   # even length, match
assert solution('ab')       == False  # even length, no match
assert solution('abcba')    == True   # odd length, classic
assert solution('abcca')    == False  # almost — center differs
assert solution('abba')     == True   # even length palindrome
print('All Pass!')
```
