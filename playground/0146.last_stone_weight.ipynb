## Last Stone Weight (LC 1046)

Array of stone weights. Each round: smash the two heaviest together.
- Equal weights → both destroyed
- Unequal → heavier survives with weight `y − x`

Return the weight of the last stone, or `0` if none remain.

**Pattern:** Max-heap simulation.  
Python's `heapq` is a **min-heap** — negate weights to simulate a max-heap.  
Each round: pop two largest, push difference if nonzero.  
When ≤1 stone left, done.

```python
%run _timer.py
```

```python
import heapq

def solution(stones):
    heap = [-s for s in stones]    # negate → min-heap behaves like max-heap
    heapq.heapify(heap)

    while len(heap) > 1:
        y = -heapq.heappop(heap)   # heaviest
        x = -heapq.heappop(heap)   # second heaviest
        if y != x:
            heapq.heappush(heap, -(y - x))   # fragment survives

    return -heap[0] if heap else 0


assert solution([2,7,4,1,8,1]) == 1
assert solution([1])           == 1
assert solution([2,2])         == 0    # equal — both destroyed
assert solution([3,3,3])       == 3    # two cancel, one left
assert solution([1,3])         == 2
assert solution([1,1,1,1,1,1]) == 0    # all cancel in pairs
print('All Pass!')
```
