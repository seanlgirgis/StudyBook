## Subsets

Given an integer array `nums` of **unique** elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the subsets in any order.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(2^n)** will fit within the execution time limit.

### Example

- For `nums = [1,2,3]`, the output should be `[[],[1],[2],[3],[1,2],[1,3],[2,3],[1,2,3]]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums`
- **[output]** `array.array.integer`

```python
def solution(nums):
    result = []
    
    def backtrack(i, path):
        if i == len(nums):
            result.append(path[:])
            return
        backtrack(i + 1, path)           # skip
        path.append(nums[i])
        backtrack(i + 1, path)           # include
        path.pop()

        
    backtrack(0, [])
    return result


def check(result, expected):
    return sorted(sorted(s) for s in result) == sorted(sorted(s) for s in expected)

assert check(solution([1,2,3]), [[],[1],[2],[3],[1,2],[1,3],[2,3],[1,2,3]])
assert check(solution([0]),     [[],[0]])
assert check(solution([1,2]),   [[],[1],[2],[1,2]])
print('All Pass!')
```

```python
from itertools import combinations

def solution(nums):
    result = []
    for r in range(len(nums) + 1):
        result.extend(combinations(nums, r))
    return [list(c) for c in result]


    
def check(result, expected):
    return sorted(sorted(s) for s in result) == sorted(sorted(s) for s in expected)

assert check(solution([1,2,3]), [[],[1],[2],[3],[1,2],[1,3],[2,3],[1,2,3]])
assert check(solution([0]),     [[],[0]])
assert check(solution([1,2]),   [[],[1],[2],[1,2]])
print('All Pass!')
```
