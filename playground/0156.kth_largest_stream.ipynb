## Kth Largest Element in a Stream (LC 703)

Design a class that finds the `k`th largest element in a stream after each `add`.

**Pattern:** Min-heap of size `k`.
- The `k`th largest = the **smallest** element in a size-k max collection
- Keep a min-heap of size exactly `k` — `heap[0]` is always the answer
- On `add`: push, then if size > k, pop the smallest (it's no longer in the top k)

```python
%run _timer.py
```

```python
import heapq

class KthLargest:
    def __init__(self, k, nums):
        self.k    = k
        self.heap = []
        for n in nums:
            self.add(n)

    def add(self, val):
        heapq.heappush(self.heap, val)
        if len(self.heap) > self.k:
            heapq.heappop(self.heap)    # evict smallest — keep only top k
        return self.heap[0]             # min of top-k = kth largest overall


kl = KthLargest(3, [4, 5, 8, 2])
assert kl.add(3)  == 4
assert kl.add(5)  == 5
assert kl.add(10) == 5
assert kl.add(9)  == 8
assert kl.add(4)  == 8

kl2 = KthLargest(1, [])
assert kl2.add(-3) == -3
assert kl2.add(-2) == -2
assert kl2.add(-4) == -2
print('All Pass!')
```
