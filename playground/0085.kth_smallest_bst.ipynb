## Kth Smallest Element in a BST

Given the root of a BST and an integer `k`, return the `k`th smallest value (1-indexed) among all node values.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

```
    3          5
   / \        / \
  1   4      3   6
   \         /
    2        2
             /
            1
```
- Tree 1, `k=1` → `1`
- Tree 2, `k=3` → `3`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `TreeNode` `root`, `integer` `k`
- **[output]** `integer`

````python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val   = val
        self.left  = left
        self.right = right

def make_tree(vals):
    if not vals or vals[0] is None:
        return None
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root


def solution(root, k):
    stack = []
    curr = root
    while curr:
        stack.append(curr)
        curr = curr.left
    while stack:
        node = stack.pop()
        k -= 1
        if k == 0:
            return node.val
        curr = node.right
        while curr:
            stack.append(curr)
            curr = curr.left
    return None


assert solution(make_tree([3,1,4,None,2]),   1) == 1
assert solution(make_tree([5,3,6,2,4,None,None,1]), 3) == 3
assert solution(make_tree([3,1,4,None,2]),   2) == 2
assert solution(make_tree([3,1,4,None,2]),   3) == 3
assert solution(make_tree([3,1,4,None,2]),   4) == 4
assert solution(make_tree([1]),              1) == 1
assert solution(make_tree([5,3,7,2,4,6,8]), 4) == 5   # median of 7-node BST
print('All Pass!')
````
