## Subarray Sum Equals K

Given an array of integers `nums` and an integer `k`, return the **number of contiguous subarrays** whose sum equals `k`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [1,1,1]`, `k = 2`, the output should be `solution(nums, k) = 2`.

  **Explanation:** `[1,1]` at indices `[0,1]` and `[1,2]` both sum to `2`.

- For `nums = [1,2,3]`, `k = 3`, the output should be `solution(nums, k) = 2`.

  **Explanation:** `[3]` at index `2` and `[1,2]` at indices `[0,1]` both sum to `3`.

- For `nums = [1,-1,1]`, `k = 1`, the output should be `solution(nums, k) = 3`.

  **Explanation:** `[1]` at index 0, `[1,-1,1]` full array, and `[1]` at index 2.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — Array of integers (may include negatives).  
  Guaranteed constraints: `1 ≤ nums.length ≤ 2·10^4`, `-10^3 ≤ nums[i] ≤ 10^3`
- **[input]** `integer` `k`
- **[output]** `integer` — Count of subarrays summing to `k`.

```python
def solution(nums, k):
    count = 0
    for i in range(len(nums)):
        total = 0
        for j in range(i, len(nums)):
            total += nums[j]
            if total == k:
                count += 1
    return count


assert solution([1,1,1], 2)       == 2  # given example 1 — two overlapping windows
assert solution([1,2,3], 3)       == 2  # given example 2 — [3] and [1,2]
assert solution([1,-1,1], 1)      == 3  # given example 3 — negatives, full array counts
assert solution([1], 1)           == 1  # single element match
assert solution([1], 2)           == 0  # single element no match
assert solution([1,2,1,2,1], 3)   == 4  # multiple pairs
assert solution([-1,-1,1], -1)    == 3  # negative k
assert solution([0,0,0], 0)       == 6  # zeros — every subarray sums to 0
print('All Pass!')
```

```python
from collections import defaultdict

def solution(nums, k):
    counts = defaultdict(int)
    counts[0] = 1      # empty prefix — sum of 0 seen once
    total = 0
    count = 0
    for num in nums:
        total += num
        count += counts[total - k]   # how many earlier prefixes make a valid subarray
        counts[total] += 1
    return count


assert solution([1,1,1], 2)       == 2  # given example 1 — two overlapping windows
assert solution([1,2,3], 3)       == 2  # given example 2 — [3] and [1,2]
assert solution([1,-1,1], 1)      == 3  # given example 3 — negatives, full array counts
assert solution([1], 1)           == 1  # single element match
assert solution([1], 2)           == 0  # single element no match
assert solution([1,2,1,2,1], 3)   == 4  # multiple pairs
assert solution([-1,-1,1], -1)    == 3  # negative k
assert solution([0,0,0], 0)       == 6  # zeros — every subarray sums to 0
print('All Pass!')
```
