## Min Cost Climbing Stairs — Spaced Repeat

You are given an integer array `cost` where `cost[i]` is the cost of the `i`-th step. Once you pay the cost, you can climb one or two steps. You can start from index `0` or `1`.

Return the **minimum cost** to reach the top (beyond the last step).

### Example

- `cost = [10,15,20]` → `15` (step 1 costs 15, jump 2 to top)
- `cost = [1,100,1,1,1,100,1,1,100,1]` → `6`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `cost`
- **[output]** `integer`

```python
%run _timer.py
```

```python
def solution(cost):
    n = len(cost)
    if n < 2:
        return 0
    prevprev = cost[0]
    prev = cost[1]
    for i in range(2, n):
        current = cost[i] + min(prevprev, prev)
        prevprev, prev = prev, current
    return min(prev, prevprev)


assert solution([10,15,20])                   == 15
assert solution([1,100,1,1,1,100,1,1,100,1]) == 6
assert solution([0,0,0,1])                    == 0
assert solution([1,0])                        == 0
assert solution([10,15])                      == 10
print('All Pass!')
```
