## Course Schedule II — Topological Sort

Return a valid order to take all courses, or `[]` if impossible (cycle).

- `numCourses=4`, `[[1,0],[2,0],[3,1],[3,2]]` → `[0,1,2,3]` (or any valid order)
- `numCourses=2`, `[[1,0],[0,1]]` → `[]` (cycle)

**Pattern:** Same 3-state DFS as course_schedule, but append to result *after* all neighbors are done (post-order = topological order).

```python
%run _timer.py
```

```python
from collections import defaultdict

def solution(numCourses, prerequisites):
    adj = defaultdict(list)
    for course, pre in prerequisites:
        adj[pre].append(course)

    state = [0] * numCourses  # 0=unvisited, 1=visiting, 2=done
    order = []

    def dfs(i):
        if state[i] == 1: return False   # cycle
        if state[i] == 2: return True    # already processed
        state[i] = 1
        for nei in adj[i]:
            if not dfs(nei): return False
        state[i] = 2
        order.append(i)                  # post-order = topological order
        return True

    for i in range(numCourses):
        if state[i] == 0:
            if not dfs(i): return []

    return order


def valid_order(result, numCourses, prerequisites):
    if not result: return False
    pos = {c: i for i, c in enumerate(result)}
    return all(pos[pre] < pos[course] for course, pre in prerequisites)

assert valid_order(solution(4, [[1,0],[2,0],[3,1],[3,2]]), 4, [[1,0],[2,0],[3,1],[3,2]])
assert solution(2, [[1,0],[0,1]]) == []
assert valid_order(solution(2, [[1,0]]), 2, [[1,0]])
print('All Pass!')
```
