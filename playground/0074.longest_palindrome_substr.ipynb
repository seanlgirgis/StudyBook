## Longest Palindromic Substring

Given a string `s`, return the **longest substring** of `s` that is a palindrome.

If there are multiple answers of the same length, return any one of them.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `s = "babad"`, the output should be `"bab"` (or `"aba"`).
- For `s = "cbbd"`, the output should be `"bb"`.
- For `s = "a"`, the output should be `"a"`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `s`  
  Guaranteed constraints: `1 ≤ len(s) ≤ 1000`, lowercase English letters only
- **[output]** `string`

```python
def palan(i, j, n, s):
    while i >= 0 and j < n and s[i] == s[j]:
        i -= 1
        j += 1
    return s[i + 1:j]


def solution(s):
    n = len(s)
    max_pal = ''
    for i in range(n):
        pal1 = palan(i, i, n, s)
        pal2 = palan(i, i+1, n, s)
        if len(pal1) > len(max_pal): max_pal = pal1
        if len(pal2) > len(max_pal): max_pal = pal2
 
    return max_pal

assert solution("cbbd")    == "bb"              # given example 2
assert solution("babad")   in ("bab", "aba")   # given example 1 — two valid answers
assert solution("cbbd")    == "bb"              # given example 2
assert solution("a")       == "a"               # single char
assert solution("ac")      in ("a", "c")        # no palindrome longer than 1
assert solution("racecar") == "racecar"         # whole string is palindrome
assert solution("aacabdkacaa") == "aca"         # multiple candidates, pick longest
assert solution("aaaa")    == "aaaa"            # all same char
assert solution("abcba")   == "abcba"           # odd-length full palindrome
print('All Pass!')
```
