## Valid Anagram

Given two strings `s` and `t`, return `True` if `t` is an anagram of `s`, and `False` otherwise.

An **anagram** is a word formed by rearranging the letters of another word, using all the original letters exactly once.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.

### Example

- For `s = "anagram"`, `t = "nagaram"`, the output should be `solution(s, t) = True`.

  **Explanation:** Both strings contain exactly `a×3, n×1, g×1, r×1, m×1`.

- For `s = "rat"`, `t = "car"`, the output should be `solution(s, t) = False`.

  **Explanation:** `s` has `r,a,t` but `t` has `c,a,r` — `t` and `c` don't match.

- For `s = "a"`, `t = "a"`, the output should be `solution(s, t) = True`.

  **Explanation:** Same single character.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `string` `s` — A string of lowercase English letters.  
  Guaranteed constraints: `1 ≤ s.length ≤ 5·10^4`
- **[input]** `string` `t`  
  Guaranteed constraints: `1 ≤ t.length ≤ 5·10^4`
- **[output]** `boolean` — `True` if anagram, `False` otherwise.

```python
def solution(s, t):
    counts = [0] * 26
    for c in s:
        counts[ord(c) - ord('a')] += 1
    for c in t:
        counts[ord(c) - ord('a')] -= 1    
    
    return all(x == 0 for x in counts)
    
assert solution('anagram', 'nagaram') == True   # given example 1 — full rearrangement
assert solution('rat', 'car')         == False  # given example 2 — different letters
assert solution('a', 'a')             == True   # given example 3 — single char match
assert solution('ab', 'a')            == False  # different lengths — never an anagram
assert solution('listen', 'silent')   == True   # classic anagram pair
assert solution('hello', 'world')     == False  # same length, different letters
assert solution('aab', 'baa')         == True   # duplicate letters, reordered
assert solution('aab', 'bab')         == False  # same length, wrong counts
print('All Pass!')
```
