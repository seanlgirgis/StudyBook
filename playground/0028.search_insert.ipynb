## Search Insert Position

Given a sorted array of distinct integers `nums` and a target integer `target`, return the **index** of `target` if it exists. Otherwise, return the **index where it would be inserted** to keep the array sorted.

Your solution must run in **O(log n)** time.

### Example

- For `nums = [1,3,5,6]`, `target = 5`, the output should be `solution(nums, target) = 2`.

  **Explanation:** `5` exists at index `2`.

- For `nums = [1,3,5,6]`, `target = 2`, the output should be `solution(nums, target) = 1`.

  **Explanation:** `2` is not in the array. It would be inserted between `1` and `3`, at index `1`.

- For `nums = [1,3,5,6]`, `target = 7`, the output should be `solution(nums, target) = 4`.

  **Explanation:** `7` is larger than all elements — insert at the end.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A sorted array of distinct integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^4`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[input]** `integer` `target`
- **[output]** `integer` — Index of target or insertion position.

```python
def solution(nums, target):
    l, r = 0, len(nums) -1

    while r >= l:
        mid = l + (r-l)//2
        if target == nums[mid]:
            return mid
        elif target > nums[mid]:
            l = mid+1
        else:
            r = mid -1

    return l


assert solution([1,3,5,6], 5)   == 2  # given example 1 — found
assert solution([1,3,5,6], 2)   == 1  # given example 2 — insert in middle
assert solution([1,3,5,6], 7)   == 4  # given example 3 — insert at end
assert solution([1,3,5,6], 0)   == 0  # insert at start
assert solution([1], 0)         == 0  # single element, insert before
assert solution([1], 1)         == 0  # single element, found
assert solution([1], 2)         == 1  # single element, insert after
assert solution([1,3,5,6], 6)   == 3  # found at last index
print('All Pass!') 
```
