## CodeSignal — Grid Shape Stamp

Build an `m × n` grid (all zeros). Process a list of shape queries in order.
For each shape, find the **first available position** (row by row, left to right)
where the shape fits entirely within bounds AND all cells are empty.
Stamp that position with the **1-based query index**. If no position fits, skip.

### Shapes (anchor = top-left cell, offsets from anchor)

```
A:  X            (1×1 single cell)

B:  X X X        (1×3 horizontal)

C:  X X          (2×2 square)
    X X

D:  X .          (vertical bar, right bump in middle)
    X X
    X .

E:  . X .        (horizontal bar, top bump in middle)
    X X X
```

### Rules
- Scan positions row by row, left to right — use the **first** position that fits.
- A position fits if: all shape cells are **in bounds** AND **currently empty (== 0)**.
- Later shapes cannot overlap earlier ones.
- If no position fits for a shape, skip it silently.
- Return the final grid.

### Example

```
solve_grid(3, 3, ["A", "C"])
Shape A → first fit (0,0) → stamp 1
Shape C → (0,0) blocked, first fit (0,1) → stamp 2
Result:
[[1, 2, 2],
 [0, 2, 2],
 [0, 0, 0]]
```

### Hint
`canFit(r, c, offsets)` — check bounds AND `grid[r+dr][c+dc] == 0` for all offsets.
`mark_shape(offsets, j)` — scan grid, call `canFit`, stamp on first hit, return.
