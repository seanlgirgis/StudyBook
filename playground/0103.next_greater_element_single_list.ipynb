## Next Greater Element — Single List

Given a single array `nums`, for each element return the first element to its **right** that is greater. If none exists, return `-1`.

### Example

- `[2,1,2,4,3]` → `[4,2,4,-1,-1]`
- `[1,2,3]` → `[2,3,-1]`
- `[3,2,1]` → `[-1,-1,-1]`

```python
%run _timer.py
```

```python
def solution(nums):
    n = len(nums)
    ans = [-1] * n
    stack = []
    for i, val in enumerate(nums):
        while stack and val > nums[stack[-1]]:
            ans[stack.pop()] = val
        stack.append(i)
    return ans


assert solution([2,1,2,4,3]) == [4,2,4,-1,-1]
assert solution([1,2,3])     == [2,3,-1]
assert solution([3,2,1])     == [-1,-1,-1]
assert solution([1])         == [-1]
print('All Pass!')
```
