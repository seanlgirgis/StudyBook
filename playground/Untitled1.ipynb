```python
import heapq

class MinHeap:
    def __init__(self):
        self._heap = []
        
    def push(self, val):
        heapq.heappush(self._heap, val)

    def pop(self):
        return heapq.heappop(self._heap)

    def peek(self):
        return self._heap[0]

    def __len__(self):
        return len(self._heap)

    def __bool__(self):
        return bool(self._heap)
        


class MaxHeap:
    def __init__(self):
        self._heap = []
        
    def push(self, val):
        heapq.heappush(self._heap, -val)

    def pop(self):
        return -heapq.heappop(self._heap)

    def peek(self):
        return -self._heap[0]

    def __len__(self):
        return len(self._heap)

    def __bool__(self):
        return bool(self._heap)

mn = MinHeap()
mn.push(9)
mn.push(3)
mn.push(18)

while(mn):
    print(len(mn), mn.peek()  , mn.pop())

    
mx = MaxHeap()
mx.push(9)
mx.push(3)
mx.push(18)

while(mx):
    print(len(mx), mx.peek()  , mx.pop())

# --- USER TO IMPLEMENT SOLUTION BELOW ---
class MedianFinder:
    def __init__(self):
        """
        initialize your data structure here.
        """
        self.small, self
        

    def addNum(self, num: int) -> None:

            

    def findMedian(self) -> float:



        
```

```python
from typing import List, Tuple, Callable
from collections import deque
import heapq
def solution(t: int, capacity: List[int], recharge: List[int]) -> int:
    # my understanding
    # return number of batteries used to reach t time
    # if final battery does not fully deplete. Do not count it
    # if t not achieved and no batteries avilable... you return -1
    inuse, out, clock, n= None, 0, 0, len(capacity)
    ready = deque(range(n))
    charging = []                # heapified list of (ready_time, index)
    def swapper():
        if charging : print ("In Swapper   ", clock , charging[0][0])
        while charging and charging[0][0] <= clock:
            
            _, idx = heapq.heappop(charging)
            print("Swapping ", idx)
            ready.append(idx)
            
    while t > clock and ready:
        swapper()
        print("-" * 75)
        print( "Clock is " , clock)
        print( "Ready is ", ready)
        print ("charging is", charging)
        if not ready:
            return -1
        else:
            #put battery on charger and get a new one
            if inuse is not None:
                heapq.heappush(charging,(clock + recharge[inuse] ,inuse))
                out += 1
                
            idx = ready.popleft()
            print("Will Use now battery    " , idx)
            inuse = idx
            #advance time
            clock += capacity[idx]
            if clock > t:
                break
        print( "Ready is ", ready)
        print ("charging is", charging)
    return out

solution(16, [2, 5, 6], [12, 8, 4])
```

```python

class MinStack:

    def __init__(self):
        self.data = []
        self.mstack = [] 
        pass

    def push(self, val: int) -> None:
        self.data.append(val)
        
    def pop(self) -> None:
        if self.data :
            val = self.data.pop()
            if self.mstack and val == self.mstack[-1]:
                self.mstack.pop()
            
            
    def top(self) -> int:
        if self.data :
            return self.data [-1]

    def getMin(self) -> int:
        if self.mstack :
            return self.mstack[-1]

mstack = MinStack()
mstack.push(-3)
```
