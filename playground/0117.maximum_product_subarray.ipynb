## Maximum Product Subarray

Given an integer array `nums`, find the subarray with the largest product and return that product.

### Example

- `[2,3,-2,4]` → `6`  (subarray `[2,3]`)
- `[-2,0,-1]` → `0`
- `[-2,3,-4]` → `24`  (whole array: `-2*3*-4=24`)

### Hint
Track both `cur_max` and `cur_min` at each step — a negative times a negative can become the new max. Reset both to `nums[i]` if `0` is encountered.

### Input/Output
- **[input]** `array.integer` `nums`
- **[output]** `integer`

```python
%run _timer.py
```

```python
def solution(nums):
    if not nums:
        return 0
    
    res = nums[0]
    cur_max, cur_min = nums[0], nums[0]
    
    for i in range(1, len(nums)):
        n = nums[i]
        
        old_max = cur_max
        
        # tracking max and min to compensate for product of negative number
        cur_max = max(n, n * cur_max, n * cur_min)
        cur_min = min(n, n * old_max, n * cur_min)
        
        # Update the global result
        res = max(res, cur_max)
        
    return res


assert solution([2,3,-2,4])   == 6
assert solution([-2,0,-1])    == 0
assert solution([-2,3,-4])    == 24
assert solution([0,2])        == 2
assert solution([-2])         == -2
assert solution([2,-5,-2,-4,3]) == 24
print('All Pass!')
```
