## House Robber II

Same as House Robber but houses are in a **circle** — first and last are adjacent so you can't rob both.

- `[2,3,2]` → `3`
- `[1,2,3,1]` → `4`

**Trick:** Run house robber twice — once on `nums[:-1]`, once on `nums[1:]`. Take the max.

```python
%run _timer.py
```

```python
def solution(nums):
    if len(nums) == 1:
        return nums[0]

    def rob(arr):
        # Standard house robber O(1) space
        prev2, prev1 = 0, 0
        for n in arr:
            prev2, prev1 = prev1, max(prev1, prev2 + n)
        return prev1

    # Can't rob both first and last — run on each half
    return max(rob(nums[:-1]), rob(nums[1:]))


assert solution([2,3,2])   == 3
assert solution([1,2,3,1]) == 4
assert solution([1,2,3])   == 3
assert solution([1])       == 1
assert solution([2,1,1,2]) == 3
print('All Pass!')
```
