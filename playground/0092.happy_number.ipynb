## Happy Number

A **happy number** is defined by the following process:

- Starting with any positive integer, replace the number by the sum of the squares of its digits.
- Repeat until the number equals `1` (happy) or it loops forever (not happy).

Return `True` if `n` is a happy number, `False` otherwise.

### Example

- `19` → `1² + 9² = 82` → `8² + 2² = 68` → `6² + 8² = 100` → `1² + 0² + 0² = 1` ✓ happy
- `2` → eventually loops, never reaches 1 → not happy

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `integer` `n`
- **[output]** `boolean`

```python
%run _timer.py
```

```python
def solution(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        total = 0
        for digit in str(n):
            total += int(digit) ** 2
        n = total
    return n == 1


assert solution(19)  == True
assert solution(1)   == True
assert solution(2)   == False
assert solution(7)   == True
assert solution(4)   == False
assert solution(100) == True
assert solution(116) == False
print('All Pass!')
```
