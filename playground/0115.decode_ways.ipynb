## Decode Ways

A string of digits can be decoded: `'1'→A`, `'2'→B`, ..., `'26'→Z`. Count the number of ways to decode the string `s`.

### Example

- `"12"` → `2`  (`"AB"` or `"L"`)
- `"226"` → `3`  (`"BZ"`, `"VF"`, `"BBF"`  wait: `"BBF"`, `"BZ"`, `"VF"`)
- `"06"` → `0`  (leading zero — invalid)
- `"10"` → `1`  (`"J"` only — `"10"` as two digits is invalid)

### Hint
DP — `dp[i]` = ways to decode `s[:i]`. One-char decode if `s[i-1] != '0'`. Two-char decode if `10 <= int(s[i-2:i]) <= 26`.

### Input/Output
- **[input]** `string` `s`
- **[output]** `integer`

```python
def solution(s):
    if not s or s[0] == '0':
        return 0
    
    n = len(s)
    dp = [0] * (n + 1)
    
    # Base cases
    dp[0] = 1  # Dummy to start
    dp[1] = 1  # s[0] is not '0', so there's 1 way
    
    for i in range(2, n + 1):
        # One digit 
        if s[i-1] != '0':
            dp[i] += dp[i-1]
            
        # Two digit  
        two_digit = int(s[i-2:i])
        if 10 <= two_digit <= 26:
            dp[i] += dp[i-2]
            
    return dp[n]

# Verification
assert solution("12")    == 2
assert solution("226")   == 3
assert solution("06")    == 0
assert solution("10")    == 1
assert solution("1")     == 1
assert solution("11106") == 2
print('All Pass!')
```
