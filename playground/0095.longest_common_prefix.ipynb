## Longest Common Prefix

Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string `""`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n · m)** will fit within the execution time limit (n = number of strings, m = length of shortest string).

### Example

- `strs = ["flower","flow","flight"]` → `"fl"`
- `strs = ["dog","racecar","car"]` → `""`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.string` `strs`
- **[output]** `string`

```python
def solution(items):
    if not items: return ""
    n = len(items)
    if n == 1: return items[0]
    mc = ""
    for i, c in enumerate(items[0]):
        for itm in items[1:]:
            if not i < len(itm) or itm[i] != c:
                return mc
        mc += c
    return mc
        

solution(["flower","flow","flight"])
assert solution(["flower","flow","flight"]) == "fl"
assert solution(["dog","racecar","car"])     == ""
assert solution(["interview","inter","internal"]) == "inter"
assert solution(["a"])                       == "a"
assert solution(["ab","a"])                  == "a"
assert solution(["abc","abc","abc"])         == "abc"
print('All Pass!')
```
