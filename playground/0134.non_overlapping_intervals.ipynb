## Non-overlapping Intervals

Return the minimum number of intervals to **remove** so the rest don't overlap.

- `[[1,2],[2,3],[3,4],[1,3]]` → `1` (remove `[1,3]`)
- `[[1,2],[1,2],[1,2]]` → `2`

**Greedy:** Sort by end time. Keep an interval if it starts at or after the last kept end. Count removed ones.

```python
%run _timer.py
```

```python
def solution(intervals):
    intervals.sort(key=lambda x: x[1])  # sort by end time
    removed  = 0
    last_end = float('-inf')

    for start, end in intervals:
        if start >= last_end:   # no overlap — keep it
            last_end = end
        else:                   # overlap — remove it
            removed += 1

    return removed


assert solution([[1,2],[2,3],[3,4],[1,3]]) == 1
assert solution([[1,2],[1,2],[1,2]])       == 2
assert solution([[1,2],[2,3]])             == 0
print('All Pass!')
```
