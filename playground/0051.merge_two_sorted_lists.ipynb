## Merge Two Sorted Lists

You are given the heads of two **sorted** linked lists `list1` and `list2`.

Merge the two lists into one **sorted** list and return its head.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n + m)** will fit within the execution time limit.

### Example

- For `list1 = [1,2,4]` and `list2 = [1,3,4]`, the output should be `[1,1,2,3,4,4]`.
- For `list1 = []` and `list2 = []`, the output should be `[]`.
- For `list1 = []` and `list2 = [0]`, the output should be `[0]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `list1`, `ListNode` `list2`  
  Guaranteed constraints: `0 ≤ n, m ≤ 50`, `−100 ≤ val ≤ 100`
- **[output]** `ListNode` — Head of the merged sorted list.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

def to_ll(lst):
    dummy = ListNode()
    cur = dummy
    for v in lst:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def solution(list1, list2):
    dummy = ListNode()
    tail = dummy

    curr1, curr2 = list1, list2

    while curr1 and curr2:
        if curr1.val <= curr2.val:
            tail.next = curr1
            curr1 = curr1.next
        else:
            tail.next = curr2
            curr2 = curr2.next
        tail = tail.next

    # attach the remaining part of whichever list isn't exhausted
    tail.next = curr1 if curr1 else curr2

    return dummy.next


assert to_list(solution(to_ll([1,2,4]), to_ll([1,3,4])))  == [1,1,2,3,4,4]  # given example 1
assert to_list(solution(to_ll([]), to_ll([])))             == []              # given example 2
assert to_list(solution(to_ll([]), to_ll([0])))            == [0]             # given example 3
assert to_list(solution(to_ll([1]), to_ll([2])))           == [1,2]           # single nodes
assert to_list(solution(to_ll([2]), to_ll([1])))           == [1,2]           # single nodes reversed
assert to_list(solution(to_ll([1,3,5]), to_ll([2,4,6])))  == [1,2,3,4,5,6]  # interleaved
assert to_list(solution(to_ll([1,2,3]), to_ll([4,5,6])))  == [1,2,3,4,5,6]  # list1 all smaller
assert to_list(solution(to_ll([1,1,1]), to_ll([1,1,1])))  == [1,1,1,1,1,1]  # all duplicates
print('All Pass!')
```
