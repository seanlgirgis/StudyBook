## Printer Slots Round-Robin (Custom Simulation)

Printers in circular order. `printer_pages[i]` pages per use, `recharge[i]` minutes cooldown. Print `target` pages total.

**Same pattern as 0140** — deque circular scan, full use counts, partial final use doesn't, no node ready → `-1`.

```python
%run _timer.py
```

```python
from collections import deque

def solution(target, printer_pages, recharge):
    n        = len(printer_pages)
    ready_at = [0] * n
    order    = deque(range(n))
    clock    = 0
    used     = 0

    while clock < target:
        found = False
        for _ in range(n):
            i = order[0]
            order.rotate(-1)
            if ready_at[i] <= clock:
                found = True
                rem = target - clock
                if printer_pages[i] <= rem:
                    clock      += printer_pages[i]
                    used       += 1
                    ready_at[i] = clock + recharge[i]
                else:
                    return used
                break
        if not found:
            return -1

    return used


assert solution(10, [7,5],     [12,4])     == 1
assert solution(8,  [2,5],     [10,4])     == -1
assert solution(9,  [3,3,3],   [100,100,100]) == 3
assert solution(10, [4,4,4],   [1,1,1])    == 2
print('All Pass!')
```
