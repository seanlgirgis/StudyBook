## Longest Common Subsequence (LC 1143)

Length of the longest subsequence common to both strings (characters in order, not necessarily adjacent).

**Pattern:** 2D DP — same grid shape as Edit Distance.

```
if text1[i-1] == text2[j-1]:   dp[i][j] = dp[i-1][j-1] + 1      (match — grow LCS)
else:                           dp[i][j] = max(dp[i-1][j],        (skip text1 char)
                                               dp[i][j-1])        (skip text2 char)
```

**vs Edit Distance:** Edit distance uses `1 + min(...)` on mismatch — LCS uses `max(...)` with no cost.

````python
def solution(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]   # base case: empty string = 0

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1              # match — extend LCS
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])   # skip one char, take best

    return dp[m][n]


assert solution("abcde", "ace")  == 3
assert solution("abc",   "abc")  == 3
assert solution("abc",   "def")  == 0
assert solution("a",     "a")    == 1
assert solution("a",     "b")    == 0
assert solution("",      "abc")  == 0
assert solution("oxcpqrsvwf", "shmtulqrypy") == 2
print('All Pass!')
````
