## Longest Increasing Subsequence

Given an integer array `nums`, return the length of the **longest strictly increasing subsequence** (elements don't have to be contiguous).

### Example

- `[10,9,2,5,3,7,101,18]` → `4`  (`[2,3,7,101]` or `[2,5,7,101]`)
- `[0,1,0,3,2,3]` → `4`
- `[7,7,7,7]` → `1`

### Hint
`dp[i]` = length of LIS ending at index `i`. For each `i`, look back at all `j < i` where `nums[j] < nums[i]`: `dp[i] = max(dp[i], dp[j] + 1)`. Answer is `max(dp)`.

### Input/Output
- **[input]** `array.integer` `nums`
- **[output]** `integer`

```python
%run _timer.py
```

```python
def solution(nums):
    # dp[i] = length of LIS ending exactly at index i
    # Every element alone is a subsequence of length 1
    dp = [1] * len(nums)

    for i in range(1, len(nums)):
        for j in range(i):
            if nums[j] < nums[i]:              # nums[i] can extend the LIS ending at j
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)


assert solution([10,9,2,5,3,7,101,18]) == 4
assert solution([0,1,0,3,2,3])         == 4
assert solution([7,7,7,7])             == 1
assert solution([1,2,3,4,5])           == 5
assert solution([5,4,3,2,1])           == 1
print('All Pass!')
```
