## Search a 2D Matrix

You are given an `m x n` integer matrix with the following properties:

- Each row is sorted in ascending order.
- The first integer of each row is greater than the last integer of the previous row.

Given an integer `target`, return `True` if it exists in the matrix, `False` otherwise.

**Note:** Your solution must run in **O(log(m·n))** time.

### Example

- For `matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]` and `target = 3`, the output should be `True`.
- For the same matrix and `target = 13`, the output should be `False`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.array.integer` `matrix`
- **[input]** `integer` `target`
- **[output]** `boolean`

```python
%run _timer.py
```

```python
def itemAt(i, matrix, m, n):
    return matrix[i // n][i % n]


def solution(matrix, target):
    m, n = len(matrix), len(matrix[0])
    lenth = m * n
    l, r = 0, lenth - 1
    while l <= r:
        mid = l + (r - l) // 2
        val = itemAt(mid, matrix, m, n)
        if target == val:
            return True
        if target > val:
            l = mid + 1
        else:
            r = mid - 1
    return False


m = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
assert solution(m,  3) == True
assert solution(m, 13) == False
assert solution(m,  1) == True    # top-left corner
assert solution(m, 60) == True    # bottom-right corner
assert solution(m,  0) == False   # below minimum
assert solution(m, 99) == False   # above maximum
assert solution([[1]], 1) == True  # 1x1 matrix hit
assert solution([[1]], 2) == False # 1x1 matrix miss
print('All Pass!')
```
