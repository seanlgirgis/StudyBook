```python
from typing import List, Callable, Tuple
def canCompleteCircuit(gas: List[int], cost: List[int]) -> int:
    """
    Finds the starting gas station index to complete a circular circuit.
    """
    net_gain = [x-y for x,y in zip(gas,cost)]
    if sum(net_gain) < 0 :   # a whoel circle can never happen
        return -1
        
    current_balance= 0
    start_index= 0
    
    for i, amount in enumerate(net_gain):
        current_balance += amount
        #if current_balance < 0 reset
        if current_balance < 0:
            current_balance = 0
            start_index = i + 1
            
    return start_index
            
        
    
# Standard Example 1 call
gas_input = [1, 2, 3, 4, 5]
cost_input = [3, 4, 5, 1, 2]

result = canCompleteCircuit(gas_input, cost_input)

print(f"Starting Station Index: {result}")
```
