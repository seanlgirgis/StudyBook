```python
from collections import deque
import itertools
def oranges_rotting(grid):
    # your code
    rows, cols = len(grid), len(grid[0])
    combinations = list(itertools.product(range(rows), range(cols)))
    clock = 0
    fresh = 0
    q = deque()
    for r,c in combinations:
        if grid[r][c] == 2:
            q.append((r,c,0))           # coordinates of rotten Orange and time 0
        elif grid[r][c] == 1:
            fresh += 1
    while q:
        r, c ,time = q.popleft()
        for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1:
                grid[nr][nc] = 2
                fresh -= 1
                clock = time + 1
                q.append((nr,nc,clock))
    return clock if fresh == 0 else -1
    
print(oranges_rotting([[2,1,1],[1,1,0],[0,1,1]]))  # 4
print(oranges_rotting([[2,1,1],[0,1,1],[1,0,1]]))  # -1
print(oranges_rotting([[0,2]]))                    # 0
```

```python
from typing import List
def unique_paths(m: int, n: int) -> int:
    grid: list[list[int]] = []
    for i in range(m):
        grid.append([1 for I in range(n)])
    for i in range(1,m):
        for j in range(1,n):
            grid[i][j] = grid[i-1][j] + grid[i][j-1]
    return grid[-1][-1]
print(unique_paths(3, 7))  # 28
print(unique_paths(3, 2))  # 3
print(unique_paths(1, 5))  # 1
print(unique_paths(5, 1))  # 1
```

```python
# Coin Change (LeetCode 322)
#
# Problem:
# You are given a list of coin denominations `coins` and an integer `amount`.
# Return the minimum number of coins needed to make up that amount.
# If it is not possible to make that amount, return -1.
#
# Notes:
# - You can use each coin unlimited times.
# - amount can be 0 (answer should be 0).
#
# Examples:
# coins = [1,2,5], amount = 11 -> 3   (5 + 5 + 1)
# coins = [2], amount = 3 -> -1
# coins = [1], amount = 0 -> 0

from typing import List

def coin_change(coins: List[int], amount: int) -> int:
    """
    Finds the fewest number of coins needed to make up 'amount'.
    """
    inf_amt = amount + 1
    buckets = [inf_amt] * (amount + 1)
    
    buckets[0] = 0
    
    for i in range(1, len(buckets)):
        n = inf_amt
        for coin in coins:
            rem = i - coin
            if rem >= 0:
                if buckets[rem] != inf_amt:
                    n = min(n, buckets[rem] + 1)
        if n != inf_amt: buckets[i] = n
    return buckets[-1] if buckets[-1] != inf_amt else -1


# Straight run calls
print(coin_change([1,2,5], 11))          # expected: 3
print(coin_change([2], 3))               # expected: -1
print(coin_change([1], 0))               # expected: 0
print(coin_change([1], 2))               # expected: 2
print(coin_change([2,5,10,1], 27))       # expected: 4
print(coin_change([186,419,83,408], 6249))  # expected: 20
print(coin_change([2,4,6], 8))           # expected: 2
print(coin_change([2,4,6], 7))           # expected: -1
```

```python
# Longest Increasing Subsequence (LeetCode 300)
#
# Problem:
# Given an integer array nums, return the length of the longest strictly
# increasing subsequence.
#
# Notes:
# - A subsequence is derived by deleting some or no elements without changing order.
# - Strictly increasing means each next value must be > previous.
#
# Example:
# nums = [10,9,2,5,3,7,101,18] -> 4   (e.g., [2,3,7,101])

from typing import List

def lengthOfLIS(nums: List[int]) -> int:
    n = len(nums)
    if n == 0 : return 0
    if n == 1 : return 1
    longest = 0
    LIS = [1] * n
    for i in range(n -1, -1, -1):
        for j in range (i+1, n):
            if nums[i] < nums[j]:
                LIS[i] = max(LIS[i], 1 + LIS[j])
                longest  = max(longest , LIS[i])
    return max(LIS)
        



# Straight run calls
print(lengthOfLIS([10,9,2,5,3,7,101,18]))  # expected: 4
print(lengthOfLIS([0,1,0,3,2,3]))          # expected: 4
print(lengthOfLIS([7,7,7,7,7,7,7]))        # expected: 1
print(lengthOfLIS([1,3,6,7,9,4,10,5,6]))   # expected: 6
print(lengthOfLIS([1]))                     # expected: 1
print(lengthOfLIS([]))                      # expected: 0
```

```python
# Decode Ways (LeetCode 91)
#
# Problem:
# A message containing letters A-Z is encoded as:
# '1' -> 'A', ..., '26' -> 'Z'
#
# Given a string s containing only digits, return the number of ways to decode it.
#
# Rules:
# - '0' cannot be decoded alone.
# - Valid two-digit decode must be between "10" and "26".
#
# Examples:
# "12"   -> 2  ("AB", "L")
# "226"  -> 3  ("BZ", "VF", "BBF")
# "06"   -> 0

def num_decodings(s: str) -> int:
    # Base case: an empty string (end of input) counts as 1 valid decoding path
    dp = { len(s) : 1 }
    
    # Iterate backwards through the string
    for i in range(len(s) - 1, -1, -1):
        # Case 1: If current digit is '0', it cannot start a valid letter
        if s[i] == "0":
            dp[i] = 0
        else:
            # Standard case: Use current digit as a single letter
            dp[i] = dp[i + 1]
    
        # Case 2: Check if current and next digit can form a valid double-digit letter (10-26)
        if (i + 1 < len(s) and (s[i] == "1" or 
            s[i] == "2" and s[i + 1] in "0123456")):
            # Add the paths possible from the position after the pair
            dp[i] += dp[i + 2]
    
    # The result is the number of ways to decode starting from the first character
    return dp[0]


# Straight run calls
print(num_decodings("12"))      # expected: 2
print(num_decodings("226"))     # expected: 3
print(num_decodings("06"))      # expected: 0
print(num_decodings("10"))      # expected: 1
print(num_decodings("2101"))    # expected: 1
print(num_decodings("11106"))   # expected: 2
print(num_decodings("27"))      # expected: 1
print(num_decodings("0"))       # expected: 0
```
