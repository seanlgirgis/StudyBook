```python
from collections import deque
import itertools
def oranges_rotting(grid):
    # your code
    rows, cols = len(grid), len(grid[0])
    combinations = list(itertools.product(range(rows), range(cols)))
    clock = 0
    fresh = 0
    q = deque()
    for r,c in combinations:
        if grid[r][c] == 2:
            q.append((r,c,0))           # coordinates of rotten Orange and time 0
        elif grid[r][c] == 1:
            fresh += 1
    while q:
        r, c ,time = q.popleft()
        for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1:
                grid[nr][nc] = 2
                fresh -= 1
                clock = time + 1
                q.append((nr,nc,clock))
    return clock if fresh == 0 else -1
    
print(oranges_rotting([[2,1,1],[1,1,0],[0,1,1]]))  # 4
print(oranges_rotting([[2,1,1],[0,1,1],[1,0,1]]))  # -1
print(oranges_rotting([[0,2]]))                    # 0
```

```python
from typing import List
def unique_paths(m: int, n: int) -> int:
    grid: list[list[int]] = []
    for i in range(m):
        grid.append([1 for I in range(n)])
    for i in range(1,m):
        for j in range(1,n):
            grid[i][j] = grid[i-1][j] + grid[i][j-1]
    return grid[-1][-1]
print(unique_paths(3, 7))  # 28
print(unique_paths(3, 2))  # 3
print(unique_paths(1, 5))  # 1
print(unique_paths(5, 1))  # 1
```

```python
# Coin Change (LeetCode 322)
#
# Problem:
# You are given a list of coin denominations `coins` and an integer `amount`.
# Return the minimum number of coins needed to make up that amount.
# If it is not possible to make that amount, return -1.
#
# Notes:
# - You can use each coin unlimited times.
# - amount can be 0 (answer should be 0).
#
# Examples:
# coins = [1,2,5], amount = 11 -> 3   (5 + 5 + 1)
# coins = [2], amount = 3 -> -1
# coins = [1], amount = 0 -> 0

from typing import List

def coin_change(coins: List[int], amount: int) -> int:
    """
    Finds the fewest number of coins needed to make up 'amount'.
    """
    inf_amt = amount + 1
    buckets = [inf_amt] * (amount + 1)
    
    buckets[0] = 0
    
    for i in range(1, len(buckets)):
        n = inf_amt
        for coin in coins:
            rem = i - coin
            if rem >= 0:
                if buckets[rem] != inf_amt:
                    n = min(n, buckets[rem] + 1)
        if n != inf_amt: buckets[i] = n
    return buckets[-1] if buckets[-1] != inf_amt else -1


# Straight run calls
print(coin_change([1,2,5], 11))          # expected: 3
print(coin_change([2], 3))               # expected: -1
print(coin_change([1], 0))               # expected: 0
print(coin_change([1], 2))               # expected: 2
print(coin_change([2,5,10,1], 27))       # expected: 4
print(coin_change([186,419,83,408], 6249))  # expected: 20
print(coin_change([2,4,6], 8))           # expected: 2
print(coin_change([2,4,6], 7))           # expected: -1
```
