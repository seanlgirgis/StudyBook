## Task Scheduler

Given tasks and a cooldown `n`, return the minimum intervals needed. Identical tasks must be `n` intervals apart. CPU can idle.

- `tasks=["A","A","A","B","B","B"]`, `n=2` → `8` (A,B,idle,A,B,idle,A,B)
- `tasks=["A","A","A","B","B","B"]`, `n=0` → `6`

**Greedy:** Always execute the most frequent available task. Use a max-heap + cooldown queue.

```python
%run _timer.py
```

```python
import heapq
from collections import Counter, deque

def solution(tasks, n):
    count  = Counter(tasks)
    heap   = [-c for c in count.values()]   # max-heap via negation
    heapq.heapify(heap)

    time  = 0
    queue = deque()   # (negative_count, available_at_time)

    while heap or queue:
        time += 1

        if heap:
            cnt = heapq.heappop(heap) + 1   # use one of this task (-3 → -2)
            if cnt:                          # still tasks of this type left
                queue.append((cnt, time + n))

        # Release tasks whose cooldown has expired
        if queue and queue[0][1] == time:
            heapq.heappush(heap, queue.popleft()[0])

    return time


assert solution(["A","A","A","B","B","B"], 2) == 8
assert solution(["A","A","A","B","B","B"], 0) == 6
assert solution(["A","A","A","A","A","A"], 1) == 11
assert solution(["A","B","C","D"],          3) == 4
print('All Pass!')
```
