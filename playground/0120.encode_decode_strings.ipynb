## Encode and Decode Strings

Design an algorithm to encode a list of strings to a single string, and decode it back. The encoded string must be transmittable and decodable even if strings contain any character.

### Example

- `["hello","world"]` → encode → some string → decode → `["hello","world"]`
- `["a","b","c"]` → encode → decode → `["a","b","c"]`
- `["with spaces","and#special"]` → encode → decode → same
- `[""]` → encode → decode → `[""]`

### Hint
Use a length-prefix format: `"5#hello5#world"` — prepend each word's length + `#`. On decode, read length, skip `#`, slice that many chars.

### Input/Output
- **[input]** `array.string`
- **[output]** `array.string` (after encode then decode)

```python
def encode(strs):
    out= ''
    for s in strs:
        n = len(s)
        encoded = str(n) + "~" + s
        out+=(encoded)
    return out


def decode(s):
    out = []
    i = 0 

    while i < len(s):
        sep = s.find("~", i)
        n = int(s[i:sep])
        start = sep + 1
        end = start + n
        out.append(s[start:end])
        i = end
    return out


def round_trip(strs):
    return decode(encode(strs))


decode(encode(["hello","world"]))
assert round_trip(["hello","world"])          == ["hello","world"]
assert round_trip(["a","b","c"])              == ["a","b","c"]
assert round_trip(["with spaces","and#spec"]) == ["with spaces","and#spec"]
assert round_trip([""])                       == [""]
assert round_trip(["hello"])                  == ["hello"]
print('All Pass!')
```

```python
def encode(strs):
    # Format: "<length>#<word>" for each word
    # e.g. ["hello","world"] → "5#hello5#world"
    # Works even if words contain '#' or spaces
    return ''.join(f"{len(w)}#{w}" for w in strs)


def decode(s):
    res = []
    i = 0
    while i < len(s):
        j = s.index('#', i)          # find the next '#'
        length = int(s[i:j])         # everything before '#' is the length
        res.append(s[j+1 : j+1+length])  # slice exactly that many chars
        i = j + 1 + length           # advance past the word
    return res


def round_trip(strs):
    return decode(encode(strs))

assert round_trip(["hello","world"])          == ["hello","world"]
assert round_trip(["a","b","c"])              == ["a","b","c"]
assert round_trip(["with spaces","and#spec"]) == ["with spaces","and#spec"]
assert round_trip([""])                       == [""]
assert round_trip(["hello"])                  == ["hello"]
print('All Pass!')
```
