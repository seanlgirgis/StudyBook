## Sum of Two Integers (LC 371)

Add `a + b` without using `+` or `-`.

**Pattern:** XOR + carry simulation — exactly how a hardware adder works.

| Operation | What it gives |
|-----------|---------------|
| `a ^ b` | Sum bits **without** carry |
| `(a & b) << 1` | **Carry** bits shifted into next column |

Repeat until carry is zero.

**Python gotcha:** Python integers have infinite precision — a negative carry loop never terminates.  
Fix: apply a 32-bit mask `0xFFFFFFFF` on every step to stay in fixed-width land.  
After the loop, convert back if the result is a negative 32-bit number.

```python
%run _timer.py
```

```python
def solution(a, b):
    MASK = 0xFFFFFFFF   # 32-bit fence — clips everything above bit 31
    MAX  = 0x7FFFFFFF   # largest positive 32-bit int (sign bit = 0)

    while b & MASK:                        # keep going while carry bits remain
        carry = ((a & b) << 1) & MASK      # AND = where carries happen, shift left
        a     = (a ^ b) & MASK             # XOR = sum without carry
        b     = carry                      # carry becomes next round's b

    # If result > MAX, the 32-bit sign bit is set → negative number
    # ~(a ^ MASK) converts two's-complement back to Python's signed int
    return a if a <= MAX else ~(a ^ MASK)


assert solution(1, 2)       == 3
assert solution(3, 5)       == 8
assert solution(0, 0)       == 0
assert solution(-1, 1)      == 0
assert solution(-3, -5)     == -8
assert solution(1000, 1000) == 2000
print('All Pass!')
```
