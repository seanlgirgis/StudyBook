## Kth Largest Element in an Array

Return the `k`th largest element (not kth distinct).

- `[3,2,1,5,6,4]`, k=2 → `5`
- `[3,2,3,1,2,4,5,5,6]`, k=4 → `4`

```python
%run _timer.py
```

```python
import heapq

def solution(nums, k):
    # Min-heap of size k — top is always the kth largest
    heap = []
    for n in nums:
        heapq.heappush(heap, n)
        if len(heap) > k:
            heapq.heappop(heap)   # evict smallest — keeps k largest
    return heap[0]                # smallest of the k largest = kth largest


assert solution([3,2,1,5,6,4],       2) == 5
assert solution([3,2,3,1,2,4,5,5,6], 4) == 4
assert solution([1],                 1) == 1
assert solution([5,2,4,1,3,6,0],     3) == 4
print('All Pass!')
```
