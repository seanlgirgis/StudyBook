## Find Median from Data Stream (LC 295)

Support `addNum(n)` and `findMedian()` on a stream of numbers.

**Pattern:** Two heaps — split numbers into lower half and upper half.

| Heap | Type | Contains |
|------|------|----------|
| `lo` | max-heap (negate) | lower half — top = largest small number |
| `hi` | min-heap | upper half — top = smallest large number |

**Invariants after every add:**
1. `lo` top ≤ `hi` top (ordering: lower half ≤ upper half)
2. `len(lo) == len(hi)` or `len(lo) == len(hi) + 1` (lo can have one extra)

**Median:** if sizes equal → average of both tops; else → top of `lo`.

```python
"""
id: lc_0295
title: Find Median from Data Stream
source: leetcode
difficulty: hard
primary: heaps
tags: [heap, design, data-stream]
leetcode_url: https://leetcode.com/problems/find-median-from-data-stream/
status: draft
last_updated: 2026-04-23
notes:
- key idea: two heaps (max-heap for lower half, min-heap for upper half)
- time: addNum O(log n), findMedian O(1)
- space: O(n)
"""

# ============================================================================
# File: 295_lc_0295_find_median_from_data_stream_empty.py
# LeetCode 295: Find Median from Data Stream
# Difficulty: Hard
#
# PROBLEM STATEMENT:
# Implement the MedianFinder class:
# - MedianFinder() initializes the object.
# - void addNum(int num) adds the integer num from the data stream to the data structure.
# - double findMedian() returns the median of all elements so far.
#
# RULES:
# - If the size of the list is even, there is no middle value, and the median 
#   is the mean of the two middle values.
# - If the size of the list is odd, the median is the middle value.
#
# EXAMPLES:
# ["MedianFinder", "addNum", "addNum", "findMedian", "addNum", "findMedian"]
# [[], [1], [2], [], [3], []]
# Output: [null, null, null, 1.5, null, 2.0]
# ============================================================================

from typing import List
import heapq

class MedianFinder:
    def __init__(self):
        self.lo = []   # max-heap (negate) — lower half  Low has half or 1 more of the items all the time
        self.hi = []   # min-heap           — upper half

    def addNum(self, num: int) -> None:
        heapq.heappush(self.lo, -num)                             # step 1: always push to lo first
        heapq.heappush(self.hi, -heapq.heappop(self.lo))         # Move the highest for lo into hi
        if (len(self.hi) > len(self.lo) ) :                       # lo has to have half or  1 more
            heapq.heappush(self.lo, -heapq.heappop(self.hi)) 

        

    def findMedian(self) -> float:
        if len(self.lo) > len(self.hi) :
            return -self.lo[0]
        else:
            return (self.hi[0] - self.lo[0] ) /2.0

# Demo calls
mf = MedianFinder()
mf.addNum(1)
mf.addNum(2)
print(mf.findMedian())
mf.addNum(3)
print(mf.findMedian())

def test():
    sol = MedianFinder()
    sol.addNum(1)
    assert sol.findMedian() == 1.0  # single element
    sol.addNum(2)
    assert sol.findMedian() == 1.5  # even number of elements [1, 2]
    sol.addNum(3)
    assert sol.findMedian() == 2.0  # odd number of elements [1, 2, 3]
    sol.addNum(10)
    assert sol.findMedian() == 2.5  # even [1, 2, 3, 10]
    sol.addNum(0)
    assert sol.findMedian() == 2.0  # odd [0, 1, 2, 3, 10]
    sol.addNum(0)
    assert sol.findMedian() == 1.5  # even [0, 0, 1, 2, 3, 10]
    sol.addNum(100)
    assert sol.findMedian() == 2.0  # odd [0, 0, 1, 2, 3, 10, 100]
    sol.addNum(-5)
    assert sol.findMedian() == 1.5  # even [-5, 0, 0, 1, 2, 3, 10, 100]
    print('All Pass!')

test()
```

```python
import heapq

class MedianFinder:
    def __init__(self):
        self.lo = []   # max-heap (negate) — lower half
        self.hi = []   # min-heap           — upper half

    def addNum(self, num):
        heapq.heappush(self.lo, -num)                             # step 1: always push to lo first
        heapq.heappush(self.hi, -heapq.heappop(self.lo))         # step 2: move lo's max to hi (fixes order)
        if len(self.hi) > len(self.lo):                           # step 3: rebalance — lo should be >= hi in size
            heapq.heappush(self.lo, -heapq.heappop(self.hi))

    def findMedian(self):
        if len(self.lo) > len(self.hi):
            return float(-self.lo[0])                             # lo has one extra — its top is the median
        return (-self.lo[0] + self.hi[0]) / 2.0                  # equal size — average of both tops


mf = MedianFinder()
mf.addNum(1); mf.addNum(2)
assert mf.findMedian() == 1.5
mf.addNum(3)
assert mf.findMedian() == 2.0
mf.addNum(4)
assert mf.findMedian() == 2.5

mf2 = MedianFinder()
for n in [-1, -2, -3]: mf2.addNum(n)
assert mf2.findMedian() == -2.0

print('All Pass!')
```
