## Ransom Note

Given two strings `ransomNote` and `magazine`, return `True` if `ransomNote` can be constructed using the letters from `magazine`, `False` otherwise.

Each letter in `magazine` can only be used **once**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n + m)** will fit within the execution time limit.

### Example

- `ransomNote = "a"`, `magazine = "b"` → `False`
- `ransomNote = "aa"`, `magazine = "ab"` → `False`
- `ransomNote = "aa"`, `magazine = "aab"` → `True`

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `ransomNote`, `string` `magazine`
- **[output]** `boolean`

```python
from collections import Counter
def solution(ransomNote, magazine):
    ran = Counter(ransomNote)
    mag = Counter(magazine)
    ret = True
    for key, val in ran.items():
        if not key in mag or mag[key] < val:
            return False

    return True

def test():
    
    assert solution("a", "b")        == False  # given example 1
    assert solution("aa", "ab")      == False  # given example 2
    assert solution("aa", "aab")     == True   # given example 3
    assert solution("", "abc")       == True   # empty note
    assert solution("abc", "abc")    == True   # exact match
    assert solution("abc", "ab")     == False  # magazine too short
    assert solution("aab", "aab")    == True   # duplicates exact
    assert solution("abcd", "dcba")  == True   # same letters different order
    assert solution("a", "aa")       == True   # magazine has more than needed — fine
    print('All Pass!')    
test()
```
