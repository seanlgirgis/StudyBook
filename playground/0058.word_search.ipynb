## Word Search

Given an `m × n` grid of characters and a string `word`, return `True` if `word` exists in the grid.

The word must be constructed from letters of **sequentially adjacent cells** (horizontally or vertically). The **same cell may not be used more than once**.

**Note:** You are not expected to provide the most optimal solution, but a solution that correctly handles the backtracking will fit within the execution time limit.

### Example

- For `word = "ABCCED"` in the grid below, the output should be `True`:
```
A B C E
S F C S
A D E E
```
- For `word = "SEE"`, the output should be `True`.
- For `word = "ABCB"`, the output should be `False` (can't reuse `B`).

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.array.character` `board`, `string` `word`  
  Guaranteed constraints: `1 ≤ m, n ≤ 6`, `1 ≤ word length ≤ 15`
- **[output]** `boolean`

````python
%run _timer.py
````

````python
def solution(board, word):
    rows, cols = len(board), len(board[0])
    path = set()
    
    def dfs(r,c,i):
        if i == len(word):
            return True

        if r < 0 or r >= rows or c < 0 or c >= cols or word[i] != board[r][c] or (r, c) in path:
            return False
        path.add((r,c))

        res = (dfs(r-1,c, i+1) or dfs(r+1,c, i+1) or dfs(r,c-1, i+1) or dfs(r,c+1, i+1))
        
        path.remove((r,c))
        return res
        
    for r in range(rows):
        for c in range(cols):
            if dfs(r,c,0):
                return True
    return False
    

board = [['A','B','C','E'],['S','F','C','S'],['A','D','E','E']]

assert solution(board, 'ABCCED') == True   # given example 1
assert solution(board, 'SEE')    == True   # given example 2
assert solution(board, 'ABCB')   == False  # given example 3 — can't reuse B

assert solution([['A']], 'A')           == True   # single cell match
assert solution([['A']], 'B')           == False  # single cell no match
assert solution([['A','B'],['C','D']], 'ABDC') == True   # snake path
assert solution([['A','B'],['C','D']], 'ABCD') == False  # B→C not adjacent
assert solution([['A','A','A'],['A','A','A'],['A','A','A']], 'AAAAAAAAA') == True  # all same
print('All Pass!')
````
