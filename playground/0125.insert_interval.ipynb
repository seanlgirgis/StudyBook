## Insert Interval

Insert a new interval into a sorted non-overlapping list. Merge if necessary.

- `[[1,3],[6,9]]`, new=`[2,5]` → `[[1,5],[6,9]]`
- `[[1,2],[3,5],[6,7],[8,10],[12,16]]`, new=`[4,8]` → `[[1,2],[3,10],[12,16]]`

```python
%run _timer.py
```

```python
def solution(intervals, newInterval):
    res = []

    for i, interval in enumerate(intervals):
        # New interval ends before current starts — insert here, append rest
        if newInterval[1] < interval[0]:
            res.append(newInterval)
            return res + intervals[i:]
        # New interval starts after current ends — current goes first
        elif newInterval[0] > interval[1]:
            res.append(interval)
        # Overlap — merge by expanding newInterval bounds
        else:
            newInterval = [min(newInterval[0], interval[0]),
                           max(newInterval[1], interval[1])]

    res.append(newInterval)   # new interval goes at the end
    return res


assert solution([[1,3],[6,9]],                        [2,5])  == [[1,5],[6,9]]
assert solution([[1,2],[3,5],[6,7],[8,10],[12,16]],   [4,8])  == [[1,2],[3,10],[12,16]]
assert solution([],                                   [5,7])  == [[5,7]]
assert solution([[1,5]],                              [2,3])  == [[1,5]]
assert solution([[1,5]],                              [6,8])  == [[1,5],[6,8]]
print('All Pass!')
```
