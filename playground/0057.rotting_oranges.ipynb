## Rotting Oranges

You are given an `m × n` grid where each cell has one of three values:
- `0` — empty
- `1` — fresh orange
- `2` — rotten orange

Every minute, any fresh orange **4-directionally adjacent** to a rotten orange becomes rotten.

Return the **minimum number of minutes** until no fresh orange remains. If it is impossible, return `-1`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(m·n)** will fit within the execution time limit.

### Example

- For `grid = [[2,1,1],[1,1,0],[0,1,1]]`, the output should be `4`.
- For `grid = [[2,1,1],[0,1,1],[1,0,1]]`, the output should be `-1`.

  **Explanation:** The bottom-left orange `(2,0)` can never be reached.

- For `grid = [[0,2]]`, the output should be `0`.

  **Explanation:** No fresh oranges to rot.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.array.integer` `grid`  
  Guaranteed constraints: `1 ≤ m, n ≤ 10`
- **[output]** `integer`

```python
%run _timer.py
```

```python
from collections import deque

def solution(grid):
    rotten = deque()
    clock = 0
    fresh = 0
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    rows, cols = len(grid), len(grid[0])

    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 1:
                fresh += 1
            if grid[r][c] == 2:
                rotten.append((r, c, clock))   # seed all rotten at t=0

    while rotten:
        rx, cx, t = rotten.popleft()
        for dr, dc in directions:
            r = rx + dr
            c = cx + dc
            if 0 <= r < rows and 0 <= c < cols and grid[r][c] == 1:
                t1 = t + 1
                rotten.append((r, c, t1))
                fresh -= 1
                grid[r][c] = 2             # mark rotten so it won't be processed again
                clock = max(clock, t1)     # track furthest minute reached

    return clock if fresh == 0 else -1     # -1 if any fresh orange was never reached


assert solution([[2,1,1],[1,1,0],[0,1,1]])  ==  4   # given example 1
assert solution([[2,1,1],[0,1,1],[1,0,1]])  == -1   # given example 2 — unreachable orange
assert solution([[0,2]])                    ==  0   # given example 3 — no fresh
assert solution([[0,0],[0,0]])              ==  0   # all empty
assert solution([[1]])                      == -1   # fresh with no rotten
assert solution([[2]])                      ==  0   # single rotten
assert solution([[2,1],[1,1]])              ==  2   # spreads in 2 steps
assert solution([[2,1,1],[1,1,1],[0,1,2]])  ==  2   # two rotten sources
print('All Pass!')
```
