## Task Scheduler Simulation

You have a list of tasks, each with a name and a duration (in minutes). A worker processes tasks **one at a time in the order given**.

Given a list of strings `tasks` in the format `"TaskName Duration"`, return a list of strings showing when each task **completes**, in the format `"TaskName FinishTime"`.

Tasks start immediately after the previous one finishes (no idle time).

### Example

- For `tasks = ["Build 3", "Test 2", "Deploy 1"]`, the output should be:
  `["Build 3", "Test 5", "Deploy 6"]`

  **Explanation:** Build finishes at 3. Test starts at 3, finishes at 5. Deploy starts at 5, finishes at 6.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.string` `tasks`
- **[output]** `array.string`

```python
%run _timer.py
```

```python
def solution(tasks):
    runtime = 0
    ret = []
    for task in tasks:
        [name, duration] = task.split(' ')
        runtime += int(duration)
        ret.append(name + ' ' + str(runtime))
    return ret


assert solution(["Build 3","Test 2","Deploy 1"])           == ["Build 3","Test 5","Deploy 6"]
assert solution(["A 1"])                                    == ["A 1"]
assert solution(["A 1","B 1","C 1"])                       == ["A 1","B 2","C 3"]
assert solution(["Alpha 10","Beta 5"])                      == ["Alpha 10","Beta 15"]
assert solution(["X 2","Y 3","Z 4","W 1"])                 == ["X 2","Y 5","Z 9","W 10"]
assert solution(["Task 100"])                               == ["Task 100"]
assert solution(["A 1","B 2","A 3"])                       == ["A 1","B 3","A 6"]
print('All Pass!')
```

```python
from collections import defaultdict
def solution(tasks):
    runtime = 0
    sched = defaultdict(int)
    for task in tasks:
        [name, duration] = task.split(' ')
        sched[name] += (int(duration) + runtime)
        runtime += int(duration)
        print (name, sched[name])
    


solution(["Build 3","Test 2","Deploy 1"]) 
```
