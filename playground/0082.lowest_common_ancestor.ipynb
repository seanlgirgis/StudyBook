## Lowest Common Ancestor of a BST

Given a BST and two node values `p` and `q`, find their **lowest common ancestor (LCA)**.

The LCA is the deepest node that has both `p` and `q` as descendants (a node can be a descendant of itself).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

```
        6
       / \
      2   8
     / \ / \
    0  4 7  9
      / \
     3   5
```
- LCA(2, 8) = 6  (root splits them)
- LCA(2, 4) = 2  (2 is ancestor of 4)

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `TreeNode` `root`, `integer` `p`, `integer` `q`
- **[output]** `integer` — the value of the LCA node

````python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val   = val
        self.left  = left
        self.right = right

def make_tree(vals):
    if not vals or vals[0] is None:
        return None
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root


def solution(root, p, q):
    curr = root
    
    while curr:
        # If both nodes are greater than current, move right
        if p > curr.val and q > curr.val:
            curr = curr.right
        # If both nodes are smaller than current, move left
        elif p < curr.val and q < curr.val:
            curr = curr.left
        # We found the split point or one of the nodes
        else:
            return curr.val
            
    return None


t = make_tree([6,2,8,0,4,7,9,None,None,3,5])
assert solution(t, 2, 8) == 6   # root splits them
assert solution(t, 2, 4) == 2   # 2 is ancestor of 4
assert solution(t, 3, 5) == 4   # meet at 4
assert solution(t, 0, 5) == 2   # meet at 2
assert solution(t, 7, 9) == 8   # meet at 8
assert solution(t, 6, 4) == 6   # root is one of the nodes
assert solution(make_tree([2,1,3]), 1, 3) == 2
print('All Pass!')
````
