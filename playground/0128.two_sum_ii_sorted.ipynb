## Two Sum II — Sorted Array

Given a **1-indexed** sorted array, return indices of two numbers that add to target.

- `[2,7,11,15]`, target=9 → `[1,2]`
- `[2,3,4]`, target=6 → `[1,3]`

**Pattern:** two pointers — advance left if sum too small, retreat right if too large.

```python
%run _timer.py
```

```python
def solution(numbers, target):
    l, r = 0, len(numbers) - 1
    while l < r:
        s = numbers[l] + numbers[r]
        if s == target:
            return [l + 1, r + 1]   # 1-indexed
        elif s < target:
            l += 1                  # need bigger sum
        else:
            r -= 1                  # need smaller sum
    return []


assert solution([2,7,11,15], 9)  == [1,2]
assert solution([2,3,4],    6)  == [1,3]
assert solution([-1,0],    -1)  == [1,2]
assert solution([1,2,3,4,5],6)  == [1,5]
print('All Pass!')
```
