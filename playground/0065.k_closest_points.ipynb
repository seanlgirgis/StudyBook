## K Closest Points to Origin

Given an array of `points` where `points[i] = [x, y]`, and an integer `k`, return the `k` closest points to the origin `(0, 0)`.

The distance is the **Euclidean distance**: `sqrt(x² + y²)`. You do **not** need to take the square root — comparing `x² + y²` is sufficient.

The answer may be returned in **any order**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n log n)** will fit within the execution time limit.

### Example

- For `points = [[1,3],[-2,2]]` and `k = 1`, the output should be `[[-2,2]]`.

  **Explanation:** Distance of `[1,3]` = √10. Distance of `[-2,2]` = √8. Closest is `[-2,2]`.

- For `points = [[3,3],[5,-1],[-2,4]]` and `k = 2`, the output should be `[[3,3],[-2,4]]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.array.integer` `points`, `integer` `k`
- **[output]** `array.array.integer` — any order

```python
import heapq

def solution(points, k):
    heap = []
    for x, y in points:
        dist = x**2 + y**2
        heapq.heappush(heap, (-dist, [x, y]))
        if len(heap) > k:
            heapq.heappop(heap)    # remove the farthest point
    return [lst for _, lst in heap]


def check(result, expected):
    return sorted(map(tuple, result)) == sorted(map(tuple, expected))

assert check(solution([[1,3],[-2,2]], 1),              [[-2,2]])
assert check(solution([[3,3],[5,-1],[-2,4]], 2),       [[3,3],[-2,4]])
assert check(solution([[0,0]], 1),                     [[0,0]])
assert check(solution([[1,0],[0,1],[-1,0],[0,-1]], 2), [[1,0],[0,1]])
assert check(solution([[1,1],[2,2],[3,3]], 1),          [[1,1]])
assert check(solution([[1,1],[2,2],[3,3]], 3),          [[1,1],[2,2],[3,3]])
assert check(solution([[-5,4],[3,-3],[1,1]], 2),        [[3,-3],[1,1]])
print('All Pass!')
```
