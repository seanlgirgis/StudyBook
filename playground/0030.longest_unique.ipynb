## Longest Substring Without Repeating Characters

Given a string `s`, return the **length** of the longest substring that contains no repeating characters.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `s = "abcabcbb"`, the output should be `solution(s) = 3`.

  **Explanation:** The longest substring without repeating characters is `"abc"`, length `3`.

- For `s = "bbbbb"`, the output should be `solution(s) = 1`.

  **Explanation:** Every character repeats — the best you can do is a single `"b"`.

- For `s = "pwwkew"`, the output should be `solution(s) = 3`.

  **Explanation:** `"wke"` has length `3`. Note that `"pwke"` is not valid — it skips `"w"`; a substring must be contiguous.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `string` `s` — A string of English letters.  
  Guaranteed constraints: `0 ≤ s.length ≤ 5·10^4`
- **[output]** `integer` — Length of the longest substring without repeating characters.

```python
"""
id: lc_0003
title: Longest Substring Without Repeating Characters
source: leetcode
difficulty: medium
primary: sliding-window
tags: [hash-table, string, sliding-window]
leetcode_url: https://leetcode.com/problems/longest-substring-without-repeating-characters/
status: draft
last_updated: 2026-04-22
notes:
- key idea: sliding window with hash map to track character indices
- time: O(n)
- space: O(min(m, n)) where m is charset size
"""

# ============================================================================
# File: 003_lc_0003_longest_substring_without_repeating_characters_empty.py
# LeetCode 3: Longest Substring Without Repeating Characters
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Given a string s, find the length of the longest substring without repeating 
# characters.
#
# RULES:
# - Substring must be contiguous.
# - Input consists of English letters, digits, symbols, and spaces.
#
# EXAMPLES:
# Input: s = "abcabcbb" -> Output: 3 ("abc")
# Input: s = "bbbbb" -> Output: 1 ("b")
# Input: s = "pwwkew" -> Output: 3 ("wke")
# ============================================================================

from typing import List

def lengthOfLongestSubstring(s: str) -> int:
    win = set()
    n = len(s)
    l, best = 0, 0
    for r in range(n):
        while s[r] in win:
            win.remove(s[l])
            l += 1
        win.add(s[r])
        best = max(best, r-l+1)
    return best


print(lengthOfLongestSubstring("abcabcbb")) #3
print(lengthOfLongestSubstring("bbbbb")) #1

def test():
    assert lengthOfLongestSubstring("abcabcbb") == 3  # given example 1
    assert lengthOfLongestSubstring("bbbbb") == 1     # given example 2
    assert lengthOfLongestSubstring("pwwkew") == 3    # given example 3
    assert lengthOfLongestSubstring("") == 0          # empty string case
    assert lengthOfLongestSubstring(" ") == 1         # single space case
    assert lengthOfLongestSubstring("au") == 2        # two distinct chars
    assert lengthOfLongestSubstring("dvdf") == 3      # repeat with char in between
    assert lengthOfLongestSubstring("abba") == 2      # window boundary check
    assert lengthOfLongestSubstring("tmmzuxt") == 5   # complex repeat
    assert lengthOfLongestSubstring("abcdefg") == 7   # all unique
    print('All Pass!')

test()
```

```python
def solution(s):
    if not s: return 0
    l,r = 0, 1
    n = len(s)
    seen = {s[0]}
    max_length = 1
    while r < n:
        while s[r] in seen and l <= r:
            seen.remove(s[l])
            l += 1
        seen.add(s[r])
        max_length = max(max_length, r-l+1)
        r+=1
    return max_length
        
            


assert solution('abcabcbb') == 3  # given example 1 — 'abc'
assert solution('bbbbb')    == 1  # given example 2 — all same
assert solution('pwwkew')   == 3  # given example 3 — 'wke'
assert solution('')         == 0  # empty string
assert solution('a')        == 1  # single char
assert solution('abcdef')   == 6  # all unique — entire string
assert solution('aab')      == 2  # 'ab'
assert solution('dvdf')     == 3  # 'vdf' — window must shrink past first 'd'
print('All Pass!')
```
