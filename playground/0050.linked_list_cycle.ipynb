## Linked List Cycle

Given the `head` of a linked list, determine if the list has a **cycle** in it.

A cycle exists if some node's `next` pointer points back to a previous node, creating an infinite loop.

Return `True` if there is a cycle, `False` otherwise.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `head = [3,2,0,-4]` with tail connected to node at index 1, the output should be `True`.

  **Explanation:** The last node (`-4`) points back to node `2`, forming a cycle.

- For `head = [1,2]` with tail connected to node at index 0, the output should be `True`.

- For `head = [1]` with no cycle, the output should be `False`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `head`  
  Guaranteed constraints: `0 ≤ n ≤ 10^4`, `−10^5 ≤ val ≤ 10^5`
- **[output]** `boolean` — `True` if cycle exists, `False` otherwise.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def make_cycle(vals, pos):
    # builds a linked list from vals; if pos >= 0, tail points back to node at index pos
    nodes = [ListNode(v) for v in vals]
    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]
    if pos >= 0:
        nodes[-1].next = nodes[pos]   # create the cycle
    return nodes[0] if nodes else None


def solution(head):
    seen = set()
    curr = head
    while curr:
        if id(curr) in seen:
            return True
        seen.add(id(curr))
        curr = curr.next
    return False


assert solution(make_cycle([3,2,0,-4], 1)) == True   # given example 1 — cycle at index 1
assert solution(make_cycle([1,2], 0))       == True   # given example 2 — cycle at index 0
assert solution(make_cycle([1], -1))        == False  # given example 3 — no cycle
assert solution(make_cycle([], -1))         == False  # empty list
assert solution(make_cycle([1,2,3,4], -1))  == False  # no cycle, multiple nodes
assert solution(make_cycle([1,2,3,4], 0))   == True   # cycle back to head
assert solution(make_cycle([1,2,3,4], 2))   == True   # cycle to middle
assert solution(make_cycle([1,2,3,4], 3))   == True   # cycle to last (self-loop)
print('All Pass!')
```
