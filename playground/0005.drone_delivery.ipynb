## Drone Delivery

You are designing a delivery system using drones in a linear warehouse. The warehouse is represented as a number line starting at position `0` and ending at position `target` (`target > 0`). Along this line, there are charging stations placed at various positions, represented by an array `stations`, where `stations[i]` is the position of the ith charging station.

Each drone has a limited battery that allows it to travel a maximum of **10 units** after being fully charged. For example, if a drone is charged at position `12`, it can travel to positions `12, 13, 14, ...` up to position `22` (inclusive), but cannot reach position `23` or beyond without recharging.

Your delivery protocol requires the following steps:

1. From your current position, pick up the cargo and carry it **on foot** to the nearest charging station ahead of you. If there are no more stations ahead, carry the cargo on foot to the target position.
2. Deploy a fully charged drone from this station and send it with the cargo **as far as possible** toward the target.
3. If the target hasn't been reached, walk to the point where the drone landed to retrieve the cargo, then repeat from step 1.

Your task is to calculate the **total distance** over which you must carry the cargo on foot, from position `0` to position `target`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(stations.length × target)** will fit within the execution time limit.

### Example

- For `target = 21` and `stations = [7, 4, 14]`, the output should be `solution(target, stations) = 4`.
  - Starting at `0`, the nearest station ahead is `stations[1] = 4`, so you carry the cargo on foot for 4 units, then deploy a drone that travels to position `14`.
  - There is another station at `14` (`stations[2] = 14`), which you use to deploy another drone that reaches the target position `21`.
  - Therefore, you carried the cargo on foot only **4** units in total.

- For `target = 27` and `stations = [15, 7, 3, 10]`, the output should be `solution(target, stations) = 7`.
  - Starting at `0`, the nearest station ahead is `stations[2] = 3`, so you carry the cargo on foot for 3 units, then deploy a drone that travels to position `13`.
  - From position `13`, the nearest station ahead is `stations[0] = 15`, requiring 2 more units on foot, then the drone reaches position `25`.
  - There are no more stations ahead, so you carry the cargo on foot 2 more units to the target.
  - Therefore, you carried the cargo on foot `3 + 2 + 2 =` **7** units in total.

- For `target = 10` and `stations = []`, the output should be `solution(target, stations) = 10`.
  - There are no charging stations available, so you must carry the cargo on foot the entire distance.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `integer` `target` — The end position of the warehouse. Guaranteed: `target > 0`
- **[input]** `array.integer` `stations` — Positions of the charging stations along the line.
- **[output]** `integer` — Total distance carried on foot.

```python
def solution(target, stations):
    stations.sort()
    if not stations: return target
    walks = stations[0]
    pos = walks
    for d in stations:
        if pos >= target: #arrived
            break
        if pos > d:          # you are post the next drone .. you need to walk to next
            continue
        else:
            walks += d - pos
            pos = d + 10
    walks += max(0,  target - pos)                    # add more if not yet arrive .. only add positive values

    return walks


print( solution(27, [15, 7, 3, 10]) )
print( solution(21, [7, 4, 14]))           #SHould return 4
assert solution(21, [7, 4, 14])      == 4   # given example 1
assert solution(27, [15, 7, 3, 10])  == 7   # given example 2
assert solution(10, [])              == 10  # no stations — walk the whole way
assert solution(1,  [])              == 1   # minimal case, no stations
assert solution(5,  [3])             == 3   # one station, drone reaches target
assert solution(20, [10])            == 10  # walk to station, drone hits target exactly
assert solution(25, [10])            == 15  # walk to station, drone to 20, walk last 5
assert solution(16, [5, 12])         == 6   # walk to 5, drone to 15, walk last 1
print("All Done!!")
```

```python
def solution(target, stations):
    stations.sort()
    n = len(stations)
    walking = 0
    position = 0
    for i in range(n):
        if position <= stations[i]:
            dist = stations[i] - position
            walking += dist
            position += dist + 10
    if position < target:
        walking += (target-position)
    return walking
            
assert solution(21, [7, 4, 14])      == 4   # given example 1
assert solution(27, [15, 7, 3, 10])  == 7   # given example 2
assert solution(10, [])              == 10  # no stations — walk the whole way
assert solution(1,  [])              == 1   # minimal case, no stations
assert solution(5,  [3])             == 3   # one station, drone reaches target
assert solution(20, [10])            == 10  # walk to station, drone hits target exactly
assert solution(25, [10])            == 15  # walk to station, drone to 20, walk last 5
assert solution(16, [5, 12])         == 6   # walk to 5, drone to 15, walk last 1
print("All Done!!")
```
