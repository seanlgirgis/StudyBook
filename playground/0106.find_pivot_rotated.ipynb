## Find Pivot in Rotated Sorted Array

A sorted array was rotated at some pivot index. Return the **pivot index** (the index of the smallest element / rotation point).

If the array is not rotated, return `0`.

### Example

- `[3,4,5,1,2]` → `3`  (smallest element 1 is at index 3)
- `[4,5,6,7,0,1,2]` → `4`  (smallest element 0 is at index 4)
- `[11,13,15,17]` → `0`  (not rotated — smallest is at index 0)
- `[2,1]` → `1`

### Note
This is the same binary search as find_min_rotated — just return the index instead of the value.

```python
%run _timer.py
```

```python
def solution(nums):
    l, r = 0, len(nums) - 1
    
    while l < r:
        mid = (l + r) // 2

        if nums[mid] > nums[r]:
            l = mid + 1

        else:
            r = mid
            
    return l


assert solution([3,4,5,1,2])     == 3
assert solution([4,5,6,7,0,1,2]) == 4
assert solution([11,13,15,17])   == 0
assert solution([2,1])           == 1
assert solution([1])             == 0
print('All Pass!')
```
