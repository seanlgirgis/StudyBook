## Find All Anagrams in a String

Given two strings `s` and `p`, return a list of all start indices in `s` where an anagram of `p` begins. The answer may be returned in any order.

An **anagram** is a rearrangement of all characters of a string.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "cbaebabacd"`, `p = "abc"`, the output should be `[0, 6]`.

  **Explanation:** `s[0:3] = "cba"` is an anagram of `"abc"`. `s[6:9] = "bac"` is an anagram of `"abc"`.

- For `s = "abab"`, `p = "ab"`, the output should be `[0, 1, 2]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `s` — the string to search in
- **[input]** `string` `p` — the pattern to find anagrams of
- **[output]** `array.integer` — sorted list of start indices

```python
"""
id: lc_0438
title: Find All Anagrams in a String
source: leetcode
difficulty: medium
primary: sliding_window
tags: [hash_table, string, sliding_window]
leetcode_url: https://leetcode.com/problems/find-all-anagrams-in-a-string/
status: draft
last_updated: 2026-04-22
notes:
- key idea: sliding window with frequency map
- time: O(n)
- space: O(1) (max 26 characters)
"""

# ============================================================================
# File: 438_lc_0438_find_all_anagrams_in_a_string_empty.py
# LeetCode 438: Find All Anagrams in a String
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Given two strings s and p, return an array of all the start indices of p's 
# anagrams in s. You may return the answer in any order.
#
# RULES:
# - s and p consist of lowercase English letters.
# - An anagram is a string formed by rearranging the letters of another.
#
# EXAMPLES:
# Input: s = "cbaebabacd", p = "abc"
# Output: [0, 6]
#
# Input: s = "abab", p = "ab"
# Output: [0, 1, 2]
# ============================================================================

from typing import List

def findAnagrams(s: str, p: str) -> List[int]:
    def makeMeAna(s2):
        l = [0] * 26
        for c in s2: 
            l[ord(c) - ord('a')] += 1
        return tuple(l)
    k = len(p)
    n = len(s)
    out = []
    p_aragram = makeMeAna(p)
    for r in range(k, n+1):
        if makeMeAna(s[r-k:r]) == p_aragram:
            out.append(r-k)
    return out


print(findAnagrams("abab", "ab")) # [0, 1, 2]
print(findAnagrams("cbaebabacd", "abc")) #[0, 6]
def test():
    assert findAnagrams("cbaebabacd", "abc") == [0, 6]  # example 1
    assert findAnagrams("abab", "ab") == [0, 1, 2]  # example 2
    assert findAnagrams("af", "be") == []  # no match same length
    assert findAnagrams("a", "a") == [0]  # single char match
    assert findAnagrams("a", "ab") == []  # p longer than s
    assert findAnagrams("baa", "aa") == [1]  # duplicates in p
    assert findAnagrams("aaaaaaaaaa", "aaaaa") == [0, 1, 2, 3, 4, 5]  # all overlapping
    assert findAnagrams("abcde", "fgh") == []  # no common chars
    assert findAnagrams("abab", "a") == [0, 2]  # single char p
    assert findAnagrams("", "a") == []  # empty s
    print('All Pass!')

test()
```

```python
def anaMe(s):
    cnts = [0] * 26
    for c in s:
        cnts[ord(c) - ord('a')] += 1
    return tuple(cnts)
    
def solution(s, p):
    if len(s) < len(p): 
        return []
    ret = []
    p_cnt = anaMe(p)
    wl = len(p)
    for i in range(len(s) -wl+1):
        w_cnt = anaMe(s[i :i + wl])
        if p_cnt == w_cnt:
            ret.append(i)
    return ret



print(solution("cbaebabacd", "abc"))

def test():
    assert solution("cbaebabacd", "abc") == [0, 6]
    
    assert solution("aa",         "bb")  == []
    assert solution("a",          "a")   == [0]
    assert solution("ab",         "abc") == []          # p longer than s
    assert solution("baa",        "aa")  == [1]
    assert solution("aaaa",       "aa")  == [0, 1, 2]
    assert solution("abcdefg",    "gfedcba") == [0]    # full string is anagram
    assert solution("abab",       "ab")  == [0, 1, 2]
    print('All Pass!')
test()
```
