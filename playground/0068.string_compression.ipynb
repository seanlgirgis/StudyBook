## String Compression

Given a string `s`, compress it using the following rules:

- For each group of consecutive repeating characters, write the character followed by its count.
- If the count is `1`, write only the character (no `1` after it).

Return the compressed string.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "aabcccdddd"`, the output should be `"a2bc3d4"`.
- For `s = "abcd"`, the output should be `"abcd"` (all counts are 1).
- For `s = "aaa"`, the output should be `"a3"`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `string` `s`
- **[output]** `string`

```python
%run _timer.py
```

```python
def solution(s):
    ch = ''
    cnt = 0
    out = ""
    for c in s:
        if c != ch:
            if ch != '':
                out += ch + (str(cnt) if cnt > 1 else '')
            cnt = 1
            ch = c
        else:
            cnt += 1
    out += ch + (str(cnt) if cnt > 1 else '')   # flush last group
    return out


assert solution("aabcccdddd") == "a2bc3d4"
assert solution("abcd")       == "abcd"
assert solution("aaa")        == "a3"
assert solution("a")          == "a"
assert solution("aa")         == "a2"
assert solution("aabb")       == "a2b2"
assert solution("aaabba")     == "a3b2a"
assert solution("aaaaaaaaaa") == "a10"
print('All Pass!')
```
