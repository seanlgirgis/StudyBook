## Word Break

Given a string `s` and a list of words `wordDict`, return `True` if `s` can be segmented into a space-separated sequence of words from the dictionary.

### Example

- `s="leetcode"`, `wordDict=["leet","code"]` → `True`
- `s="applepenapple"`, `wordDict=["apple","pen"]` → `True`
- `s="catsandog"`, `wordDict=["cats","dog","sand","and","cat"]` → `False`

### Hint
`dp[i]` = can we segment `s[:i]`? For each `i`, check all `j < i`: if `dp[j]` is True and `s[j:i]` is in the word set → `dp[i] = True`.

### Input/Output
- **[input]** `string` `s`, `array.string` `wordDict`
- **[output]** `boolean`

```python
%run _timer.py
```

```python
def solution(s, wordDict):
 
    word_set = set(wordDict)
    n = len(s)
    
 
    dp = [False] * (n + 1)
 
    dp[0] = True
    
    for i in range(1, n + 1):
        for j in range(i):
            
            if dp[j] and s[j:i] in word_set:
                dp[i] = True
                break  # No need to check other j's for this i
                
    return dp[n]

# Verification
assert solution("leetcode",     ["leet","code"])                        == True
assert solution("applepenapple",["apple","pen"])                        == True
assert solution("catsandog",    ["cats","dog","sand","and","cat"])       == False
assert solution("a",            ["a"])                                   == True
assert solution("ab",           ["a","b"])                               == True
print('All Pass!')
```
