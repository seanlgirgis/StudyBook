## Combination Sum

Find all unique combinations that sum to `target`. Each number may be used unlimited times.

- `candidates=[2,3,6,7]`, `target=7` → `[[2,2,3],[7]]`
- `candidates=[2,3]`, `target=6` → `[[2,2,2],[3,3]]`

```python
%run _timer.py
```

```python
def solution(candidates, target):
    res = []

    def backtrack(start, path, remaining):
        if remaining == 0:
            res.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                break                     # sorted — no point continuing
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])  # i not i+1 (reuse allowed)
            path.pop()

    candidates.sort()
    backtrack(0, [], target)
    return res


def check(result, expected):
    return sorted(sorted(r) for r in result) == sorted(sorted(e) for e in expected)

assert check(solution([2,3,6,7], 7), [[2,2,3],[7]])
assert check(solution([2,3],    6), [[2,2,2],[3,3]])
assert check(solution([2],      1), [])
print('All Pass!')
```
