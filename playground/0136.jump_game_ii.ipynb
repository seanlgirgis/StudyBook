## Jump Game II

Return the **minimum number of jumps** to reach the last index.

- `[2,3,1,1,4]` → `2` (0→1→4)
- `[2,3,0,1,4]` → `2`

**Greedy BFS:** Track current window `[l, r]`. From that window find the farthest reachable. Expand window. Count jumps.

```python
%run _timer.py
```

```python
def solution(nums):
    jumps = 0
    l = r = 0           # current jump window

    while r < len(nums) - 1:
        farthest = max(i + nums[i] for i in range(l, r + 1))
        l = r + 1       # next window starts after current
        r = farthest    # next window ends at farthest reachable
        jumps += 1

    return jumps


assert solution([2,3,1,1,4]) == 2
assert solution([2,3,0,1,4]) == 2
assert solution([1,2,3])     == 2
assert solution([1])         == 0
assert solution([1,1,1,1])   == 3
print('All Pass!')
```
