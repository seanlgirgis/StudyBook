## Minimum Path Sum

Given an `m × n` grid of non-negative integers, find a path from the **top-left to the bottom-right** that **minimizes the sum** of all numbers along the path. You can only move **right or down**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(m·n)** will fit within the execution time limit.

### Example

- For `grid = [[1,3,1],[1,5,1],[4,2,1]]`, the output should be `solution(grid) = 7`.

  **Explanation:** Path `1→3→1→1→1 = 7` (right, right, down, down).

- For `grid = [[1,2,3],[4,5,6]]`, the output should be `solution(grid) = 12`.

  **Explanation:** Path `1→2→3→6 = 12`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.array.integer` `grid`  
  Guaranteed constraints: `1 ≤ m, n ≤ 200`, `0 ≤ grid[i][j] ≤ 100`
- **[output]** `integer` — Minimum path sum.

```python
def solution(grid):
    m, n = len(grid), len(grid[0])
    dp = [[0] * n for _ in range(m)]
    prefix = 0
    for i in range(n): 
        prefix += grid[0][i]
        dp[0][i] =  prefix
    prefix = 0    
    for i in range(m):
        prefix += grid[i][0]
        dp[i][0]  =  prefix        
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = grid[i][j]  + min (dp[i-1][j] , dp[i][j-1])
    return dp[-1][-1]


assert solution([[1,3,1],[1,5,1],[4,2,1]])  == 7   # given example 1
assert solution([[1,2,3],[4,5,6]])          == 12  # given example 2
assert solution([[1]])                      == 1   # 1x1
assert solution([[1,2],[3,4]])              == 7   # 2x2 — 1+2+4
assert solution([[1,3],[2,1]])              == 4   # 2x2 — 1+2+1
assert solution([[0,0,0],[0,0,0]])          == 0   # all zeros
assert solution([[5]])                      == 5   # single cell
assert solution([[1,2,5],[3,2,1]])          == 6   # 1+2+2+1
print('All Pass!')
```
