## Minimum Subarray

Given an array of integers `nums`, find the **contiguous subarray** with the smallest sum and return that sum.

A subarray is a contiguous non-empty part of the array.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [2,-3,4,-1,-2,1]`, the output should be `solution(nums) = -4`.

  **Explanation:** The subarray `[-3,4,-1,-2,1]` sums to `-1`, but `[-3,4,-1,-2]` sums to `-2`, and `[-1,-2]` sums to `-3`... the minimum is `[-3,-1,-2,1] = -5`? Let me recount: `[-3]=-3`, `[-3,4]= 1`, `[4,-1,-2,1]=2`, `[-1,-2]=-3`, `[-1,-2,1]=-2`, `[-3,4,-1,-2,1]=-1`, `[-3,4,-1,-2]=-2`. Minimum is `-3` at either `[-3]` or `[-1,-2]`.

  **Explanation:** The subarray `[-3]` or `[-1,-2]` both sum to `-3`.

- For `nums = [1,2,3]`, the output should be `solution(nums) = 1`.

  **Explanation:** All positive — the least positive single element wins.

- For `nums = [-1,-2,-3]`, the output should be `solution(nums) = -6`.

  **Explanation:** All negative — the entire array is the minimum subarray.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^5`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[output]** `integer` — The smallest subarray sum.

```python
# Kadane's Algorithm — maximum subarray sum

def min_subarray(nums):
    cursum = best = nums[0]
    
    for num in nums[1:]:
        # Either extend the current subarray or start fresh here
        cursum = min(num, cursum + num)
        
        # Track the best sum seen so far
        best = min(best, cursum)
    
    return best


assert min_subarray([2,-3,4,-1,-2,1])  == -3  # given example 1 — [-3] or [-1,-2]
assert min_subarray([1,2,3])           == 1   # given example 2 — all positive, single element wins
assert min_subarray([-1,-2,-3])        == -6  # given example 3 — all negative, entire array
assert min_subarray([1])               == 1   # single element
assert min_subarray([-5])              == -5  # single negative
assert min_subarray([2,-1])            == -1  # two elements
assert min_subarray([1,-4,2,-3,1])     == -5  # [-4,2,-3] = -5
assert min_subarray([0,0,0])           == 0   # all zeros
print('All Pass!')
```

```python
def solution(nums):
    current = best = nums[0]
    for num in nums[1:]:
        current = min(num, current + num)
        best = min(best, current)
    return best


assert solution([2,-3,4,-1,-2,1])  == -3  # given example 1 — [-3] or [-1,-2]
assert solution([1,2,3])           == 1   # given example 2 — all positive, single element wins
assert solution([-1,-2,-3])        == -6  # given example 3 — all negative, entire array
assert solution([1])               == 1   # single element
assert solution([-5])              == -5  # single negative
assert solution([2,-1])            == -1  # two elements
assert solution([1,-4,2,-3,1])     == -5  # [-4,2,-3] = -5
assert solution([0,0,0])           == 0   # all zeros
print('All Pass!')
```
