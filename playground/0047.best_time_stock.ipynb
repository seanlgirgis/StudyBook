## Best Time to Buy and Sell Stock

You are given an array `prices` where `prices[i]` is the price of a stock on day `i`.

You want to **maximize your profit** by choosing a single day to **buy** and a different day **in the future** to **sell**.

Return the **maximum profit** you can achieve. If no profit is possible, return `0`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `prices = [7,1,5,3,6,4]`, the output should be `solution(prices) = 5`.

  **Explanation:** Buy on day 2 (price=1), sell on day 5 (price=6). Profit = 6−1 = **5**.

- For `prices = [7,6,4,3,1]`, the output should be `solution(prices) = 0`.

  **Explanation:** Prices only go down — no transaction gives profit, so return `0`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `prices`  
  Guaranteed constraints: `1 ≤ n ≤ 10^5`, `0 ≤ prices[i] ≤ 10^4`
- **[output]** `integer` — Maximum profit, or `0` if none.

```python
def solution(prices):
    profit = 0
    buy_price = prices[0]


    for price in prices[1:]:
        if price < buy_price:
            buy_price = price
            continue                     #no buy and sell the same day
        profit = max(profit, price - buy_price)
    return profit


assert solution([7,1,5,3,6,4])  == 5   # given example 1 — buy 1 sell 6
assert solution([7,6,4,3,1])    == 0   # given example 2 — declining, no profit
assert solution([1,2])          == 1   # two elements, profit exists
assert solution([2,1])          == 0   # two elements, no profit
assert solution([1])            == 0   # single element
assert solution([3,3,3,3])      == 0   # all same price
assert solution([1,2,3,4,5])    == 4   # steadily rising — buy first sell last
assert solution([2,4,1,7])      == 6   # dip then spike — buy at 1 sell at 7
print('All Pass!')    
```
