## Running Sum

Given an array of integers `nums`, return a new array `result` where `result[i]` is the **sum of all elements from index `0` to `i`** (inclusive).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [1,2,3,4]`, the output should be `solution(nums) = [1,3,6,10]`.

  **Explanation:**
  - `result[0] = 1`
  - `result[1] = 1+2 = 3`
  - `result[2] = 1+2+3 = 6`
  - `result[3] = 1+2+3+4 = 10`

- For `nums = [1,1,1,1,1]`, the output should be `solution(nums) = [1,2,3,4,5]`.

- For `nums = [3,1,4,1,5]`, the output should be `solution(nums) = [3,4,8,9,14]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 1000`, `-10^6 ≤ nums[i] ≤ 10^6`
- **[output]** `array.integer` — The running sum array.

```python
def solution(nums):
    res = []
    prevsum = 0
    for n in nums:
        prevsum += n
        res.append(prevsum)
    return res
        


assert solution([1,2,3,4])     == [1,3,6,10]    # given example 1
assert solution([1,1,1,1,1])   == [1,2,3,4,5]   # given example 2 — all ones
assert solution([3,1,4,1,5])   == [3,4,8,9,14]  # given example 3
assert solution([5])            == [5]            # single element
assert solution([0,0,0])        == [0,0,0]        # all zeros
assert solution([-1,-2,-3])     == [-1,-3,-6]     # negatives
assert solution([1,-1,1,-1])    == [1,0,1,0]      # alternating — running sum bounces
assert solution([10,0,-10,5])   == [10,10,0,5]    # zero in middle resets nothing
print('All Pass!')
```

```python
from itertools import accumulate
def solution(nums):
    return list(accumulate(nums))

        


assert solution([1,2,3,4])     == [1,3,6,10]    # given example 1
assert solution([1,1,1,1,1])   == [1,2,3,4,5]   # given example 2 — all ones
assert solution([3,1,4,1,5])   == [3,4,8,9,14]  # given example 3
assert solution([5])            == [5]            # single element
assert solution([0,0,0])        == [0,0,0]        # all zeros
assert solution([-1,-2,-3])     == [-1,-3,-6]     # negatives
assert solution([1,-1,1,-1])    == [1,0,1,0]      # alternating — running sum bounces
assert solution([10,0,-10,5])   == [10,10,0,5]    # zero in middle resets nothing
print('All Pass!')
```
