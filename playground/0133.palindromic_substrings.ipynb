## Palindromic Substrings

Count all palindromic substrings in `s`.

- `"abc"` → `3` (a, b, c)
- `"aaa"` → `6` (a, a, a, aa, aa, aaa)

**Pattern:** Expand-center — same as longest palindrome but count every valid expansion instead of tracking max length.

```python
def cntEvenPal(s, i):
    l = i 
    r = i + 1
    ret = 0
    while r < len(s) and l >= 0:
        if s[r] == s[l]:
            ret +=1
            r += 1
            l -= 1
        else:
            break
    return ret
 
def cntOddPal(s, i):
    l = i-1
    r = i+1
    ret = 1

    while r < len(s) and l >= 0:
        if s[r] == s[l]:
            ret +=1
            r += 1
            l -= 1
        else:
            break
    return ret
    
def solution(s):
    ret = 0
    
    for i in range(len(s)):
        ret += cntEvenPal(s, i)
        ret += cntOddPal(s,i)
    return ret

def test():
    assert solution("abc") == 3
    assert solution("aaa") == 6
    assert solution("a")   == 1
    assert solution("aa")  == 3
    print('All Pass!')  
test()


```

```python
def solution(s):
    count = 0

    def expand(l, r):
        nonlocal count
        while l >= 0 and r < len(s) and s[l] == s[r]:
            count += 1   # every valid (l,r) is a palindrome
            l -= 1
            r += 1

    for i in range(len(s)):
        expand(i, i)       # odd-length: single center
        expand(i, i + 1)   # even-length: two centers

    return count


assert solution("abc") == 3
assert solution("aaa") == 6
assert solution("a")   == 1
assert solution("aa")  == 3
print('All Pass!')
```
