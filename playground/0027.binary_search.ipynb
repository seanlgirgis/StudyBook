## Binary Search

Given a sorted array of integers `nums` and a target integer `target`, return the **index** of `target` in the array. If `target` does not exist, return `-1`.

Your solution must run in **O(log n)** time.

### Example

- For `nums = [-1,0,3,5,9,12]`, `target = 9`, the output should be `solution(nums, target) = 4`.

  **Explanation:** `9` is at index `4`.

- For `nums = [-1,0,3,5,9,12]`, `target = 2`, the output should be `solution(nums, target) = -1`.

  **Explanation:** `2` is not in the array.

- For `nums = [5]`, `target = 5`, the output should be `solution(nums, target) = 0`.

  **Explanation:** Single element, matches target.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A sorted array of distinct integers.  
  Guaranteed constraints: `1 ≤ nums.length ≤ 10^4`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[input]** `integer` `target`
- **[output]** `integer` — Index of target, or `-1` if not found.

```python
"""
id: lc_0704
title: Binary Search
source: leetcode
difficulty: easy
primary: binary search
tags: [array, binary-search]
leetcode_url: https://leetcode.com/problems/binary-search/
status: draft
last_updated: 2026-04-23
notes:
- key idea: classic two-pointer iterative binary search
- time: O(log n)
- space: O(1)
"""

# ============================================================================
# File: 704_lc_0704_binary_search_empty.py
# LeetCode 704: Binary Search
# Difficulty: Easy
#
# PROBLEM STATEMENT:
# Given a sorted array of integers nums and a target integer target, return 
# the index of target in the array. If target does not exist, return -1.
# You must write an algorithm with O(log n) runtime complexity.
#
# RULES:
# - nums is sorted in ascending order.
# - All integers in nums are unique.
#
# EXAMPLES:
# Input: nums = [-1,0,3,5,9,12], target = 9 -> Output: 4
# Input: nums = [-1,0,3,5,9,12], target = 2 -> Output: -1
# ============================================================================

from typing import List

def solution(nums, target):
    l , r  = 0 , len(nums) -1 

    while l <= r:
        mid = l+  (r -l )//2
        if nums[mid] == target:
            return mid
        if target > nums[mid]:
            l = mid + 1
        else:
            r = mid -1
    return -1





print(solution([-1,0,3,5,9,12], 9))
print(solution([-1,0,3,5,9,12], 2))

def test():
    assert solution([-1,0,3,5,9,12], 9) == 4   # given example 1 - target exists
    assert solution([-1,0,3,5,9,12], 2) == -1  # given example 2 - target missing
    assert solution([5], 5) == 0               # single element - match
    assert solution([5], 2) == -1              # single element - no match
    assert solution([2, 5], 2) == 0            # two elements - match left
    assert solution([2, 5], 5) == 1            # two elements - match right
    assert solution([2, 5], 3) == -1           # two elements - missing between
    assert solution([1, 3, 5, 7, 9], 1) == 0   # boundary - start
    assert solution([1, 3, 5, 7, 9], 9) == 4   # boundary - end
    assert solution([1, 3, 5, 7, 9], 5) == 2   # exact middle
    print('All Pass!')

test()
```
