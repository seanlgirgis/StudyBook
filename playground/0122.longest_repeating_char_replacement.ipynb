## Longest Repeating Character Replacement

Given a string `s` and integer `k`, you can replace at most `k` characters. Return the length of the longest substring containing the same letter after replacements.

- `s="ABAB"`, `k=2` → `4` (replace both Bs → "AAAA")
- `s="AABABBA"`, `k=1` → `4`

```python
%run _timer.py
```

```python
def solution(s, k):
    count = {}       # frequency of each char in current window
    max_freq = 0     # highest frequency of any single char in window
    l = 0
    res = 0

    for r in range(len(s)):
        count[s[r]] = count.get(s[r], 0) + 1
        max_freq = max(max_freq, count[s[r]])

        # Window size - max_freq = number of chars we need to replace
        # If that exceeds k, shrink from left
        while (r - l + 1) - max_freq > k:
            count[s[l]] -= 1
            l += 1

        res = max(res, r - l + 1)

    return res


assert solution("ABAB",    2) == 4
assert solution("AABABBA", 1) == 4
assert solution("AAAA",    0) == 4
assert solution("ABCD",    0) == 1
assert solution("ABCD",    4) == 4
print('All Pass!')
```
