## Longest Repeating Character Replacement

Given a string `s` and integer `k`, you can replace at most `k` characters. Return the length of the longest substring containing the same letter after replacements.

- `s="ABAB"`, `k=2` → `4` (replace both Bs → "AAAA")
- `s="AABABBA"`, `k=1` → `4`

```python
"""
id: lc_0424
title: Longest Repeating Character Replacement
source: leetcode
difficulty: medium
primary: sliding-window
tags: [string, hash-table, sliding-window]
leetcode_url: https://leetcode.com/problems/longest-repeating-character-replacement/
status: draft
last_updated: 2026-04-22
notes:
- key idea: sliding window + frequency map; window_size - max_freq <= k
- time: O(n)
- space: O(1) (26 uppercase letters)
"""

# ============================================================================
# File: 424_lc_0424_longest_repeating_character_replacement_empty.py
# LeetCode 424: Longest Repeating Character Replacement
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# You are given a string s and an integer k. You can choose any character of 
# the string and change it to any other uppercase English character. You can 
# perform this operation at most k times.
#
# Return the length of the longest substring containing the same letter 
# after performing the above operations.
#
# RULES:
# - s consists of only uppercase English letters.
# - 0 <= k <= s.length
#
# EXAMPLES:
# s = "ABAB", k = 2 -> 4 (Replace 'BB' with 'AA')
# s = "AABABBA", k = 1 -> 4 (Replace one 'A' to get "BBBB")
# ============================================================================

from typing import List


def characterReplacement(s: str, k: int) -> int:
    if not s:  return 0
    freq = {}
    best, most_freq = 0 , 0
    l = 0
    n = len(s)
        
    for r in range(n):
        freq[s[r]] = freq.get(s[r], 0) + 1
        most_freq = max(most_freq, freq[s[r]])
        while (r-l+1 - most_freq > k):
            freq[s[l]] = freq.get(s[l], 0) - 1
            l += 1
        best = max(best, r-l+1)
    return best

print(characterReplacement("AABABBA", 1))            
print(characterReplacement("ABAB", 2))


def test():
    assert characterReplacement("ABAB", 2) == 4  # example 1
    assert characterReplacement("AABABBA", 1) == 4  # example 2
    assert characterReplacement("AAAA", 0) == 4  # k=0, all same
    assert characterReplacement("ABCDE", 1) == 2  # k=1, diverse
    assert characterReplacement("BAAAB", 2) == 5  # replace ends
    assert characterReplacement("", 0) == 0  # empty string
    assert characterReplacement("A", 0) == 1  # single char
    assert characterReplacement("ABAA", 0) == 2  # k=0, partial same
    assert characterReplacement("ABBA", 2) == 4  # replace inner or outer
    assert characterReplacement("ABBB", 2) == 4  # replace single A
    print('All Pass!')

test()
```

```python
def solution(s, k):
    count = {}       # frequency of each char in current window
    max_freq = 0     # highest frequency of any single char in window
    l = 0
    res = 0

    for r in range(len(s)):
        count[s[r]] = count.get(s[r], 0) + 1
        max_freq = max(max_freq, count[s[r]])

        # Window size - max_freq = number of chars we need to replace
        # If that exceeds k, shrink from left
        while (r - l + 1) - max_freq > k:
            count[s[l]] -= 1
            l += 1

        res = max(res, r - l + 1)

    return res


assert solution("ABAB",    2) == 4
assert solution("AABABBA", 1) == 4
assert solution("AAAA",    0) == 4
assert solution("ABCD",    0) == 1
assert solution("ABCD",    4) == 4
print('All Pass!')
```
