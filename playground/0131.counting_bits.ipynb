## Counting Bits

Return an array of length `n+1` where `ans[i]` = number of 1 bits in `i`.

- `n=2` → `[0,1,1]`
- `n=5` → `[0,1,1,2,1,2]`

**Trick:** `dp[i] = dp[i >> 1] + (i & 1)` — shift right drops last bit, add whether last bit was 1.

```python
%run _timer.py
```

```python
def solution(n):
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        dp[i] = dp[i >> 1] + (i & 1)  # right-shift + check lowest bit
    return dp


assert solution(2) == [0,1,1]
assert solution(5) == [0,1,1,2,1,2]
assert solution(0) == [0]
assert solution(1) == [0,1]
print('All Pass!')
```
