## Next Greater Node in Linked List

Given a linked list, return an array where `ans[i]` is the value of the next node with a **greater** value. If none, `ans[i] = 0`.

### Example

- `[2,1,5]` → `[5,5,0]`
- `[2,7,4,3,5]` → `[7,0,5,5,0]`

### Hint
Flatten to a list first, then apply next greater element (mono stack). Return array of same length.

### Input/Output
- **[input]** head of linked list
- **[output]** `array.integer`

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def make_list(arr):
    dummy = ListNode()
    cur = dummy
    for v in arr:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def solution(head):
    values = []
    curr = head
    while curr:
        values.append(curr.val)
        curr = curr.next

    n = len(values)
    res = [0] * n
    stack = []   # indices

    for i in range(n):
        while stack and values[stack[-1]] < values[i]:
            res[stack.pop()] = values[i]
        stack.append(i)

    return res


assert solution(make_list([2,1,5]))             == [5,5,0]
assert solution(make_list([2,7,4,3,5]))         == [7,0,5,5,0]
assert solution(make_list([1,7,5,1,9,2,5,1]))   == [7,9,9,9,0,5,0,0]   # fixed: 7 has 9 to its right
assert solution(make_list([1]))                 == [0]
print('All Pass!')
```
