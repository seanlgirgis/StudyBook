## Maximum Window Sum

Given an array of integers `nums` and an integer `k`, return the **maximum sum** of any contiguous subarray of exactly length `k`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [2,1,5,1,3,2]`, `k = 3`, the output should be `solution(nums, k) = 9`.

  **Explanation:**
  - `[2,1,5] = 8`
  - `[1,5,1] = 7`
  - `[5,1,3] = 9` ← max
  - `[1,3,2] = 6`

- For `nums = [2,3,4,1,5]`, `k = 2`, the output should be `solution(nums, k) = 7`.

  **Explanation:** `[3,4] = 7` is the best pair.

- For `nums = [1,4,2,9,3]`, `k = 3`, the output should be `solution(nums, k) = 15`.

  **Explanation:** `[4,2,9] = 15`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — A non-empty array of integers.  
  Guaranteed constraints: `1 ≤ k ≤ nums.length ≤ 10^5`, `-10^4 ≤ nums[i] ≤ 10^4`
- **[input]** `integer` `k` — Window size.
- **[output]** `integer` — Maximum sum over all windows of size `k`.

```python
"""
id: lc_0643
title: Maximum Average Subarray I (Modified: Maximum Window Sum)
source: leetcode
difficulty: easy
primary: sliding-window
tags: [arrays, sliding-window, prefix-sum]
leetcode_url: https://leetcode.com/problems/maximum-average-subarray-i/
status: draft
last_updated: 2026-04-22
notes: 
- key idea: 
- time: 
- space: 
"""

# ============================================================================
# File: 643_lc_0643_maximum_window_sum.py
# LeetCode 643: Maximum Average Subarray I (Variant: Maximum Window Sum)
# Difficulty: Easy
# 
# PROBLEM STATEMENT:
# Given an array of integers nums and an integer k, return the maximum sum 
# of any contiguous subarray of length exactly k.
#
# RULES:
# - Time complexity must be O(n) or O(n^2).
# - 1 <= k <= nums.length <= 10^5
# - -10^4 <= nums[i] <= 10^4
#
# EXAMPLES:
# Input: nums = [2,1,5,1,3,2], k = 3
# Output: 9
#
# Input: nums = [2,3,4,1,5], k = 2
# Output: 7
#
# Input: nums = [1,4,2,9,3], k = 3
# Output: 15
# ============================================================================
from typing import List

def maxWindowSum(nums: List[int], k: int) -> int:
    # First window sum
    win = sum(nums[:k])
    best = win

    # Slide window
    for i in range(k, len(nums)):
        win += nums[i] - nums[i - k]
        best = max(best, win)

    return best     


print(maxWindowSum([10, 20, 30, 40, 50], 1))
print(maxWindowSum([-5, 10, -5, 20, -30, 40], 2))

def test():
    assert maxWindowSum([2,1,5,1,3,2], 3)  == 9   # given example 1 — [5,1,3]
    assert maxWindowSum([2,3,4,1,5], 2)    == 7   # given example 2 — [3,4]
    assert maxWindowSum([1,4,2,9,3], 3)    == 15  # given example 3 — [4,2,9]
    assert maxWindowSum([5], 1)            == 5   # single element
    assert maxWindowSum([1,2,3,4,5], 5)    == 15  # k = full array
    assert maxWindowSum([1,2,3,4,5], 1)    == 5   # k = 1, just max element
    assert maxWindowSum([-1,-2,-3,-4], 2)  == -3  # all negative — least bad window
    assert maxWindowSum([3,3,3,3], 2)      == 6   # all same
    print('All Pass!')

test()
```

```python
def solution(nums, k):
    n = len(nums)
    l = 0
    r = k
    if k == 1 :return max(nums)
    if k == n: return sum(nums)
    maxWin = -float('INF')
    for r in range (r, n+1):
        maxWin = max(maxWin, sum(nums[i] for i in range(l,r) )              )
        l+=1
    return(maxWin)
        

assert solution([2,1,5,1,3,2], 3)  == 9   # given example 1 — [5,1,3]
assert solution([2,3,4,1,5], 2)    == 7   # given example 2 — [3,4]
assert solution([1,4,2,9,3], 3)    == 15  # given example 3 — [4,2,9]
assert solution([5], 1)            == 5   # single element
assert solution([1,2,3,4,5], 5)    == 15  # k = full array
assert solution([1,2,3,4,5], 1)    == 5   # k = 1, just max element
assert solution([-1,-2,-3,-4], 2)  == -3  # all negative — least bad window
assert solution([3,3,3,3], 2)      == 6   # all same
assert solution([1,2,3,4,5], 2) == 9
print('All Pass!')
```
