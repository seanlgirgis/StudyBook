## Maximum Window Sum

Given an array of integers `nums` and an integer `k`, return the **maximum sum** of any contiguous subarray of exactly length `k`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [2,1,5,1,3,2]`, `k = 3`, the output should be `solution(nums, k) = 9`.

  **Explanation:**
  - `[2,1,5] = 8`
  - `[1,5,1] = 7`
  - `[5,1,3] = 9` ← max
  - `[1,3,2] = 6`

- For `nums = [2,3,4,1,5]`, `k = 2`, the output should be `solution(nums, k) = 7`.

  **Explanation:** `[3,4] = 7` is the best pair.

- For `nums = [1,4,2,9,3]`, `k = 3`, the output should be `solution(nums, k) = 15`.

  **Explanation:** `[4,2,9] = 15`.

```python
def solution(nums, k):
    n = len(nums)
    l = 0
    r = k
    if k == 1 :return max(nums)
    if k == n: return sum(nums)
    maxWin = -float('INF')
    for r in range (r, n+1):
        maxWin = max(maxWin, sum(nums[i] for i in range(l,r) )              )
        l+=1
    return(maxWin)
        

assert solution([2,1,5,1,3,2], 3)  == 9   # given example 1 — [5,1,3]
assert solution([2,3,4,1,5], 2)    == 7   # given example 2 — [3,4]
assert solution([1,4,2,9,3], 3)    == 15  # given example 3 — [4,2,9]
assert solution([5], 1)            == 5   # single element
assert solution([1,2,3,4,5], 5)    == 15  # k = full array
assert solution([1,2,3,4,5], 1)    == 5   # k = 1, just max element
assert solution([-1,-2,-3,-4], 2)  == -3  # all negative — least bad window
assert solution([3,3,3,3], 2)      == 6   # all same
assert solution([1,2,3,4,5], 2) == 9
print('All Pass!')
```
