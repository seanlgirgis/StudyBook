## Meeting Rooms II (LC 253)

Given meeting intervals `[start, end]`, return the **minimum number of rooms** needed.

**Pattern:** Sort by start + min-heap of end times.
- Heap = rooms currently occupied, keyed by when they free up
- `heap[0]` = earliest room to free
- If `heap[0] <= new_start` → room is free, reuse it (pop + push new end)
- Otherwise → no room free, open a new one (just push)
- `len(heap)` at the end = answer

```python
%run _timer.py
```

```python
import heapq

def solution(intervals):
    if not intervals:
        return 0

    intervals.sort()             # sort by start time
    heap = []                    # min-heap of room end times

    for start, end in intervals:
        if heap and heap[0] <= start:      # earliest room is free
            heapq.heapreplace(heap, end)   # reuse it — pop old end, push new end
        else:
            heapq.heappush(heap, end)      # no free room — open a new one

    return len(heap)


assert solution([[0,30],[5,10],[15,20]])               == 2
assert solution([[7,10],[2,4]])                        == 1
assert solution([[1,5]])                               == 1
assert solution([[1,2],[2,3],[3,4]])                   == 1   # sequential
assert solution([[1,5],[1,5],[1,5]])                   == 3   # all overlap
assert solution([[0,5],[5,10]])                        == 1   # back-to-back
print('All Pass!')
```
