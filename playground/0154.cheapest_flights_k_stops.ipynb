## Cheapest Flights Within K Stops (LC 787)

Find the cheapest price from `src` to `dst` using **at most `k` stops** (k+1 edges).

**Pattern:** Bellman-Ford with exactly `k+1` relaxation rounds.
- Each round = one more edge allowed
- **Key:** use a snapshot `tmp = prices[:]` at the start of each round — prevents using the same round's updates (stops you from taking more than 1 edge per round)
- Dijkstra doesn't work here because of the hop constraint

```python
%run _timer.py
```

```python
def solution(n, flights, src, dst, k):
    prices      = [float('inf')] * n
    prices[src] = 0

    for _ in range(k + 1):             # k stops = k+1 edges allowed
        tmp = prices[:]                # snapshot — only use PREVIOUS round's values
        for u, v, w in flights:
            if prices[u] != float('inf') and prices[u] + w < tmp[v]:
                tmp[v] = prices[u] + w
        prices = tmp

    return prices[dst] if prices[dst] != float('inf') else -1


assert solution(4, [[0,1,100],[1,2,100],[2,0,100],[1,3,600],[2,3,200]], 0, 3, 1) == 700
assert solution(3, [[0,1,100],[1,2,100],[0,2,500]], 0, 2, 1) == 200
assert solution(3, [[0,1,100],[1,2,100],[0,2,500]], 0, 2, 0) == 500
assert solution(2, [[0,1,100]], 1, 0, 1) == -1   # unreachable
print('All Pass!')
```
