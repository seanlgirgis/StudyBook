## Missing Number

Given an array `nums` containing `n` distinct numbers in the range `[0, n]`, return the one number that is missing.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [3,0,1]`, the output should be `2`.
- For `nums = [0,1]`, the output should be `2`.
- For `nums = [9,6,4,2,3,5,7,0,1]`, the output should be `8`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.integer` `nums` — `n` distinct numbers from range `[0, n]`
- **[output]** `integer`

```python
%run _timer.py
```

```python
def solution(nums):
    top_val = len(nums) + 1
    seen = set(nums)
    for n in range(top_val):
        if n not in seen:
            return n
    return -1

# math one-liner: return len(nums)*(len(nums)+1)//2 - sum(nums)


assert solution([3,0,1])               == 2
assert solution([0,1])                 == 2
assert solution([9,6,4,2,3,5,7,0,1])   == 8
assert solution([0])                   == 1
assert solution([1])                   == 0
assert solution([0,1,2,3,4,5,6,7,9])   == 8
print('All Pass!')
```
