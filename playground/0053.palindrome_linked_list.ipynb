## Palindrome Linked List

Given the `head` of a singly linked list, return `True` if it is a **palindrome**, `False` otherwise.

A palindrome reads the same forward and backward.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `head = [1,2,2,1]`, the output should be `True`.
- For `head = [1,2]`, the output should be `False`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `ListNode` `head`  
  Guaranteed constraints: `1 ≤ n ≤ 10^5`, `0 ≤ val ≤ 9`
- **[output]** `boolean`

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def to_ll(lst):
    dummy = ListNode()
    cur = dummy
    for v in lst:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def solution(head):
    vals = []
    curr = head
    while curr:
        vals.append(curr.val)
        curr = curr.next
    return vals == vals[::-1]


assert solution(to_ll([1,2,2,1]))     == True   # given example 1 — even palindrome
assert solution(to_ll([1,2]))         == False  # given example 2 — not palindrome
assert solution(to_ll([1]))           == True   # single node
assert solution(to_ll([1,2,1]))       == True   # odd palindrome
assert solution(to_ll([1,2,3]))       == False  # not palindrome
assert solution(to_ll([1,1,1,1]))     == True   # all same
assert solution(to_ll([1,2,3,2,1]))   == True   # odd length palindrome
assert solution(to_ll([1,2,3,4,5]))   == False  # not palindrome
print('All Pass!')
```
