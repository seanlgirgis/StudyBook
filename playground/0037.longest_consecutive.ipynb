## Longest Consecutive Sequence

Given an unsorted array of integers `nums`, return the length of the **longest consecutive sequence** (e.g. `4,5,6,7`).

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·log n)** will fit within the execution time limit.

### Example

- For `nums = [100,4,200,1,3,2]`, the output should be `solution(nums) = 4`.

  **Explanation:** The longest consecutive sequence is `[1,2,3,4]`, length `4`.

- For `nums = [0,3,7,2,5,8,4,6,0,1]`, the output should be `solution(nums) = 9`.

  **Explanation:** `[0,1,2,3,4,5,6,7,8]`, length `9`.

- For `nums = [1]`, the output should be `solution(nums) = 1`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `0 ≤ nums.length ≤ 10^5`, `-10^9 ≤ nums[i] ≤ 10^9`
- **[output]** `integer` — Length of the longest consecutive sequence.

```python
%run _timer.py
```

```python
def solution(nums):
    longest_streak = 0
    
    def length_of_streak(n, nums):
        ret = 0
        while n in nums:
            ret+= 1
            n+=1
        return ret
    nums = set(nums)
    for n in nums:
        if (n-1) not in nums:
            longest_streak= max(longest_streak, length_of_streak(n, nums))
    return longest_streak
            
            
        


assert solution([100,4,200,1,3,2])        == 4  # given example 1 — [1,2,3,4]
assert solution([0,3,7,2,5,8,4,6,0a,1])   == 9  # given example 2 — [0..8]
assert solution([1])                      == 1  # given example 3 — single element
assert solution([])                       == 0  # empty array
assert solution([1,2,3,4,5])             == 5  # already consecutive
assert solution([5,4,3,2,1])             == 5  # reverse order
assert solution([1,3,5,7])               == 1  # no consecutive pairs
assert solution([1,2,2,3])               == 3  # duplicates — [1,2,3]
print('All Pass!')
```
