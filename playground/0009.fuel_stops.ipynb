## Fuel Stops

You are driving from position `0` to position `target` along a straight highway. Your car has a full tank at the start, and the tank allows you to travel a maximum of **10 units** before running out of fuel. At each gas station you visit, your tank is refilled completely.

Gas stations are located at various positions given in the array `stations`. You may skip stations — you only need to stop if you have to.

Your task is to find the **minimum number of fuel stops** needed to reach the target. If the target is unreachable, return `-1`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(stations.length²)** will fit within the execution time limit.

### Example

- For `target = 25` and `stations = [10, 15, 20]`, the output should be `solution(target, stations) = 2`.

  **Explanation:**
  - From `0`, the farthest reachable station is `10`. Stop there. *(1 stop)*
  - From `10`, the farthest reachable station is `20`. Stop there. *(2 stops)*
  - From `20`, the target `25` is within range (`25 - 20 = 5 ≤ 10`). Done.
  - Minimum stops: **2**

- For `target = 22` and `stations = [7, 14]`, the output should be `solution(target, stations) = 2`.

  **Explanation:**
  - From `0`, the farthest reachable station is `7`. Stop there. *(1 stop)*
  - From `7`, the farthest reachable station is `14`. Stop there. *(2 stops)*
  - From `14`, the target `22` is within range (`22 - 14 = 8 ≤ 10`). Done.
  - Minimum stops: **2**

- For `target = 12` and `stations = []`, the output should be `solution(target, stations) = -1`.

  **Explanation:**
  - No stations available and `12 > 10`. The target is unreachable.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `integer` `target` — The destination position. Guaranteed: `target > 0`
- **[input]** `array.integer` `stations` — Positions of the gas stations along the highway.
- **[output]** `integer` — Minimum number of fuel stops, or `-1` if unreachable.

```python
def solution(target, stations):
    stations.sort()
    n = len(stations)
    stops = 0
    #edge case 1

    # Check if we can even reach the first station
    if not stations or stations[0] > 10:
        return -1 if target > 10 else 0

    # Check if the last station can reach the target
    if stations[-1] + 10 < target:
        return -1
        
    pos = 0
    i = 0
    while pos < target:
        if pos + 10 >= target:   # no need to fuel we are at target
            break
        found = False
        while i < n and stations[i] <= (pos + 10):
            found = True
            i += 1
        i -= 1
        if not found:
            return -1
        else:
            pos = stations[i]
            stops += 1
    return stops

assert solution(25, [10, 15, 20])  == 2   # given example 1 — jump to farthest each time
assert solution(22, [7, 14])       == 2   # given example 2
assert solution(12, [])            == -1  # given example 3 — unreachable
assert solution(5,  [])            == 0   # within range, no stops needed
assert solution(10, [])            == 0   # exactly at range limit, no stop needed
assert solution(20, [5, 15])       == 2   # must stop twice
assert solution(20, [5])           == -1  # station too early, can't bridge the gap
assert solution(30, [10, 20])      == 2   # clean two-stop route
print('All Pass!')
```
