## Network Delay Time (LC 743)

Directed weighted graph. Signal starts at node `k`. How long until all `n` nodes receive it?
Return `-1` if any node is unreachable.

**Pattern:** Dijkstra — shortest path from single source to all nodes.
- Min-heap of `(cost, node)`
- Skip already-visited nodes
- Answer = `max(dist.values())` if all `n` nodes reached, else `-1`

```python
import heapq
from collections import defaultdict

def solution(times, n, k):
    graph = defaultdict(list)
    for u, v, w in times:
        graph[u].append((v, w))

    dist = {}                   # node → shortest distance from k
    heap = [(0, k)]             # (cost, node)

    while heap:
        cost, node = heapq.heappop(heap)
        if node in dist:
            continue            # already settled with a shorter path
        dist[node] = cost
        for neighbor, weight in graph[node]:
            if neighbor not in dist:
                heapq.heappush(heap, (cost + weight, neighbor))

    return max(dist.values()) if len(dist) == n else -1


assert solution([[2,1,1],[2,3,1],[3,4,1]], 4, 2) == 2
assert solution([[1,2,1]], 2, 1)                 == 1
assert solution([[1,2,1]], 2, 2)                 == -1   # node 1 unreachable from 2
assert solution([[1,2,1],[2,3,2],[1,3,4]], 3, 1) == 3
print('All Pass!')
```
