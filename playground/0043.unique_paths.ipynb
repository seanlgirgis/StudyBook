## Unique Paths

A robot is located at the **top-left corner** of an `m × n` grid. It can only move **right** or **down**. Return the number of **unique paths** to reach the bottom-right corner.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(m·n)** will fit within the execution time limit.

### Example

- For `m = 3`, `n = 7`, the output should be `solution(m, n) = 28`.

- For `m = 3`, `n = 2`, the output should be `solution(m, n) = 3`.

  **Explanation:** Three paths: right→down→down, down→right→down, down→down→right.

- For `m = 1`, `n = 1`, the output should be `solution(m, n) = 1`.

  **Explanation:** Already at the destination.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `integer` `m` — Number of rows. `1 ≤ m ≤ 100`
- **[input]** `integer` `n` — Number of columns. `1 ≤ n ≤ 100`
- **[output]** `integer` — Number of unique paths.

```python
%run _timer.py
```

```python
def solution(m, n):
    dp = [[1] * n for _ in range(m)]
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = dp[i-1][j] + dp[i][j-1]
    return dp[-1][-1]


assert solution(3, 7)  == 28   # given example 1
assert solution(3, 2)  == 3    # given example 2
assert solution(1, 1)  == 1    # given example 3 — already there
assert solution(1, 5)  == 1    # single row — only one path
assert solution(5, 1)  == 1    # single column — only one path
assert solution(2, 2)  == 2    # 2x2 grid
assert solution(3, 3)  == 6    # 3x3 grid
assert solution(7, 3)  == 28   # flipped — same answer
print('All Pass!')
```
