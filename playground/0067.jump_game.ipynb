## Jump Game

You are given an integer array `nums`. You are initially at index `0`. Each element `nums[i]` represents the **maximum jump length** from that position.

Return `True` if you can reach the last index, `False` otherwise.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `nums = [2,3,1,1,4]`, the output should be `True`.

  **Explanation:** Jump 1 from index 0 to 1, then 3 from index 1 to the last index.

- For `nums = [3,2,1,0,4]`, the output should be `False`.

  **Explanation:** Always land on index 3 which has jump length 0 — stuck.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums`  
  Guaranteed constraints: `1 ≤ n ≤ 10^4`, `0 ≤ nums[i] ≤ 10^5`
- **[output]** `boolean`

```python
def jumpGame(nums):
    max_reach = 0

    for i in range(len(nums)):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + nums[i])
    return True
    
        

assert jumpGame([2,3,1,1,4])  == True   # given example 1
assert jumpGame([3,2,1,0,4])  == False  # given example 2 — stuck at index 3
assert jumpGame([0])          == True   # single element — already at end
assert jumpGame([1,0])        == True   # jump from 0 to end
assert jumpGame([0,1])        == False  # can't move from index 0
assert jumpGame([2,0,0])      == True   # jump over zeros
assert jumpGame([1,1,1,1,0])  == True   # reach last index exactly
assert jumpGame([1,0,1,0])    == False  # stuck at index 1
print('All Pass!')
```

```python
def jumpGame(nums):
    max_reach = 0
    for i, jump in enumerate(nums):
        if i > max_reach:
            return False
        max_reach = max(max_reach, i + jump)
    return True
        
jumpGame([3,2,2,0,4])  

assert jumpGame([2,3,1,1,4])  == True   # given example 1
assert jumpGame([3,2,1,0,4])  == False  # given example 2 — stuck at index 3
assert jumpGame([0])          == True   # single element — already at end
assert jumpGame([1,0])        == True   # jump from 0 to end
assert jumpGame([0,1])        == False  # can't move from index 0
assert jumpGame([2,0,0])      == True   # jump over zeros
assert jumpGame([1,1,1,1,0])  == True   # reach last index exactly
assert jumpGame([1,0,1,0])    == False  # stuck at index 1
print('All Pass!')
```

```python
def solution(nums):
    max_needed = len(nums)-1
    max_reach = 0
    for i in range(max_needed):
        if i > max_reach:
            return False
        max_reach = max(max_reach , nums[i]+i)
    return max_reach >= max_needed


assert solution([2,3,1,1,4])  == True   # given example 1
assert solution([3,2,1,0,4])  == False  # given example 2 — stuck at index 3
assert solution([0])          == True   # single element — already at end
assert solution([1,0])        == True   # jump from 0 to end
assert solution([0,1])        == False  # can't move from index 0
assert solution([2,0,0])      == True   # jump over zeros
assert solution([1,1,1,1,0])  == True   # reach last index exactly
assert solution([1,0,1,0])    == False  # stuck at index 1
print('All Pass!')
```
