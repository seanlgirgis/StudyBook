## Largest Rectangle in Histogram

Given an array `heights` representing bar widths=1, find the largest rectangle area you can form.

**Pattern:** Monotonic increasing stack.  
- When a shorter bar arrives, it *blocks* taller bars from extending right  
- Pop each taller bar → compute its area using how far left it could have extended  
- Remaining bars in stack extend all the way to the right end

Key insight: store `(start_index, height)` — when you pop, `start` tells you how far left that height can reach.

```python
%run _timer.py
```

```python
def solution(heights):
    stack    = []   # monotonic increasing — stores (start_idx, height)
    max_area = 0

    for i, h in enumerate(heights):
        start = i                        # how far left bar i can extend

        # Pop every bar taller than h — it can't extend past i
        while stack and stack[-1][1] > h:
            idx, height = stack.pop()
            max_area = max(max_area, height * (i - idx))   # width = i - start_idx
            start    = idx               # current bar can extend back to where popped bar started

        stack.append((start, h))

    # Remaining bars extend to the right end of the array
    for idx, height in stack:
        max_area = max(max_area, height * (len(heights) - idx))

    return max_area


assert solution([2,1,5,6,2,3])     == 10
assert solution([2,4])             == 4
assert solution([1])               == 1
assert solution([6,2,5,4,5,1,6])   == 12
assert solution([1,1])             == 2
assert solution([5,5,5,5])         == 20
print('All Pass!')
```
