## Largest Rectangle in Histogram

Given an array `heights` representing bar widths=1, find the largest rectangle area you can form.

**Pattern:** Monotonic increasing stack.  
- When a shorter bar arrives, it *blocks* taller bars from extending right  
- Pop each taller bar → compute its area using how far left it could have extended  
- Remaining bars in stack extend all the way to the right end

Key insight: store `(start_index, height)` — when you pop, `start` tells you how far left that height can reach.

```python
"""
id: lc_0084
title: Largest Rectangle in Histogram
source: leetcode
difficulty: hard
primary: monotonic stack
tags: [stack, array, monotonic-stack]
leetcode_url: https://leetcode.com/problems/largest-rectangle-in-histogram/
status: draft
last_updated: 2026-04-24
notes:
- key idea: Monotonic increasing stack of (index, height)
- time: O(n)
- space: O(n)
"""

# ============================================================================
# File: 084_lc_0084_largest_rectangle_in_histogram_empty.py
# LeetCode 84: Largest Rectangle in Histogram
# Difficulty: Hard
#
# PROBLEM STATEMENT:
# Given an array of integers heights representing the histogram's bar height 
# where the width of each bar is 1, return the area of the largest rectangle 
# in the histogram.
#
# RULES:
# - Heights are non-negative.
# - You must find the maximum area possible.
#
# EXAMPLES:
# Input: heights = [2,1,5,6,2,3]
# Output: 10 (The rectangle is formed by heights 5 and 6 with area 2 * 5 = 10)
# ============================================================================

from typing import List

def largestRectangleArea(heights: List[int]) -> int:
    best = 0 
    stack = []                  #mono increasing stack ( h , i)
    n = len(heights)
    for i , h in enumerate(heights):
        inh_ind = i
        while stack and h < stack[-1][0]:
            popped_height, popped_ind = stack.pop()
            inh_ind = popped_ind
            best = max(best, popped_height * (i -popped_ind ))
        stack.append((h, inh_ind ))
    while stack:
        popped_height, popped_ind = stack.pop()
        best = max(best, popped_height * (n - popped_ind))

    return best
        
        
    

print(largestRectangleArea([2, 1, 5, 6, 2, 3]))
print(largestRectangleArea([2, 4]))

def test():
    assert largestRectangleArea([2, 1, 5, 6, 2, 3]) == 10  # standard case
    assert largestRectangleArea([2, 4]) == 4              # two bars increasing
    assert largestRectangleArea([1, 1]) == 2              # uniform height
    assert largestRectangleArea([0, 9]) == 9              # zero height presence
    assert largestRectangleArea([4, 2, 0, 3, 2, 5]) == 6  # disconnected components
    assert largestRectangleArea([5, 4, 3, 2, 1]) == 9     # decreasing sequence
    assert largestRectangleArea([1, 2, 3, 4, 5]) == 9     # increasing sequence
    assert largestRectangleArea([2, 1, 2]) == 3           # valley shape
    assert largestRectangleArea([1]) == 1                 # single element
    assert largestRectangleArea([]) == 0                  # empty list
    print('All Pass!')

test()
```

```python
"""
id: lc_0084
title: Largest Rectangle in Histogram
source: leetcode
difficulty: hard
primary: monotonic stack
tags: [stack, array]
leetcode_url: https://leetcode.com/problems/largest-rectangle-in-histogram/
status: draft
last_updated: 2026-04-23
notes:
- key idea: maintain a monotonic increasing stack of (index, height)
- time: O(n)
- space: O(n)
"""

# ============================================================================
# File: 084_lc_0084_largest_rectangle_in_histogram_empty.py
# LeetCode 84: Largest Rectangle in Histogram
# Difficulty: Hard
#
# PROBLEM STATEMENT:
# Given an array of integers heights representing the histogram's bar height 
# where the width of each bar is 1, return the area of the largest rectangle 
# in the histogram.
#
# RULES:
# - You can only form rectangles with the given heights.
# - The width of each bar is constant at 1.
#
# EXAMPLES:
# Input: heights = [2,1,5,6,2,3]
# Output: 10
# Explanation: The largest rectangle is formed by bars [5,6] with area = 5 * 2 = 10.
# ============================================================================

from typing import List

def largestRectangleArea(heights: List[int]) -> int:
    stack = []   #tuples  inherited index and height
    best = 0
    n = len(heights)
    for i, h in enumerate(heights):
        inh_ind = i
        while  stack and  stack[-1][1] > h:
            popped_ind, popped_height = stack.pop()
            inh_ind  = popped_ind
            area = popped_height* (i - popped_ind)
            best = max(best, area)
        stack.append((inh_ind, h))
    while stack:
        popped_ind, poped_height = stack.pop()
        area =  poped_height * ( n - popped_ind)
        best = max(best,area)

    return best
        

print(largestRectangleArea([2, 1, 5, 6, 2, 3]))  #10
print(largestRectangleArea([2, 4]))   # 4

def test():
    assert largestRectangleArea([2, 1, 5, 6, 2, 3]) == 10  # given example 1
    assert largestRectangleArea([2, 4]) == 4               # given example 2
    assert largestRectangleArea([1, 1]) == 2               # minimal flat
    assert largestRectangleArea([5]) == 5                  # single element
    assert largestRectangleArea([2, 1, 2]) == 3            # valley shape
    assert largestRectangleArea([1, 2, 3, 4, 5]) == 9      # strictly increasing
    assert largestRectangleArea([5, 4, 3, 2, 1]) == 9      # strictly decreasing
    assert largestRectangleArea([2, 2, 2, 2]) == 8         # all same height
    assert largestRectangleArea([0, 9]) == 9               # zero height inclusion
    assert largestRectangleArea([10, 0, 10]) == 10         # separated peaks
    print('All Pass!')

test()
```

```python
def solution(heights):
    stack    = []   # monotonic increasing — stores (start_idx, height)
    max_area = 0

    for i, h in enumerate(heights):
        start = i                        # how far left bar i can extend

        # Pop every bar taller than h — it can't extend past i
        while stack and stack[-1][1] > h:
            idx, height = stack.pop()
            max_area = max(max_area, height * (i - idx))   # width = i - start_idx
            start    = idx               # current bar can extend back to where popped bar started

        stack.append((start, h))

    # Remaining bars extend to the right end of the array
    for idx, height in stack:
        max_area = max(max_area, height * (len(heights) - idx))

    return max_area


assert solution([2,1,5,6,2,3])     == 10
assert solution([2,4])             == 4
assert solution([1])               == 1
assert solution([6,2,5,4,5,1,6])   == 12
assert solution([1,1])             == 2
assert solution([5,5,5,5])         == 20
print('All Pass!')
```
