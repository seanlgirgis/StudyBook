## Two Sum

Given an array of integers `nums` and an integer `target`, return the **indices** `[i, j]` (with `i < j`) of the two numbers that add up to `target`.

You may assume that exactly one solution exists, and you may not use the same element twice.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n²)** will fit within the execution time limit.

### Example

- For `nums = [2, 7, 11, 15]`, `target = 9`, the output should be `solution(nums, target) = [0, 1]`.

  **Explanation:** `nums[0] + nums[1] = 2 + 7 = 9`.

- For `nums = [3, 2, 4]`, `target = 6`, the output should be `solution(nums, target) = [1, 2]`.

  **Explanation:** `nums[1] + nums[2] = 2 + 4 = 6`. Note that `3 + 3` is not valid — you cannot reuse index 0.

- For `nums = [3, 3]`, `target = 6`, the output should be `solution(nums, target) = [0, 1]`.

  **Explanation:** The only two elements are both `3`, and they are at different indices.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `nums` — Array of integers.  
  Guaranteed constraints: `2 ≤ nums.length ≤ 10^4`, `-10^9 ≤ nums[i] ≤ 10^9`
- **[input]** `integer` `target`
- **[output]** `array.integer` — `[i, j]` with `i < j`.

```python
def solution(nums, target):
    seen = {}
    for i, num in enumerate(nums):
         if (target - num) in seen:
             return [seen[target - num], i]
         seen[num] = i
    return []


assert solution([2,7,11,15], 9)       == [0,1]  # given example 1 — 2+7
assert solution([3,2,4], 6)           == [1,2]  # given example 2 — 2+4, not 3+3
assert solution([3,3], 6)             == [0,1]  # given example 3 — duplicate values, different indices
assert solution([1,2,3,4,5], 9)       == [3,4]  # answer at the end — 4+5
assert solution([0,4,3,0], 0)         == [0,3]  # zeros — 0+0
assert solution([-1,-2,-3,-4,-5], -8) == [2,4]  # negatives — -3+(-5)
assert solution([1,5,3,7,2], 8)       == [1,2]  # 1+7, earlier index wins
assert solution([2,5,5,11], 10)       == [1,2]  # duplicate mid-array — 5+5
print('All Pass!')
```
