## Merge Intervals

Given an array of intervals where `intervals[i] = [start, end]`, merge all overlapping intervals and return the result.

Two intervals overlap if one starts before the other ends.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n log n)** will fit within the execution time limit.

### Example

- For `[[1,3],[2,6],[8,10],[15,18]]`, the output should be `[[1,6],[8,10],[15,18]]`.
- For `[[1,4],[4,5]]`, the output should be `[[1,5]]` — touching counts as overlapping.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.array.integer` `intervals`
- **[output]** `array.array.integer`

```python
%run _timer.py
```

```python
def solution(intervals):

    out = []
    intervals.sort()
    def mergeIt(interv):
        if interv[0] <= out[-1][1]:
            out[-1][1] = max(out[-1][1] , interv[1])
        else:
            out.append(interv)

    out.append(intervals[0])
    
    for interv in intervals[1:]:
        mergeIt(interv)
    
    return out
    


assert solution([[1,3],[2,6],[8,10],[15,18]]) == [[1,6],[8,10],[15,18]]
assert solution([[1,4],[4,5]])                == [[1,5]]   # touching merges
assert solution([[1,4],[0,4]])                == [[0,4]]   # unsorted input
assert solution([[1,4],[2,3]])                == [[1,4]]   # one inside another
assert solution([[1,2],[3,4],[5,6]])          == [[1,2],[3,4],[5,6]]  # no overlap
assert solution([[1,10],[2,3],[4,5]])         == [[1,10]]  # all inside first
assert solution([[1,4]])                      == [[1,4]]   # single interval
print('All Pass!')
```
