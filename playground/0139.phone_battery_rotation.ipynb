## Phone Battery Rotation (Custom Simulation)

N batteries in circular order. Battery `i` has `capacity[i]` minutes and needs `recharge[i]` minutes after full drain. Keep phone on for `t` minutes.

- Count a battery only if fully drained.
- If no battery is ready at current time → return `-1`.
- If final battery only partially used → don't count it.

**Pattern:** Circular pointer. At each step, scan one full circle for a ready battery. Full drain → count + update ready_at. Partial → return count.

```python
def solution(t, capacity, recharge):
    n = len(capacity)
    ready_at  = [0] * n   # time when each battery is available again
    cur_time  = 0
    full_used = 0
    start_ptr = 0

    while cur_time < t:
        found = False
        for k in range(n):               # one full circular scan
            i = (start_ptr + k) % n
            if ready_at[i] <= cur_time:  # battery is ready
                found = True
                remaining = t - cur_time
                if capacity[i] <= remaining:      # full drain
                    cur_time  += capacity[i]
                    full_used += 1
                    ready_at[i] = cur_time + recharge[i]
                    start_ptr   = (i + 1) % n     # next search starts after i
                else:                             # partial — reaches t
                    return full_used
                break
        if not found:
            return -1                    # gap: no battery available

    return full_used


assert solution(8,  [2,5],  [10,4])      == -1
assert solution(10, [7,5],  [12,4])      == 1
assert solution(6,  [3,3],  [1,1])       == 2
assert solution(1,  [5],    [3])         == 0
assert solution(5,  [5],    [3])         == 1
print('All Pass!')
```
