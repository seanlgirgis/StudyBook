## Group Anagrams

Given an array of strings `words`, group the anagrams together and return a list of groups. Each group contains all strings from the input that are anagrams of each other. The order of groups and the order within each group does not matter.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n·k·log k)** will fit within the execution time limit, where `k` is the maximum length of a string.

### Example

- For `words = ["eat","tea","tan","ate","nat","bat"]`, the output should be `solution(words)` grouped as:
  `[["eat","tea","ate"], ["tan","nat"], ["bat"]]` (any order).

  **Explanation:** `eat/tea/ate` are anagrams, `tan/nat` are anagrams, `bat` is alone.

- For `words = [""]`, the output should be `solution(words) = [[""]]`.

- For `words = ["a"]`, the output should be `solution(words) = [["a"]]`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.string` `words` — A non-empty array of lowercase strings.  
  Guaranteed constraints: `1 ≤ words.length ≤ 10^4`, `0 ≤ words[i].length ≤ 100`
- **[output]** `array.array.string` — Groups of anagrams (order within groups and order of groups does not matter).

```python
def makeAna(s):
    lookup = [0] * 26
    for c in s:
        lookup[ord(c) - ord('a')] += 1
    return tuple(lookup)
    
def solution(words):
    grps = {}
    for s in words:
        grps.setdefault(makeAna(s), []).append(s)
    return grps.values()
    



def check(result, expected):
    return sorted([sorted(g) for g in result]) == sorted([sorted(g) for g in expected])

assert check(solution(["eat","tea","tan","ate","nat","bat"]),
             [["eat","tea","ate"],["tan","nat"],["bat"]])   # given example 1
assert check(solution([""]),          [[""]])               # given example 2 — empty string
assert check(solution(["a"]),         [["a"]])              # given example 3 — single word
assert check(solution(["a","b"]),     [["a"],["b"]])        # no anagrams
assert check(solution(["ab","ba"]),   [["ab","ba"]])        # one group
assert check(solution(["abc","bca","cab","xyz"]),
             [["abc","bca","cab"],["xyz"]])                  # three anagrams + loner
assert check(solution(["a","a"]),     [["a","a"]])          # duplicate words
assert check(solution(["cat","act","dog","god","tac"]),
             [["cat","act","tac"],["dog","god"]])           # two groups
print('All Pass!')
```
