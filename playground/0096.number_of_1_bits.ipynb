## Number of 1 Bits

Write a function that takes an unsigned integer and returns the number of `1` bits it has (also known as the **Hamming weight**).

### Example

- `n = 11` → `3` (binary: `1011` — three 1s)
- `n = 128` → `1` (binary: `10000000` — one 1)
- `n = 4294967293` → `31` (binary: 32 bits, 31 ones)

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `integer` `n`
- **[output]** `integer`

```python
def solution(n):
    count = 0
    while n:
        # Check if the last bit is 1
        if n & 1:
            count += 1
        n >>= 1
    return count


assert solution(11)         == 3
assert solution(128)        == 1
assert solution(1)          == 1
assert solution(0)          == 0
assert solution(4294967293) == 31
assert solution(2147483645) == 30
print('All Pass!')
```

```python
# n & (n-1) clears the lowest set bit — loop only runs as many times as there are 1s
def solution_b(n):
    count = 0
    while n:
        n &= n - 1
        count += 1
    return count

assert solution_b(11)         == 3
assert solution_b(128)        == 1
assert solution_b(0)          == 0
assert solution_b(4294967293) == 31
print('solution_b pass')
```

```python
# Python one-liner using bin() — converts to string like '0b1011', count the '1' chars
def solution_c(n):
    return bin(n).count('1')

assert solution_c(11)         == 3
assert solution_c(128)        == 1
assert solution_c(0)          == 0
assert solution_c(4294967293) == 31
print('solution_c pass')
```
