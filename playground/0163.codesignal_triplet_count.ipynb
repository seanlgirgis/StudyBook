## CodeSignal — Triplet Count

Given a string `s`, count the number of triplets (consecutive substrings of length 3)
where the **first and last characters are equal**, case-insensitive.

### Examples

```
"aXa"       → 1    (a == a)
"axAxbXb"   → 4    (axA ✓, xAx ✓, xbX ✓, bXb ✓)
"abc"       → 0
""          → 0
"AaA"       → 1    (A == A, case-insensitive)
"aaaa"      → 2    (overlapping: aaa, aaa)
```

### Rules
- A triplet is any 3 **consecutive** characters `s[i], s[i+1], s[i+2]`.
- Match rule: `s[i].lower() == s[i+2].lower()`
- The middle character does not matter.
- Return `0` if `len(s) < 3`.

### Hint
Slide a window of size 3 across the string. One pass, O(n).

````python
"""
id: cs_q1
title: CodeSignal — Triplet Count
source: codesignal
difficulty: easy
primary: string
tags: [sliding-window, string, case-insensitive]
leetcode_equivalent: Pattern in LC 1180, LC 467
status: draft
last_updated: 2026-04-24
notes:
- key idea: s[i].lower() == s[i+2].lower() for each window of 3
- time: O(n)
- space: O(1)
"""

def triplet_count(s: str) -> int:
    n = len(s)
    count = 0
    for i in range(n-2):
        if s[i].lower() == s[i+2].lower():
            count += 1
    return count


print(triplet_count("aXa"))      # 1
print(triplet_count("axAxbXb"))  # 4
print(triplet_count("abc"))      # 0
print(triplet_count(""))         # 0
print(triplet_count("aaaa"))     # 2


def test():
    assert triplet_count("aXa")      == 1, "single triplet"
    assert triplet_count("axAxbXb")  == 4, "axA✓ xAx✓ xbX✓ bXb✓ = 4"
    assert triplet_count("abc")      == 0, "no match"
    assert triplet_count("")         == 0, "empty string"
    assert triplet_count("a")        == 0, "too short"
    assert triplet_count("ab")       == 0, "too short 2"
    assert triplet_count("AaA")      == 1, "case insensitive A==a"
    assert triplet_count("aaa")      == 1, "all same"
    assert triplet_count("aaaa")     == 2, "overlapping"
    assert triplet_count("xXx")      == 1, "x==X"
    assert triplet_count("abcba")    == 1, "bcb only"
    assert triplet_count("AbCbA")    == 1, "bCb only"
    print("All Pass!")

test()
````
