## Min Stack

Design a stack that supports the following operations in **O(1)** time each:

- `push(val)` — push `val` onto the stack
- `pop()` — remove the top element
- `top()` — return the top element without removing it
- `get_min()` — return the **minimum** element in the stack

**Note:** `pop`, `top`, and `get_min` will always be called on a non-empty stack.

### Example

```
push(-2), push(0), push(-3)
get_min() → -3
pop()
top()    → 0
get_min() → -2
```

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- Implement the `MinStack` class below.

````python
"""
id: lc_0155
title: Min Stack
source: leetcode
difficulty: medium
primary: stack
tags: [stack, design]
leetcode_url: https://leetcode.com/problems/min-stack/
status: draft
last_updated: 2026-04-23
notes:
- key idea: Maintain an auxiliary stack to store the minimum value at each state.
- time: O(1) for all operations
- space: O(n)
"""

# ============================================================================
# File: 155_lc_0155_min_stack_empty.py
# LeetCode 155: Min Stack
# Difficulty: Medium
#
# PROBLEM STATEMENT:
# Design a stack that supports push, pop, top, and retrieving the minimum 
# element in constant time.
#
# RULES:
# - push(val) -- Push element val onto stack.
# - pop() -- Removes the element on the top of the stack.
# - top() -- Get the top element.
# - getMin() -- Retrieve the minimum element in the stack.
# - Methods pop, top and getMin operations will always be called on non-empty stacks.
#
# EXAMPLES:
# MinStack minStack = new MinStack();
# minStack.push(-2);
# minStack.push(0);
# minStack.push(-3);
# minStack.getMin(); // return -3
# minStack.pop();
# minStack.top();    // return 0
# minStack.getMin(); // return -2
# ============================================================================

from typing import List

class MinStack:
    def __init__(self):
        self.data = []

    def push(self, val: int) -> None:
        if not self.data:
            itm = (val, val)
        else:
            m = min(val, self.data[-1][1])
            itm = (val, m)
        self.data.append(itm)

    def pop(self) -> None:
        self.data.pop()

    def top(self) -> int:
        return self.data[-1][0]

    def getMin(self) -> int:
        return self.data[-1][1]

# Demo
obj = MinStack()
print(obj.push(-2), obj.push(0), obj.push(-3), obj.getMin())
print(obj.pop(), obj.top(), obj.getMin())

def test():
    ms = MinStack()
    ms.push(-2)
    ms.push(0)
    ms.push(-3)
    assert ms.getMin() == -3  # initial min after pushes
    ms.pop()
    assert ms.top() == 0      # top after popping -3
    assert ms.getMin() == -2  # min reverts to -2
    ms.push(-1)
    assert ms.getMin() == -2  # push larger than current min
    ms.push(-5)
    assert ms.getMin() == -5  # new absolute min
    ms.pop()
    assert ms.getMin() == -2  # min reverts after popping -5
    ms.push(10)
    assert ms.top() == 10     # check top identity
    assert ms.getMin() == -2  # min stays same after pushing large
    print('All Pass!')

test()
````

````python
class MinStack:
    def __init__(self):
        self.stack = []     # will store a tuple of a value and min so far

    def push(self, val):
        if not self.stack: 
            self.stack.append((val,val))
            return
        self.stack.append((val,min(val, self.stack[-1][1]))) 

    def pop(self):
        self.stack.pop()

    def top(self):
        return self.stack[-1][0]

    def get_min(self):
        return self.stack[-1][1]


s = MinStack()
s.push(-2); s.push(0); s.push(-3)
assert s.get_min() == -3   # given example — min after three pushes
s.pop()
assert s.top()     == 0    # given example — top after pop
assert s.get_min() == -2   # given example — min after pop removes -3

s2 = MinStack()
s2.push(5)
assert s2.get_min() == 5   # single element
assert s2.top()     == 5
s2.push(3)
assert s2.get_min() == 3   # new min pushed
s2.push(7)
assert s2.get_min() == 3   # larger push doesn't change min
s2.pop()
assert s2.get_min() == 3   # min unchanged after popping non-min
s2.pop()
assert s2.get_min() == 5   # min restored after popping 3
print('All Pass!')
````
