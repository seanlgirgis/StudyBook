## Min Stack

Design a stack that supports the following operations in **O(1)** time each:

- `push(val)` — push `val` onto the stack
- `pop()` — remove the top element
- `top()` — return the top element without removing it
- `get_min()` — return the **minimum** element in the stack

**Note:** `pop`, `top`, and `get_min` will always be called on a non-empty stack.

### Example

```
push(-2), push(0), push(-3)
get_min() → -3
pop()
top()    → 0
get_min() → -2
```

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- Implement the `MinStack` class below.

````python
class MinStack:
    def __init__(self):
        self.stack = []     # will store a tuple of a value and min so far

    def push(self, val):
        if not self.stack: 
            self.stack.append((val,val))
            return
        self.stack.append((val,min(val, self.stack[-1][1]))) 

    def pop(self):
        self.stack.pop()

    def top(self):
        return self.stack[-1][0]

    def get_min(self):
        return self.stack[-1][1]


s = MinStack()
s.push(-2); s.push(0); s.push(-3)
assert s.get_min() == -3   # given example — min after three pushes
s.pop()
assert s.top()     == 0    # given example — top after pop
assert s.get_min() == -2   # given example — min after pop removes -3

s2 = MinStack()
s2.push(5)
assert s2.get_min() == 5   # single element
assert s2.top()     == 5
s2.push(3)
assert s2.get_min() == 3   # new min pushed
s2.push(7)
assert s2.get_min() == 3   # larger push doesn't change min
s2.pop()
assert s2.get_min() == 3   # min unchanged after popping non-min
s2.pop()
assert s2.get_min() == 5   # min restored after popping 3
print('All Pass!')
````
