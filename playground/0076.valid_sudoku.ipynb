## Valid Sudoku

Determine if a `9 x 9` Sudoku board is valid. Only the **filled cells** need to be validated according to these rules:

1. Each row must contain the digits `1–9` with no repeats.
2. Each column must contain the digits `1–9` with no repeats.
3. Each of the nine `3 x 3` sub-boxes must contain the digits `1–9` with no repeats.

Empty cells are represented by `'.'`.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(1)** (the board is always 9×9) will fit within the execution time limit.

### Example

The following board is **valid**:
```
[["5","3",".",".","7",".",".",".","."],
 ["6",".",".","1","9","5",".",".","."],
 [".","9","8",".",".",".",".","6","."],
 ["8",".",".",".","6",".",".",".","3"],
 ["4",".",".","8",".","3",".",".","1"],
 ["7",".",".",".","2",".",".",".","6"],
 [".","6",".",".",".",".","2","8","."],
 [".",".",".","4","1","9",".",".","5"],
 [".",".",".",".","8",".",".","7","9"]]
```

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[input]** `array.array.string` `board` — always 9×9, cells are `'1'–'9'` or `'.'`
- **[output]** `boolean`

````python
def validlist(l):
    seen = set()
    for s in l:
        if s == ".":
            continue
        if s in seen:
            return False
        seen.add(s)
    return True
    
def validSquare(board,r,c):
    seen = set()
    for i in range(r, r+3):
        for j in range(c, c+3):
            s = board[i][j]
            if s == ".":
                continue
            if s in seen:
                return False
            seen.add(s)
    return True  
        
def solution(board):
    inner_vertics = [(0,0), (0,3), (0,6),(3,0), (3,3), (3,6),(6,0), (6,3), (6,6),]
    for r,c in inner_vertics:
        if not validSquare(board,r,c):
            return False
    for row in board:
        if not  validlist(row):
            return False

    for col in zip(*board):
        if not validlist(col):
            return False
    return True



valid = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
invalid_row = [
    ["5","3",".",".","7",".",".",".","5"],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
invalid_col = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    ["5",".",".",".","8",".",".","7","9"]
]
invalid_box = [
    ["5","3",".",".","7",".",".",".","."],
    ["6","5",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]

assert solution(valid)       == True
assert solution(invalid_row) == False
assert solution(invalid_col) == False
assert solution(invalid_box) == False
print('All Pass!')
````
