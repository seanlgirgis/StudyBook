## Coin Change

You are given an array of coin denominations `coins` and an integer `amount`. Return the **fewest number of coins** needed to make up that amount. If it is impossible, return `-1`.

You may use each coin denomination **any number of times**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(amount × coins.length)** will fit within the execution time limit.

### Example

- For `coins = [1,2,5]`, `amount = 11`, the output should be `solution(coins, amount) = 3`.

  **Explanation:** `5 + 5 + 1 = 11` — three coins.

- For `coins = [2]`, `amount = 3`, the output should be `solution(coins, amount) = -1`.

  **Explanation:** Cannot make `3` with only `2`s.

- For `coins = [1]`, `amount = 0`, the output should be `solution(coins, amount) = 0`.

  **Explanation:** Zero coins needed to make amount `0`.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `array.integer` `coins`  
  Guaranteed constraints: `1 ≤ coins.length ≤ 12`, `1 ≤ coins[i] ≤ 2^31 - 1`
- **[input]** `integer` `amount` — `0 ≤ amount ≤ 10^4`
- **[output]** `integer` — Fewest coins, or `-1` if impossible.

```python
%run _timer.py
```

```python
def solution(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for i in range(1, amount + 1):
        for c in coins:
            if i - c >= 0:
                dp[i] = min(dp[i], dp[i - c] + 1)
    return dp[-1] if dp[-1] != float('inf') else -1


assert solution([1,2,5], 11)   == 3   # given example 1 — 5+5+1
assert solution([2], 3)        == -1  # given example 2 — impossible
assert solution([1], 0)        == 0   # given example 3 — zero amount
assert solution([1], 1)        == 1   # single coin exact
assert solution([1], 5)        == 5   # only ones
assert solution([2,5,10], 6)   == 3   # 2+2+2
assert solution([1,5,10,25],100) == 4 # 25×4
assert solution([3,7], 5)      == -1  # impossible with these coins
print('All Pass!')
```
