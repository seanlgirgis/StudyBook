## Valid Parentheses

Given a string `s` containing only the characters `(`, `)`, `{`, `}`, `[`, `]`, return `True` if the string is valid.

A string is valid if:
1. Every open bracket is closed by the **same type** of bracket.
2. Open brackets are closed in the **correct order**.
3. Every close bracket has a **corresponding open bracket**.

**Note:** You are not expected to provide the most optimal solution, but a solution with time complexity not worse than **O(n)** will fit within the execution time limit.

### Example

- For `s = "()"`, the output should be `solution(s) = True`.
- For `s = "()[]{}"`, the output should be `solution(s) = True`.
- For `s = "(]"`, the output should be `solution(s) = False`.
- For `s = "([)]"`, the output should be `solution(s) = False`.

  **Explanation:** Brackets are interleaved — `[` closes before `(`, which is invalid.

### Input/Output

- **[execution time limit]** 4 seconds (py3)
- **[memory limit]** 1 GB
- **[input]** `string` `s` — A non-empty string of bracket characters only.  
  Guaranteed constraints: `1 ≤ s.length ≤ 10^4`
- **[output]** `boolean` — `True` if valid, `False` otherwise.

```python
def solution(s):
    lookup = { '}':'{',
                ']':'[',
               ')':'(', }
    stack = []
    for c in s:
        if c in lookup:    #it is a closing
            if not stack :
                return False
            if lookup[c] != stack[-1]:
                return False
            stack.pop()
        else:
            stack.append(c)     #    add an opening

    return not stack



assert solution('()')      == True   # given example 1
assert solution('()[]{}')  == True   # given example 2 — multiple types
assert solution('(]')      == False  # given example 3 — mismatched types
assert solution('([)]')    == False  # given example 4 — interleaved
assert solution('{[]}')    == True   # nested, correct order
assert solution('(')       == False  # unclosed open
assert solution(')')       == False  # close with nothing open
assert solution('((')      == False  # two unclosed
print('All Pass!')
```
