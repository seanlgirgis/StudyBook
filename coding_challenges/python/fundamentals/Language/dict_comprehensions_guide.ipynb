## 6. 🧩 Pattern 6: Dict Comprehensions — {k: v for ...} — LC 1, 49, 128, 217, 242, 347

---

```
PROBLEM:
  LC 1   — Two Sum: build val→index map in one line; check complement in O(1)
  LC 49  — Group Anagrams: char→count frequency map per word
  LC 128 — Longest Consecutive: val→index map for O(1) membership
  LC 217 — Contains Duplicate: frequency map; any count > 1 → duplicate
  LC 242 — Valid Anagram: char freq map; compare two dicts
  LC 347 — Top K Frequent: val→freq map, then heap

SYNTAX — four core patterns:
  {k: v for k, v in pairs}                  ← rebuild from pairs
  {v: k for k, v in d.items()}              ← invert a dict (unique values only)
  {ch: s.count(ch) for ch in set(s)}        ← freq map one-liner
  {val: i for i, val in enumerate(nums)}    ← index map one-liner

FREQ MAP ONE-LINER (vs Counter):
  freq = {ch: s.count(ch) for ch in set(s)}    ← O(n²) — each .count is O(n)
  freq = Counter(s)                             ← O(n) — one pass, cleaner
  Use dict comp when: transforming, filtering, or building non-freq maps.
  Use Counter when: pure frequency counting — faster and cleaner.

INVERT A DICT:
  d = {1: "a", 2: "b", 3: "c"}
  inv = {v: k for k, v in d.items()}
  step 1: v="a", k=1 → {"a": 1}
  step 2: v="b", k=2 → {"a":1, "b":2}
  step 3: v="c", k=3 → {"a":1, "b":2, "c":3}
  ⚠  duplicate values → last one wins, earlier keys silently dropped

INDEX MAP FROM enumerate (LC 1, 128):
  nums = [4, 7, 2, 9]
  index_map = {val: i for i, val in enumerate(nums)}
  step 1: i=0, val=4 → {4:0}
  step 2: i=1, val=7 → {4:0, 7:1}
  step 3: i=2, val=2 → {4:0, 7:1, 2:2}
  step 4: i=3, val=9 → {4:0, 7:1, 2:2, 9:3}
  O(1) lookup: is 7 in nums, and where? → index_map[7] = 1

KEY INSIGHT:
  Dict comp builds the lookup table. The real speed gain is using it later —
  turning every O(n) linear search into O(1) hash lookup.

TIME / SPACE:
  Time:  O(n) — one pass to build the dict
  Space: O(n) — at most n key-value pairs stored
```

````python
# Pattern 6: Dict Comprehensions
# Four core one-liners every interview expects you to reach for immediately.

# 1. freq map one-liner
s = "banana"
freq_comp = {ch: s.count(ch) for ch in set(s)}   # O(n^2) — .count is O(n) each
from collections import Counter
freq_counter = dict(Counter(s))                    # O(n) — one pass, cleaner
print(f"freq via comp   : {freq_comp}")
print(f"freq via Counter: {freq_counter}")

# Use dict comp for: transforming/filtering; Counter for pure freq.

# 2. invert a dict
d = {1: "a", 2: "b", 3: "c"}
inv = {v: k for k, v in d.items()}
# slow motion on d = {1:"a", 2:"b", 3:"c"}:
# step 1: k=1, v="a" → {"a":1}
# step 2: k=2, v="b" → {"a":1, "b":2}
# step 3: k=3, v="c" → {"a":1, "b":2, "c":3}
print(f"original  : {d}")
print(f"inverted  : {inv}")

# 3. index map from enumerate (LC 1, 128 pattern)
nums = [4, 7, 2, 9]
index_map = {val: i for i, val in enumerate(nums)}
# slow motion:
# i=0 val=4 → {4:0}
# i=1 val=7 → {4:0, 7:1}
# i=2 val=2 → {4:0, 7:1, 2:2}
# i=3 val=9 → {4:0, 7:1, 2:2, 9:3}
print(f"index map : {index_map}")
print(f"where is 7? → index {index_map[7]}")

# 4. char → index map (LC 242 / anagram pattern)
word = "hello"
char_index = {ch: i for i, ch in enumerate(word)}   # last position wins on dup
print(f"char→last-index: {char_index}")


def contains_duplicate(nums: list) -> bool:
    """
    LC 217 — Contains Duplicate
    Approach: freq map via dict comp; any count > 1 → duplicate.
    Args:
        nums (list[int]): integer array.
    Returns:
        bool: True if any value appears at least twice.
    Time:  O(n^2) — dict comp calls .count which is O(n) per unique value
    Space: O(n) — freq map stores at most n entries
    """
    freq = {val: nums.count(val) for val in set(nums)}  # freq map one-liner

    # slow motion on nums = [1, 2, 3, 1]:
    # set(nums) = {1, 2, 3}
    # val=1: count=2 → {1:2}
    # val=2: count=1 → {1:2, 2:1}
    # val=3: count=1 → {1:2, 2:1, 3:1}
    # any count > 1? → yes (1:2) → True

    return any(count > 1 for count in freq.values())   # one-liner check


def valid_anagram(s: str, t: str) -> bool:
    """
    LC 242 — Valid Anagram
    Approach: char freq maps for both strings via dict comp; compare dicts.
    Args:
        s (str): first string.
        t (str): second string.
    Returns:
        bool: True if t is an anagram of s.
    Time:  O(n) — two passes (one per string)
    Space: O(1) — at most 26 keys (lowercase letters)
    """
    freq_s = {ch: s.count(ch) for ch in set(s)}   # char → freq in s
    freq_t = {ch: t.count(ch) for ch in set(t)}   # char → freq in t

    # slow motion on s="anagram", t="nagaram":
    # freq_s = {a:3, n:1, g:1, r:1, m:1}
    # freq_t = {n:1, a:3, g:1, r:1, m:1}
    # freq_s == freq_t → True

    return freq_s == freq_t                        # dict equality: same keys + values


def test_harness(fn):
    tests = [
        ([1, 2, 3, 1],           True),    # duplicate: 1
        ([1, 2, 3, 4],           False),   # no duplicate
        ([1, 1, 1, 3, 3, 4, 3], True),    # multiple duplicates
        ([1],                    False),   # single element
    ]
    passed = 0
    for *inputs, expected in tests:
        got = fn(*inputs)
        status = "PASSED" if got == expected else "FAILED"
        if status == "FAILED":
            print(f"{status} | input={inputs} | expected={expected} | got={got}")
        passed += (got == expected)
    print(f"{passed}/{len(tests)} tests passed")


test_harness(contains_duplicate)

print(valid_anagram("anagram", "nagaram"))   # True
print(valid_anagram("rat", "car"))           # False

print("dict_comprehensions defined.")
````
