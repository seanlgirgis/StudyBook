# 🥞 Python `stack` — The Master Guide
### *From Zero to Interview-Ready*

---

> **Mental Model First:**  
> A `stack` is a **pile of plates**.  
> You can only add or remove from the **RIGHT end** (the top of the pile).  
> Last plate you put down is the first one you pick up — **LIFO**.  
> Last In, First Out.  
> Python has no dedicated stack class — you use a plain **list**, and only touch the right end.

## 1. 🏗️ What Is a Stack? The Visual Model

---

```
                        STACK OF PLATES

                    ┌──────────────────┐
                    │      'D'         │  ← RIGHT end (index -1)
                    ├──────────────────┤     append() pushes here
                    │      'C'         │     pop()    pulls from here
                    ├──────────────────┤
                    │      'B'         │
                    ├──────────────────┤
                    │      'A'         │  ← index 0  (bottom — don't touch)
                    └──────────────────┘

  stack = ['A', 'B', 'C', 'D']

  stack.append('E')  →  ['A', 'B', 'C', 'D', 'E']   push
  stack.pop()        →  ['A', 'B', 'C', 'D']         pop  returns 'E'
  stack[-1]          →  'D'                           peek — no removal

  Rule: ONLY touch the RIGHT end.
  Touching the left (insert(0) or pop(0)) makes it NOT a stack.
```

---

### LIFO — Why It Matters

```
  LIFO = Last In, First Out

  You pushed:  A → B → C → D
  You pop:     D → C → B → A   (reversed order)

  This reversal property is the ENTIRE reason stacks exist.
  Any problem that needs to process things in reverse order
  — or needs to remember "what came before" —
  is a stack problem.
```

## 2. 🔧 Creating a Stack

````python
# Python has no Stack class — just use a list and only touch the RIGHT end

# --- Empty stack ---
stack = []
print("Empty stack:", stack)

# --- Push items ---
stack.append(10)
stack.append(20)
stack.append(30)
print("After pushing 10, 20, 30:", stack)

# --- Peek (look at top without removing) ---
top = stack[-1]
print("Peek top (stack[-1]):", top)
print("Stack unchanged:", stack)

# --- Pop (remove from top) ---
val = stack.pop()
print(f"Popped: {val}  |  Stack now: {stack}")

# --- Check if empty ---
print("Is empty:", not stack)
print("Size:", len(stack))

# --- Pre-loaded stack ---
stack2 = list("ABC")   # ['A', 'B', 'C'] — C is on top
print("Pre-loaded:", stack2, " top is:", stack2[-1])
````

## 3. 🛠️ The Core API — Every Operation

---

```
OPERATION           COMPLEXITY   WHAT IT DOES
─────────────────────────────────────────────────────────────
stack.append(x)     O(1)*        Push x onto the top (RIGHT end)
stack.pop()         O(1)         Remove & return top element
stack[-1]           O(1)         Peek top — no removal
len(stack)          O(1)         Number of elements
not stack           O(1)         True if empty  (Pythonic empty check)
if stack:           O(1)         True if NOT empty
stack[0]            O(1)         Peek BOTTOM — rarely needed
stack[i]            O(1)         Random access — lists allow this
─────────────────────────────────────────────────────────────
* amortized

THINGS YOU DO NOT DO WITH A STACK:
  ❌  stack.insert(0, x)   — O(n), not a stack operation
  ❌  stack.pop(0)         — O(n), use deque.popleft() if you need this
  ❌  stack[3] = x         — random write, defeats the stack contract

If you need to access the middle — you probably want a list, not a stack.
```

````python
stack = []

# Push
for val in [5, 10, 15, 20]:
    stack.append(val)
    print(f"push({val:>2})  stack: {stack}")

print()

# Peek
print(f"peek     top = {stack[-1]}   stack unchanged: {stack}")
print()

# Pop
while stack:
    val = stack.pop()
    print(f"pop()  → {val:>2}   stack: {stack}")
````

## 4. 🗺️ LIFO — Your Decision Map

---

```
  SIGNAL IN THE PROBLEM              WHAT THE STACK DOES
  ─────────────────────────────────────────────────────────────
  Matching open/close pairs          Push opens, pop on close, verify match
  e.g. brackets, HTML tags

  "Next greater / smaller element"   Monotonic stack — push and hold
  for each position                  until a bigger/smaller triggers a pop

  Process items in reverse order     Push them all, then pop in order

  Undo / backtrack                   Push states, pop to go back

  Evaluate an expression             Push operands, pop on operator
  (RPN, calculator)

  DFS tree / graph traversal         Push neighbors, pop to visit
  (iterative, not recursive)

  "Running history" / previous        Push as you go, peek or pop
  state matters for current calc     when current affects past
  ─────────────────────────────────────────────────────────────

  THE PATTERN RECOGNITION TRICK:
  If you catch yourself saying "I need to remember what came before"
  while scanning left to right — that's a stack.
```

## 5. ⚡ Stack vs List vs Deque — When to Use What

---

```
  NEED                                USE
  ─────────────────────────────────────────────────────────────
  LIFO only (push/pop one end)        list  (stack pattern)
  FIFO queue (two ends)               deque
  Sliding window / recent-K           deque(maxlen=K)
  Random access + stack ops           list  (lists support both)
  Thread-safe stack                   queue.LifoQueue
  ─────────────────────────────────────────────────────────────

  OPERATION          LIST (stack)   DEQUE         NOTES
  ─────────────────────────────────────────────────────────────
  push  (right end)  O(1)*          O(1)           Tie
  pop   (right end)  O(1)           O(1)           Tie
  peek  (right end)  O(1)           O(1)           Tie
  push  (left end)   O(n) ❌        O(1) ✅        Deque wins
  pop   (left end)   O(n) ❌        O(1) ✅        Deque wins
  random access [i]  O(1) ✅        O(n) ❌        List wins
  ─────────────────────────────────────────────────────────────
  * amortized

  For stack problems: plain list is correct and idiomatic Python.
  No need to import anything.
```

## 6. 🔤 Pattern 1: Valid Parentheses / Bracket Matching

---

**LC #20** — the most classic stack problem.

```
  GOAL: Given a string of brackets, return True if every open
        bracket has a matching close bracket in the right order.

  KEY IDEA: Push every OPEN bracket.
            When you see a CLOSE bracket — peek the stack.
            If the top matches — pop it. If not — invalid.
            At the end — stack must be empty.

  s = "({[]})"

  char '('  → push     stack: ['(']
  char '{'  → push     stack: ['(', '{']
  char '['  → push     stack: ['(', '{', '[']
  char ']'  → top='['  match ✅  pop   stack: ['(', '{']
  char '}'  → top='{'  match ✅  pop   stack: ['(']
  char ')'  → top='('  match ✅  pop   stack: []
  End: stack empty → VALID ✅

  s = "([)]"
  char '('  → push     stack: ['(']
  char '['  → push     stack: ['(', '[']
  char ')'  → top='['  NO MATCH ❌  → INVALID immediately
```

````python
def isValid(s: str) -> bool:
    """
    LC #20. Valid Parentheses.
    Push opens. On close, check stack top matches.
    Stack empty at end = valid.
    Time:  O(n)
    Space: O(n)
    """
    stack = []
    match = {')': '(', '}': '{', ']': '['}   # close → expected open

    for ch in s:
        if ch in '([{':
            stack.append(ch)                 # push open bracket
        else:
            # it's a closing bracket
            if not stack or stack[-1] != match[ch]:
                return False                 # no match — invalid
            stack.pop()                      # matched — consume it

    return not stack                         # valid only if nothing left


def run_tests(fn):
    tests = [
        ("()",      True),
        ("()[]{}",  True),
        ("(]",      False),
        ("([)]",    False),
        ("{[]}",    True),
        ("",        True),
        ("[",       False),
        ("}{",      False),
    ]
    for s, expected in tests:
        got = fn(s)
        status = "✅" if got == expected else "❌"
        print(f"{status}  '{s}'  → {got}")

run_tests(isValid)
````

## 7. 🌡️ Pattern 2: Monotonic Stack — Next Greater Element

---

```
  GOAL: For each element, find the next element to its right
        that is strictly greater than it.

  KEY IDEA: The Waiting Room.
  Elements sit in the stack WAITING for something bigger to arrive.
  When a bigger element arrives — it resolves everyone smaller than it.
  Stack stays in DECREASING order (monotonic decreasing).

  nums = [2, 1, 2, 4, 3]

  i=0  val=2   stack empty → push 0          stack: [0]        (vals: [2])
  i=1  val=1   1 < 2 → push 1               stack: [0,1]       (vals: [2,1])
  i=2  val=2   2 > nums[1]=1 → pop 1        ans[1]=2
               2 = nums[0]=2 → stop         stack: [0,2]       (vals: [2,2])
  i=3  val=4   4 > nums[2]=2 → pop 2        ans[2]=4
               4 > nums[0]=2 → pop 0        ans[0]=4
               stack empty → push 3         stack: [3]         (vals: [4])
  i=4  val=3   3 < 4 → push 4               stack: [3,4]       (vals: [4,3])
  End: remaining in stack → no answer → -1
               ans[3] = -1,  ans[4] = -1

  Result: [4, 2, 4, -1, -1]

  STACK DISCIPLINE: decreasing values left to right.
  Pop when the new value BEATS the top.
  The new value IS the answer for everything it pops.
```

````python
from typing import List

def nextGreaterElement(nums: List[int]) -> List[int]:
    """
    For each element, return the next greater element to its right.
    -1 if none exists.
    Monotonic decreasing stack — stores indices.
    Time:  O(n) — each index enters and exits once
    Space: O(n)
    """
    n = len(nums)
    ans = [-1] * n        # default: no greater element found
    stack = []            # stores indices, decreasing by value

    for i, val in enumerate(nums):
        # pop everything the current value beats
        while stack and nums[stack[-1]] < val:
            idx = stack.pop()
            ans[idx] = val      # current val is the answer for idx
        stack.append(i)

    return ans


def run_tests(fn):
    tests = [
        ([2, 1, 2, 4, 3],     [4, 2, 4, -1, -1]),
        ([1, 2, 3, 4],        [2, 3, 4, -1]),
        ([4, 3, 2, 1],        [-1, -1, -1, -1]),
        ([1],                 [-1]),
        ([5, 4, 3, 2, 1, 6],  [6, 6, 6, 6, 6, -1]),
    ]
    for nums, expected in tests:
        got = fn(nums)
        status = "✅" if got == expected else "❌"
        print(f"{status}  {nums}")
        if got != expected:
            print(f"   Expected: {expected}")
            print(f"   Got:      {got}")

run_tests(nextGreaterElement)
````

## 8. 📊 Pattern 3: Largest Rectangle in Histogram

---

**LC #84** — hardest stack problem. High interview frequency.

```
  GOAL: Given bar heights, find the largest rectangle
        you can draw using one or more consecutive bars.

  heights = [2, 1, 5, 6, 2, 3]

  ┌─┐
  │ │  ┌─┬─┐
  │ │  │ │ │  ┌─┐
  │ │  │ │ │  │ │  ┌─┐
  │ ├─┤│ │ ├─┤ ├─┤ │ │
  └─┴─┴─┴─┴─┴─┴─┴─┴─┘
   2  1  5  6  2  3

  Largest rectangle = 10  (height 2, spanning bars 0-4)
  Or:                  10  (height 5, spanning bars 2-3)

  KEY IDEA: Monotonic INCREASING stack.
  A bar can extend LEFT as long as bars to its left are >= its height.
  When a shorter bar arrives — it cuts off taller bars behind it.
  At that moment: calculate area for each popped bar.

  Width = current_index - index_of_new_stack_top - 1
  Height = popped bar's height
  Area = height × width

  Trick: append 0 at the end to flush all remaining bars.
```

````python
from typing import List

def largestRectangleArea(heights: List[int]) -> int:
    """
    LC #84. Largest Rectangle in Histogram.
    Monotonic increasing stack stores indices.
    Pop when shorter bar arrives — calculate area at that moment.
    Sentinel 0 at end flushes the stack.
    Time:  O(n)
    Space: O(n)
    """
    heights = heights + [0]   # sentinel — forces full flush at end
    stack = []                # indices, increasing by height value
    max_area = 0

    for i, h in enumerate(heights):
        # shorter bar arrived — flush taller bars from stack
        while stack and heights[stack[-1]] > h:
            height = heights[stack.pop()]          # height of popped bar
            # width: from current i back to new stack top
            width  = i if not stack else i - stack[-1] - 1
            max_area = max(max_area, height * width)
        stack.append(i)

    return max_area


def run_tests(fn):
    tests = [
        ([2, 1, 5, 6, 2, 3],  10),
        ([2, 4],               4),
        ([1],                  1),
        ([0],                  0),
        ([3, 3, 3, 3],        12),
        ([1, 2, 3, 4, 5],     9),
    ]
    for heights, expected in tests:
        got = fn(heights)
        status = "✅" if got == expected else "❌"
        print(f"{status}  heights={heights}  → area={got}")
        if got != expected:
            print(f"   Expected: {expected}")

run_tests(largestRectangleArea)
````

### 🔍 Why the Width Formula Works

```
  When we pop index `j` because bar `i` is shorter:

  Everything between stack[-1] and i (exclusive) is >= heights[j]
  because those bars were already popped earlier (they were taller
  than j but shorter than something to their left).

  So bar j can legally extend from  stack[-1]+1  to  i-1  inclusive.

  width = i - stack[-1] - 1

  If stack is empty after pop — bar j can extend all the way to index 0.
  width = i
```

## 9. 🌡️ Pattern 4: Daily Temperatures

---

**LC #739** — the most common monotonic stack interview problem.

```
  GOAL: For each day, how many days until a warmer temperature?
        0 if no warmer day exists.

  temps = [73, 74, 75, 71, 69, 72, 76, 73]

  Same pattern as Next Greater Element — just measuring DISTANCE
  (index difference) instead of the value itself.

  i=0  73  push 0           stack: [0]
  i=1  74  74>73 pop 0      ans[0] = 1-0 = 1     stack: [1]
  i=2  75  75>74 pop 1      ans[1] = 2-1 = 1     stack: [2]
  i=3  71  71<75 push 3     stack: [2,3]
  i=4  69  69<71 push 4     stack: [2,3,4]
  i=5  72  72>69 pop 4      ans[4] = 5-4 = 1
           72>71 pop 3      ans[3] = 5-3 = 2
           72<75 stop       stack: [2,5]
  i=6  76  76>72 pop 5      ans[5] = 6-5 = 1
           76>75 pop 2      ans[2] = 6-2 = 4
           stack empty      stack: [6]
  i=7  73  73<76 push 7     stack: [6,7]
  End: remaining → 0        ans[6]=0, ans[7]=0

  Result: [1, 1, 4, 2, 1, 1, 0, 0]
```

````python
from typing import List

def dailyTemperatures(temperatures: List[int]) -> List[int]:
    """
    LC #739. Daily Temperatures.
    Monotonic decreasing stack stores indices.
    On warmer day: pop and record distance (i - popped_index).
    Time:  O(n)
    Space: O(n)
    """
    ans   = [0] * len(temperatures)
    stack = []   # indices, decreasing by temperature value

    for i, temp in enumerate(temperatures):
        while stack and temperatures[stack[-1]] < temp:
            idx = stack.pop()
            ans[idx] = i - idx        # days until warmer
        stack.append(i)

    return ans   # remaining in stack stay 0 — no warmer day


def run_tests(fn):
    tests = [
        ([73,74,75,71,69,72,76,73], [1,1,4,2,1,1,0,0]),
        ([30,40,50,60],             [1,1,1,0]),
        ([30,60,90],                [1,1,0]),
        ([90,80,70,60],             [0,0,0,0]),
        ([70],                      [0]),
    ]
    for temps, expected in tests:
        got = fn(temps)
        status = "✅" if got == expected else "❌"
        print(f"{status}  {temps}")
        if got != expected:
            print(f"   Expected: {expected}")
            print(f"   Got:      {got}")

run_tests(dailyTemperatures)
````

## 10. 📌 Pattern 5: Min Stack — O(1) Minimum Tracking

---

**LC #155**

```
  GOAL: Design a stack that supports push, pop, and getMin
        — all in O(1).

  NAIVE PROBLEM: After a pop, you'd have to scan the whole stack
                 to find the new min — that's O(n).

  KEY IDEA: Run a SECOND stack in parallel — the min_stack.
            It tracks the current minimum at every level.
            Both stacks are always the same size.

  Push 5:  main: [5]        min_stack: [5]       min=5
  Push 3:  main: [5,3]      min_stack: [5,3]     min=3
  Push 7:  main: [5,3,7]    min_stack: [5,3,3]   min=3  ← 7>3 so repeat 3
  Push 2:  main: [5,3,7,2]  min_stack: [5,3,3,2] min=2
  pop()    main: [5,3,7]    min_stack: [5,3,3]   min=3  ← 2 is gone, 3 returns
  pop()    main: [5,3]      min_stack: [5,3]     min=3
  pop()    main: [5]        min_stack: [5]       min=5

  getMin() always returns min_stack[-1] — O(1). Always.
```

````python
class MinStack:
    """
    LC #155. Min Stack.
    Two parallel stacks — main stack and min tracking stack.
    min_stack[-1] is always the current minimum.
    Time:  O(1) for all operations
    Space: O(n)
    """

    def __init__(self):
        self.stack     = []    # main stack
        self.min_stack = []    # tracks current min at every level

    def push(self, val: int) -> None:
        self.stack.append(val)
        # new min is the smaller of val and current min
        cur_min = val if not self.min_stack else min(val, self.min_stack[-1])
        self.min_stack.append(cur_min)

    def pop(self) -> None:
        self.stack.pop()
        self.min_stack.pop()   # both stacks always in sync

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]   # O(1) — always ready


# Test it
ms = MinStack()
for val in [5, 3, 7, 2, 8]:
    ms.push(val)
    print(f"push({val})  top={ms.top()}  min={ms.getMin()}  "
          f"main={ms.stack}  min_stack={ms.min_stack}")

print()
while ms.stack:
    print(f"pop()   removed={ms.top()}  ", end="")
    ms.pop()
    if ms.stack:
        print(f"new top={ms.top()}  new min={ms.getMin()}")
    else:
        print("stack empty")
````

## 11. 🧮 Pattern 6: Evaluate Reverse Polish Notation

---

**LC #150**

```
  GOAL: Evaluate an expression in Reverse Polish Notation.
        Operands come first, operator applies to last two.

  tokens = ["2", "1", "+", "3", "*"]
  Means:   (2 + 1) * 3 = 9

  KEY IDEA:
  - Number → push it
  - Operator → pop two numbers, apply operator, push result
  - Final answer is the one remaining value on the stack

  token  "2"  → push      stack: [2]
  token  "1"  → push      stack: [2, 1]
  token  "+"  → pop 1, pop 2  → 2+1=3  push 3    stack: [3]
  token  "3"  → push      stack: [3, 3]
  token  "*"  → pop 3, pop 3  → 3*3=9  push 9    stack: [9]
  End: return stack[0] = 9

  ⚠️  Division truncates TOWARD ZERO (not floor division).
      Use int(a / b) not a // b for negatives.
      int(-3 / 2) = -1   (truncate toward zero)
      -3 // 2    = -2    (floor — rounds toward negative infinity)
```

````python
from typing import List

def evalRPN(tokens: List[str]) -> int:
    """
    LC #150. Evaluate Reverse Polish Notation.
    Number → push. Operator → pop two, compute, push result.
    Division truncates toward zero: int(a/b).
    Time:  O(n)
    Space: O(n)
    """
    stack = []
    ops = {'+', '-', '*', '/'}

    for token in tokens:
        if token not in ops:
            stack.append(int(token))    # push number
        else:
            b = stack.pop()             # second operand (top)
            a = stack.pop()             # first operand
            if   token == '+': stack.append(a + b)
            elif token == '-': stack.append(a - b)
            elif token == '*': stack.append(a * b)
            elif token == '/': stack.append(int(a / b))  # truncate toward zero

    return stack[0]


def run_tests(fn):
    tests = [
        (["2","1","+","3","*"],              9),
        (["4","13","5","/","+"],             6),
        (["10","6","9","3","+","-11","*","/","*","17","+","5","+"], 22),
        (["3","4","-"],                      -1),
        (["6","-3","/"],                      -2),   # truncate toward zero
    ]
    for tokens, expected in tests:
        got = fn(tokens)
        status = "✅" if got == expected else "❌"
        print(f"{status}  {tokens}  → {got}")
        if got != expected:
            print(f"   Expected: {expected}")

run_tests(evalRPN)
````

## 12. 🚗 Pattern 7: Car Fleet

---

**LC #853** — the trickiest stack setup. The stack is conceptual.

```
  GOAL: Cars drive toward a target. Faster cars behind slower ones
        catch up and form a FLEET. Count the number of fleets.

  KEY IDEA:
  Sort cars by starting position (closest to target first).
  Calculate time each car takes to reach target.
  A car behind that arrives FASTER — catches up and joins the fleet ahead.
  A car behind that arrives SLOWER (or same time) — forms its own fleet.

  The stack holds ARRIVAL TIMES of fleet leaders.
  If the new car's time <= stack top → it joins that fleet (don't push).
  If the new car's time >  stack top → it forms a new fleet (push).

  target=12  position=[10,8,0,5,3]  speed=[2,4,1,1,3]

  Sort by position descending (closest to target first):
  pos:   [10,  8,  5,  3,  0]
  speed: [ 2,  4,  1,  3,  1]
  time:  [1.0, 1.0, 7.0, 3.0, 12.0]

  Process:
  time=1.0   stack empty → push          stack: [1.0]
  time=1.0   1.0 <= 1.0  → joins fleet   stack: [1.0]   (no push)
  time=7.0   7.0 > 1.0   → new fleet     stack: [1.0, 7.0]
  time=3.0   3.0 <= 7.0  → joins fleet   stack: [1.0, 7.0]
  time=12.0  12.0 > 7.0  → new fleet     stack: [1.0, 7.0, 12.0]

  Answer = len(stack) = 3 fleets
```

````python
from typing import List

def carFleet(target: int, position: List[int], speed: List[int]) -> int:
    """
    LC #853. Car Fleet.
    Sort by position descending. Compute arrival times.
    Stack holds fleet leader times.
    New time > stack top → new fleet. Otherwise joins existing.
    Time:  O(n log n) — sort dominates
    Space: O(n)
    """
    # pair and sort by position descending (closest to target first)
    cars = sorted(zip(position, speed), reverse=True)

    stack = []   # arrival times of fleet leaders

    for pos, spd in cars:
        time = (target - pos) / spd     # time this car takes alone
        if not stack or time > stack[-1]:
            stack.append(time)          # new fleet — this car leads it
        # else: time <= stack[-1] → catches up, joins fleet ahead, don't push

    return len(stack)


def run_tests(fn):
    tests = [
        (12, [10,8,0,5,3],  [2,4,1,1,3],  3),
        (10, [3],           [3],           1),
        (100,[0,2,4],       [2,1,3],       1),
        (10, [6,8],         [3,2],         2),
    ]
    for target, pos, spd, expected in tests:
        got = fn(target, pos, spd)
        status = "✅" if got == expected else "❌"
        print(f"{status}  target={target}  pos={pos}  spd={spd}  → {got} fleets")

run_tests(carFleet)
````

## 13. 🗺️ The Monotonic Stack Decision Map

---

```
  Most hard stack interview problems are monotonic stack problems.
  Here is how to identify which flavor you need:

  ─────────────────────────────────────────────────────────────
  QUESTION TYPE                    STACK ORDER       POP WHEN
  ─────────────────────────────────────────────────────────────
  Next GREATER element             Decreasing        new > top
  Next SMALLER element             Increasing        new < top
  Previous GREATER element         Decreasing        scan right-to-left
  Largest rectangle / max area     Increasing        new < top (shorter bar)
  Trapping rain water              Decreasing        new > top
  Stock span / online span         Decreasing        new >= top
  ─────────────────────────────────────────────────────────────

  THE TWO FLAVORS:

  MONOTONIC DECREASING (most common)
  Stack values go: big → small → ... (left to right)
  Waiting room for "next bigger thing"
  Pop when new value BEATS the top
  → Next Greater Element, Daily Temperatures, Stock Span

  MONOTONIC INCREASING
  Stack values go: small → big → ... (left to right)
  Waiting room for "next smaller thing"
  Pop when new value IS SHORTER THAN the top
  → Largest Rectangle, Trapping Rain Water (alternative approach)

  ─────────────────────────────────────────────────────────────
  IN BOTH CASES:
  - Stack stores INDICES (not values)
  - Each element enters once, exits once → O(n)
  - The `while` loop is not O(n²) — it's amortized O(1) per element
  ─────────────────────────────────────────────────────────────
```

## 14. 📋 Interview Cheat Sheet

---

### When to reach for a stack:

| Signal in the Problem | What to Do |
|---|---|
| Matching brackets / pairs | Push opens, pop on close |
| Next greater / next smaller | Monotonic stack (decreasing / increasing) |
| Largest area / max rectangle | Monotonic increasing stack |
| Evaluate expression (RPN) | Push numbers, pop on operator |
| O(1) min or max tracking | Parallel aux stack |
| DFS iterative | Push neighbors, pop to visit |
| Undo / backtrack | Push state, pop to reverse |
| "How far back" type questions | Stack of indices + distance calc |

---

### The O(1) operations — memorize these:

```python
stack = []
stack.append(x)    # push     O(1)
stack.pop()        # pop      O(1)
stack[-1]          # peek     O(1)
not stack          # empty?   O(1)
len(stack)         # size     O(1)
```

### Monotonic stack templates:

```python
# NEXT GREATER ELEMENT (monotonic decreasing)
ans = [-1] * len(nums)
stack = []   # indices
for i, val in enumerate(nums):
    while stack and nums[stack[-1]] < val:
        ans[stack.pop()] = val
    stack.append(i)

# LARGEST RECTANGLE (monotonic increasing)
heights.append(0)   # sentinel flush
stack = []
for i, h in enumerate(heights):
    while stack and heights[stack[-1]] > h:
        height = heights[stack.pop()]
        width  = i if not stack else i - stack[-1] - 1
        max_area = max(max_area, height * width)
    stack.append(i)

# BRACKET MATCHING
stack = []
match = {')':'(', '}':'{', ']':'['}
for ch in s:
    if ch in '([{': stack.append(ch)
    elif not stack or stack[-1] != match[ch]: return False
    else: stack.pop()
return not stack

# MIN STACK
stack, min_stack = [], []
# push:
stack.append(val)
min_stack.append(min(val, min_stack[-1]) if min_stack else val)
# pop:
stack.pop(); min_stack.pop()
# getMin:
return min_stack[-1]
```

### Gotchas to not forget:

```
❌  Pop from empty stack → IndexError. Always check: if stack:
❌  RPN division: use int(a/b) not a//b for negative numbers
❌  Monotonic stack: the while loop is O(n) total, NOT O(n) per step
✅  Histogram: append 0 sentinel to force flush of remaining bars
✅  Car Fleet: sort descending by position before processing
✅  Stack stores INDICES — lets you calculate distances and widths
✅  Parallel min_stack: both stacks always same size, always in sync
```
