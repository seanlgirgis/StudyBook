# Graph Programming — 01: Fundamentals

A **graph** is a collection of **nodes** (vertices) connected by **edges**.  
Almost every real-world network problem maps to a graph: roads, social networks, dependencies, pipelines.

---

## Key Vocabulary

| Term | Meaning |
|---|---|
| Node / Vertex | A point in the graph |
| Edge | A connection between two nodes |
| Directed | Edges have direction (A → B, not B → A) |
| Undirected | Edges go both ways (A — B) |
| Weighted | Edges carry a cost/distance |
| Cycle | A path that starts and ends at the same node |
| Connected | Every node is reachable from every other node |
| DAG | Directed Acyclic Graph — directed, no cycles |

---

## Two Ways to Represent a Graph

### 1. Adjacency List  ← Use this 95% of the time
- Dictionary: `node → [list of neighbors]`
- Space: O(V + E)
- Fast neighbor lookup

### 2. Adjacency Matrix
- 2D grid: `matrix[i][j] = 1` means edge from i to j
- Space: O(V²)
- Good when you need to check "does edge (i,j) exist?" in O(1)

````python
# === ADJACENCY LIST ===
# Undirected graph:
#   0 -- 1 -- 2
#   |         |
#   3 ------- 4

graph_undirected = {
    0: [1, 3],
    1: [0, 2],
    2: [1, 4],
    3: [0, 4],
    4: [2, 3]
}

# Directed graph (course prerequisites):
#   0 → 1 → 3
#   0 → 2 → 3

graph_directed = {
    0: [1, 2],
    1: [3],
    2: [3],
    3: []
}

print("Neighbors of node 0 (undirected):", graph_undirected[0])
print("Neighbors of node 0 (directed):", graph_directed[0])
````

````python
# === ADJACENCY MATRIX ===
# Same undirected graph above

n = 5
matrix = [[0] * n for _ in range(n)]

edges = [(0,1),(1,2),(2,4),(3,4),(0,3)]
for u, v in edges:
    matrix[u][v] = 1
    matrix[v][u] = 1  # undirected: both directions

print("Adjacency Matrix:")
for row in matrix:
    print(row)

print("\nIs there an edge 0→1?", bool(matrix[0][1]))
print("Is there an edge 0→2?", bool(matrix[0][2]))
````

````python
# === BUILD GRAPH FROM EDGE LIST (common in LeetCode) ===
from collections import defaultdict

def build_graph(n, edges, directed=False):
    """Build adjacency list from edge list."""
    graph = defaultdict(list)
    for u, v in edges:
        graph[u].append(v)
        if not directed:
            graph[v].append(u)
    return graph

# LC-style: numCourses=4, prerequisites=[[1,0],[2,0],[3,1],[3,2]]
# means: to take 1, you need 0 first → edge 0→1
prereqs = [[1,0],[2,0],[3,1],[3,2]]
g = build_graph(4, [(v, u) for u, v in prereqs], directed=True)
print(dict(g))
# 0→[1,2], 1→[3], 2→[3]
````

## Grid as a Graph

Many LeetCode problems give you a **2D grid** (matrix) instead of an explicit edge list.  
Each cell `(r, c)` is a node. Its neighbors are the 4 adjacent cells (up/down/left/right).

```
NEIGHBORS of (r, c):
  (r-1, c)  ← up
  (r+1, c)  ← down
  (r, c-1)  ← left
  (r, c+1)  ← right
```

This is the core pattern for **Islands** problems.

````python
# === GRID NEIGHBOR PATTERN (memorize this) ===
DIRECTIONS = [(0,1),(0,-1),(1,0),(-1,0)]  # right, left, down, up

def get_neighbors(grid, r, c):
    rows, cols = len(grid), len(grid[0])
    neighbors = []
    for dr, dc in DIRECTIONS:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols:  # in-bounds check
            neighbors.append((nr, nc))
    return neighbors

grid = [
    [1,1,0],
    [0,1,0],
    [1,0,1]
]

print("Neighbors of (1,1):", get_neighbors(grid, 1, 1))
print("Neighbors of (0,0):", get_neighbors(grid, 0, 0))  # corner — only 2 neighbors
````

## Summary

| Representation | When to use |
|---|---|
| Adjacency list | Sparse graphs, most LeetCode problems |
| Adjacency matrix | Dense graphs, O(1) edge existence check |
| Grid (implicit) | 2D grid problems (Islands, BFS shortest path) |
