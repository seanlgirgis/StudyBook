## Official Problem Statement

Given a **1-indexed** array of integers `numbers` that is already
sorted in non-decreasing order, find two numbers such that they
add up to a specific `target` number.

Return the indices of the two numbers, `[index1, index2]`,
added by one as integers: `index1 < index2`.

The tests are generated such that there is **exactly one** solution.
You **may not** use the same element twice.
Your solution must use only constant extra space.

**Constraints:**
- `2 <= numbers.length <= 3 * 10^4`
- `-1000 <= numbers[i] <= 1000`
- `numbers` is sorted in non-decreasing order.
- `-1000 <= target <= 1000`
- The tests are generated with exactly one solution.

## What This Is Actually Asking

Find two distinct positions whose values sum to `target` in a
sorted array and return their 1-based indices.
A hash map would work but violates the O(1) space constraint.
The sorted property means if the current sum is too small, we
must increase it — move the left pointer right.
If the sum is too large, decrease it — move the right pointer left.
Exactly one solution is guaranteed, so we always find it.

## Walk Through an Example by Hand

```
numbers = [2, 7, 11, 15], target = 9

left=0, right=3  → sum = 2+15 = 17 > 9  → right--
left=0, right=2  → sum = 2+11 = 13 > 9  → right--
left=0, right=1  → sum = 2+7  =  9 == 9 → return [1, 2]

numbers = [2, 3, 4], target = 6

left=0, right=2  → sum = 2+4 = 6 == 6   → return [1, 3]
```

## The Picture

```
 index:  0    1    2    3
 value: [2,   7,  11,  15]   target = 9
         ▲                ▲
        left            right

  sum=2+15=17  too big  → move right ←

 index:  0    1    2    3
 value: [2,   7,  11,  15]
         ▲         ▲
        left     right

  sum=2+11=13  too big  → move right ←

 index:  0    1    2    3
 value: [2,   7,  11,  15]
         ▲    ▲
        left right

  sum=2+7=9  MATCH  → return [left+1, right+1] = [1, 2]
```

## When To Use This Pattern

- When the array is **sorted** and you need a pair summing to a
  target, think **two pointers from both ends**.
- When the problem says O(1) space (no hash map), think two
  pointers over a sorted input.
- When you need to **squeeze** a window or range inward based on
  a comparison, think converging two pointers.
- When a brute-force O(n²) loop feels obvious, check if sorting
  first + two pointers gives O(n log n) total.
- Container With Most Water (LC 11) uses the exact same pointer
  logic — move the shorter side inward.

## The Approach

Place one pointer at the leftmost index and one at the rightmost.
Compute their sum: if it equals the target we are done.
If the sum is less than the target, the only way to increase it
is to advance the left pointer to a larger value.
If the sum is greater, retreat the right pointer to a smaller value.
The pointers converge and since a unique solution exists we
always find it before they cross.

````python
from typing import List
````

````python
def test_harness(func):
    cases = [
        ([2, 7, 11, 15], 9,  [1, 2]),
        ([2, 3, 4],       6,  [1, 3]),
        ([-1, 0],         -1, [1, 2]),
        ([1, 2, 3, 4, 4, 9, 56, 90], 8, [4, 5]),
        ([5, 25, 75],    100, [2, 3]),
    ]
    passed = 0
    for numbers, target, expected in cases:
        result = func(numbers[:], target)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        print(
            f"{status} | numbers={numbers}, target={target}"
            f" | expected={expected}, got={result}"
        )
    print(f"\n{passed}/{len(cases)} tests passed.")
````

````python
def twoSum(
    nums: List[int], target: int
) -> List[int]:
    l, r = 0, len(nums)-1
    while l <= r:
        s = nums[l] + nums[r]
        if s == target:
            return [l+1, r+1]
        if s > target:
            r-=1
        else:
            l+=1
    return []

print(twoSum([2, 7, 11, 15], 9))   # [1, 2]
print(twoSum([2, 3, 4], 6))        # [1, 3]
print(twoSum([-1, 0], -1))         # [1, 2]
print(twoSum([0, 0], 0))           # [1, 2]
test_harness(twoSum)       
````

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute force (nested loops) | O(n²) | O(1) | Check every pair |
| Hash map (unsorted) | O(n) | O(n) | Can't use: O(1) required |
| Two pointers (optimal) | O(n) | O(1) | Uses sorted property |
