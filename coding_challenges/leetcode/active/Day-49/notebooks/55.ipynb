## Official Problem Statement

You are given an integer array `nums`. You are initially positioned
at the first index of the array. Each element `nums[i]` represents
your maximum jump length at that position. Return `true` if you
can reach the last index, or `false` otherwise.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 10^5`

## What This Is Actually Asking

At each position you have a "fuel" value — the max steps you can
jump forward. You want to know if the fuel is ever enough to reach
the end. You don't need the exact path, just reachability. A zero
in the array is only a trap if you're forced to land on it. The
greedy trick: always track the furthest index reachable so far.

## When To Use This Pattern

- When each element gives a **local capability** (steps, fuel,
  reach), think greedy max-reach tracking.
- When the question is **can you reach X** (not how), a single
  forward pass maintaining a frontier is enough.
- When you see zeros as potential blockers, think about whether
  they can be bypassed by prior momentum.
- When backtracking feels expensive, check if a greedy "keep
  the best option" forward scan works.
- When array values represent ranges or intervals, think greedy
  interval-cover / reach extension.

## The Approach

Maintain `max_reach`, the furthest index reachable so far. Scan
left to right: if the current index exceeds `max_reach`, you're
stranded — return False. Otherwise update `max_reach` greedily
to the max of itself and `i + nums[i]`. If you finish the loop,
every index was reachable, so return True.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute Force (backtrack) | O(2^n) | O(n) | Try every path |
| DP (bottom-up) | O(n^2) | O(n) | Mark each reachable |
| **Greedy (optimal)** | **O(n)** | **O(1)** | Single pass |

The greedy approach is strictly optimal — one pass, no extra space.
