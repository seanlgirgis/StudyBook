## Official Problem Statement

You are given a 0-indexed array of integers `nums` of length `n`.
You are initially positioned at `nums[0]`. Each element `nums[i]`
represents the maximum length of a forward jump from index `i`.

Return the **minimum number of jumps** to reach `nums[n-1]`.
The test cases are generated such that you can always reach
`nums[n-1]`.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 1000`
- The answer is guaranteed to exist.

## What This Is Actually Asking

This is Jump Game I's harder sibling — instead of asking if
you can reach the end, it asks for the fewest jumps to get
there. You must count each time you are forced to jump. The
greedy insight: process positions in "BFS levels" where each
level is one jump's worth of reach. Within a level, always
extend to the farthest possible next position.

## Walk Through an Example by Hand

```
nums = [2, 3, 1, 1, 4]
idx:    0  1  2  3  4

jumps=0, cur_end=0, farthest=0

i=0: farthest = max(0, 0+2) = 2
     i==cur_end(0) => jumps=1, cur_end=2

i=1: farthest = max(2, 1+3) = 4
i=2: farthest = max(4, 2+1) = 4
     i==cur_end(2) => jumps=2, cur_end=4

i=3: farthest = max(4, 3+1) = 4
     (i=3 != cur_end=4, no jump yet)

Loop ends (we stop at n-2=3, no need to jump from last)
Return jumps = 2
```

## The Picture

```
nums = [2, 3, 1, 1, 4]
idx:    0  1  2  3  4

Level 0 (start):   [0]
  Jump 1 reaches:  [1][2]        (from 0, max=2)

Level 1 (window 1..2):
  from idx 1: 1+3=4
  from idx 2: 2+1=3
  farthest = 4
  Jump 2 reaches:  [3][4]        (cur_end moves to 4)

Level 2 (window 3..4): includes last index => done!

Jumps = 2

Timeline:
  [0] --jump1--> [1 or 2] --jump2--> [4]
   |                 |
  reach=2         reach=4
```

## When To Use This Pattern

- When asked for the **minimum number of steps/hops**, think
  greedy BFS-level simulation.
- When each element defines a **range of reachable positions**,
  think expanding windows.
- When you must count "rounds" or "batches" of actions, the
  cur_end boundary trick counts them cheaply.
- When a BFS would be too memory-heavy (O(n) queue), check if
  the greedy level approach gives O(1) space instead.
- When the problem guarantees a solution exists, skip the
  infeasibility check and focus on minimising count.

## The Approach

Use three variables: `jumps`, `cur_end` (end of current jump
level), and `farthest` (best reach seen in this level). Scan
indices 0 to n-2 (no jump needed from the last index). At each
step, update `farthest`. When `i == cur_end`, we must jump —
increment `jumps` and set `cur_end = farthest`. Return `jumps`.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute Force (recursion) | O(n^n) | O(n) | All paths |
| DP (bottom-up) | O(n^2) | O(n) | dp[i] = min jumps |
| BFS (queue) | O(n) | O(n) | Level-by-level |
| **Greedy (optimal)** | **O(n)** | **O(1)** | No queue needed |

Greedy achieves BFS optimality without the O(n) queue overhead.
