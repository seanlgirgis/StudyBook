## Official Problem Statement

You are given an integer array `nums`. You are initially positioned at
the **first index** of the array. Each element in the array represents
your **maximum jump length** at that position.

Return `true` if you can reach the last index, or `false` otherwise.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 10^5`

## What This Is Actually Asking

You start at position 0. Each number tells you the *maximum* steps
forward you can take from that spot. You want to know if there is
any sequence of hops that lands you at the last position.

You do not need to find the actual path — just a yes/no answer.
A zero blocks progress only if you are forced to land on it with
no way around it. The greedy trick is to track the furthest reachable
index at every step.

## Walk Through an Example by Hand

`nums = [2, 3, 1, 1, 4]`

```
i=0: nums[0]=2 → max_reach = max(0, 0+2) = 2
i=1: 1 <= 2 OK → max_reach = max(2, 1+3) = 4
i=2: 2 <= 4 OK → max_reach = max(4, 2+1) = 4
i=3: 3 <= 4 OK → max_reach = max(4, 3+1) = 4
i=4: 4 <= 4 OK → max_reach = max(4, 4+4) = 8
Loop ends — last index (4) was always reachable → True
```

`nums = [3, 2, 1, 0, 4]`

```
i=0: max_reach = 3
i=1: max_reach = max(3, 3) = 3
i=2: max_reach = max(3, 3) = 3
i=3: max_reach = max(3, 3) = 3
i=4: 4 > 3 → return False
```

## The Picture

```
nums = [2, 3, 1, 1, 4]
idx     0  1  2  3  4

From 0 (jump=2):  ──────►
                  0──1──2

From 1 (jump=3):     ──────────►
                     1──2──3──4   ← reaches end!

max_reach timeline:
  i=0 │ reach=2  [====]
  i=1 │ reach=4  [========]
  i=2 │ reach=4  [========]
  i=3 │ reach=4  [========]
       0    1    2    3    4
                          ^ last index always inside reach window

Blocked example: nums = [3, 2, 1, 0, 4]
  i=3 │ reach=3  [======]
  i=4 │ 4 > reach=3  ← STUCK
       0    1    2    3  | 4
                         wall
```

## When To Use This Pattern

- When you see **"can you reach position X from 0"**, think
  *greedy max-reach sweep*.
- When each element gives a **maximum range** (not exact), think
  *track furthest reachable, not exact path*.
- When a **zero** might block progress, think *check if index
  exceeds current reach*.
- When asked for **yes/no reachability** (not min jumps), think
  *this simpler O(n) greedy, not BFS/DP*.
- When you see the follow-up LC 45 (min jumps), think
  *extend this pattern with a level counter*.

## The Approach

Keep a single variable `max_reach` initialized to 0 representing the
farthest index reachable so far. Scan every index from left to right.
If the current index `i` is greater than `max_reach`, no sequence of
jumps can get us here, so return False immediately. Otherwise, update
`max_reach` to be the maximum of itself and `i + nums[i]`. After the
loop, return True — we never got stuck.

## Real World Connection

At Citi, deployment pipelines are staged — each stage must be reachable
from the previous without a gap in environment availability. The jump
game models whether a release can propagate from dev → test → staging
→ prod given variable window sizes per environment.

On AWS, Step Functions orchestrate workflows where each Lambda's
timeout defines how far downstream it can trigger. If a Lambda times
out (jump=0), the chain breaks — exactly the blocked-zero scenario.

In data engineering, partitioned ETL jobs must hand off to the next
partition window. A failed partition (no records = jump 0) can stall
the entire daily load if no retry range covers the gap.
