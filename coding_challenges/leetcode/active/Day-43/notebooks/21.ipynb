## Official Problem Statement

You are given the heads of two sorted linked lists `list1`
and `list2`. Merge the two lists into one **sorted** list.
The list should be made by splicing together the nodes of
the first two lists. Return the head of the merged linked list.

**Constraints:**
- The number of nodes in both lists is in the range `[0, 50]`.
- `-100 <= Node.val <= 100`
- Both `list1` and `list2` are sorted in **non-decreasing** order.

## What This Is Actually Asking

Two already-sorted chains exist in memory — do not copy values
into a new array and re-sort. Instead, re-wire the existing
nodes by adjusting their `next` pointers to interleave them
in sorted order. Think of it as merging two sorted stacks of
cards by always picking the smaller top card.

## Walk Through an Example by Hand

```
list1: 1 -> 2 -> 4
list2: 1 -> 3 -> 4
```

Create `dummy -> None`, `tail = dummy`.

**Step 1:** `list1.val=1 == list2.val=1`, pick `list1`.
  `tail.next = list1(1)`, advance `list1` to 2, `tail` to 1.

**Step 2:** `list1.val=2 > list2.val=1`, pick `list2`.
  `tail.next = list2(1)`, advance `list2` to 3, `tail` to 1.

**Step 3:** `list1.val=2 < list2.val=3`, pick `list1`.
  `tail.next = list1(2)`, advance `list1` to 4, `tail` to 2.

**Step 4:** `list1.val=4 > list2.val=3`, pick `list2`.
  `tail.next = list2(3)`, advance `list2` to 4, `tail` to 3.

**Step 5:** `list1.val=4 == list2.val=4`, pick `list1`.
  `tail.next = list1(4)`, advance `list1` to None, `tail` to 4.

**Step 6:** `list1=None`, exit loop. `tail.next = list2(4)`.

Result: `1 -> 1 -> 2 -> 3 -> 4 -> 4`

## The Picture

```
list1:  [1] --> [2] --> [4] --> None
list2:  [1] --> [3] --> [4] --> None

```

## When To Use This Pattern

- When you need to **merge two sorted sequences**, think
  **dummy head + tail pointer**.
- When handling linked list problems where the **head node is
  uncertain**, think **dummy sentinel to avoid edge cases**.
- When a problem asks you to **combine sorted data** from
  multiple sources, think **merge sort's merge step**.
- When you need to **merge K sorted lists**, think **extend
  this exact two-pointer merge with a heap**.
- When **one input could be empty**, think **attach the rest
  directly after the loop**.

## The Approach

Create a `dummy` node and a `tail` pointer pointing to it.
While both lists have nodes, compare the current heads and
attach the smaller one to `tail.next`, then advance that list
and `tail`. Once either list is exhausted, attach the remaining
nodes of the other list directly to `tail.next`. Return
`dummy.next` as the head of the merged list.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list(vals):
    dummy = ListNode(0)
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


def make_list_with_cycle(vals, pos):
    # pos=-1 means no cycle
    nodes = [ListNode(v) for v in vals]
    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]
    if pos >= 0:
        nodes[-1].next = nodes[pos]
    return nodes[0] if nodes else None
````
