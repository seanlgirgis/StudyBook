## Official Problem Statement

Given the `head` of a singly linked list, reverse the list, and
return the reversed list.

**Constraints:**
- The number of nodes in the list is in the range `[0, 5000]`.
- `-5000 <= Node.val <= 5000`

**Follow up:** A linked list can be reversed either iteratively
or recursively. Could you implement both?

## What This Is Actually Asking

You have a chain of nodes where each node points to the next one.
Your job is to make every node point to the previous one instead.
The old tail becomes the new head, and the old head must point
to `None` (it is now the tail).
You must do this in-place — no converting to an array and back.

## Walk Through an Example by Hand

Input: `1 -> 2 -> 3 -> 4 -> 5 -> None`

Start: `prev=None`, `curr=1`

**Step 1:** `nxt=2`, flip `1.next=None`, `prev=1`, `curr=2`
  List so far: `None <- 1   2 -> 3 -> 4 -> 5`

**Step 2:** `nxt=3`, flip `2.next=1`, `prev=2`, `curr=3`
  List so far: `None <- 1 <- 2   3 -> 4 -> 5`

**Step 3:** `nxt=4`, flip `3.next=2`, `prev=3`, `curr=4`

**Step 4:** `nxt=5`, flip `4.next=3`, `prev=4`, `curr=5`

**Step 5:** `nxt=None`, flip `5.next=4`, `prev=5`, `curr=None`

Loop ends. Return `prev` (node 5) as the new head.

Output: `5 -> 4 -> 3 -> 2 -> 1 -> None`

## The Picture

```
BEFORE:

  head
   |
  [1] --> [2] --> [3] --> [4] --> [5] --> None

DURING (after step 2):

  prev    curr
   |       |
None <-[1]<-[2]  [3] --> [4] --> [5] --> None
                  |
                 nxt

AFTER:

                                       prev
                                        |
None <--[1] <--[2] <--[3] <--[4] <--[5]
                                        |
                                     new head

KEY: Save nxt BEFORE flipping, or you lose the rest of the list!
```

## When To Use This Pattern

- When you need to **reverse a linked list** or a portion of one,
  think **prev/curr pointer swap**.
- When a problem says **"in-place"** on a linked list, think
  **pointer reassignment, not array conversion**.
- When you see **palindrome check on a linked list**, think
  **reverse the second half**.
- When you need to **rotate or reorder** a linked list, think
  **find midpoint + reverse**.
- When the follow-up asks for recursion, think **reverse rest
  first, then attach current node at the tail**.

## The Approach

Walk the list with two pointers: `prev` (starts at `None`) and
`curr` (starts at `head`). At each step, save `curr.next` before
overwriting it, flip `curr.next` to `prev`, then advance both
pointers one step forward. When `curr` reaches `None`, every
pointer has been flipped and `prev` holds the new head.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list(vals):
    dummy = ListNode(0)
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result


def make_list_with_cycle(vals, pos):
    # pos=-1 means no cycle
    nodes = [ListNode(v) for v in vals]
    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]
    if pos >= 0:
        nodes[-1].next = nodes[pos]
    return nodes[0] if nodes else None
````
