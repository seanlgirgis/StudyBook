## Official Problem Statement

Given an array of intervals `intervals` where `intervals[i] = [start, end]`,
return the **minimum number of intervals you need to remove** to make the
rest non-overlapping.

- `1 <= intervals.length <= 10^5`
- `-5 * 10^4 <= start_i < end_i <= 5 * 10^4`

**Examples:**
```
Input:  [[1,2],[2,3],[3,4],[1,3]]
Output: 1   # remove [1,3]

Input:  [[1,2],[1,2],[1,2]]
Output: 2   # keep one, remove two

Input:  [[1,2],[2,3]]
Output: 0   # already non-overlapping
```

## What This Is Actually Asking

You want to **keep as many non-overlapping intervals as possible**.
The answer is then `total - kept`.

Two intervals overlap when the next one **starts before** the previous
one ends (strictly: `next_start < prev_end`).
Touching at a single point (`start == end`) is **not** an overlap.

The greedy key insight: always prefer the interval that **ends earliest**
— it leaves the most room for future intervals.

## Walk Through an Example by Hand

```
intervals = [[1,2],[2,3],[3,4],[1,3]]

Step 1 — sort by end:
  [[1,2], [2,3], [1,3], [3,4]]

Step 2 — greedy scan (prev_end starts at -inf):
  [1,2]: start(1) >= prev_end(-inf) => KEEP  prev_end=2  removed=0
  [2,3]: start(2) >= prev_end(2)    => KEEP  prev_end=3  removed=0
  [1,3]: start(1) <  prev_end(3)    => REMOVE            removed=1
  [3,4]: start(3) >= prev_end(3)    => KEEP  prev_end=4  removed=1

Answer: 1
```

## The Picture

```
Timeline (sorted by end):

  [1,2]   |--|
  [2,3]      |--|
  [1,3]   |-----| <-- overlaps with [2,3]; REMOVE
  [3,4]         |--|

Axis:  1  2  3  4

Kept (non-overlapping):  [1,2]  [2,3]  [3,4]
Removed: 1

Why sort by END?
  Ending earliest => fewest future conflicts.
  A long interval that ends late blocks more of the future.
```

## The Approach

1. **Sort** intervals by their **end** value (ascending).
2. Track `prev_end = float('-inf')` and `removed = 0`.
3. For each interval `[s, e]`:
   - If `s >= prev_end` → **keep it**, update `prev_end = e`.
   - Else → **remove it**, increment `removed`.
4. Return `removed`.

**Why this is optimal:** By always keeping the interval that ends
soonest, we maximise the remaining timeline for future intervals
(classic Activity Selection / Exchange Argument proof).
