## Official Problem Statement

Given the `root` of a binary tree, invert the tree,
and return its root.

**Example 1:**
```
Input:          Output:
    4               4
   / \             / \
  2   7           7   2
 / \ / \         / \ / \
1  3 6  9       9  6 3  1
```
```
Input:  root = [4,2,7,1,3,6,9]
Output: [4,7,2,9,6,3,1]
```
**Example 2:**
```
Input:  root = [2,1,3]
Output: [2,3,1]
```

**Constraints:**
- `0 <= number of nodes <= 100`
- `-100 <= Node.val <= 100`

## What This Is Actually Asking

Mirror the tree left-to-right — like flipping it in
front of a mirror.
Every node's left child becomes its right child and
vice versa.
This must happen at every level, all the way down.

## Walk Through an Example by Hand

```
Input:
    4
   / \
  2   7
 / \ / \
1  3 6  9

invertTree(4):
  swap: left=7, right=2   (node 4's children swapped)
  invertTree(7):           (recurse into new left)
    swap: left=9, right=6
    invertTree(9): None + None -> return
    invertTree(6): None + None -> return
  invertTree(2):           (recurse into new right)
    swap: left=3, right=1
    invertTree(3): return
    invertTree(1): return

Result:
    4
   / \
  7   2
 / \ / \
9  6 3  1
```

## The Picture

```
Think of holding the tree upside-down by the root
and shaking it — every left/right pair swaps.

Before:         After:
    4               4
   / \             / \
  2   7    ->     7   2
 /\  /\          /\  /\
1 3  6 9        9 6  3 1

At each node:
  node.left, node.right = node.right, node.left
  then recurse into both children

None -> return None  (base case, stops recursion)
```

## When To Use This Pattern

- When you see **"mirror", "invert", "flip"** on a tree,
  think **swap children at every node via recursion**
- When the same operation applies to every node,
  think **recursion — write it once, it runs everywhere**
- When the base case is a missing node, think
  **`if not root: return None`**
- When you need an iterative version, think
  **BFS queue: pop node, swap, push children**

## The Approach

If the node is None, return None.
Swap its left and right children in one Python line.
Recurse into the (now swapped) left child, then the
right child.
Return the node — the root comes back with the whole
tree inverted beneath it.

````python
from collections import deque  # for build_tree and BFS alternative
from typing import Optional
````

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(vals: list) -> Optional[TreeNode]:
    """Build tree from level-order list."""
    if not vals or vals[0] is None:
        return None
    root = TreeNode(vals[0])
    q = deque([root])
    i = 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root


def to_list(root: Optional[TreeNode]) -> list:
    """Serialize tree back to level-order list."""
    if not root:
        return []
    result, q = [], deque([root])
    while q:
        node = q.popleft()
        if node:
            result.append(node.val)
            q.append(node.left)
            q.append(node.right)
    return result


def test_harness(func):
    tests = [
        # (input tree, expected inverted tree)
        ([4,2,7,1,3,6,9], [4,7,2,9,6,3,1]),
        ([2,1,3],         [2,3,1]),
        ([],              []),
        ([1],             [1]),
        ([1,2,None],      [1,None,2]),
        ([1,2,3,4,5],     [1,3,2,5,4]),
    ]

    passed = 0
    for i, (inp, expected) in enumerate(tests):
        root = build_tree(inp)
        result_root = func(root)
        result = to_list(result_root)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | input={inp} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Recursive DFS | O(n) | O(h) |
| Iterative BFS | O(n) | O(w) |

Both visit every node once — choose iterative BFS
if the tree may be deeply skewed to avoid stack
overflow.
