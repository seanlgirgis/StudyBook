## Official Problem Statement

Given an integer array `nums`, return **all triplets**
`[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`,
`j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

The solution set must **not contain duplicate triplets**.

**Constraints:**
- `3 <= nums.length <= 3000`
- `-10^5 <= nums[i] <= 10^5`

## What This Is Actually Asking

Find every *unique* combination of three numbers that add to zero.
"Unique" is the trap — the same values appearing at different
indices must not be double-counted.

Key observations:
- Order inside a triplet does not matter (it's a set of values).
- After sorting, skipping repeated anchor values is enough to
  prevent duplicate triplets — no hash set needed on the result.
- If `nums[i] > 0` after sorting, no triplet can sum to 0
  (early exit).

## Walk Through an Example by Hand

`nums = [-1, 0, 1, 2, -1, -4]`

After sort: `[-4, -1, -1, 0, 1, 2]`

```
i=0, anchor=-4  left=1(-1)  right=5(2)  sum=-3 < 0 → left++
                left=2(-1)  right=5(2)  sum=-3 < 0 → left++
                left=3(0)   right=5(2)  sum=-2 < 0 → left++
                left=4(1)   right=5(2)  sum=-1 < 0 → left++
                left=5 >= right=5 → stop

i=1, anchor=-1  left=2(-1)  right=5(2)  sum=0  → add [-1,-1,2]
                skip dups: left→3, right→4
                left=3(0)   right=4(1)  sum=0  → add [-1,0,1]
                skip dups: left→4, right→3
                left=4 >= right=3 → stop

i=2, anchor=-1  same as i=1 (dup) → skip

i=3, anchor=0   left=4(1)   right=5(2)  sum=3 > 0 → right--
                left=4 >= right=4 → stop
```

Result: `[[-1,-1,2], [-1,0,1]]`

## The Picture

Sorted array with anchor `i`, and two pointers `L`, `R`:

```
index:  0    1    2    3    4    5
value: -4   -1   -1    0    1    2
        i    L              →   R
             ←————————————→
          sweep this window

Rule:
  sum < 0  →  move L right  (need bigger value)
  sum > 0  →  move R left   (need smaller value)
  sum = 0  →  record, skip dups on both sides

Anchor i advances; skip if nums[i] == nums[i-1] (dup anchor).
Stop early if nums[i] > 0 (can't reach 0 sum).
```

## When To Use This Pattern

Use the **Sort + Two Pointers** pattern when:
- You need to find pairs/triplets with a specific sum.
- The array can be sorted without losing problem meaning.
- You want O(n²) instead of O(n³) brute force.
- Duplicate results must be avoided without extra memory.

Classic family: 2Sum (sorted), 3Sum, 4Sum, 3Sum Closest,
Container With Most Water.

## The Approach

1. **Sort** `nums`.
2. **Iterate anchor** `i` from `0` to `n-3`:
   - Early exit: if `nums[i] > 0`, break.
   - Skip dup anchor: if `i > 0` and `nums[i] == nums[i-1]`,
     continue.
3. **Two pointers** `left = i+1`, `right = n-1`:
   - Compute `s = nums[i] + nums[left] + nums[right]`.
   - `s == 0`: append triplet, advance left past dups,
     retreat right past dups.
   - `s < 0`: `left += 1`.
   - `s > 0`: `right -= 1`.
4. Return accumulated result list.

## Complexity

| | |
|---|---|
| **Time** | O(n²) — sorting is O(n log n); the outer loop runs O(n) times, each with an O(n) two-pointer sweep |
| **Space** | O(1) extra — sort is in-place; the result list is output, not auxiliary space |

**Brute force comparison:** O(n³) with three nested loops.
Sorting + two pointers cuts one dimension by maintaining
order and exploiting monotonicity.

## Real World Connection

**Financial reconciliation:** Given a ledger of transactions,
find all sets of three entries that net to zero — identifying
circular transfers or balanced adjustments. Sorting by amount
and using two pointers lets auditors scan millions of records
in O(n²) rather than O(n³).

**Chemistry / molecular matching:** Find three molecular
weights whose combined delta from a reference is zero —
used in mass spectrometry peak identification.
