## Official Problem Statement

Given an integer array `nums` of length `n` and an integer
`target`, find three integers in `nums` such that the sum is
**closest to** `target`.

Return the **sum** of the three integers.

You may assume that each input would have exactly one solution.

**Constraints:**
- `3 <= nums.length <= 500`
- `-1000 <= nums[i] <= 1000`
- `-10^4 <= target <= 10^4`

## What This Is Actually Asking

Unlike 3Sum (exact zero), here we want the *closest possible*
sum to `target` — and we return the **sum value**, not the
triplet itself.

Key observations:
- There is always exactly one answer (guaranteed by constraints).
- If we ever hit `sum == target` exactly, we can return
  immediately — it can't get any closer.
- Sorting enables the greedy two-pointer move:
  `sum < target` → need larger → move left right;
  `sum > target` → need smaller → move right left.

## Walk Through an Example by Hand

`nums = [-1, 2, 1, -4]`, `target = 1`

After sort: `[-4, -1, 1, 2]`

```
closest = -4 + -1 + 1 = -4  (initial)

i=0, anchor=-4
  L=1(-1)  R=3(2)  s=-4+-1+2=-3  |−3−1|=4 > |−4−1|=5 → closest=-3
  s=-3 < target=1 → left++
  L=2(1)   R=3(2)  s=-4+1+2=-1   |-1-1|=2 < |-3-1|=4 → closest=-1
  s=-1 < target=1 → left++
  left=3 >= right=3 → stop

i=1, anchor=-1
  L=2(1)   R=3(2)  s=-1+1+2=2    |2-1|=1 < |-1-1|=2 → closest=2
  s=2 > target=1 → right--
  right=2 >= left=2 → stop

Return closest = 2
```

## The Picture

Sorted array with anchor `i`, and two pointers `L`, `R`.
We track `closest` across all sweeps:

```
index:  0    1    2    3
value: -4   -1    1    2     target = 1
        i    L         R
             ←————————→
          sweep window

At each step:
  s = nums[i] + nums[L] + nums[R]
  if |s - target| < |closest - target|:  closest = s
  if s == target:  return s immediately
  if s < target:   L++   (push sum higher)
  if s > target:   R--   (push sum lower)

No duplicate-skip needed — we return a single number,
not a list of triplets.
```

## When To Use This Pattern

Use **Sort + Two Pointers with running best** when:
- You need the triplet (or k-tuple) *closest* to a target.
- An exact match may not exist, so you track a running minimum
  absolute difference.
- Early exit on exact match can save work.

Related: 3Sum, 4Sum, Minimum Difference in Sorted Array,
Closest Pair of Points.

## The Approach

1. **Sort** `nums`.
2. **Initialize** `closest = nums[0] + nums[1] + nums[2]`.
3. **Iterate anchor** `i` from `0` to `n-3`:
   - Set `left = i+1`, `right = n-1`.
   - While `left < right`:
     - `s = nums[i] + nums[left] + nums[right]`.
     - If `abs(s - target) < abs(closest - target)`:
       update `closest = s`.
     - If `s == target`: return `s` immediately.
     - If `s < target`: `left += 1`.
     - Else: `right -= 1`.
4. Return `closest`.

## Complexity

| | |
|---|---|
| **Time** | O(n²) — O(n log n) sort; outer loop O(n), inner two-pointer sweep O(n) per anchor |
| **Space** | O(1) extra — sort in-place; only one scalar `closest` tracked |

**Note:** No duplicate-skipping logic is required here because
we return a single numeric answer, not a list of triplets.
The two-pointer sweep terminates naturally.
