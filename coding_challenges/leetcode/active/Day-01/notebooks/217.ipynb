## Official Problem Statement

Given an integer array `nums`, return `true` if any
value appears at least twice in the array, and
return `false` if every element is distinct.

**Example 1:**
```
Input:  nums = [1,2,3,1]
Output: true
```
**Example 2:**
```
Input:  nums = [1,2,3,4]
Output: false
```
**Example 3:**
```
Input:  nums = [1,1,1,3,3,4,3,2,4,2]
Output: true
```

**Constraints:**
- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

## What This Is Actually Asking

Scan the list and check whether any number shows
up more than once.
Return True the moment you find the first repeat.
If you reach the end without finding one, return
False — every number is unique.

## The Picture

```
Think of the set as a stamp-card.
Each number you see gets stamped.
If a number asks to be stamped and it's already
stamped — duplicate found.

nums: [ 1 | 2 | 3 | 1 ]

Step 1: see 1  -> stamp 1  card = {1}
Step 2: see 2  -> stamp 2  card = {1,2}
Step 3: see 3  -> stamp 3  card = {1,2,3}
Step 4: see 1  -> ALREADY STAMPED -> True

One-liner shortcut:
   len(nums) != len(set(nums))
   If the set is smaller, something got deduplicated.
   That means there was a duplicate.

   [1,2,3,1]  -> set = {1,2,3}  len 4 != 3  -> True
   [1,2,3,4]  -> set = {1,2,3,4} len 4 == 4  -> False
```

## When To Use This Pattern

- When you see **"has any element appeared before?"**,
  think **HashSet — O(1) membership check**
- When you need to detect the **first duplicate**
  early, think **add-and-check loop** (stops sooner
  than the one-liner)
- When memory is tight and n is small, think
  **sort first, check adjacent** — O(1) space
- When you only need a yes/no answer (not the
  duplicate's index), think **one-liner set compare**

## The Approach

Create an empty set to track numbers seen so far.
Walk through the array: for each number, check if
it is already in the set — if yes, return True.
If not, add it to the set and move on.
If the loop finishes without finding a match,
return False.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check every pair | O(n²) | O(1) |
| Sort then check adjacent | O(n log n) | O(1) |
| HashSet — one pass | O(n) | O(n) |

HashSet wins on time at the cost of O(n) space —
the right trade-off when speed matters more than
memory.

At Citi, the telemetry pipeline must detect duplicate

````python
def containsDuplicate(nums: List[int]) -> bool:
    """
    Return True if any value appears at least twice.

    Walk the array with a set tracking seen values.
    For each number, check membership in O(1). If
    already seen, return True immediately. Otherwise
    add to set and continue. Return False if no
    duplicate found after full traversal.

    Time:  O(n) — one pass, O(1) set ops each step
    Space: O(n) — set stores up to n distinct values
    """
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False


# Quick debug — run this cell while building
print(containsDuplicate([1,2,3,1]))   # True
print(containsDuplicate([1,2,3,4]))   # False
print(containsDuplicate([0,0]))        # True
print(containsDuplicate([1]))          # False
test_harness(containsDuplicate)
````
