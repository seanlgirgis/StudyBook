## Official Problem Statement

Given an integer array `nums`, return an array
`answer` such that `answer[i]` is equal to the
product of all the elements of `nums` except
`nums[i]`.

The product of any prefix or suffix of `nums` is
guaranteed to fit in a 32-bit integer.

You must write an algorithm that runs in `O(n)`
time and without using the division operation.

**Example 1:**
```
Input:  nums = [1,2,3,4]
Output: [24,12,8,6]
```
**Example 2:**
```
Input:  nums = [-1,1,0,-3,3]
Output: [0,0,9,0,0]
```

**Constraints:**
- `2 <= nums.length <= 10^5`
- `-30 <= nums[i] <= 30`
- The product of any prefix/suffix fits in 32-bit int
- **Follow up:** Can you solve it in O(1) extra space?

## What This Is Actually Asking

For each position in the array, compute the product
of every other number — skip the one at that position.
You cannot divide (because zeros would break it).
The output array is the same length as the input,
where each slot holds the product of all the others.

## Walk Through an Example by Hand

```
nums   = [ 1,  2,  3,  4]
index:     0   1   2   3

--- Forward pass: build left products ---
result[i] = product of nums[0..i-1]

prefix=1
i=0: result[0]=1      prefix = 1*1 = 1
i=1: result[1]=1      prefix = 1*2 = 2
i=2: result[2]=2      prefix = 2*3 = 6
i=3: result[3]=6      prefix = 6*4 = 24

result so far = [1, 1, 2, 6]

--- Backward pass: multiply in right products ---
result[i] *= product of nums[i+1..end]

suffix=1
i=3: result[3] = 6*1  = 6   suffix = 1*4 = 4
i=2: result[2] = 2*4  = 8   suffix = 4*3 = 12
i=1: result[1] = 1*12 = 12  suffix = 12*2 = 24
i=0: result[0] = 1*24 = 24  suffix = 24*1 = 24

Final result = [24, 12, 8, 6]
```

## The Picture

```
nums = [1, 2, 3, 4]

For index 2 (value=3), the answer is:
   everything LEFT of 2  x  everything RIGHT of 2
   = (1 * 2)             x  (4)
   =    2                x   4   =  8  ✓

Visualised as two sweeps:

  LEFT products (forward sweep ->):
  index:  0    1    2    3
  left:  [1]  [1]  [2]  [6]   (product of all before me)

  RIGHT products (backward sweep <-):
  right: [24] [12]  [4]  [1]  (product of all after me)

  result = left × right:
  [1*24, 1*12, 2*4, 6*1] = [24, 12, 8, 6]

Space trick: store left products in result[],
then multiply right products in on the fly
using a single 'suffix' variable. O(1) extra.
```

## When To Use This Pattern

- When you see **"product of all except self"** and
  **"no division"**, think **prefix × suffix pass**
- When each output depends on everything to the left
  AND everything to the right, think **two passes —
  forward then backward**
- When zeros are in the input, think **prefix/suffix
  handles them naturally** (division would not)
- When asked for O(1) extra space, think **reuse the
  output array for the left pass, use a scalar for
  the right pass**

## The Approach

Initialise the result array to all ones.
Do a forward pass: at each index, set result[i] to
the running prefix product (everything to the left),
then update the running product.
Do a backward pass: multiply each result[i] by a
running suffix product (everything to the right),
then update the suffix.
The result array now holds the correct answers
without any division and with O(1) extra space.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — nested loops | O(n²) | O(1) |
| Prefix + suffix arrays | O(n) | O(n) |
| Prefix pass + suffix scalar | O(n) | O(1) |

Reusing the output array for the prefix pass and
a single scalar for the suffix drops extra space
to O(1) while keeping the same O(n) time.

## Real World Connection

At Citi, risk-contribution calculations require each
position's exposure as a fraction of the total
portfolio product — the value for each asset
depends on the product of all other assets' weights.
The prefix-suffix pattern computes these contribution
scores in a single linear pass over the position
array, avoiding the O(n²) pairwise multiplication
that would be too slow for a large portfolio.
On the ETL side, the same pattern calculates
cumulative scaling factors for telemetry normalisation:
each server's normalised metric is its raw value
divided by the product of all other servers' baselines
— computable without division using two passes.

````python
def productExceptSelf(nums: List[int]) -> List[int]:
    """
    Return array where output[i] = product of all
    nums except nums[i]. No division. O(1) extra space.

    Forward pass: result[i] = product of nums[0..i-1].
    Backward pass: multiply result[i] by suffix product
    (nums[i+1..end]), tracked in a single variable.
    Output array is the only extra structure used.

    Time:  O(n) — two passes over the array
    Space: O(1) extra — output array not counted
    """
    """ 
        we will work in teo passes.. Going right multiplying so far numbers
        Then coming back from right to left
    """
    res = [1] * len(nums)
    rtol = [1] * len(nums)
    multr = 1
    multl = 1
    for i in range ( len(nums)):
        res[i] = multl
        multl = multl * nums[i]
    for i in range ( len(nums) -1, -1, -1):
        res[i] = res[i] * multr
        multr = multr * nums[i]
        
    return res


# Quick debug — run this cell while building
print(productExceptSelf([1,2,3,4]))       # [24,12,8,6]
print(productExceptSelf([-1,1,0,-3,3]))   # [0,0,9,0,0]
print(productExceptSelf([2,3]))            # [3,2]
print(productExceptSelf([1,0]))            # [0,1]
test_harness(productExceptSelf)
````
