## Official Problem Statement

Given an array of integers `nums` and an integer
`target`, return indices of the two numbers such
that they add up to `target`.

You may assume that each input would have exactly
one solution, and you may not use the same element
twice.

You can return the answer in any order.

**Example 1:**
```
Input:  nums = [2,7,11,15], target = 9
Output: [0,1]
```
**Example 2:**
```
Input:  nums = [3,2,4], target = 6
Output: [1,2]
```
**Example 3:**
```
Input:  nums = [3,3], target = 6
Output: [0,1]
```

**Constraints:**
- `2 <= nums.length <= 10^4`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- Only one valid answer exists

## What This Is Actually Asking

Find two numbers in the list that add up to a given
total and return their positions.
You cannot use the same number twice (even if the
same value appears twice, they must be at different
positions).
There is always exactly one correct answer.

## Walk Through an Example by Hand

```
nums = [2, 7, 11, 15]   target = 9
seen = {}    (empty at start)

i=0  num=2   complement = 9-2 = 7
     7 not in seen
     store: seen = {2: 0}

i=1  num=7   complement = 9-7 = 2
     2 IS in seen -> seen[2]=0
     return [0, 1]   <- indices of 2 and 7

Done in 2 steps.
Brute force would have checked: (2,7),(2,11),(2,15),(7,11)...
HashMap stops as soon as the pair is found.
```

## The Picture

```
target = 9

nums:  [  2  |  7  |  11  |  15  ]
index:    0     1      2      3

At each step, ask: "Does my missing piece exist?"

   See 2  ->  need 7  ->  7 not seen yet  ->  store 2
   See 7  ->  need 2  ->  2 IS stored!    ->  done

seen map = a locker room:
   Each number drops its index tag in a locker.
   When you walk in, you check if YOUR missing piece
   already left a tag. If yes — done. If no — tag up.

   num   |  complement  |  action
   ------|--------------|---------
    2    |      7       |  store {2:0}
    7    |      2       |  FOUND! return [0,1]
```

## When To Use This Pattern

- When you see **"two numbers that sum to target"**,
  think **store complement in a HashMap**
- When a brute-force would scan left and right for
  each element (O(n²)), think **can a map give me
  O(1) lookup instead?**
- When you need the **index of the found element**
  (not just the value), think **map: value → index**
- When the array is unsorted, think **HashMap over
  Two Pointers** (Two Pointers needs sorted input)

## The Approach

Walk through the array once, keeping a map of every
number seen so far alongside its index.
For each new number, compute what its complement
would need to be (target minus the current number).
Check if that complement is already in the map — if
yes, return both indices immediately.
If not, add the current number to the map and move on.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check every pair | O(n²) | O(1) |
| HashMap — one pass | O(n) | O(n) |

The HashMap trades O(n) space for O(n) time savings —
O(1) lookup per element replaces the inner loop.

## Real World Connection

At Citi, reconciliation jobs search for debit and
credit pairs that cancel each other to zero — a
direct Two Sum with target=0 across millions of
settlement records.
A brute-force pairwise scan across a daily batch of
10 million transactions is impractical; the HashMap
approach reduces it to a single linear pass.
The same pattern runs in the ETL pipeline to match
trade confirmations from two counterparty systems:
store one side in a map, look up each record from
the other side in O(1).

````python
def twoSum(nums: List[int], target: int) -> List[int]:
    """
    Return indices of the two numbers that sum to target.

    Walk the array once. For each number, compute its
    complement (target - num). If complement is in the
    seen map, return both indices. Otherwise store the
    current number and its index in the map and continue.

    Time:  O(n) — one pass, O(1) map lookup each step
    Space: O(n) — map stores up to n elements
    """
    hash ={}
    for idx, num in enumerate(nums):
        complement = target - num
        if complement in hash:
            return [hash[complement], idx]
        hash[num]=idx
            



# Quick debug — run this cell while building
print(twoSum([2,7,11,15], 9))   # [0,1]
print(twoSum([3,2,4], 6))        # [1,2]
print(twoSum([3,3], 6))          # [0,1]
print(twoSum([-1,-2,-3,-4], -7)) # [2,3]
test_harness(twoSum)
````
