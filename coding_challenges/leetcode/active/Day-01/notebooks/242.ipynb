## Official Problem Statement

Given two strings `s` and `t`, return `true` if `t`
is an anagram of `s`, and `false` otherwise.

An anagram is a word or phrase formed by rearranging
the letters of a different word or phrase, typically
using all the original letters exactly once.

**Example 1:**
```
Input:  s = "anagram", t = "nagaram"
Output: true
```
**Example 2:**
```
Input:  s = "rat", t = "car"
Output: false
```

**Constraints:**
- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters

## What This Is Actually Asking

Check if string t is just a shuffled version of
string s — same letters, same counts, different
order.
Different lengths means they cannot be anagrams.
Every character used in s must appear exactly the
same number of times in t.

## Walk Through an Example by Hand

```
s = "anagram"   t = "nagaram"

Count characters in s:
  a:3  n:1  g:1  r:1  m:1

Count characters in t:
  n:1  a:3  g:1  r:1  m:1

Compare: {a:3,n:1,g:1,r:1,m:1} == {a:3,n:1,g:1,r:1,m:1}
True -> anagram

---

s = "rat"   t = "car"

Count s:  r:1  a:1  t:1
Count t:  c:1  a:1  r:1

Compare: {r:1,a:1,t:1} != {c:1,a:1,r:1}
  t has 'c', s does not.  s has 't', t does not.
False -> not an anagram
```

## The Picture

```
Think of each string as a bag of letter tiles.
Two bags are anagrams if they hold the exact same
tiles in the same quantities.

s = "anagram"
tiles:  a a a  n  g  r  m

t = "nagaram"
tiles:  n  a a a  g  r  m

Bags match -> True

---

s = "rat"       t = "car"
tiles: r a t     tiles: c a r

't' is in s bag but not t bag  -> mismatch -> False
'c' is in t bag but not s bag  -> mismatch -> False

---

Frequency map approach (single count array):
  count[c] += 1 for c in s
  count[c] -= 1 for c in t
  if all zeros -> True
  any non-zero -> False
```

## When To Use This Pattern

- When you see **"same characters, different order"**,
  think **frequency map / Counter comparison**
- When you need to check if two collections have
  identical contents, think **Counter(a) == Counter(b)**
- When only lowercase letters are involved, think
  **O(1) space** (at most 26 keys in the map)
- When the input could include Unicode or any chars,
  think **Counter still works** — keys become those
  characters

## The Approach

First check lengths — if they differ, return False
immediately since anagrams must use the same letters.
Build a frequency map for s, counting how many times
each character appears.
Walk through t, decrementing the count for each
character — if any count drops below zero, t has a
character s does not, so return False.
If the loop finishes cleanly, return True.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Sort both strings, compare | O(n log n) | O(n) |
| Frequency map (manual) | O(n) | O(1) |
| Counter comparison | O(n) | O(1) |

Frequency map is O(n) time and O(1) space for
lowercase-only input — the 26-character alphabet
bounds the map size regardless of string length.

````python
def isAnagram(s: str, t: str) -> bool:
    """
    Return True if t is an anagram of s.

    Early exit if lengths differ. Build a frequency
    map from s (count up). Walk t and decrement each
    character count. If any count goes negative,
    t has a char s doesn't -> False. All zeros -> True.

    Time:  O(n) — n = len(s), one pass each string
    Space: O(1) — at most 26 keys (lowercase letters)
    """
    if len(s) != len(t):
        return False
#    occur = [0] * (ord('z') - ord('a') + 1)
    occur = [0] * 26
    for c in s:
        occur[ord(c) - ord('a')] += 1
    for c in t:
        occur[ord(c) - ord('a')] -= 1    
    return all(x == 0 for x in occur)


# Quick debug — run this cell while building
print(isAnagram("anagram", "nagaram"))  # True
print(isAnagram("rat", "car"))           # False
print(isAnagram("listen", "silent"))     # True
print(isAnagram("aa", "a"))              # False
test_harness(isAnagram)
````
