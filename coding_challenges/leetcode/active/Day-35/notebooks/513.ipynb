## Official Problem Statement

Given the `root` of a binary tree, return the **leftmost value**
in the **last row** of the tree.

**Constraints:**
- The number of nodes in the tree is in the range `[1, 10^4]`.
- `-2^31 <= Node.val <= 2^31 - 1`
- The tree is **guaranteed non-empty** (root ≠ None).

## What This Is Actually Asking

Find the deepest level of the tree.
Among all nodes at that deepest level, return the value of the
leftmost one.
If the deepest level has only one node, that node is the answer.
The tree is always non-empty, so there is always an answer.
"Leftmost" means the node you would reach first in a
left-to-right scan at that depth.

## Walk Through an Example by Hand

Tree: `[2, 1, 3]`
```
    2
   / \
  1   3
```
Level 0: [2]  → first = 2
Level 1: [1, 3] → first = 1   ← last level
Answer: 1

---
Tree: `[2, 1, 3, None, None, None, 4, None, 5]`
```
      2
     / \
    1   3
         \
          4
           \
            5
```
Level 0: [2]  → first = 2
Level 1: [1, 3] → first = 1
Level 2: [4]  → first = 4
Level 3: [5]  → first = 5   ← last level
Answer: 5

## The Picture

```
         1           <- Level 0  first=1
        / \
       2   3         <- Level 1  first=2
      /   / \
     4   5   6       <- Level 2  first=4
        /
       7             <- Level 3  first=7  ← deepest

Answer: 7

BFS Queue States:
┌──────────────────────────────────────────────────┐
│ Start      │ [1]                                 │
│ After L0   │ [2,3]      first_in_level = 1       │
│ After L1   │ [4,5,6]    first_in_level = 2       │
│ After L2   │ [7]        first_in_level = 4       │
│ After L3   │ []         first_in_level = 7  ✓    │
└──────────────────────────────────────────────────┘

Trick: overwrite a variable with level[0] each level.
       After BFS ends, that variable holds the answer.
```

## When To Use This Pattern

- When you need "the deepest X" in a tree → BFS; the last level
  processed is the deepest.
- When "leftmost" at any depth is requested → BFS ensures
  left-to-right order; index 0 of each level is leftmost.
- When DFS would require tracking depth explicitly
  → BFS makes depth tracking implicit and cleaner.
- When the answer is a single node from the last row
  → overwrite a variable each level; read it after BFS.

## The Approach

Run the standard level-order BFS with a level_size snapshot.
At the start of each level, record the value of the first node
dequeued (index 0 of the level).
Keep overwriting that variable for every level.
When the queue is empty, the variable holds the leftmost value
of the last (deepest) level — return it.

````python
from collections import deque        # BFS queue
from typing import Optional, List    # type hints


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def make_tree(vals):
    """Build a binary tree from a level-order list.
    None in the list means no node at that position.
    """
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = deque([root])
    i = 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force (DFS, track max_depth + leftmost) | O(n) | O(h) |
| Optimal BFS (level_size + overwrite first)    | O(n) | O(n) |

- **Time O(n):** every node visited exactly once.
- **Space O(n):** deque holds at most the widest level (≤ n/2).
- DFS is O(h) space but requires explicit depth bookkeeping;
  BFS is conceptually cleaner for "last level" problems.
