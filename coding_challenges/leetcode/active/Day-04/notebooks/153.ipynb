**Pattern:** Binary Search on Rotated Array — Compare Mid to Right

## Official Problem Statement

Suppose an array of length `n` sorted in ascending
order is **rotated** between `1` and `n` times.

Given the sorted rotated array `nums` of **unique**
elements, return the **minimum element** of this
array.

You must write an algorithm that runs in
`O(log n)` time.

**Example 1:**
```
Input:  nums = [3,4,5,1,2]
Output: 1
```
**Example 2:**
```
Input:  nums = [4,5,6,7,0,1,2]
Output: 0
```
**Example 3:**
```
Input:  nums = [11,13,15,17]
Output: 11
```

**Constraints:**
- `n == nums.length`
- `1 <= n <= 5000`
- `-5000 <= nums[i] <= 5000`
- All integers of `nums` are unique
- `nums` is sorted and rotated between 1 and n times

## What This Is Actually Asking

A sorted array was cut at some point and the two
pieces were swapped: [1,2,3,4,5] might become
[3,4,5,1,2]. Find the smallest number in O(log n).
The minimum is always at the rotation point —
the place where the sequence drops.

## Walk Through an Example by Hand

```
nums = [4, 5, 6, 7, 0, 1, 2]
idx:    0  1  2  3  4  5  6

lo=0  hi=6

Step 1:
  mid = (0+6)//2 = 3
  nums[3]=7  nums[6]=2
  7 > 2 -> min is in RIGHT half
  lo = mid+1 = 4

Step 2:
  lo=4  hi=6
  mid = (4+6)//2 = 5
  nums[5]=1  nums[6]=2
  1 < 2 -> min is in LEFT half (includes mid)
  hi = mid = 5

Step 3:
  lo=4  hi=5
  mid = (4+5)//2 = 4
  nums[4]=0  nums[5]=1
  0 < 1 -> min in LEFT half
  hi = mid = 4

lo=4 == hi=4 -> loop ends
Answer: nums[4] = 0
```

## The Picture

```
Original sorted: [0, 1, 2, 4, 5, 6, 7]
After rotation:  [4, 5, 6, 7, 0, 1, 2]

  7     |  |
  6     |  |  |
  5  |  |  |  |
  4  |  |  |  |
  2  .  .  .  .  .  .  |
  1  .  .  .  .  .  |  |
  0  .  .  .  .  |  |  |
     0  1  2  3  4  5  6
                 ^
              min here (rotation point)

Decision rule at each mid:

  nums[mid] > nums[right]
    -> mid is in the LEFT (high) part
    -> min MUST be in right half
    -> lo = mid + 1

  nums[mid] <= nums[right]
    -> mid is in the RIGHT (low) part
    -> min is at mid or left of mid
    -> hi = mid   (keep mid in range)

When lo == hi: that's the minimum.
```

## When To Use This Pattern

- When an array is sorted then rotated, think
  **binary search comparing mid to right boundary**
- When `nums[mid] > nums[right]`, think
  **min is in right half — lo = mid + 1**
- When `nums[mid] <= nums[right]`, think
  **min is at mid or left — hi = mid**
- When `lo == hi` at loop exit, think
  **return nums[lo] — that's the minimum**
- When the array is not rotated at all, think
  **still works — nums[0] is always found**

## The Approach

Set lo to 0 and hi to the last index.
While lo is less than hi, compute mid.
If nums[mid] is greater than nums[hi], the rotation
point is in the right half so move lo to mid plus
one.
Otherwise the minimum is at mid or to its left
so move hi to mid.
Return nums[lo] when the loop ends.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Linear scan | O(n) | O(1) |
| Binary search on rotated array | O(log n) | O(1) |

Binary search works here because even in a rotated
array one of the two halves is always fully sorted,
which gives us enough information to discard half
the space at every step.

## Real World Connection

At Citi, the telemetry ingestion pipeline stores

circular array that overwrites the oldest entry
when full. The buffer is a rotated sorted sequence

entry — the start of the window — uses the rotated
binary search: O(log n) instead of scanning all

O(log n) before executing the time-range filter.
