## Official Problem Statement

Given an array of integers `nums` which is sorted
in ascending order, and an integer `target`, write
a function to search `target` in `nums`. If
`target` exists, then return its index. Otherwise,
return `-1`.

You must write an algorithm with `O(log n)` runtime
complexity.

**Example 1:**
```
Input:  nums = [-1,0,3,5,9,12], target = 9
Output: 4
```
**Example 2:**
```
Input:  nums = [-1,0,3,5,9,12], target = 2
Output: -1
```

**Constraints:**
- `1 <= nums.length <= 10^4`
- `-10^4 < nums[i], target < 10^4`
- All integers in `nums` are unique
- `nums` is sorted in ascending order

## What This Is Actually Asking

The list is already sorted smallest to largest.
Find the position of a given number.
You must not scan every element — that would be
O(n). Instead, halve the search space each step.
Return -1 if the number is not in the list.

## Walk Through an Example by Hand

```
nums   = [-1,  0,  3,  5,  9, 12]
index:     0   1   2   3   4   5
target = 9

lo=0  hi=5
  mid = (0+5)//2 = 2   nums[2]=3
  3 < 9 -> target in right half  lo = mid+1 = 3

lo=3  hi=5
  mid = (3+5)//2 = 4   nums[4]=9
  9 == 9 -> FOUND  return 4

---
target = 2

lo=0  hi=5
  mid=2  nums[2]=3  3>2 -> hi = mid-1 = 1

lo=0  hi=1
  mid=0  nums[0]=-1  -1<2 -> lo = mid+1 = 1

lo=1  hi=1
  mid=1  nums[1]=0  0<2 -> lo = mid+1 = 2

lo=2 > hi=1 -> loop ends -> return -1
```

## The Picture

```
nums = [-1, 0, 3, 5, 9, 12]   target = 9
        lo            hi

Round 1:  lo=0  hi=5  mid=2  nums[2]=3  < 9
          |---------|  |----------|
           discard        keep

Round 2:  lo=3  hi=5  mid=4  nums[4]=9  == 9  DONE!
                      |---|

Each round: search space cuts in HALF.

  n=1,000,000 -> at most 20 rounds (log2(1M) ≈ 20)
  n=1,000,000,000 -> at most 30 rounds

Bug trap — mid calculation:
  WRONG:  mid = (lo + hi) // 2  (overflows in C/Java)
  RIGHT:  mid = lo + (hi - lo) // 2  (safe)
  Python: int is arbitrary precision -> both fine here
```

## When To Use This Pattern

- When you see **sorted array + find element**,
  think **binary search — O(log n)**
- When the problem says **O(log n) required**,
  think **binary search is the only option**
- When target < mid, think **search left — hi = mid-1**
- When target > mid, think **search right — lo = mid+1**
- When lo > hi, think **not found — return -1**

## The Approach

Set left to 0 and right to the last index.
While left does not exceed right, compute the
middle index and compare the middle value to target.
If equal return the index; if target is smaller
move right to mid minus one; if larger move left
to mid plus one.
If the loop ends without a match, return -1.

````python
from typing import List  # type hints for the solution
````

````python
def test_harness(func):
    tests = [
        # (nums, target, expected)
        ([-1,0,3,5,9,12],  9,   4),
        ([-1,0,3,5,9,12],  2,  -1),
        ([5],              5,   0),   # single element found
        ([5],              3,  -1),   # single element miss
        ([-1,0,3,5,9,12], -1,   0),  # first element
        ([-1,0,3,5,9,12], 12,   5),  # last element
        ([1,2,3,4,5,6,7],  1,   0),
        ([1,2,3,4,5,6,7],  7,   6),
        ([1,2,3,4,5,6,7],  4,   3),  # middle
        ([1,3,5,7,9],      6,  -1),  # not in array
    ]

    passed = 0
    for i, (nums, target, expected) in enumerate(tests):
        result = func(nums[:], target)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"nums={nums} target={target} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def search(nums: List[int], target: int) -> int:
    """
    Return index of target in sorted nums, or -1.

    Classic binary search. Set lo=0, hi=len-1. Each
    step compute mid = lo+(hi-lo)//2. If nums[mid]==
    target return mid. If target < nums[mid] search
    left (hi=mid-1), else search right (lo=mid+1).
    Return -1 when lo > hi.

    Time:  O(log n) — halves search space each step
    Space: O(1) — two pointers only
    """
    l, r = 0, len(nums) -1
    while l<=r:
        mid = l + (r - l) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            l= mid+1
        else:
            r = mid-1
    return -1

r'''
4
-1
0
-1
Test 1: PASSED | nums=[-1, 0, 3, 5, 9, 12] target=9 | expected=4 | got=4
Test 2: PASSED | nums=[-1, 0, 3, 5, 9, 12] target=2 | expected=-1 | got=-1
Test 3: PASSED | nums=[5] target=5 | expected=0 | got=0
Test 4: PASSED | nums=[5] target=3 | expected=-1 | got=-1
Test 5: PASSED | nums=[-1, 0, 3, 5, 9, 12] target=-1 | expected=0 | got=0
Test 6: PASSED | nums=[-1, 0, 3, 5, 9, 12] target=12 | expected=5 | got=5
Test 7: PASSED | nums=[1, 2, 3, 4, 5, 6, 7] target=1 | expected=0 | got=0
Test 8: PASSED | nums=[1, 2, 3, 4, 5, 6, 7] target=7 | expected=6 | got=6
Test 9: PASSED | nums=[1, 2, 3, 4, 5, 6, 7] target=4 | expected=3 | got=3
Test 10: PASSED | nums=[1, 3, 5, 7, 9] target=6 | expected=-1 | got=-1

10/10 tests passed

'''



# Quick debug — run this cell while building
print(search([-1,0,3,5,9,12], 9))   # 4
print(search([-1,0,3,5,9,12], 2))   # -1
print(search([5], 5))                # 0
print(search([1,3,5,7,9], 6))       # -1
test_harness(search)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Linear scan | O(n) | O(1) |
| Binary search | O(log n) | O(1) |

Binary search requires the array to be sorted —
that prerequisite is what enables halving the
search space and achieving O(log n).
