# LC 33 — Search in Rotated Sorted Array
**Difficulty:** Medium &nbsp;|&nbsp; **Category:** Binary Search
**Pattern:** Binary Search — One Half Always Sorted

<div style="border-left:4px solid purple; padding:10px 16px;
            background:#f5f0ff; margin-top:12px;">
<strong>Core Insight:</strong> In a rotated array,
one of the two halves is always fully sorted.
Check which half is sorted, then decide if the
target falls inside that sorted half. If yes,
search it; if no, search the other half.
</div>

## Official Problem Statement

There is an integer array `nums` sorted in
ascending order (with **distinct** values).

Prior to being passed to your function, `nums` is
**possibly rotated** at an unknown pivot index `k`
such that the resulting array is
`[nums[k], nums[k+1], ..., nums[n-1], nums[0],
nums[1], ..., nums[k-1]]`.

Given the array `nums` and an integer `target`,
return the **index** of `target` if it is in
`nums`, or `-1` if it is not in `nums`.

You must write an algorithm with `O(log n)` runtime
complexity.

**Example 1:**
```
Input:  nums = [4,5,6,7,0,1,2], target = 0
Output: 4
```
**Example 2:**
```
Input:  nums = [4,5,6,7,0,1,2], target = 3
Output: -1
```
**Example 3:**
```
Input:  nums = [1], target = 0
Output: -1
```

**Constraints:**
- `1 <= nums.length <= 5000`
- `-10^4 <= nums[i], target <= 10^4`
- All values of `nums` are unique
- `nums` is sorted and possibly rotated

## What This Is Actually Asking

A sorted array was cut somewhere and the two pieces
were swapped. Find a target value and return its
index, or -1 if it is absent. You must do it in
O(log n) — no linear scan allowed.

## Walk Through an Example by Hand

```
nums   = [4, 5, 6, 7, 0, 1, 2]
idx:      0  1  2  3  4  5  6
target = 0

lo=0  hi=6

Step 1:
  mid=3  nums[3]=7
  Left half [4,5,6,7] sorted? nums[lo]=4 <= nums[mid]=7 YES
  Is target 0 in [4..7]? NO (0 < 4)
  Search right half:  lo = mid+1 = 4

Step 2:
  lo=4  hi=6  mid=5  nums[5]=1
  Left half [0,1] sorted? nums[lo]=0 <= nums[mid]=1 YES
  Is target 0 in [0..1]? YES (0 >= 0 and 0 <= 1)
  Search left half:  hi = mid-1 = 4

Step 3:
  lo=4  hi=4  mid=4  nums[4]=0
  0 == target -> return 4
```

## The Picture

```
nums = [4, 5, 6, 7, 0, 1, 2]   target = 0

  7        |
  6     |  |
  5  |  |  |
  4  |  |  |  |
  2  .  .  .  .  .  .  |
  1  .  .  .  .  .  |  |
  0  .  .  .  .  |  |  |
     0  1  2  3  4  5  6
             ^
           mid=3 (left half sorted: 4,5,6,7)
           target 0 NOT in [4..7] -> go right

Key insight: one half is ALWAYS sorted.

If nums[lo] <= nums[mid]:  LEFT half is sorted
  target in [nums[lo]..nums[mid]]? -> search left
  else                              -> search right

Else:                      RIGHT half is sorted
  target in [nums[mid]..nums[hi]]? -> search right
  else                              -> search left
```

## When To Use This Pattern

- When an array is sorted then rotated and you
  need to find a value, think **binary search with
  sorted-half detection**
- When `nums[lo] <= nums[mid]`, think
  **left half is sorted — check range**
- When `nums[lo] > nums[mid]`, think
  **right half is sorted — check range**
- When target is out of the sorted half's range,
  think **search the other half**
- When `nums[mid] == target`, think **return mid**

## The Approach

Set lo and hi to the array boundaries. While lo
does not exceed hi, compute mid and return mid if
it matches.
Determine which half is sorted by comparing
nums[lo] to nums[mid].
If the left half is sorted and target falls within
its bounds, search left; otherwise search right.
If the right half is sorted and target falls within
its bounds, search right; otherwise search left.
Return -1 if the loop ends without finding target.

````python
from typing import List  # type hints for the solution
````

````python
def test_harness(func):
    tests = [
        # (nums, target, expected)
        ([4,5,6,7,0,1,2],  0,  4),
        ([4,5,6,7,0,1,2],  3, -1),
        ([1],              0, -1),
        ([1],              1,  0),
        ([3,1],            1,  1),
        ([3,1],            3,  0),
        ([5,1,3],          3,  2),
        ([4,5,6,7,0,1,2],  4,  0),  # first element
        ([4,5,6,7,0,1,2],  2,  6),  # last element
        ([1,3,5],          5,  2),   # no rotation
    ]

    passed = 0
    for i, (nums, target, expected) in enumerate(tests):
        result = func(nums[:], target)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"nums={nums} target={target} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def search(nums: List[int], target: int) -> int:
    """
    Return index of target in rotated sorted array, -1.

    Binary search: at each mid check which half is
    sorted (nums[lo] <= nums[mid] -> left sorted).
    If target in sorted half's range search that half;
    else search the other. Return mid on match, -1
    when lo > hi.

    Time:  O(log n) — halves search space each step
    Space: O(1) — two pointers only
    """
    l, r = 0, len(nums) -1
    while l <= r:
        mid = (l+r) //2
        if nums[mid] == target:
            return mid
        if nums[l] <= nums[mid]:      # left side sorted
            if target > nums[mid] or target < nums[l]:
                l = mid + 1 
            else:
                r = mid - 1
        else:                    #right side sorted
            if target < nums[mid] or target > nums[r]:
                r = mid - 1 
            else:
                l = mid + 1
    return -1


# Quick debug — run this cell while building
print(search([4,5,6,7,0,1,2], 0))   # 4
print(search([4,5,6,7,0,1,2], 3))   # -1
print(search([1], 0))                # -1
print(search([3,1], 1))             # 1
test_harness(search)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Linear scan | O(n) | O(1) |
| Binary search on rotated array | O(log n) | O(1) |

The rotation does not break the halving guarantee
because at least one half is always sorted.
That sorted half gives a definitive yes/no on
whether the target lies within it.

## Real World Connection

At Citi, log files are stored in a circular buffer
indexed by timestamp. When a new day begins the
index wraps — the buffer is effectively a rotated
sorted array. The on-call engineer needs to locate
a specific error event by timestamp in O(log n)
without downloading the entire buffer.
The rotated binary search finds the event directly,
cutting lookup time from a linear scan of 6,000
entries to under 13 comparisons.
On AWS CloudTrail, event lookup across rotated
partition indices uses the same technique: one
partition half is always in order, enabling
O(log n) event retrieval at the storage layer.
