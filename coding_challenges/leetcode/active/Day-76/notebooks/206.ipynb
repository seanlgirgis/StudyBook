## Official Problem Statement
Given the `head` of a singly linked list, reverse the list and return
the reversed list's head.

**Example 1:**
```
Input:  head = [1, 2, 3, 4, 5]
Output: [5, 4, 3, 2, 1]
```
**Example 2:**
```
Input:  head = [1, 2]
Output: [2, 1]
```
**Example 3:**
```
Input:  head = []
Output: []
```
**Constraints:**
- Number of nodes: `[0, 5000]`
- `-5000 <= Node.val <= 5000`

## What This Is Actually Asking
You have a chain of nodes where each node points to the next.
Your job is to make every node point to the *previous* node instead.

- You cannot allocate a new list — flip pointers in-place.
- After reversal the original `head` becomes the new `tail`
  (its `next` should be `None`).
- The original last node becomes the new `head` (return it).

Think of it as a chain of dominoes: you walk along picking up each
domino and re-laying it in the opposite direction.

## Walk Through an Example by Hand
List: `1 → 2 → 3 → None`

**Initial state:**
```
prev = None
curr = 1
```

**Step 1** (curr = 1):
```
nxt  = curr.next  # nxt = 2
curr.next = prev  # 1.next = None
prev = curr       # prev = 1
curr = nxt        # curr = 2
```

**Step 2** (curr = 2):
```
nxt  = 3
curr.next = 1     # 2 → 1
prev = 2
curr = 3
```

**Step 3** (curr = 3):
```
nxt  = None
curr.next = 2     # 3 → 2
prev = 3
curr = None  ← loop ends
```

Return `prev` (= 3). Result: `3 → 2 → 1 → None`

## The Picture
```
BEFORE:

  head
   |
  [1] --> [2] --> [3] --> [4] --> [5] --> None


DURING (reversing at node 3):

  prev    curr    nxt
   |       |       |
  [2]     [3]     [4] --> [5] --> None
   ^
  [1] <-- [2]
            (already reversed)

  curr.next = prev  -->  [3] --> [2]


AFTER:

  new_head
     |
    [5] --> [4] --> [3] --> [2] --> [1] --> None
```

Key: `prev` always tracks the already-reversed portion.
`curr` is the node being rewired. `nxt` saves the forward link
before we overwrite it.

## When To Use This Pattern
Use the **iterative three-pointer reversal** pattern when:

- You need to reverse a linked list (or a sub-section of one).
- You want **O(1) space** (no stack, no extra list).
- The problem involves re-wiring `.next` pointers in-place.

This pattern is a **building block** inside harder problems:
- Reverse a sub-list (LC 92)
- Reorder list (LC 143)
- Palindrome linked list (LC 234)
- K-group reverse (LC 25)

## The Approach
**Algorithm: Iterative In-Place Reversal**

1. Initialize `prev = None`, `curr = head`.
2. While `curr` is not `None`:
   - Save `nxt = curr.next` (so we don't lose the rest of the list).
   - Flip: `curr.next = prev`.
   - Advance: `prev = curr`, `curr = nxt`.
3. Return `prev` — it points to the new head.

**Why it works:**
- We never lose the rest of the list because we save `nxt` first.
- `prev` accumulates the growing reversed sub-list.
- When `curr` reaches `None`, `prev` is the last original node
  — now the new head.

**Complexity:** O(n) time, O(1) space.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val; self.next = next


def make_list(vals):
    dummy = ListNode(0); cur = dummy
    for v in vals:
        cur.next = ListNode(v); cur = cur.next
    return dummy.next


def to_list(head):
    r = []
    while head:
        r.append(head.val); head = head.next
    return r
````

## Real World Connection
**Undo history / browser back-stack:**
Many undo systems store edits as a singly linked list (most
recent → oldest). To replay edits forward you need to reverse
that list. The same three-pointer technique applies: no extra
memory is allocated; you just rewire the existing chain.

**Version control (e.g. Git):** Commit history is a linked
list. Tools that display history in chronological order must
traverse it in reverse, and the reversal logic maps directly
to this algorithm.

**Network packet re-assembly:** Packets arriving out of order
are linked in arrival sequence. Reversing that linked list
gives the correct transmission order for reassembly.
