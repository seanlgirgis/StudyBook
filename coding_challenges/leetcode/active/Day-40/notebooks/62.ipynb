<strong>Core Insight:</strong> Every cell can only be reached from the
cell above it or the cell to its left. So the number of unique paths
to any cell is simply the sum of paths to those two neighbors.
The entire first row and first column are always 1 — there is only
one way to reach any cell on the border.

## Official Problem Statement

There is a robot on an `m x n` grid. The robot is initially located
at the **top-left corner** (i.e., `grid[0][0]`). The robot tries to
move to the **bottom-right corner** (i.e., `grid[m-1][n-1]`).
The robot can only move either **down** or **right** at any point
in time.

Given the two integers `m` and `n`, return the number of possible
unique paths that the robot can take to reach the bottom-right corner.

**Constraints:**
- `1 <= m, n <= 100`
- The answer will be less than or equal to `2 * 10^9`.

## What This Is Actually Asking

You have a grid and a robot that can only move right or down.
You need to count every distinct sequence of moves that gets
the robot from the top-left to the bottom-right.
Two paths are different if they differ in even one step.
The robot always makes exactly `(m-1) + (n-1)` moves total
— the only question is in what order it makes them.

## Walk Through an Example by Hand

**Input:** `m=3, n=3`

Step 1 — Initialize: first row = all 1s, first col = all 1s.
```
dp[0][0]=1  dp[0][1]=1  dp[0][2]=1
dp[1][0]=1  dp[1][1]=?  dp[1][2]=?
dp[2][0]=1  dp[2][1]=?  dp[2][2]=?
```
Step 2 — Fill `dp[1][1]`:
  dp[1][1] = dp[0][1] + dp[1][0] = 1 + 1 = **2**

Step 3 — Fill `dp[1][2]`:
  dp[1][2] = dp[0][2] + dp[1][1] = 1 + 2 = **3**

Step 4 — Fill `dp[2][1]`:
  dp[2][1] = dp[1][1] + dp[2][0] = 2 + 1 = **3**

Step 5 — Fill `dp[2][2]`:
  dp[2][2] = dp[1][2] + dp[2][1] = 3 + 3 = **6**

Answer: **6 unique paths**.

## The Picture

3x3 grid — each cell shows dp[i][j] = number of paths to reach it:

```
  col0  col1  col2
  +-----+-----+-----+
r0|  1  |  1  |  1  |
  +-----+-----+-----+
r1|  1  |  2  |  3  |
  +-----+-----+-----+
r2|  1  |  3  | [6] |  <-- answer
  +-----+-----+-----+

  dp[i][j] = dp[i-1][j] + dp[i][j-1]
  borders   = 1 (only one direction to arrive)
```

## When To Use This Pattern

- When you move through a grid with **restricted directions**,
  think **2D DP where each cell depends on its neighbors**.
- When a problem asks to **count all distinct paths**,
  think **accumulate sums across the grid**.
- When the grid has a fixed start and end,
  think **bottom-up DP from start to end**.
- When `m` or `n` can be large but constraints still allow O(m*n),
  think **1D rolling array to optimize space**.
- When recursion leads to overlapping subproblems,
  think **memoization or tabulation**.

## The Approach

Build a 2D table where `dp[i][j]` stores the number of unique paths
to reach row `i`, column `j`. Initialize the entire first row and
first column to 1 because there is only one way to travel along
any border edge. For every interior cell, add the value from the
cell above and the cell to the left. The answer lives at
`dp[m-1][n-1]`.

````python
def unique_paths(m: int, n: int) -> int:
    """
    Count unique paths from top-left to bottom-right
    of an m x n grid moving only right or down.

    Args:
        m: number of rows
        n: number of columns

    Returns:
        Number of unique paths (int)

    Strategy:
        - Build dp table of size m x n
        - dp[0][*] = 1, dp[*][0] = 1 (borders)
        - dp[i][j] = dp[i-1][j] + dp[i][j-1]
        - return dp[m-1][n-1]

    Time:  O(m * n)
    Space: O(m * n) — can reduce to O(n) with 1D array
    """
    # Debug: print dimensions
    print(f"[DEBUG] Grid size: {m} rows x {n} cols")

    # Build the DP table
    dp = [[1] * n for _ in range(m)]
    print(f"[DEBUG] Initialized dp (borders=1):")
    for row in dp:
        print(f"  {row}")

    # Fill interior cells
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = dp[i-1][j] + dp[i][j-1]

    print(f"[DEBUG] Filled dp table:")
    for row in dp:
        print(f"  {row}")
    print(f"[DEBUG] Answer at dp[{m-1}][{n-1}] = {dp[m-1][n-1]}")

    pass  # replace with: return dp[m-1][n-1]
````

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute Force (recursion) | O(2^(m+n)) | O(m+n) stack | Recomputes subproblems |
| Memoization (top-down) | O(m * n) | O(m * n) | Caches results |
| Tabulation 2D (optimal) | O(m * n) | O(m * n) | Fills table bottom-up |
| Tabulation 1D (optimal) | O(m * n) | O(n) | Rolling row array |
