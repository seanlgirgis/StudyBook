## Official Problem Statement

Given the `root` of a binary tree and an integer `targetSum`,
return `true` if the tree has a **root-to-leaf path** such that
adding up all the values along the path equals `targetSum`.

A **leaf** is a node with no children.

**Constraints:**
- The number of nodes in the tree is in the range `[0, 5000]`.
- `-1000 <= Node.val <= 1000`
- `-1000 <= targetSum <= 1000`

## What This Is Actually Asking

Does any single root-to-leaf path exist whose node values sum
to a given target? We must reach a leaf — stopping at an
internal node does not count.
Negative values are allowed, so we cannot prune early by
checking if the running sum exceeds the target.
An empty tree always returns False because there is no leaf
to complete any path.
We only need existence — not the actual path itself.

## Walk Through an Example by Hand

Tree: `[5,4,8,11,None,13,4,7,2,None,None,None,1]`
Target: `22`

```
         5
        / \
       4   8
      /   / \
     11  13   4
    /  \       \
   7    2       1
```

Trace with subtraction approach:
1. node=5, remaining=22 → pass 22-5=17 to children
2. node=4, remaining=17 → pass 17-4=13 to children
3. node=11, remaining=13 → pass 13-11=2 to children
4. node=7, remaining=2, is leaf → 7 != 2 → False
5. node=2, remaining=2, is leaf → 2 == 2 → **True!**
   Path: 5 → 4 → 11 → 2 = 22 ✓

## The Picture

DFS explores every root-to-leaf path; we stop early on True:

```
         5  (target=22, pass 17 down)
        / \
       4   8  (L:pass 13, R:pass 9)
      /   / \
     11  13   4  (pass 2, 9 is leaf!=9, pass 8)
    /  \       \
   7    2       1
  leaf  leaf    leaf
  7!=2  2==2*   1!=8
        FOUND!

Winning path marked with *:
  5 -> 4 -> 11 -> 2  (sum = 22)

DFS short-circuits: once left side returns True at node 11,
right side of node 4 is never explored.
```

## When To Use This Pattern

- When a problem involves **root-to-leaf paths**, think DFS
  with a running accumulated value.
- When you need to **check existence** (not enumerate all
  paths), use `or` short-circuit to stop early.
- When values can be **negative**, avoid greedy pruning —
  always recurse to a leaf.
- When the problem has a **leaf condition**, handle it
  explicitly: `if not root.left and not root.right`.
- When you see "sum equals target", think **subtract-as-you-
  go** to avoid needing a separate accumulator variable.

## The Approach

Use recursive DFS. If root is None, return False (empty tree or
missing child). If we reach a leaf, check if its value equals
the remaining target. Otherwise recurse into both children,
passing `targetSum - root.val` down, and return True if either
side finds a valid path.

````python
from collections import deque
from typing import Optional


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def make_tree(vals):
    """BFS builder: construct TreeNode tree from level-order list."""
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = deque([root])
    i = 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root
````

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Enumerate all paths explicitly | O(n) | O(n) | Stores all paths |
| Recursive DFS subtract (optimal) | O(n) | O(h) | Short-circuits on True |
| Iterative DFS with stack | O(n) | O(h) | Same asymptotic, explicit stack |

**n** = nodes, **h** = height (O(log n) balanced, O(n) skewed).
