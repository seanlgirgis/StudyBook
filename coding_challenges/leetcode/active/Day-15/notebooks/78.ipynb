## Official Problem Statement

Given an integer array `nums` of **unique**
elements, return all possible subsets (the power
set).

The solution set **must not** contain duplicate
subsets. Return the solution in **any order**.

**Example 1:**
```
Input:  nums = [1,2,3]
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]
```
**Example 2:**
```
Input:  nums = [0]
Output: [[],[0]]
```

**Constraints:**
- `1 <= nums.length <= 10`
- `-10 <= nums[i] <= 10`
- All integers are unique

## What This Is Actually Asking

Return every possible selection of elements from
the array, including the empty selection and the
full set. For 3 elements there are 2³ = 8 subsets.
No duplicates — all inputs are unique.

## Walk Through an Example by Hand

```
nums = [1, 2, 3]

backtrack(start=0, path=[]):
  add [] to result

  choose 1: path=[1]
    backtrack(start=1, path=[1]):
      add [1]
      choose 2: path=[1,2]
        backtrack(start=2, path=[1,2]):
          add [1,2]
          choose 3: path=[1,2,3]
            backtrack(start=3): add [1,2,3]
          undo 3: path=[1,2]
        return
      undo 2: path=[1]
      choose 3: path=[1,3]
        backtrack(start=3): add [1,3]
      undo 3: path=[1]
    return
  undo 1: path=[]

  choose 2: path=[2]  ... add [2],[2,3]
  choose 3: path=[3]  ... add [3]

Result: [[],[1],[1,2],[1,2,3],[1,3],[2],[2,3],[3]]
```

## The Picture

```
nums = [1, 2, 3]

Decision tree — at each level pick from remaining:

        []
      /    \    \
    [1]    [2]   [3]
   /   \     \
 [1,2] [1,3] [2,3]
  |
[1,2,3]

Every node in this tree is a valid subset.
Add path to result at EVERY call (not just leaves).

Key: pass `start` index to avoid reusing earlier
  elements. At each call loop from start to end:

  for i in range(start, len(nums)):
      path.append(nums[i])
      backtrack(i + 1, path)
      path.pop()               # UNDO
```

## When To Use This Pattern

- When asked for **all subsets / power set**,
  think **backtrack with start index, add at
  every level**
- When each element can only appear once per
  subset, think **pass i+1 as start to children**
- When saving a path, think
  **always `path[:]` — never the reference**
- When undoing, think
  **`path.pop()` after the recursive call**

## The Approach

Use a recursive helper with a start index and
the current path. At each call, append a copy of
the current path to results (every partial
selection is valid). Then loop from start to the
end: add the element, recurse with start+1, then
remove the element to restore state.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Iterative bit masking | O(n * 2^n) | O(n * 2^n) |
| Backtracking | O(n * 2^n) | O(n) recursion |

Both produce 2^n subsets. Backtracking uses O(n)
stack space versus O(n * 2^n) output — the output
storage dominates either way.
