## Official Problem Statement

Given an array of intervals `intervals` where
`intervals[i] = [starti, endi]`, return the
**minimum** number of intervals you need to remove
to make the rest of the intervals non-overlapping.

**Note** that intervals which only touch at a
point are **non-overlapping**.

**Example 1:**
```
Input:  intervals = [[1,2],[2,3],[3,4],[1,3]]
Output: 1
Explanation: Remove [1,3] — then no overlaps remain.
```
**Example 2:**
```
Input:  intervals = [[1,2],[1,2],[1,2]]
Output: 2
```
**Example 3:**
```
Input:  intervals = [[1,2],[2,3]]
Output: 0
```

**Constraints:**
- `1 <= intervals.length <= 10^5`
- `intervals[i].length == 2`
- `-5 * 10^4 <= starti < endi <= 5 * 10^4`

## What This Is Actually Asking

Some time ranges in your list overlap each other.
Remove as few ranges as possible so that no two
remaining ranges overlap. Return how many you had
to remove. Touching at a single point is fine.

## Walk Through an Example by Hand

```
intervals = [[1,2],[2,3],[3,4],[1,3]]

Step 1 — sort by END:
  [[1,2],[2,3],[1,3],[3,4]]

Step 2 — greedy keep:
  kept=1  prev_end=2  (keep [1,2])

  [2,3]: start=2 >= prev_end=2  -> no overlap
         keep it  kept=2  prev_end=3

  [1,3]: start=1 < prev_end=3   -> OVERLAP
         skip it (remove)

  [3,4]: start=3 >= prev_end=3  -> no overlap
         keep it  kept=3  prev_end=4

Total = 4 intervals, kept = 3
Removed = 4 - 3 = 1
```

## The Picture

```
Before sort:
  [1---2]
  [1------3]
      [2--3]
          [3--4]

Sort by END:
  [1,2]  [2,3]  [1,3]  [3,4]
   end=2  end=3  end=3  end=4

Greedy: always keep the interval that ends EARLIEST.
Why? It leaves the most room for future intervals.

  Keep [1,2]  prev_end=2
  [2,3]: 2 >= 2 -> no overlap  Keep  prev_end=3
  [1,3]: 1 <  3 -> overlap     SKIP  (remove this)
  [3,4]: 3 >= 3 -> no overlap  Keep  prev_end=4

Kept: 3    Removed: 4-3 = 1

Overlap check: next_start < prev_end  (strict <)
Touch is OK:   next_start == prev_end -> keep
```

## When To Use This Pattern

- When asked for **minimum removals to eliminate
  overlaps**, think **sort by END, greedy keep**
- When `next_start < prev_end`, think
  **overlap — skip (count as removed)**
- When `next_start >= prev_end`, think
  **no overlap — keep, update prev_end**
- When converting keep count to remove count, think
  **removed = n − kept**
- When sorting by end (not start), think
  **earliest-ending = most room for future = greedy**

## The Approach

Sort intervals by their end time. Track the end
time of the last kept interval (start with the
first interval).
For each subsequent interval, if its start is
before the last end, it overlaps — count it as
removed. Otherwise keep it and update the last
end.
Return the total removed count.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Dynamic programming | O(n²) | O(n) |
| Greedy sort by end | O(n log n) | O(1) |

Sorting by end and greedily keeping intervals is
provably optimal — the interval ending earliest
always gives the most remaining timeline for future
intervals, so no DP is needed.
