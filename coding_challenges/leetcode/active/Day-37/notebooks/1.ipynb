## Official Problem Statement

Given an array of integers `nums` and an integer `target`, return
the **indices** of the two numbers such that they add up to
`target`. You may assume that each input would have **exactly
one solution**, and you may not use the same element twice.
You can return the answer in any order.

**Constraints:**
- `2 <= nums.length <= 10^4`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- Only one valid answer exists.

## What This Is Actually Asking

Find two positions in the array (0-based indices) whose values
add up to the target. The array is NOT sorted, so the sorted
two-pointer trick doesn't apply directly. We need a fast way to
check for each element whether its complement already exists.
Return indices, not the values themselves.

## Walk Through an Example by Hand

```

i=1, num=7:  complement = 9-7 = 2   IN seen!     => return [0, 1]

---
nums = [3, 2, 4], target = 6

seen = {}
i=0, num=3:  complement = 3   not in seen  => store {3:0}
i=1, num=2:  complement = 4   not in seen  => store {3:0, 2:1}
i=2, num=4:  complement = 2   IN seen!     => return [1, 2]
```

## The Picture

```
nums:   [2,  7,  11,  15]    target = 9
         0   1   2    3

Walk left to right. At each step:

  +-------+          +------------------+
  | num=2 | -------> | need 9-2=7?  NO  | => store {2:0}
  +-------+          +------------------+

  +-------+          +------------------+
  | num=7 | -------> | need 9-7=2?  YES | => return [seen[2], i]
  +-------+          +------------------+
                             ^
                             hash map lookup: O(1)

The hash map is a "have I seen the complement?" oracle.
We build and query it in a single left-to-right pass.
```

## When To Use This Pattern

- When you need to find a pair summing to a target in an
  **unsorted** array, think **hash map for complement lookup**.
- When a brute-force O(n²) scan seems obvious, think
  **can a hash map reduce this to O(n)?**
- When you need to return **indices** (not values), think
  **store value→index in the map, not just presence**.
- When each element's "partner" can be derived mathematically,
  think **compute complement, look it up**.

## The Approach

Create an empty hash map `seen` mapping values to their index.
Iterate through the array once. For each number, compute its
complement (`target - num`). If the complement is already in
`seen`, return `[seen[complement], current_index]`. Otherwise,
store the current number and its index into `seen`.

## Complexity

| Approach      | Time   | Space | Notes                          |
|---------------|--------|-------|--------------------------------|
| Brute force   | O(n²)  | O(1)  | Nested loops check every pair  |
| Sort + 2-ptr  | O(n log n) | O(1) | Loses original indices       |
| Hash map      | O(n)   | O(n)  | Optimal for unsorted + indices |

## Real World Connection

The hash map complement lookup is the engine behind database
hash joins — one table is probed, the other is built into a hash
table, and matching keys are found in O(1). At Citi, transaction
matching (find a payment that offsets a charge) uses exactly this
logic. In AWS Glue and PySpark, broadcast hash joins are the
go-to strategy when one side is small enough to fit in memory,
leveraging the same O(1) complement lookup principle.
