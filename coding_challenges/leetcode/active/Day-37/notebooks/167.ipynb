## Official Problem Statement

Given a **1-indexed** array of integers `numbers` that is already
sorted in non-decreasing order, find two numbers such that they
add up to a specific `target`. Return the indices of the two
numbers as an integer array `[index1, index2]` of length 2,
where `1 <= index1 < index2 <= numbers.length`.

**Constraints:**
- `2 <= numbers.length <= 3 * 10^4`
- `-1000 <= numbers[i] <= 1000`
- `numbers` is sorted in non-decreasing order
- `-1000 <= target <= 1000`
- The tests are generated such that there is **exactly one** solution

## What This Is Actually Asking

Find the pair of numbers in a sorted list that sum to `target`.
The array is sorted, so we know small values are on the left and
large values are on the right. We need 1-based indices in the
answer. There is always exactly one answer, so we never need to
handle the no-solution case.

## Walk Through an Example by Hand

```
numbers = [2, 7, 11, 15], target = 9

Step 1: left=0, right=3  => 2 + 15 = 17 > 9  => move right inward
Step 2: left=0, right=2  => 2 + 11 = 13 > 9  => move right inward
Step 3: left=0, right=1  => 2 +  7 =  9 == 9 => FOUND!

Return [left+1, right+1] = [1, 2]
```

Each step shrinks the window. If sum too big → right--.
If sum too small → left++. Sorted order guarantees correctness.

## The Picture

```
Sorted array:
  Index:  0    1    2    3
  Value: [2]  [7] [11] [15]
          L              R      sum=17 > 9  => R moves left
          L         R           sum=13 > 9  => R moves left
          L    R                sum= 9 == 9 => DONE!

Decision rule:
  sum < target  => left++   (need bigger value)
  sum > target  => right--  (need smaller value)
  sum == target => return [left+1, right+1]

  L -----> ... <----- R
  squeeze from both ends until they meet the target
```

## When To Use This Pattern

- When the array is **sorted** and you need pairs with a target
  sum, think **two pointers from both ends**.
- When you see O(1) space required with a sorted input, think
  **two pointers instead of a hash map**.
- When a brute-force O(n²) loop seems obvious but the input is
  sorted, think **can two pointers get this to O(n)?**
- When each pointer move provably eliminates a chunk of the
  search space, think **two pointer squeeze**.

## The Approach

Place one pointer at the start (`left=0`) and one at the end
(`right=len-1`). Compute their sum each iteration. If the sum
equals the target, return 1-based indices. If too small, move
left rightward to get a larger number. If too large, move right
leftward to get a smaller number. The sorted order guarantees
we will always find the unique answer.

````python
from typing import List
````

````python
def test_harness(func):
    tests = [
        # (numbers, target, expected)
        ([2, 7, 11, 15], 9,   [1, 2]),
        ([2, 3, 4],      6,   [1, 3]),
        ([-1, 0],        -1,  [1, 2]),
        ([1, 2, 3, 4, 5], 9,  [4, 5]),
        ([0, 0],          0,  [1, 2]),   # edge: both zeros
    ]
    passed = 0
    for nums, target, expected in tests:
        result = func(nums, target)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        print(f"{status} | nums={nums}, target={target} "
              f"| got={result}, expected={expected}")
    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def twoSum(nums: List[int], target: int) -> List[int]:
    """
    returns are 1 indexed array .. return l+1 r+1
    """
    
    l, r = 0, len(nums)-1
    while l <= r:
        s = nums[l] + nums[r]
        if s == target:
            return [l+1, r+1]
        if s > target :
            r-=1
        else:
            l+1
# Quick debug — run this cell while building
print(twoSum([2, 7, 11, 15], 9))   # [1, 2]
print(twoSum([2, 3, 4], 6))        # [1, 3]
print(twoSum([-1, 0], -1))         # [1, 2]
print(twoSum([0, 0], 0))           # [1, 2]
test_harness(twoSum)
````

## Complexity

| Approach      | Time   | Space | Notes                          |
|---------------|--------|-------|--------------------------------|
| Brute force   | O(n²)  | O(1)  | Check every pair               |
| Hash map      | O(n)   | O(n)  | Unsorted; stores complements   |
| Two pointers  | O(n)   | O(1)  | Optimal; uses sorted property  |
