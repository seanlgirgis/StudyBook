## Official Problem Statement

Given an integer array `nums`, return all the triplets
`[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`,
`j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

The solution set must not contain duplicate triplets.

**Constraints:**
- `3 <= nums.length <= 3000`
- `-10^5 <= nums[i] <= 10^5`

## What This Is Actually Asking

Find every unique group of three numbers in the array that sums
to zero. Duplicates in the output are not allowed even if the
same value appears multiple times in the input. The order inside
a triplet and the order of triplets in the result does not matter.
We must handle negative, zero, and positive numbers.

## Walk Through an Example by Hand

```
nums = [-1, 0, 1, 2, -1, -4]
After sort: [-4, -1, -1, 0, 1, 2]

i=0, nums[i]=-4  left=1, right=5
  -4 + (-1) + 2 = -3 < 0  => left++
  -4 + (-1) + 2 = -3 < 0  => left++
  -4 +   0  + 2 = -2 < 0  => left++
  -4 +   1  + 2 = -1 < 0  => left++
  left >= right, stop

i=1, nums[i]=-1  left=2, right=5
  -1 + (-1) + 2 = 0 == 0  => add [-1,-1,2], skip dups
  -1 +   0  + 1 = 0 == 0  => add [-1, 0, 1], skip dups
  left >= right, stop

i=2, nums[i]=-1  == nums[1]=-1  => SKIP (duplicate i)

i=3, nums[i]=0   left=4, right=5
  0 + 1 + 2 = 3 > 0  => right--
  left >= right, stop

Result: [[-1,-1,2], [-1,0,1]]
```

## The Picture

```
Sorted: [-4, -1, -1,  0,  1,  2]
          ^   ^                ^
          i  left            right    fix i, squeeze left/right

For each fixed i:
  need = -nums[i]   (two numbers on the right must sum to need)

  [ i ][ left -----> ... <----- right ]
        two-pointer on this subarray

Duplicate skipping rules:
  1. Skip i if nums[i] == nums[i-1]  (same outer number)
  2. After a hit, skip left  while nums[left] == nums[left-1]
  3. After a hit, skip right while nums[right] == nums[right+1]

Early exit: if nums[i] > 0, all remaining sums are positive.
```

## When To Use This Pattern

- When you need k-sum (k >= 3) with no duplicates, think
  **sort + fix outer + two-pointer inner**.
- When the problem says "all unique triplets", think
  **sort to make duplicate skipping easy**.
- When you reduce a 3-variable problem to a 2-variable one,
  think **fix one, apply known pattern on the rest**.
- When you see O(n²) is acceptable for n≤3000, think
  **outer loop + two-pointer = O(n²)**.

## The Approach

Sort the array. Iterate index `i` from 0 to n-3; skip if
`nums[i]` equals the previous value (dedup). For each `i`, run
a two-pointer scan from `i+1` to `n-1`. When the three values
sum to zero, record the triplet and skip past duplicate values
on both pointers before continuing. Stop early once `nums[i] > 0`
since no positive triple can sum to zero.

| Approach           | Time   | Space | Notes                       |
|--------------------|--------|-------|-----------------------------||
| Brute force        | O(n³)  | O(1)  | Check every triple          |
| Hash set (unsorted)| O(n²)  | O(n)  | Fix two, hash the third     |
