## Official Problem Statement

Given two strings `s` and `t`, return `true` if `t` is
an anagram of `s`, and `false` otherwise.

An anagram is a word or phrase formed by rearranging
the letters of a different word or phrase, using all
the original letters exactly once.

**Example 1:**
```
Input: s = "anagram", t = "nagaram"
Output: true
```
**Example 2:**
```
Input: s = "rat", t = "car"
Output: false
```

**Constraints:**
- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters

## What This Is Actually Asking

Two words are anagrams if you can scramble one to
make the other.
That means every letter must appear the same number
of times in both words.
Different lengths — immediately false.
Same length — check if the letter counts match.

```
Count down using t:

  a -> count[a] = 3-1 = 2   ok
  g -> count[g] = 1-1 = 0   ok
  a -> count[a] = 2-1 = 1   ok
  r -> count[r] = 1-1 = 0   ok
  a -> count[a] = 1-1 = 0   ok
  m -> count[m] = 1-1 = 0   ok

No count went negative -> return True

s = "rat"    t = "car"
  Count up: r:1 a:1 t:1
  Count down on "car":
    c -> count[c] = 0-1 = -1  NEGATIVE! return False
```

## The Picture

```
Think of each letter as a tile in a bag.

s = "anagram":
  bag:  [a][a][a][n][g][r][m]

t = "nagaram":
  try removing each tile t needs:
  remove n -> [a][a][a][ ][g][r][m]
  remove a -> [a][a][ ][ ][g][r][m]
  remove g -> [a][a][ ][ ][ ][r][m]
  remove a -> [a][ ][ ][ ][ ][r][m]
  remove r -> [a][ ][ ][ ][ ][ ][m]
  remove a -> [ ][ ][ ][ ][ ][ ][m]
  remove m -> [ ][ ][ ][ ][ ][ ][ ]

  Bag empty -> ANAGRAM
  Tile missing (count goes negative) -> NOT ANAGRAM
```

## When To Use This Pattern

- When you see **"same letters, different order"**,
  think **character frequency HashMap**
- When you see **different lengths**, think
  **return False immediately — no check needed**
- When you see **two strings to compare by content**,
  think **Counter(s) == Counter(t) as a one-liner**
- When the follow-up asks about **Unicode**, think
  **HashMap still works — just more possible keys**

## The Approach

First check lengths — if different, return False.
Use a single dictionary to count characters in `s`
(increment) then subtract counts using `t` (decrement).
If any count goes below zero during the second pass,
`t` has a character `s` doesn't — return False.
If all counts stay non-negative, return True.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Sort both strings, compare | O(n log n) | O(n) |
| HashMap frequency count | O(n) | O(1) |

The HashMap approach trades the sort cost for two
linear passes — and space is bounded at 26 keys.
