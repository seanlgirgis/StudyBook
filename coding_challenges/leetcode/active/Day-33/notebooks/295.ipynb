## Official Problem Statement

The **median** is the middle value in an ordered integer list.
If the list size is even, the median is the mean of the two
middle values.

Implement the `MedianFinder` class:
- `MedianFinder()` initializes the object.
- `void addNum(int num)` adds integer `num` to the data structure.
- `double findMedian()` returns the median of all elements so far.

**Constraints:**
- `-10^5 <= num <= 10^5`
- There will be at least one element before `findMedian` is called.
- At most `5 * 10^4` calls to `addNum` and `findMedian`.

## What This Is Actually Asking

Numbers arrive one at a time and you must always be able to
report the median of everything seen so far.
You cannot re-sort from scratch on every query — too slow.
The trick is to maintain a pre-sorted structure that exposes
the middle at O(1), updating it in O(log n) per insertion.
Two heaps meeting in the middle give you exactly that.

## Walk Through an Example by Hand

```
addNum(1): lo=[-1], hi=[]      → findMedian() = 1.0
addNum(2): lo=[-1], hi=[2]     → findMedian() = (1+2)/2 = 1.5
addNum(3): lo=[-2], hi=[2,3] ← unbalanced!
           len(lo)=1 < len(hi)=2 → rebalance:
           push -hi[0]=-2 to lo, pop hi[0]=2
           lo=[-2,-1], hi=[3]   → findMedian() = 2.0

Balance rule: |len(lo) - len(hi)| <= 1, lo size >= hi size.
```

## The Picture

```
Stream so far: 1, 2, 3, 4, 5

lo (max-heap, store negated)   hi (min-heap)
holds LOWER half               holds UPPER half

   lo stores: [-3, -1, -2]        hi stores: [4, 5]
   top of lo: -lo[0] = 3          top of hi: hi[0] = 4

   Sorted view:  [1, 2, | 3 | 4, 5]
                            ^
                      median = 3.0

Even count example: 1, 2, 3, 4
   lo: [-2, -1]   hi: [3, 4]
   median = (-lo[0] + hi[0]) / 2 = (2 + 3) / 2 = 2.5

Key invariant:
  all(lo_val <= hi_val for lo_val, hi_val in zip(lo, hi))
  len(lo) == len(hi)  OR  len(lo) == len(hi) + 1
```

## When To Use This Pattern

- When you see "streaming median" or "running median",
  think **two heaps**.
- When you need O(1) access to the middle of a dynamic set.
- When data arrives incrementally and you can't batch-sort.
- When percentile queries (p50, p95) must be answered live.
- When the problem involves splitting data at a pivot value.

## The Approach

Maintain two heaps: `lo` is a max-heap for the lower half
(simulated with negation) and `hi` is a min-heap for the upper.
On each `addNum`, push to `lo` first, then push-pop to `hi` to
guarantee ordering; rebalance sizes so `lo` is never smaller
than `hi`. For `findMedian`, return `lo` top if sizes differ,
or the average of both tops if equal.

## Complexity

| Approach                  | Time (addNum) | Time (findMedian) | Space |
|---------------------------|---------------|-------------------|-------|
| Brute force (sort each)   | O(n log n)    | O(1)              | O(n)  |
| Two heaps                 | O(log n)      | O(1)              | O(n)  |
| Sorted list + bisect      | O(n)          | O(1)              | O(n)  |

- **n** = total numbers added so far.
- Two heaps is optimal: O(log n) insert, O(1) query.
- Sorted list insert is O(n) due to shifting — worse for large n.
