## Official Problem Statement

Given an integer array `nums` and an integer `k`, return the `k`
most frequent elements. You may return the answer in any order.

**Constraints:**
- `1 <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`
- `k` is in the range `[1, the number of unique elements in nums]`
- The answer is guaranteed to be unique.

## What This Is Actually Asking

You have a list of numbers where some appear more than others.
Your job is to find the `k` numbers that show up the most.
Order does not matter — just identify the right elements.
Think of it like a popularity contest: return the top-k winners.
Counting first, then selecting, is the key two-step idea.

## Walk Through an Example by Hand

```
nums = [1, 1, 1, 2, 2, 3],  k = 2

Step 1 — Count frequencies:
  {1: 3,  2: 2,  3: 1}

Step 2 — Build min-heap of size k=2:
  Push (3, 1) → heap = [(3, 1)]
  Push (2, 2) → heap = [(2, 2), (3, 1)]
  Push (1, 3) → size > k, so pop min:
    Pop (1, 3) → heap = [(2, 2), (3, 1)]

Step 3 — Extract elements from heap:
  Result = [2, 1]   (order doesn't matter)
```

## The Picture

```
nums = [1,1,1,2,2,3]   k = 2

Frequency map:
  1 → ███ (3)
  2 → ██  (2)
  3 → █   (1)

Min-heap evolves (stores (freq, num)):

  After (3,1):        After (2,2):        After (1,3) + pop:
  ┌───────┐           ┌───────┐           ┌───────┐
  │ (3,1) │           │ (2,2) │           │ (2,2) │  ← min root
  └───────┘           │ (3,1) │           │ (3,1) │
                      └───────┘           └───────┘
                                           popped (1,3)

Final heap elements: num=2, num=1  → answer [1, 2]
```

## When To Use This Pattern

- When you see "top K" or "K most/least frequent", think **heap**.
- When you need only K results from N items, think **min-heap size K**.
- When sorting everything would be O(n log n) but K << N,
  think **heap for O(n log k)**.
- When frequencies or counts drive selection, **Counter + heap**.
- When K equals N, bucket sort gives O(n) time.

## The Approach

First, count how often each number appears using a hash map.
Then maintain a min-heap of size exactly k — the root always
holds the least frequent of your current top-k candidates.
When a new element's frequency beats the root, swap it in.
At the end, whatever is left in the heap is your answer.

## Complexity

| Approach      | Time         | Space  |
|---------------|--------------|--------|
| Brute force (sort by freq) | O(n log n) | O(n) |
| Min-heap size k | O(n log k) | O(n)  |
| Bucket sort   | O(n)         | O(n)   |

- **n** = length of nums, **k** = requested top count.
- Heap approach shines when k << n.
- Bucket sort is optimal but uses extra index space.
