## Official Problem Statement

Given an integer array `nums` and an integer `k`, return the
`k`th largest element in the array.

Note that it is the `k`th largest element in sorted order,
not the `k`th distinct element.

Can you solve it without sorting?

**Constraints:**
- `1 <= k <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

## What This Is Actually Asking

If you sorted the array in descending order, which value sits
at position k (1-indexed)? That is the kth largest.
Duplicates count as separate positions in the sorted order.
The goal is to find that value efficiently — ideally O(n log k)
rather than O(n log n) from a full sort.

## Walk Through an Example by Hand

```
nums = [3, 2, 1, 5, 6, 4],  k = 2

Goal: 2nd largest → sorted desc = [6,5,4,3,2,1] → answer = 5

Min-heap of size 2:
  Process 3: heap=[3]
  Process 2: heap=[2,3]
  Process 1: push→[1,3,2], len>k → pop min(1) → [2,3]
  Process 5: push→[2,3,5], len>k → pop min(2) → [3,5]
  Process 6: push→[3,5,6], len>k → pop min(3) → [5,6]
  Process 4: push→[4,6,5], len>k → pop min(4) → [5,6]

heap[0] = 5  ← the kth largest
```

## The Picture

```
nums = [3, 2, 1, 5, 6, 4]   k = 2

Heap after each element (min at top, size capped at k=2):

  +-----+   +-----+   +-----+   +-----+   +-----+   +-----+
  |  3  |   |  2  |   |  2  |   |  3  |   |  5  |   |  5  |
  +-----+   |  3  |   |  3  |   |  5  |   |  6  |   |  6  |
             +-----+   +-----+   +-----+   +-----+   +-----+
             (1 popped)           (2 popped)(3 popped)(4 popped)

  num=3   num=2     num=1     num=5     num=6     num=4

Heap root = 5 = answer (kth largest, i.e. 2nd largest)

Intuition: heap holds the TOP-K largest seen so far.
           Root = smallest of that group = kth largest.
```

## When To Use This Pattern

- When you see "kth largest/smallest", think **min-heap size k**.
- When k << n and sorting everything is overkill.
- When the array is a stream and you can't see all data upfront.
- When the answer is a single value, not a list of k values.
- When O(n log k) beats O(n log n) for large n, small k.

## The Approach

Iterate through every number in the array and push it onto a
min-heap. Whenever the heap grows beyond size k, pop the
minimum — this evicts the smallest element, keeping only the
k largest values seen so far. After processing all numbers,
the heap root is the kth largest element.

## Complexity

| Approach              | Time       | Space  |
|-----------------------|------------|--------|
| Brute force (sort)    | O(n log n) | O(1)   |
| Min-heap size k       | O(n log k) | O(k)   |
| Quickselect (avg)     | O(n)       | O(1)   |

- **n** = length of nums, **k** = target rank.
- Heap is simpler to code reliably; quickselect is fastest avg.
- Heap degrades to O(n log n) only when k ≈ n.
