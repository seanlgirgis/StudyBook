<strong>Core Insight:</strong> To reach step n you must
have come from step n-1 (one step) or step n-2
(two steps) — so the number of ways to reach n
is exactly ways(n-1) + ways(n-2): Fibonacci.

## Official Problem Statement

You are climbing a staircase. It takes `n` steps to
reach the top.

Each time you can either climb `1` or `2` steps.
In how many distinct ways can you climb to the top?

**Example 1:**
```
Input:  n = 2
Output: 2
Explanation: 1+1, 2
```
**Example 2:**
```
Input:  n = 3
Output: 3
Explanation: 1+1+1, 1+2, 2+1
```

**Constraints:**
- `1 <= n <= 45`

## What This Is Actually Asking

You are on the ground floor and want to reach the
top, which is n steps up.
Every move is either one step or two steps.
Count every distinct sequence of moves that gets
you there — the order matters.
For n=4: 1+1+1+1, 2+1+1, 1+2+1, 1+1+2, 2+2 = 5 ways.

## Walk Through an Example by Hand

```
n = 5

dp[i] = number of distinct ways to reach step i

dp[1] = 1    (only: 1)
dp[2] = 2    (1+1, 2)
dp[3] = dp[2] + dp[1] = 2+1 = 3   (1+1+1, 1+2, 2+1)
dp[4] = dp[3] + dp[2] = 3+2 = 5
dp[5] = dp[4] + dp[3] = 5+3 = 8

Answer: 8

Why dp[i] = dp[i-1] + dp[i-2]?
  If last move was 1 step -> came from dp[i-1]
  If last move was 2 steps -> came from dp[i-2]
  Total = both possibilities added together
```

## The Picture

```
Steps:   1    2    3    4    5    6    7 ...
Ways:    1    2    3    5    8   13   21 ...

         ^--- this IS the Fibonacci sequence
              (shifted by 1)

Each box = sum of the two boxes before it:

  [ 1 ] + [ 2 ] = [ 3 ]
         [ 2 ] + [ 3 ] = [ 5 ]
                [ 3 ] + [ 5 ] = [ 8 ]

Space-optimized: only keep the last two values.
You never need the whole row — just a 2-slot window.

  prev2  prev1  ->  curr = prev1 + prev2
    1      2          3
    2      3          5
    3      5          8
```

## When To Use This Pattern

- When the answer at position i depends only on
  the previous 1 or 2 values, think **Fibonacci DP**
- When you can take 1 or 2 steps to reach a target,
  think **dp[i] = dp[i-1] + dp[i-2]**
- When asked to count distinct sequences to a goal,
  think **DP — not combinatorics**
- When dp[i] only uses the last two values, think
  **drop the array — use two variables instead**

## The Approach

Handle the base cases: n=1 gives 1 way, n=2 gives
2 ways.
For every step from 3 to n, the number of ways equals
the sum of the previous two values — because you
arrived from one step back or two steps back.
Keep only two variables (prev2 and prev1) instead of
the whole array — slide them forward each iteration.

````python
# No imports needed — pure arithmetic
````

````python
def test_harness(func):
    tests = [
        # (n, expected)
        (1,  1),
        (2,  2),
        (3,  3),
        (4,  5),
        (5,  8),
        (10, 89),
        (20, 10946),
        (45, 1836311903),  # max constraint
    ]

    passed = 0
    for i, (n, expected) in enumerate(tests):
        result = func(n)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"n={n} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def climbStairs(n: int) -> int:
    """
    Count distinct ways to climb n stairs taking 1 or 2
    steps at a time.

    Base cases: n=1 -> 1 way, n=2 -> 2 ways. For each
    step i from 3 to n, ways[i] = ways[i-1] + ways[i-2]
    because you arrived from step i-1 or step i-2.
    Keep only two variables — the full array is not
    needed since each step uses only the last two.

    Time:  O(n) — one pass from 3 to n
    Space: O(1) — two variables, no array
    """
    if n == 1: return 1
    if n == 2 :return 2
    prev, pprev = 2, 1
    for i in range (3, n+1):
        res = prev + pprev
        pprev = prev 
        prev = res
    return (res)
r'''
1
2
3
8
Test 1: PASSED | n=1 | expected=1 | got=1
Test 2: PASSED | n=2 | expected=2 | got=2
Test 3: PASSED | n=3 | expected=3 | got=3
Test 4: PASSED | n=4 | expected=5 | got=5
Test 5: PASSED | n=5 | expected=8 | got=8
Test 6: PASSED | n=10 | expected=89 | got=89
Test 7: PASSED | n=20 | expected=10946 | got=10946
Test 8: PASSED | n=45 | expected=1836311903 | got=1836311903

8/8 tests passed
'''


# Quick debug — run this cell while building
print(climbStairs(1))   # 1
print(climbStairs(2))   # 2
print(climbStairs(3))   # 3
print(climbStairs(5))   # 8
test_harness(climbStairs)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Naive recursion (no memo) | O(2^n) | O(n) |
| DP array | O(n) | O(n) |
| Two-variable space-optimized | O(n) | O(1) |

The two-variable version is the interview answer —
same time as the DP array but constant space because
each step only looks back two positions.
