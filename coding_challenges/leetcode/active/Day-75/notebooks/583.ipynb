# LC 583 — Delete Operation for Two Strings
**Difficulty:** Medium | **Pattern:** 2D DP on Strings (LCS)

<strong>Core Insight:</strong> The minimum deletions to make two
strings equal equals the characters that are <em>not</em> in their
Longest Common Subsequence. Formally:
<code>min_deletes = len(word1) + len(word2) - 2 * LCS(word1, word2)</code>.
Keep the LCS; delete everything else.
</div>

## Official Problem Statement

Given two strings `word1` and `word2`, return the **minimum number
of steps** required to make `word1` and `word2` the same.

In one step, you can delete **exactly one character** in either
string.

**Example:**
```
Input:  word1 = "sea", word2 = "eat"
Output: 2
# Delete 's' from "sea" and 't' from "eat" → both become "ea"
```
**Constraints:** `1 <= word1.length, word2.length <= 500`

## What This Is Actually Asking

Find the **fewest total character deletions** (from either string)
so the two strings become identical.

**Key insight:** The best outcome is both strings reduced to their
Longest Common Subsequence (LCS). Every character outside the LCS
must be deleted.

```
word1 = "sea"   LCS = "ea"   delete 's'   → 1 deletion
word2 = "eat"   LCS = "ea"   delete 't'   → 1 deletion
Total = 2

Formula: (m - LCS) + (n - LCS) = m + n - 2*LCS
```

So this problem **reduces to finding the LCS** (LC 1143).

## Walk Through an Example by Hand

`word1 = "sea"`, `word2 = "eat"`

**Step 1 — find LCS:**

- `s` vs `e,a,t` → no match worth extending early
- `e` vs `e` → match! LCS so far: `"e"`
- `a` vs `a` → match! LCS so far: `"ea"`
- `a` appears at index 1 in word2; `t` doesn't match `a`
- LCS = `"ea"`, length = **2**

**Step 2 — compute deletions:**
```
m = 3, n = 3, lcs = 2
min_deletes = 3 + 3 - 2*2 = 6 - 4 = 2  ✓
```

## The Picture

LCS DP table for `word1="sea"`, `word2="eat"`:

```
       ""   e   a   t
  ""  [ 0,  0,  0,  0 ]
  s   [ 0,  0,  0,  0 ]  's' matches nothing in "eat"
  e   [ 0,  1,  1,  1 ]  'e'=='e' → diagonal + 1
  a   [ 0,  1,  2,  2 ]  'a'=='a' → diagonal + 1
              ^
              LCS = dp[3][3] = 2

  MATCH:    dp[i][j] = dp[i-1][j-1] + 1
  NO MATCH: dp[i][j] = max(dp[i-1][j], dp[i][j-1])

  min_deletes = m + n - 2 * dp[m][n]
              = 3 + 3 - 2*2 = 2
```

## When To Use This Pattern

Use **LCS-based reduction** when a problem asks for minimum
operations on two strings that boil down to "keep the common
part, remove the rest".

**Signals:**
- "Minimum deletions to make two strings equal"
- "Make strings identical using only deletions"
- Answers can be expressed as `total_chars - 2 * LCS`

**Contrast with Edit Distance (LC 72):** Edit distance allows
insertions, deletions, and substitutions. This problem only
allows deletions, so LCS is the cleanest reduction.

**Related:** LC 1143 (pure LCS), LC 1092 (Shortest Common
Supersequence), LC 72 (Edit Distance).

## The Approach

**Two-phase strategy:**

**Phase 1 — compute LCS via 2D DP:**
1. Create `dp` table `(m+1) x (n+1)`, all zeros
2. For `i` in `1..m`, `j` in `1..n`:
   - If `word1[i-1] == word2[j-1]`:
     `dp[i][j] = dp[i-1][j-1] + 1`
   - Else:
     `dp[i][j] = max(dp[i-1][j], dp[i][j-1])`

**Phase 2 — convert LCS to answer:**
```
return m + n - 2 * dp[m][n]
```

This works because we must delete every character not in the
LCS from both strings.

## Complexity

| | Value | Reason |
|---|---|---|
| **Time** | O(m × n) | Fill every cell of the 2D LCS table |
| **Space** | O(m × n) | Store the full DP table |

**Space optimisation:** Only the previous row is needed to
compute the current row → O(min(m, n)) with two 1D arrays.
The final answer requires just `dp[m][n]`, so no back-tracking
is needed.

Where `m = len(word1)`, `n = len(word2)`.

## Real World Connection

**Minimum edit by deletion only appears in:**

- **Version control / diff tools** — `git diff` computes the
  minimum set of deletions and insertions between file versions;
  when only deletions are counted, this exact formula applies
- **Data deduplication** — comparing log files or configs to
  find the minimum edits to synchronise them
- **Compiler / linter suggestions** — autocorrect systems
  estimating how many characters to strip from a misspelled
  token to match a known keyword
- **Genomics** — measuring how many base-pair deletions
  separate two gene sequences (a restricted edit distance)

The key takeaway: recognising that "minimum deletions" reduces
to LCS unlocks a clean O(mn) solution from a first-principles
observation.
