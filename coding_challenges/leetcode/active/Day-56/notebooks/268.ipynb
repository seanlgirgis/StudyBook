## Official Problem Statement

Given an array `nums` containing `n` distinct numbers in the range
`[0, n]`, return the only number in the range that is missing from
the array.

**Example 1:**
```
Input:  nums = [3, 0, 1]
Output: 2
```

**Example 2:**
```
Input:  nums = [0, 1]
Output: 2
```

**Example 3:**
```
Input:  nums = [9,6,4,2,3,5,7,0,1]
Output: 8
```

**Constraints:**
- `n == nums.length`
- `1 <= n <= 10^4`
- `0 <= nums[i] <= n`
- All numbers in `nums` are unique.

**Follow up:** O(1) extra space and O(n) runtime.

## What This Is Actually Asking

You have n numbers from the range [0..n] but one is missing.
Find it.

The brute-force set approach uses O(n) extra space. The Gauss
sum approach (`n*(n+1)//2 - sum(nums)`) is O(1) space but can
overflow for very large n in some languages.

The XOR approach is overflow-safe, O(1) space, and demonstrates
the elegant self-cancelling property of XOR: every number that
appears in both the index list and the array cancels to 0.

## Walk Through an Example by Hand

nums = [3, 0, 1],  n = 3,  missing = 2

```
XOR approach:
  Start with result = 0
  XOR all indices 0..n:  0^1^2^3 = 0
  XOR all values:        0^3^0^1 = ...

  Combined step by step:
  result = 0
  i=0: result ^= 0 ^ nums[0] = 0 ^ 0 ^ 3 = 3
  i=1: result ^= 1 ^ nums[1] = 3 ^ 1 ^ 0 = 2
  i=2: result ^= 2 ^ nums[2] = 2 ^ 2 ^ 1 = 1
  i=n: result ^= n           = 1 ^ 3     = 2
  Answer: 2  (correct!)

Gauss sum approach:
  expected = 3*4//2 = 6
  actual   = 3+0+1  = 4
  missing  = 6-4    = 2  (correct!)
```

## The Picture

```
XOR cancellation visualised for nums = [3, 0, 1]:

  Indices as a "bag":  { 0, 1, 2, 3 }
  Values  as a "bag":  { 3, 0, 1 }

  XOR both bags together:
  0 XOR 0 = 0  <- cancelled
  1 XOR 1 = 0  <- cancelled
  2           <- NO matching value, survives!
  3 XOR 3 = 0  <- cancelled
  ─────────────
  Result = 2  (the survivor)

  Property used: a XOR a = 0,  a XOR 0 = a
  So XOR is its own inverse — duplicate values annihilate.

Gauss sum visualised:

  0  1  2  3  4  ...  n      <- triangle sum = n(n+1)/2
  ●  ●     ●         <- present numbers
        ^--- gap = missing number = expected - actual
```

## When To Use This Pattern

- When you need to find **one missing or duplicate** in a range,
  think XOR (or Gauss sum).
- When an element appears an **odd number of times** and all
  others appear even times, think XOR all elements.
- When the problem says "find the one that doesn't pair up",
  think XOR.
- When O(1) space is required and sum/XOR of a known range
  can be computed directly, think math trick.
- When you see constraints `0 <= nums[i] <= n` with n = len,
  think pigeonhole / Gauss / XOR.

## The Approach

**XOR method:** XOR together every index from 0 to n, and every
value in nums. Each number present in nums appears once as an
index and once as a value — they cancel via XOR. The missing
number appears only as an index and is left in the result.

**Gauss method (alternative):** The expected sum of [0..n] is
`n*(n+1)//2`. Subtract the actual sum of nums. The difference
is the missing number — elegant and equally O(n) time, O(1) space.

````python
# Imports
from typing import List, Callable
````

````python
# --------------- Test Harness ---------------

def test_harness(func: Callable[[List[int]], int]) -> None:
    """Run test cases for LC 268 and report PASSED/FAILED."""
    cases = [
        ([3, 0, 1],               2, "LC example 1"),
        ([0, 1],                  2, "LC example 2: missing at end"),
        ([9,6,4,2,3,5,7,0,1],    8, "LC example 3: large"),
        ([0],                     1, "single element, missing 1"),
        ([1],                     0, "single element, missing 0"),
        (list(range(1, 101)),     0, "missing 0 from [1..100]"),
        (list(range(0, 100)),   100, "missing 100 from [0..99]"),
    ]

    passed = 0
    for nums, expected, label in cases:
        result = func(list(nums))  # pass a copy
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        else:
            print(f"  {status} [{label}]")
            print(f"    got:      {result}")
            print(f"    expected: {expected}")
        print(f"  {status} [{label}]")

    total = len(cases)
    print(f"\n{'='*40}")
    print(f"Results: {passed}/{total} passed")
    if passed == total:
        print("All tests PASSED!")
    else:
        print(f"{total - passed} test(s) FAILED.")
````

````python
def missingNumber(nums: List[int]) -> int:
    """
    LC 268 — Missing Number

    Strategy (XOR):
        result = 0
        for i, v in enumerate(nums):
            result ^= i ^ v
        result ^= len(nums)   # XOR the last index n
        return result

    Alternative (Gauss sum):
        n = len(nums)
        return n*(n+1)//2 - sum(nums)

    Parameters
    ----------
    nums : List[int] — n distinct integers from range [0, n]

    Returns
    -------
    int — the one missing number from the range [0, n]

    Time : O(n)
    Space: O(1)
    """
    res = len(nums)
    for i, n in enumerate(nums):
        res ^= i ^ n
    return (res)
    
    # Debug: print(f"missingNumber({nums})")
test_harness(missingNumber)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — sort and scan | O(n log n) | O(1) |
| HashSet — mark present numbers | O(n) | O(n) |
| Gauss sum (optimal) | O(n) | O(1) |
| XOR (optimal) | O(n) | O(1) |

Both Gauss sum and XOR are optimal. XOR is preferred in languages
where integer overflow is a concern (Java/C++) since XOR has no
overflow risk. In Python, integers are unbounded so both are safe.

## Real World Connection

At Citi, reconciliation jobs compare expected trade sequences
against received confirmations — finding the missing trade ID in
a consecutive range is exactly this problem at scale.

AWS S3 uses checksums and parity bits (a generalisation of XOR)
to detect missing or corrupted data chunks in multipart uploads.

In data engineering, detecting gaps in auto-increment primary key
sequences (e.g., missing event IDs in a Kafka stream) is a daily
task — the Gauss sum approach gives O(1) space over a full scan.

RAID-5 storage systems use XOR parity across disk blocks: if one
disk fails, XOR-ing the remaining disks reconstructs the missing
block — the same self-cancelling XOR property in hardware.
