# LC 239 — Sliding Window Maximum

<strong>Core Insight:</strong> A monotonic decreasing deque of
indices keeps track of the maximum in the current window. Smaller
elements behind the new element can never be the max while the
new element is in the window — evict them immediately.
</div>

## Official Problem Statement

Given an array `nums` and integer `k`, return the max of each
sliding window of size `k` as the window moves left to right.

**Example 1:**
```
Input:  nums=[1,3,-1,-3,5,3,6,7], k=3
Output: [3,3,5,5,6,7]
```
**Example 2:**
```
Input:  nums=[1], k=1
Output: [1]
```
**Constraints:**
- `1 <= k <= n <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

## What This Is Actually Asking

For each of the `n - k + 1` windows, report the maximum. A naive
scan of each window is O(nk) — too slow for large inputs.

The key observation: if element `a` is to the left of element `b`
and `a <= b`, then `a` can never be the window maximum as long as
`b` is in the window. So we discard `a` immediately.

This gives us a deque that is always monotonically decreasing.
The front of the deque is always the index of the maximum for
the current window.

## Walk Through an Example by Hand

```
nums = [1, 3, -1, -3, 5, 3, 6, 7],  k=3
        0  1   2   3  4  5  6  7

i=0: deque=[]  → append 0 → deque=[0]
i=1: nums[1]=3 >= nums[0]=1 → pop 0, append 1 → deque=[1]
     i>=k-1? No
i=2: nums[2]=-1 < nums[1]=3 → append 2 → deque=[1,2]
     i>=k-1? Yes → output nums[deque[0]]=nums[1]=3

i=3: front=1, i-k=0, 1>0 OK. nums[3]=-3 < -1 → append
     deque=[1,2,3] → output nums[1]=3

i=4: nums[4]=5. Pop 3(-3<5),pop 2(-1<5),pop 1(3<5)
     deque=[4] → output nums[4]=5

i=5: nums[5]=3 < 5 → append. deque=[4,5]
     front=4, i-k=2, 4>2 OK → output nums[4]=5

i=6: nums[6]=6. Pop 5(3<6), pop 4(5<6). deque=[6]
     front=6, i-k=3, 6>3 OK → output nums[6]=6

i=7: nums[7]=7. Pop 6(6<7). deque=[7]
     → output nums[7]=7

Result: [3, 3, 5, 5, 6, 7] ✓
```

## The Picture

```
nums = [1,  3, -1, -3,  5,  3,  6,  7]
        0   1   2   3   4   5   6   7

Window [0..2]: max=3   deque stores: [1]     (idx of 3)
Window [1..3]: max=3   deque stores: [1,2,3] (3,-1,-3)
Window [2..4]: max=5   deque stores: [4]     (idx of 5)
Window [3..5]: max=5   deque stores: [4,5]   (5,3)
Window [4..6]: max=6   deque stores: [6]     (idx of 6)
Window [5..7]: max=7   deque stores: [7]     (idx of 7)

Deque invariant (decreasing by value):

  FRONT (max of window)     BACK (most recent)
  ┌───┬───┬───┐
  │ 4 │ 5 │   │   values: nums[4]=5 > nums[5]=3
  └───┴───┴───┘
    ↑
  max of current window

  When new i arrives:
    1. Evict front if index <= i - k (out of window)
    2. Pop back while nums[back] <= nums[i] (useless)
    3. Append i
    4. If i >= k-1: record nums[deque[0]]
```

## When To Use This Pattern

- When you need **min or max in every sliding window**, think
  **monotonic deque** (not a heap — deque is O(n) total).
- When an element to the left is strictly dominated by a newer
  element, think **evict from back of deque**.
- When a window moves forward and old elements expire, think
  **evict from front of deque** using index checks.
- When you need O(n) over a sliding window with order statistics,
  think **monotonic deque over indices**.
- When input is large (n=10^5) and k is variable, think deque
  over heap (heap is O(n log k), deque is O(n)).

## The Approach

Iterate `i` from 0 to n-1. Before appending `i` to the deque:
remove expired front indices (`deque[0] <= i - k`), then pop
from the back all indices whose values are ≤ `nums[i]` — they
are dominated and can never be future window maximums.

Append `i`. Once `i >= k - 1` (first full window formed), append
`nums[deque[0]]` to the result. The deque front always holds the
index of the current window's maximum.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute force (each window) | O(nk) | O(1) | TLE for large k |
| Max-heap | O(n log k) | O(k) | Heap size bounded by k |
| Monotonic deque (optimal) | O(n) | O(k) | Each elem pushed/popped once |
| Segment tree | O(n log n) | O(n) | Overkill here |
