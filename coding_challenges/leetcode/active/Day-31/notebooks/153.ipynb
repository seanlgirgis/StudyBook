## Official Problem Statement

Suppose an array of length `n` sorted in ascending order is **rotated**
between `1` and `n` times. For example, the array
`nums = [0,1,2,4,5,6,7]` might become `[4,5,6,7,0,1,2]` if it was
rotated `4` times.

Given the sorted rotated array `nums` of **unique** elements, return
the **minimum element** of this array.

You must write an algorithm that runs in `O(log n)` time.

**Constraints:**
- `n == nums.length`
- `1 <= n <= 5000`
- `-5000 <= nums[i] <= 5000`
- All the integers of `nums` are **unique**.
- `nums` is sorted and rotated between `1` and `n` times.

## What This Is Actually Asking

Imagine a sorted list that has been cut at some point and the two
pieces swapped — like rotating a deck of cards. The minimum element
sits at that cut point. You cannot use a simple scan because the array
is no longer in order from start to finish. Instead, binary search can
still find the cut point in O(log n) by asking which half is "out of
order" at each step.

## Walk Through an Example by Hand

```
nums = [4, 5, 6, 7, 0, 1, 2]

Step 1: lo=0, hi=6, mid=3
        nums[mid]=7, nums[hi]=2
        7 > 2  => min is RIGHT of mid  => lo = mid+1 = 4

Step 2: lo=4, hi=6, mid=5
        nums[mid]=1, nums[hi]=2
        1 <= 2 => min is in [lo..mid]  => hi = mid = 5

Step 3: lo=4, hi=5, mid=4
        nums[mid]=0, nums[hi]=2
        0 <= 2 => min is in [lo..mid]  => hi = mid = 4

lo == hi == 4  => return nums[4] = 0
```

## The Picture

```
nums = [4,  5,  6,  7,  0,  1,  2]
idx  =  0   1   2   3   4   5   6
                         ^ rotation point (minimum)

Round 1:
  lo=0                hi=6
  [4,  5,  6,  7,  0,  1,  2]
               ^mid=3
  nums[3]=7 > nums[6]=2  => min is RIGHT  => lo=4

Round 2:
              lo=4     hi=6
  [ _,  _,  _,  _,  0,  1,  2]
                          ^mid=5
  nums[5]=1 <= nums[6]=2 => keep left half => hi=5

Round 3:
              lo=4  hi=5
  [ _,  _,  _,  _,  0,  1,  _]
                      ^mid=4
  nums[4]=0 <= nums[5]=1 => keep left half => hi=4

  lo == hi == 4  =>  nums[4] = 0  MINIMUM FOUND
```

## When To Use This Pattern

- When the array is sorted but **rotated** at an unknown pivot.
- When asked to find the **minimum** or the **pivot index** in O(log n).
- When `nums[mid] > nums[hi]` is the key comparison — signals the
  left half is fully sorted and the break is on the right.
- When `lo < hi` (not `lo <= hi`) is the loop condition — you want
  to converge to a single element, not overshoot.
- When all values are **unique** (duplicates require a different variant).

## The Approach

Use binary search with `lo` and `hi` converging toward the minimum.
At each step compare `nums[mid]` to `nums[hi]`. If `nums[mid]` is
greater, the left portion is sorted and the minimum must be to the
right of `mid`, so set `lo = mid + 1`. Otherwise the minimum is at
`mid` or somewhere to its left, so set `hi = mid` (keep mid in range).
Loop while `lo < hi`; when they meet, `nums[lo]` is the answer.
