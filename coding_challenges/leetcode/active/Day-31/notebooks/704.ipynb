<strong>Core Insight:</strong> When the array is sorted, you can eliminate
half of the remaining candidates each step by comparing the middle

## Official Problem Statement

Given an array of integers `nums` which is sorted in ascending order,
and an integer `target`, write a function to search `target` in `nums`.
If `target` exists, return its index. Otherwise, return `-1`.

You must write an algorithm with `O(log n)` runtime complexity.

**Constraints:**
- `1 <= nums.length <= 10^4`
- `-10^4 < nums[i], target < 10^4`
- All the integers in `nums` are **unique**.
- `nums` is sorted in ascending order.

## What This Is Actually Asking

You have a sorted list of numbers and you need to find where a specific
number lives. Think of it like looking up a word in a dictionary — you
open to the middle, decide if your word comes before or after, and
repeat on the relevant half. You keep cutting the search space in half
until you either find the number or confirm it is not there.

## Walk Through an Example by Hand

```
nums = [-1, 0, 3, 5, 9, 12],  target = 9

Step 1: lo=0, hi=5, mid=2  nums[2]=3  3<9 => lo=3
Step 2: lo=3, hi=5, mid=4  nums[4]=9  9==9 => return 4
```

Answer: index **4**

## The Picture

```
nums = [-1,  0,  3,  5,  9, 12],  target = 9
idx  =   0   1   2   3   4   5

Round 1:
  lo=0          hi=5
  [-1,  0,  3,  5,  9, 12]
            ^mid=2
  nums[mid]=3 < 9  => move lo to mid+1=3

Round 2:
        lo=3     hi=5
  [ _,  _,  _,  5,  9, 12]
                    ^mid=4
  nums[mid]=9 == 9  => FOUND at index 4
```

## When To Use This Pattern

- When the input array is **sorted** — binary search requires order.
- When you see `O(log n)` required — that is the classic hint.
- When you need to find an **exact value** or its insertion point.
- When the search space can be **halved** on each decision.
- When a linear scan works but feels too slow for the given constraints.

## The Approach

Maintain two pointers, `lo` and `hi`, that bracket the active search
window. On each iteration compute `mid = (lo + hi) // 2` and compare
`nums[mid]` to the target. If they match, return `mid`. If the target
is larger, the left half is useless so advance `lo` to `mid + 1`. If
the target is smaller, shrink `hi` to `mid - 1`. When the window
collapses (`lo > hi`) the target is absent, so return `-1`.

````python
from typing import List  # standard type hints
````

````python
def test_harness(func):
    """Run test cases for LC 704 Binary Search."""
    tests = [
        # (nums, target, expected)
        ([-1, 0, 3, 5, 9, 12], 9,  4),   # standard find
        ([-1, 0, 3, 5, 9, 12], 2, -1),   # not present
        ([5],                  5,  0),   # single element found
        ([5],                  3, -1),   # single element missing
        ([-1, 0, 3, 5, 9, 12], -1, 0),  # first element
        ([-1, 0, 3, 5, 9, 12], 12, 5),  # last element
    ]
    passed = 0
    for nums, target, expected in tests:
        result = func(nums, target)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        else:
            print(
                f"{status} | nums={nums} target={target} "
                f"| got={result} expected={expected}"
            )
    print(f"\n{passed}/{len(tests)} tests passed.")
````

````python
def search(nums: List[int], target: int) -> int:
    """
    Binary search in a sorted array.

    Args:
        nums:   Sorted list of unique integers.
        target: Value to find.

    Returns:
        Index of target if found, else -1.

    Time:  O(log n)
    Space: O(1)
    """
    l , r = 0, len(nums) -1
    while l <= r:
        mid = l + (r - l )//2
        if nums[mid] == target:
            return mid
        elif target > nums[mid]:
            l = mid +1
        else:
            r = mid -1
    return -1



# --- Debug prints (remove before final submission) ---
# Expected: 4
print(search([-1, 0, 3, 5, 9, 12], 9))

# Expected: -1
print(search([-1, 0, 3, 5, 9, 12], 2))

# Expected: 0  (single element, found)
print(search([5], 5))

# Expected: 0  (first element)
print(search([-1, 0, 3, 5, 9, 12], -1))

# Expected: 5  (last element)
print(search([-1, 0, 3, 5, 9, 12], 12))
test_harness(search)
````

## Real World Connection

At Citi, with over 6,000 API endpoints registered in a service catalog,
looking up an endpoint configuration by name with a linear scan would
be slow. If the catalog is sorted, binary search finds any entry in
about 13 comparisons instead of up to 6,000. AWS Parameter Store uses
the same principle internally when resolving hierarchical config keys —
sorted key spaces mean lookups stay fast regardless of how many
parameters accumulate over time.
