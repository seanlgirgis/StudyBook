## Official Problem Statement

Given an integer array `nums` and an integer `k`, return the **k-th largest
element** in the array.

Note: it is the k-th largest in sorted order, **not** the k-th distinct
element.

You must solve it without full sorting (expected O(n log k)).

**Constraints:**
- `1 <= k <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

**Examples:**
```
Input:  nums = [3,2,1,5,6,4], k = 2   Output: 5
Input:  nums = [3,2,3,1,2,4,5,5,6], k = 4   Output: 4
```

## What This Is Actually Asking

If you sorted the array in **descending** order, which element sits at
index `k-1`?

- k=1 → the maximum
- k=2 → second largest
- k=len(nums) → the minimum

The challenge: find that value **without** a full O(n log n) sort.
A min-heap of size k lets us do it in **O(n log k)**.

## Walk Through an Example by Hand

`nums = [3, 2, 1, 5, 6, 4]`, `k = 2`

| Step | num | heap (min at top) | action            |
|------|-----|-------------------|-------------------|
| 1    | 3   | [3]               | push              |
| 2    | 2   | [2, 3]            | push              |
| 3    | 1   | [2, 3]            | push then pop 1   |
| 4    | 5   | [3, 5]            | push then pop 2   |
| 5    | 6   | [5, 6]            | push then pop 3   |
| 6    | 4   | [5, 6]            | push then pop 4   |

Answer: `heap[0]` = **5** ✓

The heap always holds the **k largest** elements seen; its min is
the k-th largest.

## The Picture

After processing `[3, 2, 1, 5, 6, 4]` with k=2:

```
Min-heap (size k=2):

        5          ← heap[0] = answer (k-th largest)
       / \
      6   (empty)

Timeline of heap root (k-th largest so far):

  push 3 → [3]       root=3
  push 2 → [2,3]     root=2
  push 1 → [1,2,3] → pop 1 → [2,3]   root=2
  push 5 → [2,3,5] → pop 2 → [3,5]   root=3
  push 6 → [3,5,6] → pop 3 → [5,6]   root=5
  push 4 → [4,5,6] → pop 4 → [5,6]   root=5  ✓

Key: heap root = smallest of the k largest = k-th largest
```

## When To Use This Pattern

Use a **fixed-size min-heap** when:

- You need the **k-th largest** (or top-k) from a stream or large array
- Full sorting is too expensive or data arrives incrementally
- You want O(n log k) instead of O(n log n)

**Signals in the problem:**
- "k-th largest / smallest"
- "top k elements"
- "streaming / online" data

**Related problems:** LC 703 (stream variant), LC 347 (top-k freq)

## The Approach

**Min-Heap of Size k**

```
1. Initialize empty min-heap.
2. For each number in nums:
   a. heappush(heap, num)
   b. If len(heap) > k: heappop(heap)  ← evict smallest
3. Return heap[0]  ← smallest of top-k = k-th largest
```

Why this works: after every eviction the heap contains exactly the
k largest elements seen so far. The root (minimum of those k) is
by definition the k-th largest.

**Time:** O(n log k) | **Space:** O(k)

## Complexity

| Dimension | Value      | Reason                                    |
|-----------|------------|-------------------------------------------|
| Time      | O(n log k) | n pushes, each heap op costs O(log k)     |
| Space     | O(k)       | heap holds at most k+1 elements at once   |

**Comparison:**

| Approach          | Time       | Space  | Notes                   |
|-------------------|------------|--------|-------------------------|
| Full sort         | O(n log n) | O(n)   | simple but slower       |
| Min-heap size k   | O(n log k) | O(k)   | optimal for large n     |
| QuickSelect       | O(n) avg   | O(1)   | best avg, O(n²) worst   |

When k << n, the heap approach is significantly faster than sorting.

## Real World Connection

**Leaderboards & ranking systems** use this exact pattern.

- A gaming platform tracks millions of player scores in real-time.
  To display "Top 100 players" it maintains a min-heap of size 100.
  Each new score: push → if heap > 100 → pop minimum. The heap root
  is the current cutoff score for the leaderboard.

- **Monitoring dashboards** (Datadog, CloudWatch) identify the top-k
  slowest API endpoints from millions of requests per minute without
  sorting the entire dataset.

- **Recommendation engines** surface the k most-relevant items from
  a candidate pool of millions using a bounded priority queue.

The fixed-size heap is the production-grade alternative to sorting
when you only need the extremes, not the full order.
