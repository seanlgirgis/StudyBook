## Official Problem Statement

Given the `head` of a singly linked list, reverse
the list, and return the reversed list.

**Example 1:**
```
Input:  head = [1,2,3,4,5]
Output: [5,4,3,2,1]
```
**Example 2:**
```
Input:  head = [1,2]
Output: [2,1]
```
**Example 3:**
```
Input:  head = []
Output: []
```

**Constraints:**
- `0 <= number of nodes <= 5000`
- `-5000 <= Node.val <= 5000`

## What This Is Actually Asking

A chain of nodes each pointing to the next. Flip
all the arrows so each node points to the previous
one instead. The old tail becomes the new head.
Return the new head.

## Walk Through an Example by Hand

```
Input: 1 -> 2 -> 3 -> 4 -> 5 -> None

prev=None  curr=1

Step 1:
  nxt  = curr.next = 2
  curr.next = prev = None    (1 -> None)
  prev = curr = 1
  curr = nxt  = 2

Step 2:
  nxt  = 3
  curr.next = prev = 1       (2 -> 1)
  prev = 2  curr = 3

Step 3:
  nxt  = 4
  curr.next = 2              (3 -> 2)
  prev = 3  curr = 4

Step 4:
  nxt  = 5
  curr.next = 3              (4 -> 3)
  prev = 4  curr = 5

Step 5:
  nxt  = None
  curr.next = 4              (5 -> 4)
  prev = 5  curr = None

curr == None -> loop ends
Return prev = 5
Result: 5 -> 4 -> 3 -> 2 -> 1 -> None
```

## The Picture

```
Before:  None <- 1 -> 2 -> 3 -> 4 -> 5 -> None
              prev curr nxt

At each step, flip one arrow:

         None <- 1    2 -> 3 -> 4 -> 5
                  prev curr

         None <- 1 <- 2    3 -> 4 -> 5
                      prev curr

         None <- 1 <- 2 <- 3    4 -> 5
                           prev curr

Three pointers — order matters:
  1. Save next   (nxt = curr.next)
  2. Flip arrow  (curr.next = prev)
  3. Advance     (prev = curr; curr = nxt)

Must save nxt BEFORE flipping — otherwise you
lose the rest of the list.
```

## When To Use This Pattern

- When reversing a linked list, think
  **three pointers: prev, curr, nxt**
- When touching curr.next, think
  **save nxt = curr.next FIRST**
- When loop ends (`curr is None`), think
  **prev is the new head**
- When asked for O(1) space, think
  **iterative only — recursion is O(n) stack**

## The Approach

Start with prev as None and curr as head. While
curr is not None: save curr.next, redirect
curr.next to prev, then advance prev to curr and
curr to the saved next. When the loop ends, prev
is pointing at the new head — return it.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list(vals):
    """Build a linked list from a Python list."""
    dummy = ListNode(0)
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def to_list(head):
    """Convert a linked list to a Python list."""
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Iterative (three pointers) | O(n) | O(1) |
| Recursive | O(n) | O(n) call stack |

Iterative is always preferred for space. The
recursive version is elegant but risks stack
overflow for very long lists.

## Real World Connection

At Citi, log replay systems process events in
reverse chronological order to reconstruct the
state before a failure. The audit log is stored
as a linked list of timestamped events; reversing
it gives a clean replay stack.
The three-pointer reversal runs in O(n) with no
additional memory — critical when the log has
millions of entries and memory is constrained.
On AWS, DynamoDB Streams reversal for PITR
(point-in-time recovery) uses the same in-place
pointer-swap pattern on the change log chain.
