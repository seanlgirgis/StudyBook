## Official Problem Statement

You are given the heads of two sorted linked lists
`list1` and `list2`.

Merge the two lists into one **sorted** list.
The list should be made by splicing together the
nodes of the first two lists.

Return the head of the merged linked list.

**Example 1:**
```
Input:  list1 = [1,2,4], list2 = [1,3,4]
Output: [1,1,2,3,4,4]
```
**Example 2:**
```
Input:  list1 = [], list2 = []
Output: []
```
**Example 3:**
```
Input:  list1 = [], list2 = [0]
Output: [0]
```

**Constraints:**
- `0 <= number of nodes <= 50`
- `-100 <= Node.val <= 100`
- Both lists are sorted in non-decreasing order

## What This Is Actually Asking

You have two already-sorted chains of numbers.
Interleave them into one sorted chain using the
existing nodes (no new nodes). Return the head
of the combined sorted chain.

## Walk Through an Example by Hand

```
list1: 1 -> 2 -> 4
list2: 1 -> 3 -> 4

dummy -> ?
tail = dummy

l1=1  l2=1:  1==1, pick l1  tail->1  l1=2
l1=2  l2=1:  1<2,  pick l2  tail->1  l2=3
l1=2  l2=3:  2<3,  pick l1  tail->2  l1=4
l1=4  l2=3:  3<4,  pick l2  tail->3  l2=4
l1=4  l2=4:  4==4, pick l1  tail->4  l1=None
l1=None: attach rest of l2  tail->4

Result: dummy.next = 1->1->2->3->4->4
```

## The Picture

```
list1: 1 --> 2 --> 4 --> None
list2: 1 --> 3 --> 4 --> None

dummy --> [result chain being built]
tail  ^

Always compare list1.val vs list2.val.
Append the smaller; advance that pointer.

  Compare 1 vs 1: tie -> pick list1  result: 1
  Compare 2 vs 1: 1 wins            result: 1->1
  Compare 2 vs 3: 2 wins            result: 1->1->2
  Compare 4 vs 3: 3 wins            result: 1->1->2->3
  Compare 4 vs 4: tie -> pick list1 result: 1->1->2->3->4
  list1 exhausted -> attach list2    result: 1->1->2->3->4->4

Dummy head trick:
  Lets you always do tail.next = node
  without special-casing the very first node.
  Return dummy.next as the real head.
```

## When To Use This Pattern

- When merging two sorted linked lists, think
  **dummy head + tail pointer**
- When both lists have nodes, think
  **compare heads, append smaller, advance**
- When one list is exhausted, think
  **tail.next = the remaining list — no loop needed**
- When building a new list from scratch, think
  **dummy head eliminates first-node edge case**

## The Approach

Create a dummy node and point tail to it. While
both lists have nodes, compare their heads and
attach the smaller to tail, then advance that
list's pointer and tail. When either list runs
out, attach the remainder of the other list
directly. Return dummy.next.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list(vals):
    dummy = ListNode(0)
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Convert to arrays, merge, rebuild | O(m+n) | O(m+n) |
| In-place pointer manipulation | O(m+n) | O(1) |

In-place is optimal — no extra memory is allocated.
The dummy head is the only extra node and is O(1).
Each node is visited and linked exactly once.
