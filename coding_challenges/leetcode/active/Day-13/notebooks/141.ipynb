## Official Problem Statement

Given `head`, the head of a linked list, determine
if the linked list has a cycle in it.

There is a cycle in a linked list if there is
some node in the list that can be reached again
by continuously following the `next` pointer.
Return `true` if there is a cycle, else `false`.

**Example 1:**
```
Input:  head = [3,2,0,-4], pos = 1
        (tail connects back to node at index 1)
Output: true
```
**Example 2:**
```
Input:  head = [1,2], pos = 0
Output: true
```
**Example 3:**
```
Input:  head = [1], pos = -1
Output: false
```

**Constraints:**
- `0 <= number of nodes <= 10^4`
- `-10^5 <= Node.val <= 10^5`
- `pos` is -1 or a valid index

## What This Is Actually Asking

Walk along the linked list following next pointers.
Does the path ever loop back to a node you have
already visited? If yes there is a cycle, if you
reach None there is no cycle. Do it in O(1) space.

## Walk Through an Example by Hand

```
List: 3 -> 2 -> 0 -> -4
                      |
                      v
           2 <--------+   (cycle back to node 2)

slow=3   fast=3

Step 1:  slow=2   fast=0
Step 2:  slow=0   fast=2   (fast loops: -4 -> 2)
Step 3:  slow=-4  fast=-4  MEET! -> cycle -> True

No cycle example: 1 -> 2 -> 3 -> None

slow=1  fast=1
Step 1: slow=2  fast=3
Step 2: slow=3  fast=None  -> fast hit None -> False
```

## The Picture

```
No cycle:               Cycle:
  1 -> 2 -> 3 -> None     1 -> 2 -> 3
  slow --->                         |
  fast ------->                     v
  fast hits None = no cycle    5 <- 4
                               ^
                               slow/fast meet here

Think of it as a race track:
  If there is NO loop, the faster runner falls
  off the end of the track.

  If there IS a loop, the faster runner laps the
  slower one — they must meet somewhere inside
  the loop.

Guard: check fast and fast.next before advancing
       (fast takes 2 steps — need both to exist)
```

## When To Use This Pattern

- When asked to **detect a cycle in a linked list
  in O(1) space**, think **fast/slow pointers**
- When slow moves 1 step and fast moves 2 steps,
  think **if they meet, cycle exists**
- When fast or fast.next is None, think
  **no cycle — list is finite**
- When asked for the **cycle entry point**, think
  **reset slow to head after meeting; both move
  1 step; they meet at entry (Floyd's Part 2)**

## The Approach

Initialise slow and fast to head. While fast and
fast.next are not None, advance slow one step and
fast two steps. If slow and fast point to the
same node, a cycle exists — return True. If the
loop exits because fast reached None, return False.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list_with_cycle(vals, pos):
    """Build list; tail connects to node at index pos."""
    if not vals:
        return None
    nodes = [ListNode(v) for v in vals]
    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]
    if pos >= 0:
        nodes[-1].next = nodes[pos]  # create cycle
    return nodes[0]
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| HashSet of visited nodes | O(n) | O(n) |
| Fast / slow pointers | O(n) | O(1) |

Both are O(n) time. Fast/slow wins on space —
O(1) vs O(n). For any list with a cycle of length
k, fast catches slow within k steps after both
enter the cycle.
