## Official Problem Statement

Given the `head` of a linked list, remove the
`n`th node from the end of the list and return
its head.

**Example 1:**
```
Input:  head = [1,2,3,4,5], n = 2
Output: [1,2,3,5]
```
**Example 2:**
```
Input:  head = [1], n = 1
Output: []
```
**Example 3:**
```
Input:  head = [1,2], n = 1
Output: [1]
```

**Constraints:**
- `1 <= number of nodes <= 30`
- `0 <= Node.val <= 100`
- `1 <= n <= number of nodes`

## What This Is Actually Asking

Count from the tail: remove the nth node from the
end. For n=2 in a 5-node list, remove node 4
(the 2nd from last). Do it in one pass without
knowing the list length upfront.

## Walk Through an Example by Hand

```
head = 1 -> 2 -> 3 -> 4 -> 5 -> None    n=2

dummy -> 1 -> 2 -> 3 -> 4 -> 5 -> None
slow=dummy  fast=dummy

Step 1: advance fast n=2 times:
  fast -> 1  (step 1)
  fast -> 2  (step 2)

Step 2: advance both until fast.next is None:
  slow=dummy  fast=2  -> fast.next=3 not None
    slow=1  fast=3
  slow=1     fast=3  -> fast.next=4 not None
    slow=2  fast=4
  slow=2     fast=4  -> fast.next=5 not None
    slow=3  fast=5
  slow=3     fast=5  -> fast.next=None -> STOP

Step 3: remove slow.next (node 4):
  slow.next = slow.next.next  (3 -> 5)

Result: 1 -> 2 -> 3 -> 5 -> None
Return dummy.next
```

## The Picture

```
dummy -> 1 -> 2 -> 3 -> 4 -> 5 -> None    n=2

Phase 1: fast leaps n steps ahead of slow:
  slow=dummy    fast=2   (gap of 2)

Phase 2: move both until fast.next = None:
  slow=dummy    fast=2
  slow=1        fast=3
  slow=2        fast=4
  slow=3        fast=5   <- fast.next=None  STOP

  slow is now at the node BEFORE the target.
  target = slow.next = node 4

Step 3: slow.next = slow.next.next
  3 -> 5  (node 4 skipped)

Dummy head trick:
  Handles removing the FIRST node cleanly.
  When n == list length, slow stays at dummy
  and dummy.next skips the head.
```

## When To Use This Pattern

- When working from the end of a linked list,
  think **two pointers with an n-step gap**
- When fast pointer needs an n-step head start,
  think **advance fast n times before moving both**
- When removing a node, think
  **find the node BEFORE it, then skip: prev.next
  = prev.next.next**
- When head itself might be removed, think
  **dummy head — slow starts there**

## The Approach

Create a dummy node pointing to head. Set both
slow and fast to dummy. Advance fast n steps.
Then move both one step at a time until fast.next
is None. At this point slow is just before the
target node. Set slow.next to slow.next.next to
remove the target. Return dummy.next.

````python
from typing import Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def make_list(vals):
    dummy = ListNode(0)
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next


def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Two-pass (measure length, then remove) | O(n) | O(1) |
| One-pass two-pointer n-gap | O(n) | O(1) |

Both are O(n) time and O(1) space. The one-pass
approach is cleaner — it never needs to know the
total list length, making it more general.
