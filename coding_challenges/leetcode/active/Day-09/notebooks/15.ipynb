## Official Problem Statement

Given an integer array `nums`, return all the triplets
`[nums[i], nums[j], nums[k]]` such that `i != j`,
`i != k`, and `j != k`, and
`nums[i] + nums[j] + nums[k] == 0`.

Notice that the solution set must not contain
duplicate triplets.

**Example 1:**
```
Input:  nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
```
**Example 2:**
```
Input:  nums = [0,1,1]
Output: []
```
**Example 3:**
```
Input:  nums = [0,0,0]
Output: [[0,0,0]]
```

**Constraints:**
- `3 <= nums.length <= 3000`
- `-10^5 <= nums[i] <= 10^5`

## What This Is Actually Asking

Find every group of 3 numbers in the list that
add up to zero.
No group may appear twice, even in different order.
The list can have duplicates, so you must skip
them carefully.

## Walk Through an Example by Hand

```
nums = [-1, 0, 1, 2, -1, -4]

Sort: [-4, -1, -1, 0, 1, 2]

i=0  fixed = -4
  L=1 R=5  -4+(-1)+2 = -3 < 0  -> L++
  L=2 R=5  -4+(-1)+2 = -3 < 0  -> L++
  L=3 R=5  -4+0+2   = -2 < 0  -> L++
  L=4 R=5  -4+1+2   = -1 < 0  -> L++
  L=5 -> stop (L not < R)

i=1  fixed = -1  (nums[1]=-1)
  L=2 R=5  -1+(-1)+2 = 0   -> FOUND [-1,-1,2]  L++ R--
  L=3 R=4  -1+0+1   = 0   -> FOUND [-1,0,1]   L++ R--
  L=4 -> stop

i=2  nums[2]=nums[1]=-1  -> SKIP (duplicate fixed)

i=3  fixed = 0
  L=4 R=5  0+1+2 = 3 > 0  -> R--
  L=4 R=4 -> stop

Result: [[-1,-1,2], [-1,0,1]]
```

## The Picture

```
Sorted: [-4, -1, -1,  0,  1,  2]
          i   L            R

Fix i. Run Two Sum II between L and R.

  i fixed here    |<-- two pointers squeeze -->|

Three skip rules to eliminate duplicates:
  1. Outer: if nums[i] == nums[i-1], skip i
  2. Inner: after finding a triplet, skip L
     while nums[L] == nums[L+1]
  3. Inner: after finding a triplet, skip R
     while nums[R] == nums[R-1]

Early exit: if nums[i] > 0, no triplet sums to 0
            (sorted, so rest is also positive).
```

## When To Use This Pattern

- When you see **three numbers summing to a target**,
  think **sort + fix one + two pointers**
- When you see **no duplicate triplets**, think
  **skip equal values at i, L, and R after each find**
- When the fixed element `nums[i] > 0`, think
  **early exit — sorted array means nothing to the
  right can sum to zero**
- When extending to kSum, think **recurse: reduce
  k-Sum to (k-1)-Sum with one fixed element**

## The Approach

Sort the array first — this enables two pointers
and makes duplicate-skipping straightforward.
Iterate with index i as the fixed element; skip if
it's the same as the previous to avoid duplicate
triplets.
Run two pointers from i+1 to the end: if the three
values sum to zero, record the triplet and skip over
duplicates for both pointers before continuing.

````python
def test_harness(func):
    def normalize(triplets):
        # sort each triplet then sort the list for comparison
        return sorted([sorted(t) for t in triplets])

    tests = [
        ([-1, 0, 1, 2, -1, -4], [[-1,-1,2],[-1,0,1]]),
        ([0, 1, 1],              []),
        ([0, 0, 0],              [[0,0,0]]),
        ([],                     []),
        ([1, 2, 3],              []),        # no zero sum
        ([-2, 0, 0, 2, 2],      [[-2,0,2]]),# duplicates in input
        ([-4,-2,-2,-2,0,1,2,2,2,3,3,4,4,6,6], [
            [-4,-2,6],[-4,0,4],[-4,1,3],[-4,2,2],
            [-2,-2,4],[-2,0,2]
        ]),
    ]

    passed = 0
    for i, (nums, expected) in enumerate(tests):
        result = func(nums[:])
        ok = normalize(result) == normalize(expected)
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | nums={nums} | "
            f"expected={normalize(expected)} | "
            f"got={normalize(result)}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check all triples | O(n³) | O(1) |
| HashSet dedup | O(n²) | O(n) |
| Sort + fix + two pointers | O(n²) | O(1) |

Sort + two pointers matches HashSet time but uses
O(1) extra space — and duplicate skipping is built
in without any set overhead.
