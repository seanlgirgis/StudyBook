## Official Problem Statement

Given a 1-indexed array of integers `numbers` that is
already sorted in non-decreasing order, find two
numbers such that they add up to a specific `target`
number. Return the indices of the two numbers,
`index1` and `index2`, added by one as an integer
array `[index1, index2]` of length 2.
The tests are generated such that there is exactly
one solution. You may not use the same element twice.
Your solution must use only constant extra space.

**Example 1:**
```
Input:  numbers = [2,7,11,15], target = 9
Output: [1,2]
```
**Example 2:**
```
Input:  numbers = [2,3,4], target = 6
Output: [1,3]
```
**Example 3:**
```
Input:  numbers = [-1,0], target = -1
Output: [1,2]
```

**Constraints:**
- `2 <= numbers.length <= 3 * 10^4`
- `-1000 <= numbers[i] <= 1000`
- `numbers` is sorted in non-decreasing order
- `-1000 <= target <= 1000`
- Exactly one solution exists

## What This Is Actually Asking

The list is already sorted — smallest on the left,
largest on the right.
Find two numbers that add up to the target.
Return their positions in the list (starting at 1,
not 0).
You cannot use a hashmap — only O(1) extra space.

## Walk Through an Example by Hand

```
numbers = [2, 7, 11, 15]   target = 9
           L            R

Step 1:  L=0 R=3   sum = 2+15 = 17 > 9  -> move R left
           L     R
         [2, 7, 11, 15]

Step 2:  L=0 R=2   sum = 2+11 = 13 > 9  -> move R left
           L  R
         [2, 7, 11, 15]

Step 3:  L=0 R=1   sum = 2+7 = 9 == 9   -> FOUND!
         return [0+1, 1+1] = [1, 2]
```

## The Picture

```
Sorted array — squeeze from both ends:

  [2]  [7]  [11]  [15]
   L                R     sum=17  too big  -> R moves left
   L         R            sum=13  too big  -> R moves left
   L    R                 sum=9   just right -> done!

Rule:
  sum < target  ->  L moves right  (need bigger number)
  sum > target  ->  R moves left   (need smaller number)
  sum == target ->  return [L+1, R+1]

Sorted order makes this safe — every move eliminates
elements that cannot possibly be part of the answer.
```

## When To Use This Pattern

- When you see **sorted array + find pair summing to
  target**, think **opposite-end two pointers**
- When you see **O(1) space constraint**, think
  **two pointers over a HashMap**
- When sum is **too small**, think **move left pointer
  right** to increase the sum
- When sum is **too large**, think **move right pointer
  left** to decrease the sum

## The Approach

Place left at the start and right at the end.
Add the two values — if the sum equals the target,
return their 1-indexed positions.
If the sum is too small, move left one step right
to get a larger value.
If the sum is too large, move right one step left
to get a smaller value.

````python
from typing import List  # type hints for the solution
````

````python
def test_harness(func):
    tests = [
        # (numbers, target, expected)
        ([2, 7, 11, 15],   9,   [1, 2]),
        ([2, 3, 4],        6,   [1, 3]),
        ([-1, 0],         -1,   [1, 2]),
        ([1, 2, 3, 4, 5], 9,   [4, 5]),  # last two
        ([1, 2, 3, 4, 5], 3,   [1, 2]),  # first two
        ([-3, -1, 0, 2],  -1,  [1, 4]),  # negatives
        ([1, 3, 5, 7, 9], 8,   [1, 4]),  # skip middle
        ([0, 0],           0,   [1, 2]),  # zeros
    ]

    passed = 0
    for i, (numbers, target, expected) in enumerate(tests):
        result = func(numbers[:], target)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"numbers={numbers} target={target} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def twoSum(nums: List[int], target: int) -> List[int]:
    """
    Find two indices in a sorted array whose values
    sum to target. Return 1-indexed.

    Left pointer at start, right pointer at end.
    Sum too small: left moves right. Sum too large:
    right moves left. Return when sum equals target.
    Sorted order guarantees no valid pair is skipped.

    Time:  O(n) — each pointer moves at most n steps
    Space: O(1) — two pointers, no extra structure
    """
    l,r = 0, len(nums) -1
    while l < r:
        s = nums[l] + nums[r]
        if s == target:
            return [l+1, r+1]
        if s > target:
            r-=1
        else:
            l+=1
r'''
[1, 2]
[1, 3]
[1, 2]
[1, 2]
Test 1: PASSED | numbers=[2, 7, 11, 15] target=9 | expected=[1, 2] | got=[1, 2]
Test 2: PASSED | numbers=[2, 3, 4] target=6 | expected=[1, 3] | got=[1, 3]
Test 3: PASSED | numbers=[-1, 0] target=-1 | expected=[1, 2] | got=[1, 2]
Test 4: PASSED | numbers=[1, 2, 3, 4, 5] target=9 | expected=[4, 5] | got=[4, 5]
Test 5: PASSED | numbers=[1, 2, 3, 4, 5] target=3 | expected=[1, 2] | got=[1, 2]
Test 6: PASSED | numbers=[-3, -1, 0, 2] target=-1 | expected=[1, 4] | got=[1, 4]
Test 7: PASSED | numbers=[1, 3, 5, 7, 9] target=8 | expected=[1, 4] | got=[1, 4]
Test 8: PASSED | numbers=[0, 0] target=0 | expected=[1, 2] | got=[1, 2]

8/8 tests passed


'''



# Quick debug — run this cell while building
print(twoSum([2, 7, 11, 15], 9))   # [1, 2]
print(twoSum([2, 3, 4], 6))        # [1, 3]
print(twoSum([-1, 0], -1))         # [1, 2]
print(twoSum([0, 0], 0))           # [1, 2]
test_harness(twoSum)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check all pairs | O(n²) | O(1) |
| HashMap (unsorted Two Sum) | O(n) | O(n) |
| Two pointers (sorted input) | O(n) | O(1) |

Sorted input is the unlock — it lets two pointers
achieve O(n) time AND O(1) space simultaneously.

## Real World Connection

At Citi, daily server P&L deltas are stored in a
sorted ledger. A reconciliation job must find two
entries that net to zero — a debit and a credit that
cancel each other out.
Because the ledger is already sorted by amount, two
pointers finds the pair in O(n) without loading the
whole dataset into a hashmap.
The same pattern runs in the ETL pipeline when
validating that sorted ingestion timestamps bracket
a target processing window — left and right close in
until the exact window boundaries are found.
