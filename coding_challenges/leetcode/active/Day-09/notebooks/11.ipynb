# LC 11 — Container With Most Water
**Difficulty:** Medium &nbsp;|&nbsp; **Category:** Two Pointers
**Pattern:** Greedy Two Pointers — Move the Shorter Side

<div style="border-left:4px solid purple; padding:10px 16px;
            background:#f5f0ff; margin-top:12px;">
<strong>Core Insight:</strong> Width shrinks every step.
The only way to possibly find more water is a taller
line. Always move the shorter pointer inward —
the taller one can never help with less width.
</div>

## Official Problem Statement

You are given an integer array `height` of length `n`.
There are `n` vertical lines drawn such that the two
endpoints of the `i`th line are `(i, 0)` and
`(i, height[i])`.

Find two lines that together with the x-axis form a
container, such that the container contains the most
water.

Return the maximum amount of water a container can
store.

**Notice** you may not slant the container.

**Example 1:**
```
Input:  height = [1,8,6,2,5,4,8,3,7]
Output: 49
```
**Example 2:**
```
Input:  height = [1,1]
Output: 1
```

**Constraints:**
- `n == height.length`
- `2 <= n <= 10^5`
- `0 <= height[i] <= 10^4`

## What This Is Actually Asking

Pick any two vertical bars from the list.
Water fills up to the shorter bar's height.
Width is the distance between them.
Volume = shorter height × distance.
Find the pair that gives the biggest volume.

```
     |  |  |  |  |  |  |  |  |  |

     L=0                          R=8
     Water = min(1,7) * 8 = 8

     L=1                          R=8
     Water = min(8,7) * 7 = 49  <- MAX

Greedy: the shorter bar IS the ceiling.
Moving the TALLER bar inward can only hurt
(smaller width, same or smaller height).
Moving the SHORTER bar inward might find a taller one.
```

## When To Use This Pattern

- When you see **two endpoints, area between them**,
  think **opposite-end two pointers**
- When width decreases with each step, think
  **only height improvement can compensate**
- When deciding which pointer to move, think
  **always move the shorter one** — it's the only
  side with upside potential
- When both heights are equal, think
  **move either — both are equally worth exploring**

## The Approach

Start left at index 0 and right at the last index.
Compute the water volume: minimum of the two heights
times the distance between them.
Track the maximum seen so far.
Move the shorter of the two pointers inward — if
left is shorter move left right, else move right left.
Stop when the pointers meet.

````python
from typing import List  # type hints for the solution
````

````python
def test_harness(func):
    tests = [
        # (height, expected)
        ([1,8,6,2,5,4,8,3,7], 49),
        ([1,1],               1),
        ([4,3,2,1,4],         16),  # matching ends
        ([1,2,1],             2),
        ([1,2,4,3],           4),
        ([2,3,4,5,18,17,6],   17), # tall bars near end
        ([1,0,0,0,1],         4),  # zeroes in middle
        ([3,1,2,4,0,1,3,2],   18), # general case
    ]

    passed = 0
    for i, (height, expected) in enumerate(tests):
        result = func(height[:])
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"height={height} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def maxArea(height: List[int]) -> int:
    """
    Find two lines forming a container with max water.

    Opposite-end two pointers. Each step: compute
    min(h[L], h[R]) * (R-L), update max. Move the
    shorter pointer inward — it's the only side that
    might yield a larger area. Stop when L >= R.

    Time:  O(n) — each pointer moves at most n steps
    Space: O(1) — two pointers, one max variable
    """
    res = 0
    l, r = 0, len(height)-1
    while l < r:
        area = (r-l) * min (height[l], height[r])
        res = max(res, area)
        if height[l] < height[r]:
            l+=1
        else:
            r-=1
    return res
r'''
49
1
16
4
Test 1: PASSED | height=[1, 8, 6, 2, 5, 4, 8, 3, 7] | expected=49 | got=49
Test 2: PASSED | height=[1, 1] | expected=1 | got=1
Test 3: PASSED | height=[4, 3, 2, 1, 4] | expected=16 | got=16
Test 4: PASSED | height=[1, 2, 1] | expected=2 | got=2
Test 5: PASSED | height=[1, 2, 4, 3] | expected=4 | got=4
Test 6: PASSED | height=[2, 3, 4, 5, 18, 17, 6] | expected=17 | got=17
Test 7: PASSED | height=[1, 0, 0, 0, 1] | expected=4 | got=4
Test 8: PASSED | height=[3, 1, 2, 4, 0, 1, 3, 2] | expected=18 | got=18

8/8 tests passed

'''

# Quick debug — run this cell while building
print(maxArea([1,8,6,2,5,4,8,3,7]))  # 49
print(maxArea([1,1]))                 # 1
print(maxArea([4,3,2,1,4]))           # 16
print(maxArea([1,0,0,0,1]))           # 4
test_harness(maxArea)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check all pairs | O(n²) | O(1) |
| Greedy two pointers | O(n) | O(1) |

The greedy choice — always move the shorter pointer
— is provably optimal: no skipped pair could have
beaten the current maximum.

## Real World Connection

At Citi, each server cluster has a maximum ingest
capacity — the bottleneck is the weakest node, just
like the shorter bar is the water ceiling.
The capacity planner needs to find the two time
windows where the combined available headroom (width
× minimum capacity) is maximised for scheduling a
heavy ETL batch.
The greedy pointer strategy maps directly: always
explore the window with the lower capacity ceiling
first — the higher-capacity window is already doing
its best at the current width.
