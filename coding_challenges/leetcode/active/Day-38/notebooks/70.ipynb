<strong>Core Insight:</strong> To reach step <code>i</code>, you came

exactly <code>dp[i-1] + dp[i-2]</code> — this is the Fibonacci
sequence in disguise.

## Official Problem Statement

You are climbing a staircase. It takes `n` steps to reach the top.
Each time you can either climb `1` or `2` steps. In how many
distinct ways can you climb to the top?

**Constraints:**
- `1 <= n <= 45`

## What This Is Actually Asking

Count the number of distinct sequences of 1s and 2s that sum to `n`.
Each sequence represents a unique path to the top of the staircase.
The answer only depends on the two previous steps, so we never need
to store the full history — just two values at a time.
This is the classic Fibonacci structure: each state builds on the
previous two.

## Walk Through an Example by Hand

**Input:** `n = 5`

Base cases:
- `dp[1] = 1` (only one way: step 1)
- `dp[2] = 2` (ways: [1,1] or [2])

Fill the rest:
- `dp[3] = dp[2] + dp[1] = 2 + 1 = 3`
  - Paths: [1,1,1], [1,2], [2,1]
- `dp[4] = dp[3] + dp[2] = 3 + 2 = 5`
  - Paths: [1,1,1,1],[1,1,2],[1,2,1],[2,1,1],[2,2]
- `dp[5] = dp[4] + dp[3] = 5 + 3 = 8`

**Answer:** `8`

## The Picture

```
n =  1   2   3   4   5   6   7
    ----+---+---+---+---+---+---
dp=  1   2   3   5   8  13  21
    ----+---+---+---+---+---+---
         ^   ^^
         |   |+-- dp[3] = dp[2]+dp[1] = 2+1 = 3
         |+------ dp[2] = 2 (base case)

Rolling variables (space optimised):

  step:  a=1  b=2
  i=3:   a=2  b=3   (a,b = b, a+b)
  i=4:   a=3  b=5
  i=5:   a=5  b=8   <-- return b
```

## When To Use This Pattern

- When each state depends on a **fixed number of prior states**,
  think rolling-variable DP.
- When the recurrence is `f(n) = f(n-1) + f(n-2)`, think Fibonacci.
- When choices at each step are small and independent, think 1D DP.
- When `n` is small (≤ 45) and you see "how many ways",
  think bottom-up DP.
- When space matters and lookback window is fixed,
  think two-variable optimisation.

## The Approach

Start with two base cases: one way to reach step 1, two ways to
reach step 2. For every step from 3 to n, the number of ways equals
the sum of the previous two values. Instead of an array, keep two
rolling variables `a` and `b` to use O(1) space. Return `b` which
holds the answer for step `n`.

````python
def test_harness(func):
    """Run test cases for climbStairs."""
    cases = [
        # (input_n, expected)
        (1,  1),   # base case
        (2,  2),   # base case
        (3,  3),
        (4,  5),
        (5,  8),
        (10, 89),
        (45, 1836311903),  # largest allowed n
    ]
    passed = 0
    for n, expected in cases:
        result = func(n)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        else:
            print(
                f"  {status}: n={n} => got {result}, "
                f"expected {expected}"
            )
    total = len(cases)
    print(f"\nSummary: {passed}/{total} tests passed.")
    if passed == total:
        print("All tests PASSED!")
````

````python
def climbStairs(n: int) -> int:
    """
    Count distinct ways to climb n stairs taking 1 or 2 steps.

    Strategy:
        dp[i] = dp[i-1] + dp[i-2]
        Use two rolling variables a, b for O(1) space.

    Args:
        n: Number of stairs (1 <= n <= 45)

    Returns:
        Number of distinct ways to reach the top.

    Examples:
        >>> climbStairs(2)
        2
        >>> climbStairs(3)
        3
    """
    pprev, prev, res  = 1, 2 ,  0
    if n ==2 : return prev
    if n == 1: return pprev

    for i in range (3, n+1):
        res = prev + pprev
        pprev = prev
        prev = res
    return res
        
    # Debug: show input
    print(f"[DEBUG] n = {n}")

    # Debug: base case check
    if n <= 2:
        print(f"[DEBUG] base case => return {n}")
        pass

    # Debug: rolling variables before loop
    print("[DEBUG] a=1, b=2, loop from 3 to n")

    # Debug: first iteration placeholder
    print("[DEBUG] each step: a, b = b, a+b")

    # Debug: expected return
    print("[DEBUG] return b after loop")

test_harness(climbStairs)
````

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Naive recursion | O(2^n) | O(n) | Exponential, TLE |
| Memoisation | O(n) | O(n) | Top-down, memo dict |
| Bottom-up DP array | O(n) | O(n) | dp array size n+1 |
| **Rolling variables** | **O(n)** | **O(1)** | **Optimal** |
