## Official Problem Statement

Given an array of integers `nums` and an integer `k`, return the total
number of subarrays whose sum equals to `k`.

A subarray is a contiguous non-empty sequence of elements within
an array.

**Constraints:**
- `1 <= nums.length <= 2 * 10^4`
- `-1000 <= nums[i] <= 1000`
- `-10^7 <= k <= 10^7`

## What This Is Actually Asking

You have a list of numbers. Find every stretch of consecutive
numbers that adds up to k.

A stretch can start and end anywhere. It just has to be
consecutive — no skipping.

Count how many such stretches exist. That's the answer.

## Walk Through an Example by Hand

Input: nums = [1, 1, 1], k = 2

We keep a running total called prefix_sum.
We also keep a map of how many times each prefix_sum appeared.
Start: map = {0: 1}, prefix_sum = 0, count = 0

Step 1 — num = 1:
  prefix_sum = 0 + 1 = 1
  Look for (1 - 2) = -1 in map → not found
  map = {0:1, 1:1}, count = 0

Step 2 — num = 1:
  prefix_sum = 1 + 1 = 2
  Look for (2 - 2) = 0 in map → found 1 time
  count = 0 + 1 = 1
  map = {0:1, 1:1, 2:1}

Step 3 — num = 1:
  prefix_sum = 2 + 1 = 3
  Look for (3 - 2) = 1 in map → found 1 time
  count = 1 + 1 = 2
  map = {0:1, 1:1, 2:1, 3:1}

Answer: 2  (subarrays [1,1] at index 0-1 and 1-2)

## The Picture

Think of prefix sums as mile markers on a road.

```
nums:        [ 1,  1,  1 ]
index:          0   1   2

prefix:   0 -> 1 -> 2 -> 3
marker:   P0   P1   P2   P3
```

Distance from P0 to P2 = 2 - 0 = 2  --> k=2 HIT
Distance from P1 to P3 = 3 - 1 = 2  --> k=2 HIT

Each time you reach a new marker, ask:
  "How many earlier markers are exactly k miles behind me?"
That count is how many subarrays end here and sum to k.

## When To Use This Pattern

- When asked to count subarrays with a target sum
- When the array has negative numbers (sliding window breaks)
- When you need cumulative totals across a sequence
- When "how many stretches sum to X" appears in the prompt

## The Approach

Walk through the array once, keeping a running total.
At each step, ask: how many times have I seen
(running_total - k) before?
Each such occurrence means a valid subarray ends here.
Store every running total in a map so lookups are instant.

## Complexity

| Approach          | Time          | Space  |
|-------------------|---------------|--------|
| Brute Force       | O(n^2)        | O(1)   |
| Prefix Sum + Map  | O(n)          | O(n)   |

The brute force checks every pair of start/end indices.
The HashMap lets us skip all that by turning "did I see
this prefix before?" into an O(1) lookup.
