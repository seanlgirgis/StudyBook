## Official Problem Statement

Given an array of strings `strs`, group the anagrams together.
You can return the answer in any order.

An anagram is a word or phrase formed by rearranging the letters
of a different word or phrase, typically using all the original
letters exactly once.

**Constraints:**
- `1 <= strs.length <= 10^4`
- `0 <= strs[i].length <= 100`
- `strs[i]` consists of lowercase English letters

## What This Is Actually Asking

You have a list of words. Some words use the exact same letters
just in a different order — those are anagrams of each other.

Put all anagrams into the same bucket. Return all the buckets.

"eat", "tea", "ate" all go in one bucket. "tan", "nat" in
another. A word with no matches gets its own bucket.

## Walk Through an Example by Hand

Input: ["eat", "tea", "tan", "ate", "nat", "bat"]

Sort each word to find its key:
  "eat" -> sorted = "aet"  -> map["aet"] = ["eat"]
  "tea" -> sorted = "aet"  -> map["aet"] = ["eat","tea"]
  "tan" -> sorted = "ant"  -> map["ant"] = ["tan"]
  "ate" -> sorted = "aet"  -> map["aet"] = ["eat","tea","ate"]
  "nat" -> sorted = "ant"  -> map["ant"] = ["tan","nat"]
  "bat" -> sorted = "abt"  -> map["abt"] = ["bat"]

Final map values:
  [["eat","tea","ate"], ["tan","nat"], ["bat"]]

Answer: 3 groups.

## The Picture

Think of each word as a bag of Scrabble tiles.
Two bags are the same if they hold the exact same tiles.
Sorting the tiles gives you a "fingerprint" for the bag.

```
 WORD     SORT     FINGERPRINT
 ----     ----     -----------
 eat   -> aet  -+
 tea   -> aet  -+-> GROUP 1: [eat, tea, ate]
 ate   -> aet  -+

 tan   -> ant  -+
 nat   -> ant  -+-> GROUP 2: [tan, nat]

 bat   -> abt  ---> GROUP 3: [bat]
```

Same fingerprint = same group. One pass through the list.

## When To Use This Pattern

- When grouping strings by shared letter content
- When two things are "the same" but look different
- When you need a canonical form to use as a key
- When the prompt says "group" or "cluster" by equivalence

## The Approach

For each word, sort its letters to get a fingerprint.
Use that fingerprint as a key in a HashMap.
Append the original word to the list at that key.
At the end, return all the lists stored in the map.

## Complexity

| Approach              | Time             | Space    |
|-----------------------|------------------|----------|
| Brute Force (compare) | O(n^2 * k)       | O(n * k) |
| Sort + HashMap        | O(n * k * log k) | O(n * k) |

Brute force compares every pair of words — slow at scale.
Sorting once per word and grouping by key is far cheaper
because k (word length) is bounded and small in practice.

## Real World Connection
