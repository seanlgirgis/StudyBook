## Official Problem Statement

Given an array of integers `nums` and an integer `target`,
return *indices* of the two numbers such that they add up
to `target`.

You may assume that each input would have **exactly one
solution**, and you may not use the same element twice.
You can return the answer in any order.

**Constraints:**
- `2 <= nums.length <= 10^4`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- Only one valid answer exists.

## What This Is Actually Asking

Find two positions in the array whose values sum to
the target — return the positions, not the values.
The brute-force checks every pair (O(n²)), but we can
do better by remembering what we have already seen.
A hash map lets us answer "have I seen target-num
before?" in O(1) time.
The key trade-off is space for time: one extra dict
buys us a single linear pass.

## Walk Through an Example by Hand

```
nums = [2, 7, 11, 15]   target = 9

i=0  num=2   complement=7   map={}        -> 7 not in map
             map becomes {2:0}

i=1  num=7   complement=2   map={2:0}     -> 2 IS in map!
             return [map[2], 1] = [0, 1]
```

Second example:
```
nums = [3, 2, 4]   target = 6

i=0  num=3   complement=3   map={}        -> not in map
             map = {3:0}

             map = {3:0, 2:1}

i=2  num=4   complement=2   map={3:0,2:1} -> 2 IS in map!
             return [map[2], 2] = [1, 2]
```

## The Picture

```
nums  = [ 2,  7, 11, 15 ]    target = 9
index =   0   1   2   3

Scan ->

  Step 1:  num=2
           need = 9 - 2 = 7
           HashMap: { }   <- 7 not here yet
           Store: {2: 0}

  Step 2:  num=7
           need = 9 - 7 = 2
           HashMap: {2: 0}  <- FOUND at index 0!
           Return [0, 1]  ✓

  HashMap acts as a "memory" of past numbers:

  num=X arrives
       |
       v
   is (target-X) in map? --YES--> return [map[target-X], i]
       |
       NO
       |
       v
   store map[X] = i
```

## When To Use This Pattern

- When you need to find **two elements** that satisfy a
  condition, think **complement + hash map**.
- When brute force is O(n²) pair-checking, think
  **one-pass hash map** to reach O(n).
- When a problem asks "does X exist in what I've seen?",
  think **set or dict lookup**.
- When indices (not values) are the answer, think
  **map value → index** rather than just a set.
- When the array is unsorted and sorting would cost O(n
  log n), think **hash map** over two-pointer.

## The Approach

Iterate through `nums` once, keeping a dictionary that
maps each value to its index. At each step, compute the
complement `target - num` and check if it already lives
in the dictionary. If yes, we found our pair and return
both indices immediately. If no, store the current number
and index for future lookups.

````python
from typing import List
from collections import defaultdict, deque
````

````python
def test_harness(func):
    cases = [
        # (nums, target, expected)
        ([2, 7, 11, 15], 9,  [0, 1]),
        ([3, 2, 4],       6,  [1, 2]),
        ([3, 3],          6,  [0, 1]),
        ([-1, -2, -3, -4, -5], -8, [2, 4]),
        ([0, 4, 3, 0],   0,  [0, 3]),
    ]
    passed = 0
    for nums, target, expected in cases:
        result = func(nums, target)
        # order-independent check
        ok = sorted(result) == sorted(expected)
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        else:
            print(
                f"{status} | nums={nums} target={target}"
                f" | got={result} expected={expected}"
            )
    print(
        f"\nResults: {passed}/{len(cases)} passed"
    )
    if passed == len(cases):
        print("All tests PASSED!")
````

````python
def two_sum(nums: List[int], target: int) -> List[int]:
    """
    Return indices of two numbers that sum to target.

    Strategy: one-pass hash map.
      - For each num at index i:
          complement = target - num
          if complement in seen -> return [seen[complement], i]
          else seen[num] = i

    Args:
        nums:   list of integers
        target: integer target sum

    Returns:
        list of two indices [i, j] where nums[i]+nums[j]==target
    """
    # Debug: inspect inputs
    print(f"[DEBUG] nums={nums}, target={target}")

    seen = {}  # value -> index

    for i, num in enumerate(nums):
        complement = target - num
        print(f"[DEBUG] i={i} num={num} "
              f"complement={complement} seen={seen}")

        if complement in seen:
            print(f"[DEBUG] Found! indices "
                  f"[{seen[complement]}, {i}]")
            return [seen[complement], i]

        seen[num] = i

    return []  # problem guarantees one solution exists


pass
````
