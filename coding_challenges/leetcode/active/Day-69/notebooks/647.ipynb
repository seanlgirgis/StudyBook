## Official Problem Statement

Given a string `s`, return *the number of **palindromic substrings** in it*.

A string is a palindrome when it reads the same backward as forward.

A **substring** is a contiguous sequence of characters within the string.

**Constraints:**
- `1 <= s.length <= 1000`
- `s` consists of lowercase English letters.

## What This Is Actually Asking

Count every substring that is a palindrome — including single characters.
Single characters always count (each is trivially a palindrome).
We need to count overlapping palindromes too, e.g. in `aaa` we count
`a`, `a`, `a`, `aa`, `aa`, and `aaa` — total 6.
Brute force is O(n³); expanding from centers is O(n²) and very clean.

## Walk Through an Example by Hand

Input: `s = "abc"`

```
Center i=0 ('a'):
  Odd  expand (0,0): s[0]==s[0] -> 'a'  count=1
  Even expand (0,1): s[0]='a' != s[1]='b' -> stop

Center i=1 ('b'):
  Odd  expand (1,1): 'b'=='b' -> count=2
  Even expand (1,2): s[1]='b' != s[2]='c' -> stop

Center i=2 ('c'):
  Odd  expand (2,2): 'c'=='c' -> count=3
  Even expand (2,3): right out of bounds -> stop

Result: 3
```

Input: `s = "aaa"`
```
i=0: odd(0,0)->1, even(0,1)->'a'=='a'->2
i=1: odd(1,1)->3, even(1,2)->'a'=='a'->4
       odd expands: (0,2)->'a'=='a'->5
i=2: odd(2,2)->6
Result: 6
```

## The Picture

```
s = "racecar"
     0123456

Odd center at i=3 ('e'):
  L=3, R=3  -> 'e'=='e'  +1  [e]
  L=2, R=4  -> 'c'=='c'  +1  [cec]
  L=1, R=5  -> 'a'=='a'  +1  [aceca]
  L=0, R=6  -> 'r'=='r'  +1  [racecar]
  L=-1 stop

Even center between i=3,4 ('e','c'):
  L=3, R=4  -> 'e' != 'c'  stop

Expand pattern:

  L <--- center ---> R
  |                  |
  v                  v
[ r  a  c  e  c  a  r ]
        ^--^  match!
     ^--------^  match!
  ^--------------^  match!

Every matching pair = +1 palindrome found.
```

## When To Use This Pattern

- When asked to **count or find palindromic substrings**, think
  **expand around center**.
- When the problem involves symmetry around a middle index, think
  **two-pointer from center**.
- When overlap between substrings matters, think **center expansion
  not sliding window**.
- When O(n²) is acceptable and you want to avoid a full DP table,
  think **center expansion** (space O(1) vs O(n²) for DP table).
- When you see both odd-length and even-length palindromes needed,
  think **two separate expansions per index**.

## The Approach

For every index `i`, treat it as the center of an odd-length palindrome
and expand left/right while characters match, incrementing count each
time. Then treat the gap between `i` and `i+1` as the center of an
even-length palindrome and do the same. A helper `expand(l, r)` handles
both cases cleanly: while `l >= 0`, `r < n`, and `s[l] == s[r]`,
add 1 to count, move `l` left and `r` right.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute Force (check all substrings) | O(n³) | O(1) | Too slow n=1000 |
| DP table (dp[i][j]=palindrome) | O(n²) | O(n²) | Table stores all |
| **Expand Around Center** | **O(n²)** | **O(1)** | **Optimal in space** |
| Manacher's Algorithm | O(n) | O(n) | Complex, rarely needed |

- **n** = length of string `s`
- Expand-around-center does 2n expansions, each up to O(n) steps.
