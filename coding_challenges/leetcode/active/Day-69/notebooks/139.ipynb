## Official Problem Statement

Given a string `s` and a dictionary of strings `wordDict`, return
`true` if `s` can be segmented into a space-separated sequence of
one or more dictionary words.

**Note:** The same word in the dictionary may be reused multiple
times in the segmentation.

**Constraints:**
- `1 <= s.length <= 300`
- `1 <= wordDict.length <= 1000`
- `1 <= wordDict[i].length <= 20`
- `s` and `wordDict[i]` consist of lowercase English letters.
- All strings in `wordDict` are **unique**.

## What This Is Actually Asking

Can we slice the string into pieces where every piece is in the
dictionary? Words can repeat and order matters.
We don't need to find the actual segmentation — just a yes/no answer.
This is a classic reachability problem: can we "reach" the end of
the string by hopping through valid dictionary words?
DP turns overlapping subproblems into a single O(n²) pass.

## When To Use This Pattern

- When asked if a string **can be fully segmented** into valid tokens,
  think **1D DP over string prefixes**.
- When a problem has overlapping sub-decisions about prefixes/suffixes,
  think **DP with a set for O(1) lookups**.
- When words can be reused and you need reachability, think
  **forward DP from position 0 to n**.
- When the dictionary is large, think **convert to set first** to
  avoid O(k) lookup cost inside the double loop.
- When you see "can be segmented / split / partitioned", think
  **reachability DP** similar to Jump Game.

## The Approach

Convert `wordDict` to a set for O(1) lookups. Create a boolean array
`dp` of length `n+1` with `dp[0] = True` (empty string is always
valid). For each end position `i` from 1 to n, scan all split points
`j` from 0 to i: if `dp[j]` is True and `s[j:i]` is in the word set,
set `dp[i] = True` and break early. Return `dp[n]`.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
| Brute Force (recursion, no memo) | O(2ⁿ) | O(n) | Exponential |
| Recursion + Memoization | O(n²) | O(n) | Top-down DP |
| **Bottom-up DP (1D array)** | **O(n²)** | **O(n)** | **Optimal** |
| DP + word-length pruning | O(n·L) | O(n) | L=max word len |

- **n** = length of string `s`
- Substring slicing `s[j:i]` is O(n) in Python, making naive
  inner loop O(n³); using a set lookup and careful slicing keeps
  it O(n²) in practice for the given constraints.
