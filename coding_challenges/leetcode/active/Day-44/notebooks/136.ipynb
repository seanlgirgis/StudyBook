## Official Problem Statement

Given a **non-empty** array of integers `nums`, every element appears
**twice** except for one. Find that single one.

You must implement a solution with **linear runtime complexity** and
use only **constant extra space**.

**Constraints:**
- `1 <= nums.length <= 3 * 10^4`
- `-3 * 10^4 <= nums[i] <= 3 * 10^4`
- Each element in the array appears twice except for exactly one
  element which appears only once.

## What This Is Actually Asking

You have a list where every number shows up exactly twice — except
one number, which shows up once. Find that odd-one-out number.

A naive approach (HashSet or sort) works but uses O(n) space or
O(n log n) time. The constraint forces you toward a bit trick.

XOR is the key: two identical numbers XOR to zero. So XOR-ing the
entire array eliminates all pairs, leaving only the single number.

The order of XOR doesn't matter — it's commutative and associative,
so pairs cancel regardless of position in the array.

## Walk Through an Example by Hand

`nums = [4, 1, 2, 1, 2]`

```
Step 1: result = 0

Step 2: result ^ 4
  0000
^ 0100
------
  0100  → result = 4

Step 3: result ^ 1
  0100
^ 0001
------
  0101  → result = 5

Step 4: result ^ 2
  0101
^ 0010
------
  0111  → result = 7

Step 5: result ^ 1  (pair with step 3's 1)
  0111
^ 0001
------
  0110  → result = 6

Step 6: result ^ 2  (pair with step 4's 2)
  0110
^ 0010
------
  0100  → result = 4  ✓
```

Pairs (1,1) and (2,2) cancelled. Only 4 survives.

## The Picture

```
nums = [4,  1,  2,  1,  2]

Binary representation:
  4 = 100
  1 = 001
  2 = 010
  1 = 001
  2 = 010

XOR column-by-column:
  Bit 2:  1^0^0^0^0 = 1
  Bit 1:  0^0^1^0^1 = 0
  Bit 0:  0^1^0^1^0 = 0

Result = 100 = 4  ✓

Key properties used:
  a ^ a = 0   (same number cancels)
  a ^ 0 = a   (zero is identity)
  XOR is commutative: a^b = b^a
  XOR is associative: (a^b)^c = a^(b^c)

So: 4^1^2^1^2
  = 4 ^ (1^1) ^ (2^2)
  = 4 ^ 0 ^ 0
  = 4
```

## When To Use This Pattern

- When you need to find a **unique element among pairs**, think XOR.
- When a problem asks for **O(1) space and O(n) time** with integers,
  think bitwise tricks.
- When you see **"every element appears twice except one"**, think
  `result ^= n`.
- When you need to **toggle flags** without extra state, think XOR.
- When you see **symmetric difference** of two sets, think XOR on
  combined elements.

## The Approach

Initialize a result variable to zero. Iterate through every number
in the array and XOR it into result. Because XOR is commutative and
associative, all pairs cancel each other out (producing 0) and the
single number XOR-ed with 0 returns itself. Return result.

````python
from typing import List
````

````python
def test_harness(func):
    """Run test cases for single_number."""
    cases = [
        # (input,          expected, label)
        ([2, 2, 1],        1,        "basic-3"),
        ([4, 1, 2, 1, 2],  4,        "basic-5"),
        ([1],              1,        "single-elem"),
        ([0, 0, 5],        5,        "zero-pair"),
        ([-1, -1, 7],      7,        "neg-pair"),
        ([99],             99,       "single-large"),
        ([3, 3, 7, 7, 9],  9,        "end-single"),
    ]
    passed = 0
    for nums, expected, label in cases:
        got = func(nums)
        status = "PASSED" if got == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        print(
            f"[{status}] {label}: "
            f"got={got}, expected={expected}"
        )
    total = len(cases)
    print(f"\nResult: {passed}/{total} passed")
````

````python
def single_number(nums: List[int]) -> int:
    """
    Find the element that appears only once.

    Strategy: XOR all numbers together.
      - Pairs cancel: a ^ a = 0
      - Identity:     0 ^ a = a
      - Result is the lone survivor.

    Args:
        nums: List where every elem appears twice
              except one.

    Returns:
        The element that appears exactly once.

    Time:  O(n)
    Space: O(1)
    """
    # Debug: show input
    print(f"Input: {nums}")
    res = 0 
    for num in nums:
        res ^= num
    return res

    # Debug: trace XOR accumulation
    result = 0
    for n in nums:
        before = result
        result ^= n
        print(f"  {before} ^ {n} = {result}")

    # Debug: final result
    print(f"Single number: {result}")

    pass
test_harness(single_number)
````

## Complexity

| Approach     | Time     | Space  | Notes                        |
|:-------------|:---------|:-------|:-----------------------------|
| Brute force  | O(n²)    | O(1)   | Check each elem vs all others|
| Hash set     | O(n)     | O(n)   | Store/remove seen elements   |
| Sort + scan  | O(n lgn) | O(1)   | Adjacent pairs differ        |
| **XOR**      | **O(n)** | **O(1)**| **Optimal — bit trick**     |

## Real World Connection

**Citi / Finance:** Transaction reconciliation systems compare debit
and credit entries; XOR-based checksums quickly flag unmatched
single-sided transactions in high-volume settlement pipelines.

**AWS / Data Engineering:** Network packet deduplication at scale
uses XOR checksums. An odd packet without its pair indicates loss
or corruption — detected in O(n) time without buffering the stream.

**General DE:** ETL pipelines validate symmetric data loads (e.g.,
source rows XOR'd with target rows) to detect missing records with
constant memory, even across billions of rows.
