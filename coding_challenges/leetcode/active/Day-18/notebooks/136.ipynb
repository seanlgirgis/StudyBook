<strong>Core Insight:</strong> XOR a number with itself and you get 0.

## Official Problem Statement

Given a non-empty array of integers `nums`, every element appears
twice except for one. Find that single one.

You must implement a solution with linear runtime complexity
and use only constant extra space.

**Constraints:**
- `1 <= nums.length <= 3 * 10^4`
- `-3 * 10^4 <= nums[i] <= 3 * 10^4`
- Every element appears exactly twice except for one element
  which appears exactly once

## What This Is Actually Asking

You have a list where everything has a twin — except one value.
Find the loner.

The brute force (hash map counting) works but uses O(n) space.
The elegant solution uses XOR and zero extra memory.

Think of it as: pairs "annihilate" each other, the single one
is left standing.

## Walk Through an Example by Hand

Input: `[4, 1, 2, 1, 2]`

XOR all values left to right:
```
result = 0

0 XOR 4:  0000 XOR 0100 = 0100  (4)
4 XOR 1:  0100 XOR 0001 = 0101  (5)
5 XOR 2:  0101 XOR 0010 = 0111  (7)
7 XOR 1:  0111 XOR 0001 = 0110  (6)
6 XOR 2:  0110 XOR 0010 = 0100  (4)
```
**Answer: 4**

Why it works:
- 1 XOR 1 = 0  (pair cancelled)
- 2 XOR 2 = 0  (pair cancelled)
- 0 XOR 4 = 4  (loner remains)

## The Picture

```
XOR truth table (bit by bit):
  0 XOR 0 = 0
  1 XOR 1 = 0   ← same bits cancel
  0 XOR 1 = 1   ← different bits survive
  1 XOR 0 = 1

Visualize [4, 1, 2, 1, 2]:

   4 = 0100
   1 = 0001  ─┐ pair cancels
   2 = 0010  ─┐ pair cancels
   1 = 0001  ─┘
   2 = 0010  ─┘

XOR all:
  0100
  0001
  0010
  0001  ← cancels first 0001
  0010  ← cancels first 0010
  ----
  0100  ← only 4 is left

Order does not matter — XOR is commutative and associative.
```

## When To Use This Pattern

- When you need to find the **odd one out** in a list of pairs.
- When the problem says **O(1) space** and all duplicates pair up.
- When two XOR problems stack: find 2 unique numbers
  (use XOR + bit partitioning).
- When you see "every element appears k times except one"
  — XOR is the go-to for k=2.

## The Approach

Initialize result to 0. XOR every number in the array into result.
Because XOR is associative and commutative, all duplicate pairs
cancel each other out (a ^ a = 0) and the lone number is XOR'd
with 0, which returns it unchanged. Return result.

````python
from typing import List  # standard collection types
````

````python
def test_harness(func):
    tests = [
        # (input, expected, label)
        ([2, 2, 1],          1, "basic: 1 is single"),
        ([4, 1, 2, 1, 2],    4, "standard: 4 is single"),
        ([1],                1, "edge: single element"),
        ([0, 0, 99],        99, "includes zero pair"),
        ([-1, -1, -2],      -2, "negative numbers"),
    ]
    passed = 0
    for nums, expected, label in tests:
        result = func(nums)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        print(f"  [{status}] {label}")
        if status == "FAILED":
            print(f"           got={result}, expected={expected}")
    print(f"\n  {passed}/{len(tests)} tests passed")
````

````python
def singleNumber(nums: List[int]) -> int:
    """
    Find the one element that appears exactly once;
    all others appear exactly twice.

    Approach: XOR all elements.
      a ^ a = 0 for every pair.
      0 ^ x = x for the single element.
      Result is the unique number.

    Time:  O(n)
    Space: O(1)
    """
    res = 0 
    for num in nums:
        res ^= num
    return res



# Debug prints — expected values shown in comments
print(singleNumber([2, 2, 1]))         # expected: 1
print(singleNumber([4, 1, 2, 1, 2]))   # expected: 4
print(singleNumber([1]))               # expected: 1
print(singleNumber([-1, -1, -2]))      # expected: -2
test_harness (singleNumber)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Hash map (count each element) | O(n) | O(n) |
| Sorting + scan for non-pair | O(n log n) | O(1) |
| XOR all elements | O(n) | O(1) |
