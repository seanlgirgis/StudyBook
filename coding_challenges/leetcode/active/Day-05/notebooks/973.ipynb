# LC 973 — K Closest Points to Origin
**Difficulty:** Medium &nbsp;|&nbsp; **Category:** Heap
**Pattern:** Max-Heap of Size K — K Closest by Distance

<strong>Core Insight:</strong> Keep a max-heap of

## Official Problem Statement

Given an array of `points` where
`points[i] = [xi, yi]` represents a point on the
**X-Y plane** and an integer `k`, return the `k`
closest points to the origin `(0, 0)`.

The distance between two points on the X-Y plane
is the Euclidean distance
(i.e., `sqrt(x^2 + y^2)`).

You may return the answer in **any order**. The
answer is **guaranteed** to be unique (except for
the order that it is in).

**Example 1:**
```
Input:  points = [[1,3],[-2,2]], k = 1
Output: [[-2,2]]
```
**Example 2:**
```
Input:  points = [[3,3],[5,-1],[-2,4]], k = 2
Output: [[3,3],[-2,4]]
```

**Constraints:**
- `1 <= k <= points.length <= 10^4`
- `-10^4 <= xi, yi <= 10^4`

## What This Is Actually Asking

You have a bunch of points on a map. The origin
is the centre. Find the k points that are nearest
to the centre. Order does not matter. You do not
need the actual distance — just the k nearest
points themselves.

## Walk Through an Example by Hand

```
points = [[3,3],[5,-1],[-2,4]]   k=2

Squared distances (no need for sqrt):
  [3,3]  -> 3²+3² = 18
  [5,-1] -> 5²+1² = 26
  [-2,4] -> 2²+4² = 20

Max-heap of size 2 (store negative dist for Python):

  Point [3,3]  dist=18  heap empty -> push (-18,[3,3])
  heap = [(-18,[3,3])]

  Point [5,-1] dist=26  push (-26,[5,-1])
  heap = [(-26,[5,-1]),(-18,[3,3])]  size=2 ok

  Point [-2,4] dist=20  push (-20,[-2,4])
  heap size=3 > k=2
  pop max (most negative = farthest) = (-26,[5,-1])
  heap = [(-20,[-2,4]),(-18,[3,3])]

Return remaining points: [[-2,4],[3,3]]
```

## The Picture

```
          y
     4    *  <- [-2,4]  dist²=20
     3       *  <- [3,3]   dist²=18
     2
     1
  ---+-------------- x
    -1         *  <- [5,-1] dist²=26

k=2: keep the 2 closest -> [3,3] and [-2,4]
     evict [5,-1] (farthest)

Max-heap of size k — always evict the FARTHEST:

  heap (max by dist):  [26:[5,-1], 20:[-2,4], 18:[3,3]]
                        ^
                        pop this when size > k

Why max-heap (not min)?
  We want to KICK OUT the farthest point.
  A max-heap surfaces the largest distance at top,
  so we can evict it in O(log k).

Squared distance trick:
  sqrt(x²+y²) vs sqrt(a²+b²) comparison
  == x²+y² vs a²+b² comparison  (sqrt preserves order)
  Skip sqrt — saves time and floating-point errors.
```

## When To Use This Pattern

- When you need the **k closest (or smallest)**
  items, think **max-heap of size k — evict the
  farthest/largest when full**
- When comparing distances, think
  **use squared distance — skip sqrt**
- When Python heapq is min-heap only, think
  **negate the key to simulate a max-heap**
- When order of output does not matter, think
  **return remaining heap contents directly**

## The Approach

For each point compute its squared distance from
the origin and push a (negative distance, point)
tuple onto a max-heap.
Whenever the heap grows beyond k elements, pop the
top — which is the farthest point seen so far.
After all points are processed, extract the point
coordinates from the heap and return them.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Sort all by distance | O(n log n) | O(n) |
| Max-heap of size k | O(n log k) | O(k) |
| QuickSelect | O(n) avg | O(1) |

The heap approach beats sorting when k << n.
QuickSelect is asymptotically better but harder
to implement correctly in an interview.
