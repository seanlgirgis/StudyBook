## Official Problem Statement

Given an integer array `nums` and an integer `k`,
return the `k`th largest element in the array.

Note that it is the `k`th largest element in the
sorted order, not the `k`th distinct element.

Can you solve it without sorting?

**Example 1:**
```
Input:  nums = [3,2,1,5,6,4], k = 2
Output: 5
```
**Example 2:**
```
Input:  nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
```

**Constraints:**
- `1 <= k <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

## What This Is Actually Asking

If you sorted the array from largest to smallest,
what value would be at position k?
For example k=2 means the second-largest number.
Duplicates count — [5,5,4] has 5 as both 1st and
2nd largest, and 4 as 3rd.

## Walk Through an Example by Hand

```
nums = [3, 2, 1, 5, 6, 4]   k=2

Min-heap approach — keep top 2:

  push 3  heap=[3]        size 1 <= k=2
  push 2  heap=[2,3]      size 2 == k, no pop
  push 1  heap=[1,3,2]    size 3 > k=2 -> pop min=1
          heap=[2,3]
  push 5  heap=[2,3,5]    size 3 > k=2 -> pop min=2
          heap=[3,5]
  push 6  heap=[3,5,6]    size 3 > k=2 -> pop min=3
          heap=[5,6]
  push 4  heap=[4,6,5]    size 3 > k=2 -> pop min=4
          heap=[5,6]

heap[0] = 5 = 2nd largest
Answer: 5

Sorted descending: [6, 5, 4, 3, 2, 1]
Position 2:              ^
```

## The Picture

```
nums = [3, 2, 1, 5, 6, 4]   k=2

Think of the heap as a VIP bouncer:
  Only the top-k get in.
  Everyone smaller is turned away (popped).

Final heap = [5, 6]  (the 2 largest)
              ^
              heap minimum = 2nd largest = answer

Why min at the top?
  The min of the top-k bucket is the kth largest.
  Anything smaller than the min was already kicked
  out — it can't be in the top k.

Two approaches:

  1. Min-heap of size k    O(n log k)  O(k)
     Scan all nums, push each, pop if > k.
     Return heap[0].

  2. heapq.nlargest(k, nums)  O(n log k)  O(k)
     Python built-in — same heap internally.
     Return nlargest[k-1] (last of the top-k).
```

## When To Use This Pattern

- When asked for **kth largest in an array**, think
  **min-heap of size k — O(n log k)**
- When k is small relative to n, think
  **heap beats full sort (O(n log k) < O(n log n))**
- When k equals n, think
  **just sort — heap offers no advantage**
- When the interviewer hints at O(n), think
  **QuickSelect — same idea as QuickSort pivot**

## The Approach

Push every number into a min-heap one by one.
After each push, if the heap has grown beyond k
elements, pop the smallest — that number cannot
be in the top k.
After the full scan the heap holds exactly the k
largest numbers seen, and heap[0] (the minimum of
those k) is the kth largest overall.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Sort descending, index k-1 | O(n log n) | O(1) |
| Min-heap of size k | O(n log k) | O(k) |
| QuickSelect (avg) | O(n) avg | O(1) |

The heap is the clean interview answer: guaranteed
O(n log k) with no edge-case risk. QuickSelect
is faster on average but O(n²) worst case without
randomisation.
