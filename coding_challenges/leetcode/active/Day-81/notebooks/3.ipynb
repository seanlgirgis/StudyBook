## Official Problem Statement

Given a string `s`, find the length of the **longest substring**
without repeating characters.

**Constraints:**
- `0 <= s.length <= 5 * 10^4`
- `s` consists of English letters, digits, symbols and spaces.

## What This Is Actually Asking

You have a string of characters — letters, numbers, spaces.
Find the longest stretch of characters where none repeat.
Think of it like a sliding highlighter on a page.
You want the longest highlighted section with no repeated letters.
Return just the length of that section — not the section itself.

## Walk Through an Example by Hand

Input: `s = "abcabcbb"`

```

Step 2: left=0, right=1  char='b'  window='ab'    seen={a:0,b:1}  len=2
Step 3: left=0, right=2  char='c'  window='abc'   seen={..c:2}  len=3
Step 4: left=0, right=3  char='a'  'a' seen at 0!
         Move left to 0+1=1         window='bca'
         Update seen={a:3}          len=3
Step 5: left=1, right=4  char='b'  'b' seen at 1!
         Move left to 1+1=2         window='cab'
         Update seen={b:4}          len=3
Step 6: left=2, right=5  char='c'  'c' seen at 2!
         Move left to 2+1=3         window='abc'
         Update seen={c:5}          len=3
Step 7: left=3, right=6  char='b'  'b' seen at 4  >= left=3!
         Move left to 4+1=5         window='cb'
         len=2
Step 8: left=5, right=7  char='b'  'b' seen at 6  >= left=5!
         Move left to 6+1=7         window='b'
         len=1
Answer: max seen = 3
```

## The Picture

Think of it like a rubber-band window sliding over the string.
The right end stretches forward. When it grabs a duplicate,
the left snaps forward to just past where that char last was.

```
s =  a  b  c  a  b  c  b  b
     0  1  2  3  4  5  6  7

     [L        R]          <-- window='abc', len=3, no dupes
                              'a' found at index 3!
        [L     R]          <-- left jumps to 1
                              'b' found at index 4!
           [L  R]          <-- left jumps to 2

Map tracks LAST SEEN index of each char:
+-------+-------+
| char  | index |
+-------+-------+
|  'a'  |   3   |
|  'b'  |   4   |
|  'c'  |   5   |
+-------+-------+
Left only moves RIGHT — never left.
```

## When To Use This Pattern

- When you need the **longest/shortest subarray or substring**
  satisfying a condition.
- When the problem involves **no duplicates** or **unique
  elements** in a window.
- When you need to track **last seen position** of characters.
- When brute force would be O(n^2) and you want O(n).

## The Approach

Keep two pointers — left and right — that mark the current
window. Move right one step at a time. Each time you land on
a character, check if it was seen before inside the window.
If yes, jump left to just past that character's last position.
Track the longest window size seen throughout the whole scan.

## Complexity

| Approach       | Time    | Space              |
|----------------|---------|--------------------|
| Brute Force    | O(n^2)  | O(min(n, a))       |
| Sliding Window | O(n)    | O(min(n, a))       |

The key difference: brute force checks every possible
substring start. Sliding window never re-scans — it jumps
the left pointer directly using the stored last-seen index.
