**Difficulty:** Medium &nbsp;|&nbsp; **Category:** Sliding Window
**Pattern:** Variable-Size Window — Jump Left on Repeat

<strong>Core Insight:</strong> Store each character's

## Official Problem Statement

Given a string `s`, find the length of the longest
substring without repeating characters.

**Example 1:**
```
Input:  s = "abcabcbb"
Output: 3
Explanation: "abc" is the longest, length = 3.
```
**Example 2:**
```
Input:  s = "bbbbb"
Output: 1
Explanation: "b" is the longest, length = 1.
```
**Example 3:**
```
Input:  s = "pwwkew"
Output: 3
Explanation: "wke" is the longest, length = 3.
```

**Constraints:**
- `0 <= s.length <= 5 * 10^4`
- `s` consists of English letters, digits, symbols
  and spaces

## What This Is Actually Asking

Find the longest run of characters in the string
where no character repeats.
A substring must be contiguous — you cannot skip
characters.
Return the length of that run, not the substring
itself.

## The Picture

```
s = "a b c a b c b b"
idx: 0 1 2 3 4 5 6 7

Window grows right until a repeat is found.
When repeat found, left JUMPS past the old copy.

  [a b c]  a  b  c  b  b     len=3  no repeats
   L     R

  right hits 'a' again (at index 3):
  old 'a' was at index 0 -> jump left to 1

   a [b c a]  b  c  b  b     len=3  still 3
     L     R

The key trick:
  Store char -> LAST SEEN INDEX (not just a set).
  This lets left JUMP directly instead of crawling.

  Guard: only jump if old index >= left.
  (The char might have been seen before the window.)
```

## When To Use This Pattern

- When you see **longest substring with some
  constraint**, think **variable-size sliding window**
- When a character repeat collapses the window,
  think **store last-seen index, jump left past it**
- When the constraint is about uniqueness, think
  **char → index map** (not just a set)
- When the left pointer must guard against stale
  entries, think **only jump if old_index >= left**

## The Approach

Maintain a map from each character to the last index
where it was seen, plus a left pointer starting at 0.
Walk right through the string: if the current
character is in the map and its stored index is
inside the current window, jump left to one past
that stored index.
Update the map with the current index and update
the max length at every step.

````python
def test_harness(func):
    tests = [
        # (s, expected)
        ("abcabcbb",  3),
        ("bbbbb",     1),
        ("pwwkew",    3),
        ("",          0),   # empty string
        (" ",         1),   # single space
        ("au",        2),   # two unique
        ("dvdf",      3),   # 'vdf'
        ("aab",       2),   # 'ab'
        ("abcdefg",   7),   # all unique
        ("abba",      2),   # 'ab' or 'ba'
    ]

    passed = 0
    for i, (s, expected) in enumerate(tests):
        result = func(s)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"s={s!r} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def lengthOfLongestSubstring(s: str) -> int:
    r"""
    Return length of longest substring with no repeats.

    Map each char to its last seen index. Walk right
    with a pointer. If current char was seen inside the
    current window (stored index >= left), jump left
    to one past that index. Update map and max length
    at every step.

    Time:  O(n) — each char visited at most twice
    Space: O(min(n, m)) — map bounded by alphabet size
    """
    max_length, l = 0,0
    last_seen = {}
    for i, c in enumerate (s):
        if c in last_seen and last_seen[c] >= l:
            l = last_seen[c] + 1        #Jump there all at once ..
        last_seen[c] = i
        max_length = max(max_length, (i-l+1))
    return max_length
    
 


# Quick debug — run this cell while building
print(lengthOfLongestSubstring("abcabcbb"))  # 3
print(lengthOfLongestSubstring("bbbbb"))      # 1
print(lengthOfLongestSubstring("pwwkew"))     # 3
print(lengthOfLongestSubstring("dvdf"))       # 3
test_harness(lengthOfLongestSubstring)
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Brute force — check all substrings | O(n³) | O(m) |
| Set — shrink one step at a time | O(n) | O(m) |
| Map — jump left directly | O(n) | O(m) |

m = character set size. The map version matches the
set version in time and space but avoids the inner
shrink loop by jumping left directly.

## Real World Connection

IDs in an alert sequence — no endpoint alerting
twice in a row — maps exactly to this pattern.
