## Official Problem Statement

Given a string `s` containing just the characters
`'('`, `')'`, `'{'`, `'}'`, `'['` and `']'`,
determine if the input string is valid.

An input string is valid if:
1. Open brackets must be closed by the same type
   of brackets.
2. Open brackets must be closed in the correct order.
3. Every close bracket has a corresponding open
   bracket of the same type.

**Example 1:**
```
Input:  s = "()"
Output: true
```
**Example 2:**
```
Input:  s = "()[]{}"
Output: true
```
**Example 3:**
```
Input:  s = "(]"
Output: false
```

**Constraints:**
- `1 <= s.length <= 10^4`
- `s` consists of parentheses only `'()[]{}'`

## What This Is Actually Asking

Check that every opening bracket has a matching
closing bracket, and they are properly nested.
Order matters: `([)]` is wrong even though every
bracket has a partner, because they overlap.
If the string ends with unmatched openers still on
the stack, it is also invalid.

## The Picture

```
Think of the stack as a "last opened" pile.
Each opener goes on top. Each closer must match
whatever is on top — the most recent unmatched open.

s = ( [ { } ] )

  (              push (    stack: [(]
  ( [            push [    stack: [(, []
  ( [ {          push {    stack: [(, [, {]
  ( [ { }        } matches { -> pop  stack: [(, []
  ( [ }          ] matches [ -> pop  stack: [(]
  ( )            ) matches ( -> pop  stack: []

  empty stack at end -> VALID

Failure modes:
  Closer with wrong top  -> mismatch  -> False
  Closer with empty stack -> nothing to match -> False
  Loop ends with non-empty stack -> unclosed openers -> False
```

## When To Use This Pattern

- When you see **nested or paired structures** (brackets,
  tags, quotes), think **stack — push openers**
- When a closer must match the most recent unmatched
  opener, think **check stack top, then pop**
- When validity requires empty stack at the end,
  think **`return len(stack) == 0`**
- When you see nested SQL, JSON, or XML validation,
  think **this exact pattern**

## The Approach

Build a map from each closing bracket to its matching
opening bracket.
Walk through the string: push every opening bracket
onto the stack; for every closing bracket, check
that the stack is not empty and that the top matches
— if either check fails, return False immediately.
After the loop, return True only if the stack is
empty (all openers were matched).

````python
def test_harness(func):
    tests = [
        # (s, expected)
        ("()",       True),
        ("()[]{}",   True),
        ("(]",       False),
        ("([{}])",   True),
        ("([)]",     False),
        ("{[]}",     True),
        ("(",        False),  # unclosed opener
        (")",        False),  # closer with empty stack
        ("",         True),   # empty string is valid
        ("(((",      False),
        ("(){}[]",   True),
        ("{[()]}",   True),
    ]

    passed = 0
    for i, (s, expected) in enumerate(tests):
        result = func(s)
        ok = result == expected
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        print(
            f"Test {i+1}: {status} | "
            f"s={s!r} | "
            f"expected={expected} | got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
def isValid(s: str) -> bool:
    """
    Return True if all brackets in s are validly nested.

    Map each closer to its matching opener. Walk s:
    push openers. For each closer, fail if stack is
    empty or top doesn't match. Pop on match. Return
    True iff stack is empty at the end.

    Time:  O(n) — one pass through the string
    Space: O(n) — stack holds at most n/2 openers
    """
    """ 
        never push a closing in stack. stack only for opens.
        if char not in hash map .. it is opens .. just append.
        else
        if stack and hashmap of char == top of stack  .. pop stack
        else return false
        return not stack at the end. if any is left in stack , it is false

    """
    stack = []
    hashmap = { ')':'(' , ']':'[' , '}':'{' }
    for c in s:
        #It is a close .. do I have a top of stack
        if c not in hashmap:         #It is opens
            stack.append(c)
        else:                        #it Closes
            if stack and hashmap[c] == stack[-1]:
                stack.pop()
            else:
                return False
    return not stack            

            
            
    
def testing():
    # Quick debug — run this cell while building
    print(isValid("()"))       # True
    print(isValid("()[]{}"))   # True
    print(isValid("(]"))       # False
    print(isValid("([)]"))     # False
    print(isValid(")"))        #False
    test_harness(isValid)
testing()
````

## Complexity

| Approach | Time | Space |
|---|---|---|
| Counter only (no ordering) | O(n) | O(1) |
| Stack — correct approach | O(n) | O(n) |

A simple counter fails on `([)]` — it needs the
stack to enforce ordering. The stack is the only
correct approach and it is O(n) time.
