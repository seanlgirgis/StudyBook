## Official Problem Statement

Given `n` nodes labeled `0` to `n-1` and a list of undirected edges,
return the number of connected components in the graph.

**Constraints:**
- `1 <= n <= 2000`
- `0 <= edges.length <= 5000`
- `edges[i].length == 2`
- `0 <= ai, bi < n`
- `ai != bi`
- No repeated edges

## What This Is Actually Asking

Think of nodes as islands and edges as bridges.
How many separate landmasses exist after building all bridges?

A connected component is a maximal set of nodes where every node
can reach every other node by following edges.

- If two nodes share the same "root" in Union-Find, they are
  in the same component.
- We want the total number of distinct roots.

## Walk Through an Example by Hand

```
n=5, edges=[[0,1],[1,2],[3,4]]
```

Start: parent = [0, 1, 2, 3, 4]  (5 components)

Process edge (0,1):
  find(0)=0, find(1)=1 → different → union → parent[1]=0
  parent = [0, 0, 2, 3, 4]  (4 components)

Process edge (1,2):
  find(1)=0, find(2)=2 → different → union → parent[2]=0
  parent = [0, 0, 0, 3, 4]  (3 components)

Process edge (3,4):
  find(3)=3, find(4)=4 → different → union → parent[4]=3
  parent = [0, 0, 0, 3, 3]  (2 components)

Count roots: node 0 → root 0 ✓, node 3 → root 3 ✓
Answer = 2

## The Picture

```
Graph:                Union-Find parent array:

  0 — 1 — 2   3 — 4   idx:  0  1  2  3  4
                            [0][0][0][3][3]
                                          ^
  Component 1  Component 2   roots: 0, 3

find(x) with path compression:
  find(2): parent[2]=0 → parent[0]=0 → return 0
  find(4): parent[4]=3 → parent[3]=3 → return 3

Count of i where parent[i]==i → {0, 3} → answer = 2
```

The key: after all unions, **roots are self-referential** (`parent[i]==i`).

## When To Use This Pattern

Use Union-Find (Disjoint Set Union) when:
- You need to group elements into components dynamically
- You process edges/connections one at a time
- You need to answer "are X and Y connected?" efficiently
- Cycle detection in undirected graphs
- Kruskal's minimum spanning tree

**Not ideal when:** you need the actual path between nodes
(use BFS/DFS instead).

**Signature:** `parent = [i for i in range(n)]` initializes
each node as its own component.

## The Approach

1. Initialize `parent[i] = i` for all `i` in `0..n-1`
2. Define `find(x)` with path compression:
   - If `parent[x] != x`, recurse: `parent[x] = find(parent[x])`
   - Return `parent[x]`
3. Define `union(x, y)`:
   - `rx, ry = find(x), find(y)`
   - If `rx != ry`: set `parent[rx] = ry`
4. For each edge `(u, v)`: call `union(u, v)`
5. Count `i` where `find(i) == i` → return that count

**Time:** O(n + E * α(n)) ≈ O(n + E)  
**Space:** O(n)

## Complexity

| | Value | Why |
|---|---|---|
| **Time** | O(n + E·α(n)) | Init n nodes + process E edges with near-O(1) UF ops |
| **Space** | O(n) | Parent array of size n |

α(n) is the inverse Ackermann function — effectively constant (≤5)
for any realistic input size.

**Alternative approaches:**
- BFS/DFS: O(n + E) time, O(n + E) space (need adjacency list)
- Union-Find is preferable here: no graph construction needed,
  directly processes edge list.
