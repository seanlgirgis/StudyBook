## Official Problem Statement

Balloons are represented as `[x_start, x_end]` on a number line.
An arrow shot at x bursts all balloons where
`x_start <= x <= x_end`. Arrows travel infinitely in both
directions. Return the **minimum number of arrows** needed to
burst all balloons.

**Constraints:**
- `1 <= points.length <= 10^5`
- `points[i].length == 2`
- `-2^31 <= x_start <= x_end <= 2^31 - 1`

## What This Is Actually Asking

Find the minimum number of points that collectively pierce every
interval at least once.
This is the interval point cover problem.
It is the complement of Non-overlapping Intervals:
each contiguous group of overlapping intervals needs exactly
one arrow (one point in their common overlap).
Sort by end time to greedily extend coverage as far as possible.

## Walk Through an Example by Hand

```
points = [[10,16],[2,8],[1,6],[7,12]]
Sorted by end: [[1,6],[2,8],[7,12],[10,16]]
```
- Arrow at x=6 (first balloon's end). arrows=1
- [2,8]: start=2 <= 6 → already burst by arrow at 6
- [7,12]: start=7 > 6 → new arrow at x=12. arrows=2
- [10,16]: start=10 <= 12 → burst by arrow at 12
- **Answer = 2**

## The Picture

```
Number line:
1====6         [1,6]
  2======8     [2,8]
       7====12 [7,12]
          10======16 [10,16]

  ^arrow1=6    ^arrow2=12

Sort by END. Place arrow at each interval's end.
Skip balloons whose start <= current arrow position.
When start > arrow_pos: shoot new arrow at this end.

Contrast with LC 435:
  435 counts REMOVED intervals (overlapping ones).
  452 counts KEPT groups (non-overlapping clusters).
  They are the same greedy, different output.
```

## When To Use This Pattern

- When minimizing **coverage points** for intervals, think
  sort by end + greedy arrow placement.
- When all overlapping intervals form one group needing one
  arrow, think **greedy contiguous cluster**.
- When a balloon's start exceeds the current arrow, that
  signals a new non-overlapping group — think new arrow.
- Recognize this as the **dual** of Non-overlapping Intervals
  (LC 435): same sort, count groups vs count removals.

## The Approach

Sort by end time. Place the first arrow at the first balloon's
end. For each subsequent balloon: if its start exceeds the
current arrow position, place a new arrow at this balloon's
end and increment the count.
Balloons whose start is within the current arrow range are
burst for free.
