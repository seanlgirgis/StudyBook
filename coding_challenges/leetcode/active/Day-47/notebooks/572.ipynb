## Official Problem Statement

Given the roots of two binary trees `root` and `subRoot`, return
`true` if there is a subtree of `root` with the same structure
and node values as `subRoot`, and `false` otherwise.

A subtree of a binary tree `root` is a tree consisting of a node
in `root` and all of its descendants.

**Constraints:**
- `root`: 1 to 2000 nodes, `-10^4 <= val <= 10^4`
- `subRoot`: 1 to 1000 nodes, `-10^4 <= val <= 10^4`

## What This Is Actually Asking

Walk every node in the main tree. At each node, check if the
subtree rooted there is structurally identical to `subRoot`.
Structural identity means same shape, same values, at every level.
Return True as soon as you find any matching node.
This decomposes into two separate recursive checks.

## Walk Through an Example by Hand

- isSubtree(3, 4): is_same(3,4)? 3≠4 → False
  - isSubtree(4, 4): is_same(4,4)? 4=4, check children:
    - is_same(1,1) → True, is_same(2,2) → True → **True!**
- Short-circuit returns True.

## The Picture

```
is_same(s, t) logic:
  s=None, t=None  ->  True   (both empty, match)
  s=None OR t=None -> False  (one empty, mismatch)
  s.val != t.val  ->  False  (values differ)
  else:  is_same(s.left, t.left)
     AND is_same(s.right, t.right)

isSubtree(root, sub) logic:
  root=None -> False
  is_same(root, sub) -> True  (found it!)
  isSubtree(root.left, sub)   (check left branch)
  OR isSubtree(root.right, sub) (check right branch)
```

## When To Use This Pattern

- When checking if two trees are **identical**, think recursive
  base cases: both None, one None, values differ.
- When searching every node of a tree for a **local property**,
  think DFS with a helper that checks that property.
- When a problem nests two recursive structures, think
  **outer DFS + inner DFS** (isSubtree calls is_same).
- When short-circuit evaluation matters for performance,
  use `or` to stop early.

## The Approach

Write `is_same(s, t)` that returns True only if both trees are
structurally identical with matching values at every node.
Then write `isSubtree(root, sub)`: if root is None return False;
if `is_same(root, sub)` return True; otherwise recurse into
both children and return their OR.

````python
from collections import deque
from typing import Optional


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def make_tree(vals):
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = deque([root])
    i = 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root
````
