## Official Problem Statement

There are `n` cities. Some are directly connected, some are not.
If city `a` is directly connected to `b`, and `b` to `c`, then
`a`, `b`, `c` are in the same province.

Given `isConnected[i][j] = 1` if city i and j are directly connected,
return the total number of provinces.

**Constraints:**
- `1 <= n <= 200`
- `n == isConnected.length == isConnected[i].length`
- `isConnected[i][j]` is `0` or `1`
- `isConnected[i][i] == 1`
- `isConnected[i][j] == isConnected[j][i]`

## What This Is Actually Asking

Find the number of connected components in an undirected graph
represented as an adjacency matrix.
The matrix tells you which cities are directly linked.
Provinces are transitive: indirect connections still form one group.
This is conceptually identical to Number of Islands but with an
explicit adjacency matrix instead of a grid.

## Walk Through an Example by Hand

```
isConnected = [[1,1,0],[1,1,0],[0,0,1]]
```
Nodes: 0, 1, 2
- i=0,j=1: isConnected[0][1]=1 → union(0,1). parent=[0,0,2]
- i=0,j=2: isConnected[0][2]=0 → skip
- i=1,j=2: isConnected[1][2]=0 → skip

Distinct roots: find(0)=0, find(1)=0, find(2)=2 → **2 provinces**

## The Picture

```
Adjacency Matrix:          Union-Find forest:
  0 1 2                    Province 1   Province 2
0[1,1,0]                      [0]          [2]
1[1,1,0]   --- union --->      |
2[0,0,1]                      [1]

Count distinct roots in parent array:
parent = [0, 0, 2]
         ^        ^   <- two unique roots = 2 provinces
```

## When To Use This Pattern

- When merging groups dynamically, think **Union-Find**.
- When given an adjacency matrix and asked for component count,
  think **Union-Find or DFS**.
- When "connected" is transitive (a-b-c means a-c), think
  **component grouping**.
- When you need near O(1) per union/find operation, think
  **Union-Find with path compression**.

## The Approach

Initialize a parent array where each city is its own root.
Scan upper triangle of the matrix; when `isConnected[i][j]==1`,
call `union(i, j)` to merge their components.
Finally count how many nodes are their own parent — each is
the root of one province.

| Union-Find (basic) | O(n²) | O(n) | n² union calls |
