## Official Problem Statement

There are `n` gas stations arranged in a circle. `gas[i]` is the
amount you can collect at station `i`. `cost[i]` is the fuel
needed to travel from station `i` to `i+1`.

Starting with an empty tank, find the station index you can start
from to complete the circuit. Return `-1` if impossible.
If a solution exists, it is **guaranteed to be unique**.

**Constraints:**
- `n == gas.length == cost.length`
- `1 <= n <= 10^5`
- `0 <= gas[i], cost[i] <= 10^4`

## What This Is Actually Asking

Find the starting position in a circular array where you can
maintain a non-negative cumulative sum of (gas[i] - cost[i]).
The circuit is possible only if the total sum is non-negative.
If possible, the valid start is always the station right after
the point where the running total last dropped negative.

## Walk Through an Example by Hand

```
gas  = [1,2,3,4,5]
cost = [3,4,5,1,2]
diff = [-2,-2,-2,3,3]
```
- i=0: tank = 0+(-2) = -2 < 0 → start=1, tank=0
- i=1: tank = 0+(-2) = -2 < 0 → start=2, tank=0
- i=2: tank = 0+(-2) = -2 < 0 → start=3, tank=0
- i=3: tank = 0+3 = 3 ≥ 0 → ok
- i=4: tank = 3+3 = 6 ≥ 0 → ok
- total = sum(diff) = 0 ≥ 0 → solution exists
- Answer = **3**

## The Picture

```
Running tank as we scan left-to-right:

  idx:   0    1    2    3    4
  diff: -2   -2   -2   +3   +3
  tank: -2   -2   -2   +3   +6
         ^    ^    ^
  Tank goes negative -> reset start to i+1 each time

  start resets: 0->1->2->3, then stays at 3

Greedy key:
  If we can't reach station i from start X,
  then NO station between X and i can be a valid start.
  (Because they'd all start with even less fuel.)
```

## When To Use This Pattern

- When a circular array problem asks for a valid **starting
  index**, think greedy running-sum with reset.
- When total feasibility can be checked with a **global sum**
  before finding the local optimum, do the global check first.
- When a failure at position i **eliminates all previous**
  candidates, think O(n) greedy rather than O(n²) brute force.
- When a problem guarantees uniqueness of the answer,
  the greedy choice is safe.

## The Approach

First check if total gas >= total cost; if not, return -1.
Otherwise do a single pass: maintain a running tank.
When tank drops below 0, the current start is invalid —
reset the start to i+1 and the tank to 0.
The start index remaining at the end is the answer.
