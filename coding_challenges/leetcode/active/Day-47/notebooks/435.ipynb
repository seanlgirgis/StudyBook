## Official Problem Statement

Given an array of intervals `intervals[i] = [start_i, end_i]`,
return the **minimum number of intervals you need to remove** to
make the rest non-overlapping.

**Constraints:**
- `1 <= intervals.length <= 10^5`
- `intervals[i].length == 2`
- `-5 * 10^4 <= start_i < end_i <= 5 * 10^4`

## What This Is Actually Asking

Find the largest subset of non-overlapping intervals (Activity
Selection Problem). The answer is total intervals minus that max
subset size.
Two intervals overlap if one starts before the other ends.
Sorting by end time ensures we always keep the interval that
finishes earliest, leaving the most room for future intervals.

## Walk Through an Example by Hand

```
intervals = [[1,2],[2,3],[3,4],[1,3]]
Sorted by end: [[1,2],[2,3],[1,3],[3,4]]
```
- Keep [1,2]. prev_end=2. kept=1
- [2,3]: start=2 >= prev_end=2 → keep. prev_end=3. kept=2
- [1,3]: start=1 < prev_end=3 → overlap! remove. removals=1
- [3,4]: start=3 >= prev_end=3 → keep. kept=3
- **Answer = 1 removal**

## The Picture

```
Timeline:
1---2        [1,2]  <- keep (ends earliest)
    2---3    [2,3]  <- keep (starts >= 2)
1-------3    [1,3]  <- REMOVE (overlaps [2,3])
        3---4 [3,4] <- keep (starts >= 3)

Why sort by END (not start)?
  Keeping the earliest-ending interval leaves the
  maximum remaining timeline for future intervals.
  This is the greedy proof by exchange argument.
```

## When To Use This Pattern

- When minimizing removals of overlapping intervals, think
  **sort by end + greedy keep**.
- When maximizing non-overlapping intervals (Activity Selection),
  think same greedy — answer is total minus removals.
- When intervals touch at endpoints (e.g., [1,2] and [2,3]),
  clarify whether touching counts as overlap (here it does not).
- When intervals are on a 1D timeline, think **sort + scan**.

## The Approach

Sort intervals by end time. Initialize `prev_end` to negative
infinity and `removals` to 0. For each interval: if its start
is >= prev_end, keep it (update prev_end); otherwise it overlaps
with the previously kept interval — increment removals.
Return the total removals.

## Complexity

| Approach | Time | Space | Notes |
|---|---|---|---|
