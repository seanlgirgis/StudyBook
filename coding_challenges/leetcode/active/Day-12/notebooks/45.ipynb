## Official Problem Statement

You are given a **0-indexed** array of integers
`nums` of length `n`. You are initially positioned
at `nums[0]`.

Each element `nums[i]` represents the maximum
length of a forward jump from index `i`. Return
the **minimum number of jumps** to reach
`nums[n-1]`.

The test cases are generated such that you can
always reach `nums[n-1]`.

**Example 1:**
```
Input:  nums = [2,3,1,1,4]
Output: 2
Explanation: jump from 0->1 (3 steps), then 1->4.
```
**Example 2:**
```
Input:  nums = [2,3,0,1,4]
Output: 2
```

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 1000`
- The answer will always exist

## What This Is Actually Asking

Starting at index 0, each cell tells you how far
you can jump. You always want to land as far
forward as possible. Find the minimum number of
jumps to reach the last cell.

## Walk Through an Example by Hand

```
nums = [2, 3, 1, 1, 4]
idx:    0  1  2  3  4

jumps=0  cur_end=0  farthest=0

i=0: farthest = max(0, 0+2) = 2
     i==cur_end=0 -> must jump! jumps=1  cur_end=2

i=1: farthest = max(2, 1+3) = 4
i=2: farthest = max(4, 2+1) = 4
     i==cur_end=2 -> must jump! jumps=2  cur_end=4

cur_end=4 >= last index=4 -> stop early
Answer: 2

BFS level view:
  Level 0 (start): [0]
  Level 1 (jump 1): indices 1, 2  (from 0, reach up to 2)
  Level 2 (jump 2): indices 3, 4  (from 1 or 2)
  Reached last index at level 2 -> 2 jumps
```

## The Picture

```
nums = [2, 3, 1, 1, 4]
        0  1  2  3  4

Jump levels (like BFS levels):

  Level 0:  [0]
             |
             can reach indices 1, 2 (jump up to 2)

  Level 1:  [1, 2]
             |  |
             from 1: reach 2,3,4   farthest=4
             from 2: reach 3       farthest=4

  Level 2:  [3, 4]  <- last index 4 reached!

Two variables:
  cur_end   = end of the CURRENT jump level
  farthest  = farthest reachable in NEXT jump

When i reaches cur_end:
  jumps += 1
  cur_end = farthest
  (we're committed to taking a jump now)
```

## When To Use This Pattern

- When asked **minimum jumps to reach end**,
  think **greedy BFS levels — two boundaries**
- When `i == cur_end`, think
  **jump boundary hit — jumps += 1, advance boundary**
- When tracking next reach, think
  **farthest = max(farthest, i + nums[i])**
- When `cur_end >= last index`, think
  **stop early — answer found**

## The Approach

Track two boundaries: cur_end (end of current
jump's reach) and farthest (farthest reachable
from any cell in the current jump). Walk through
the array; update farthest at each cell. When
the current index reaches cur_end, a jump must
be made — increment the count and set cur_end to
farthest. Stop when cur_end reaches the last index.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Dynamic programming | O(n²) | O(n) |
| Greedy BFS boundaries | O(n) | O(1) |

The greedy approach works because jumping to the
farthest cell within each level is always at least
as good as any other choice — no look-ahead or
backtracking is ever needed.
