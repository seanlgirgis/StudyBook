## Official Problem Statement

You are given an integer array `nums`. You are
initially positioned at the array's **first
index**, and each element in the array represents
your maximum jump length at that position.

Return `true` if you can reach the last index,
or `false` otherwise.

**Example 1:**
```
Input:  nums = [2,3,1,1,4]
Output: true
```
**Example 2:**
```
Input:  nums = [3,2,1,0,4]
Output: false
Explanation: Always reach index 3 with value 0 —
             cannot jump further.
```

**Constraints:**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 10^5`

## What This Is Actually Asking

You are standing at position 0. Each cell tells
you the maximum number of steps you can jump
forward from there. Can you reach the last cell?
A zero means you cannot move forward from there.

## Walk Through an Example by Hand

```
nums = [2, 3, 1, 1, 4]   (can reach end?)
idx:    0  1  2  3  4

max_reach = 0

i=0: 0 <= max_reach=0 OK  max_reach=max(0,0+2)=2
i=1: 1 <= max_reach=2 OK  max_reach=max(2,1+3)=4
i=2: 2 <= max_reach=4 OK  max_reach=max(4,2+1)=4
i=3: 3 <= max_reach=4 OK  max_reach=max(4,3+1)=4
i=4: 4 <= max_reach=4 OK  max_reach=max(4,4+4)=8

Loop completes without blocking -> return True

---
nums = [3, 2, 1, 0, 4]   (stuck?)

max_reach = 0

i=0: 0 <= 0 OK  max_reach=max(0,0+3)=3
i=1: 1 <= 3 OK  max_reach=max(3,1+2)=3
i=2: 2 <= 3 OK  max_reach=max(3,2+1)=3
i=3: 3 <= 3 OK  max_reach=max(3,3+0)=3
i=4: 4 > max_reach=3 -> STUCK -> return False
```

## The Picture

```
nums = [2, 3, 1, 1, 4]
        0  1  2  3  4

From index 0, jump up to 2 steps:
  can reach: 1, 2
                   max_reach = 2

From index 1, jump up to 3 steps:
  can reach: 2, 3, 4
                          max_reach = 4

max_reach >= last index (4) -> True!

---
nums = [3, 2, 1, 0, 4]
        0  1  2  3  4

  0: reach up to 3
  1: reach up to 3
  2: reach up to 3
  3: jump=0 -> cannot extend past 3
             max_reach = 3 forever
  4: i=4 > max_reach=3 -> UNREACHABLE

Greedy rule:
  If i > max_reach at any point -> stuck -> False
  Else update max_reach = max(max_reach, i + nums[i])
```

## When To Use This Pattern

- When each position gives a **max jump reach**,
  think **track max_reach greedily**
- When `i > max_reach`, think
  **current index is unreachable — return False**
- When the loop completes, think
  **all positions reachable — return True**
- When a zero is encountered, think
  **only stuck if max_reach cannot pass it**

## The Approach

Track the farthest index reachable so far. Scan
each index from left to right. If the current
index exceeds the farthest reach, the position
is unreachable — return False. Otherwise update
the farthest reach to the max of its current
value and the current index plus the jump length.
Return True if the loop ends normally.

## Complexity

| Approach | Time | Space |
|---|---|---|
| Dynamic programming (bottom-up) | O(n²) | O(n) |
| Greedy max-reach tracking | O(n) | O(1) |

Greedy works because looking back never helps:
if index i is reachable, every position up to
i + nums[i] is reachable from the same path.
No DP table needed.
