# LC 62 — Unique Paths
**Difficulty:** Medium | **Category:** 2D Dynamic Programming
**Pattern:** Grid DP — Count paths by summing neighbors

<div style="border-left:4px solid purple; padding:10px 16px;

<strong>Core Insight:</strong> Each cell can only be reached from
above or from the left. So the number of unique paths to any cell
is simply the sum of paths to the cell above it and the cell
to its left. The first row and first column are always 1.
</div>


## Official Problem Statement

There is a robot on an `m x n` grid. The robot is initially located
at the **top-left corner** (i.e., `grid[0][0]`). The robot tries to
move to the **bottom-right corner** (i.e., `grid[m-1][n-1]`).
The robot can only move either **down** or **right** at any point.

Given the two integers `m` and `n`, return the number of possible
unique paths that the robot can take to reach the bottom-right
corner.

**Constraints:**
- `1 <= m, n <= 100`
- The answer will be less than or equal to `2 * 10^9`


## What This Is Actually Asking

Imagine a grid of city blocks. You start at the top-left corner
and want to reach the bottom-right corner.

You can only walk right or down — never left or up.

How many different routes can you take?

Every intersection you reach was entered from above or from the
left, so its route count is the sum of those two neighbors.


## Walk Through an Example by Hand

**Input:** m=3, n=3

Step 1: Fill row 0 with all 1s (only one way to reach any cell
        in the first row — go straight right).
Step 2: Fill col 0 with all 1s (only one way — go straight down).
Step 3: For each remaining cell, add the cell above + cell left.

```
dp[1][1] = dp[0][1] + dp[1][0] = 1 + 1 = 2
dp[1][2] = dp[0][2] + dp[1][1] = 1 + 2 = 3
dp[2][1] = dp[1][1] + dp[2][0] = 2 + 1 = 3
dp[2][2] = dp[1][2] + dp[2][1] = 3 + 3 = 6
```

**Answer:** 6


## The Picture

DP table for m=3, n=3 (rows=m, cols=n):

```
      col0  col1  col2
row0 [  1     1     1  ]
row1 [  1     2     3  ]
row2 [  1     3    [6] ]  <-- answer
```

Arrow view for dp[2][2]:
```
  3        (from above)
  ^  
  |  
  3 --> [6]  (from left)
```

Each cell = cell above + cell to the left.
Edges are always 1 (only one straight-line route).


## When To Use This Pattern

- When you move through a grid with **restricted directions**
  (right/down only), think Grid DP.
- When the problem asks to **count paths** in a grid, think
  dp[i][j] = dp[i-1][j] + dp[i][j-1].
- When **each cell depends only on its top and left neighbors**,
  this recurrence applies directly.
- When constraints are small (m,n <= 100), a 2D table is fine.
- When you see "how many ways" + grid movement, reach for this.


## The Approach

Build an m x n DP table initialized to 1 everywhere. The first
row and column stay at 1 because there is only one way to travel
along an edge. For every interior cell, set its value to the sum
of the cell directly above and the cell directly to the left.
Return the value at dp[m-1][n-1].
