## Official Problem Statement

Given two strings `text1` and `text2`, return the length of their
**longest common subsequence**. If there is no common subsequence,
return `0`.

A **subsequence** of a string is a new string generated from the
original string with some characters (can be none) deleted without
changing the relative order of the remaining characters.

A **common subsequence** of two strings is a subsequence that is
common to both strings.

**Constraints:**
- `1 <= text1.length, text2.length <= 1000`
- `text1` and `text2` consist of only lowercase English letters


## What This Is Actually Asking

Find the longest sequence of characters that appears in both
strings in the same order, but not necessarily next to each other.

For "abcde" and "ace", the LCS is "ace" (length 3) because a, c,
and e appear in that order in both strings.

You can skip characters from either string, but you cannot reorder
them.

The key question at each pair of characters: did they match?
If yes, the LCS grows by 1. If no, take the better of two skips.


## Walk Through an Example by Hand

**Input:** text1="abcde", text2="ace"

Build dp[i][j] = LCS of text1[:i] and text2[:j].
Add a zero-row and zero-column as base case.

```
     ""  a  c  e
""  [ 0  0  0  0 ]
a   [ 0  1  1  1 ]  <- 'a'=='a': dp[0][0]+1=1
b   [ 0  1  1  1 ]  <- 'b'!='c': max(1,1)=1
c   [ 0  1  2  2 ]  <- 'c'=='c': dp[1][1]+1=2
d   [ 0  1  2  2 ]  <- 'd'!='e': max(2,2)=2
e   [ 0  1  2  3 ]  <- 'e'=='e': dp[3][2]+1=3
```

**Answer:** dp[5][3] = 3


## The Picture

DP table for text1="abcde", text2="ace":

```
       ""   a   c   e
  ""  [ 0   0   0   0 ]
   a  [ 0   1   1   1 ]
   b  [ 0   1   1   1 ]
   c  [ 0   1   2   2 ]
   d  [ 0   1   2   2 ]
   e  [ 0   1   2  [3] ]  <-- answer
```

Recurrence:
```
if text1[i-1] == text2[j-1]:
    dp[i][j] = dp[i-1][j-1] + 1   (diagonal + 1)
else:
    dp[i][j] = max(dp[i-1][j],    (skip text1 char)
                   dp[i][j-1])    (skip text2 char)
```


## When To Use This Pattern

- When asked for the **longest common subsequence** of two strings,
  think LCS DP directly.
- When characters from two sequences must appear **in order** but
  not necessarily adjacent, think dp[i][j] with diagonal logic.
- When comparing two strings character by character and the choice
  is "match" or "skip one", reach for this pattern.
- When the problem mentions diff, similarity, or alignment between
  sequences, LCS is usually the foundation.
- When n,m <= 1000, an O(n*m) table is perfectly fine.


## The Approach

Create a (len(text1)+1) x (len(text2)+1) table filled with zeros.
Iterate over every (i, j) pair. If the characters match, set
dp[i][j] to dp[i-1][j-1] + 1. If they don't match, set dp[i][j]
to max(dp[i-1][j], dp[i][j-1]). Return dp[m][n].
