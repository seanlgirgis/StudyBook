<strong>Core Insight:</strong> If characters match, no operation is

## Official Problem Statement

Given two strings `word1` and `word2`, return the minimum number
of operations required to convert `word1` to `word2`.

You have the following three operations permitted on a word:
- **Insert** a character
- **Delete** a character
- **Replace** a character

**Constraints:**
- `0 <= word1.length, word2.length <= 500`
- `word1` and `word2` consist of lowercase English letters


## What This Is Actually Asking

How many single-character edits does it take to turn word1 into
word2? Each edit is an insert, delete, or replace.

Think of it like spell-checking: "horse" to "ros" takes 3 edits.

At every pair of characters, you make the cheapest choice: match
for free, or pay 1 for whichever operation costs the least.

The DP table stores the edit distance for every prefix pair,
building up to the full strings.


## Walk Through an Example by Hand

**Input:** word1="horse", word2="ros"

Base cases:
- dp[i][0] = i  (delete all i chars from word1)
- dp[0][j] = j  (insert j chars to build word2)

```
     ""  r  o  s
""  [ 0  1  2  3 ]
h   [ 1  1  2  3 ]  'h'!='r': 1+min(1,1,0)=1
o   [ 2  2  1  2 ]  'o'=='o': dp[1][1]=1
r   [ 3  2  2  2 ]  'r'=='r': dp[2][1]=2? No:
                    'r'=='r'? word2[1]='o', no.
                    'r'!='o': 1+min(2,2,1)=2
s   [ 4  3  3  2 ]  's'=='s': dp[3][2]=2
e   [ 5  4  4  3 ]  'e'!='s': 1+min(3,3,2)=3
```

**Answer:** dp[5][3] = 3


## The Picture

DP table for word1="horse", word2="ros":

```
       ""  r  o  s
  ""  [ 0  1  2  3 ]
   h  [ 1  1  2  3 ]
   o  [ 2  2  1  2 ]
   r  [ 3  2  2  2 ]
   s  [ 4  3  3  2 ]
   e  [ 5  4  4 [3] ]  <-- answer
```

Three-way choice at dp[i][j] when chars differ:
```
  dp[i-1][j-1]  dp[i-1][j]
   (replace)     (delete)
        \           |
         \          v
  dp[i][j-1] --> dp[i][j] = 1 + min of the three
  (insert)
```


## When To Use This Pattern

- When the problem asks for **minimum edits** between two strings,
  think Edit Distance DP immediately.
- When operations have **uniform cost** (insert=delete=replace=1),
  this classic recurrence applies directly.
- When you see "transform string A into string B", think 2D DP
  with base cases = full deletion or full insertion.
- When the problem involves **spell correction, diff, or fuzzy
  matching**, edit distance is the foundation.
- When string lengths are <= 500, O(m*n) is perfectly fine.


## The Approach

Create a (m+1) x (n+1) DP table. Set the base cases: row 0 is
0..n and column 0 is 0..m. For each cell, if characters match,
copy dp[i-1][j-1]. If they differ, take 1 + the minimum of the
diagonal (replace), top (delete), and left (insert) cells.
Return dp[m][n].
