# LC 64 — Minimum Path Sum
**Difficulty:** Medium | **Category:** 2D Dynamic Programming
**Pattern:** Grid DP — Minimize cost by taking cheapest neighbor

<div style="border-left:4px solid purple; padding:10px 16px;
background:#f5f0ff; margin-top:12px;">
<strong>Core Insight:</strong> The cheapest way to reach any cell
is the cell's own value plus the cheaper of the two paths that
lead into it — from above or from the left. Fill the table once
and the answer sits in the bottom-right corner.
</div>


## Official Problem Statement

Given a `m x n` grid filled with non-negative numbers, find a
path from the **top-left** to the **bottom-right**, which
**minimizes the sum** of all numbers along its path.

**Note:** You can only move either **down** or **right** at any
point in time.

**Constraints:**
- `m == grid.length`
- `n == grid[i].length`
- `1 <= m, n <= 200`
- `0 <= grid[i][j] <= 200`


## What This Is Actually Asking

Walk from the top-left to the bottom-right of a grid of numbers.
You can only step right or down at each move.

Each cell you step on adds its value to your running total.

Find the path that collects the smallest possible total.

At each cell, the best you can do is take whichever incoming
direction (above or left) already has the lower cumulative cost.


## Walk Through an Example by Hand

**Input:**
```
grid = [[1,3,1],
        [1,5,1],
        [4,2,1]]
```

Step 1: Fill row 0 as cumulative sums (only move right).
```
dp[0] = [1, 4, 5]
```
Step 2: Fill col 0 as cumulative sums (only move down).
```
dp[1][0] = 1+1 = 2,  dp[2][0] = 2+4 = 6
```
Step 3: Fill interior cells.
```
dp[1][1] = 5 + min(4, 2) = 5+2 = 7
dp[1][2] = 1 + min(5, 7) = 1+5 = 6
dp[2][1] = 2 + min(7, 6) = 2+6 = 8
dp[2][2] = 1 + min(6, 8) = 1+6 = 7
```

**Answer:** 7  (path: 1→3→1→1→1)


## The Picture

DP table for the example grid:

```
Grid:          DP (min cost to reach):
[ 1  3  1 ]   [  1   4   5 ]
[ 1  5  1 ]   [  2   7   6 ]
[ 4  2  1 ]   [  6   8  [7] ]  <-- answer
```

Recurrence:
```
dp[i][j] = grid[i][j] + min(dp[i-1][j],   (from above)
                             dp[i][j-1])   (from left)
```
Edge rows/cols only have one incoming direction — use that one.


## When To Use This Pattern

- When asked for the **minimum cost path** in a grid (right/down
  only), think dp[i][j] = val + min(above, left).
- When each cell has a **cost or weight** and you need the cheapest
  route, this grid DP pattern applies directly.
- When movement is restricted and the problem has **optimal
  substructure**, reach for 2D DP.
- When constraints are small (m, n <= 200), a full table is fine.
- When the problem says "minimum sum" + "path" + "grid", this is
  almost certainly the right approach.


## The Approach

Create a copy of the grid as the DP table (or modify in-place).
Prefix the first row and first column with cumulative sums since
there is only one way to enter edge cells. For every interior
cell, set dp[i][j] to grid[i][j] plus the minimum of the cell
above and the cell to the left. Return dp[m-1][n-1].
