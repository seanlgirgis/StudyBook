# LC 121 — Best Time to Buy and Sell Stock
**Difficulty:** Easy | **Category:** 1D DP / Greedy
**Pattern:** Single Pass — Track Min Price

## Official Problem Statement

You are given an array `prices` where `prices[i]` is the price of a
given stock on the `i`th day.

You want to maximize your profit by choosing a **single day** to buy
one stock and choosing a **different day in the future** to sell that
stock.

Return the maximum profit you can achieve from this transaction.
If you cannot achieve any profit, return `0`.

**Constraints:**
- `1 <= prices.length <= 100,000`
- `0 <= prices[i] <= 10,000`

## What This Is Actually Asking

You have a list of stock prices, one per day.
You can buy on one day and sell on a later day.
You want to make as much money as possible.
You can only do this once.
If prices only go down, return 0 — do not trade.

## Walk Through an Example by Hand

Input: prices = [7, 1, 5, 3, 6, 4]

Start: min_price = 7, max_profit = 0

Day 0 — price=7
  min_price = min(7, 7) = 7
  profit    = 7 - 7 = 0
  max_profit = max(0, 0) = 0

Day 1 — price=1
  min_price = min(7, 1) = 1   <-- new low!
  profit    = 1 - 1 = 0
  max_profit = max(0, 0) = 0

Day 2 — price=5
  min_price = min(1, 5) = 1
  profit    = 5 - 1 = 4
  max_profit = max(0, 4) = 4

Day 3 — price=3
  min_price = min(1, 3) = 1
  profit    = 3 - 1 = 2
  max_profit = max(4, 2) = 4

Day 4 — price=6
  min_price = min(1, 6) = 1
  profit    = 6 - 1 = 5
  max_profit = max(4, 5) = 5   <-- best!

Day 5 — price=4
  min_price = min(1, 4) = 1
  profit    = 4 - 1 = 3
  max_profit = max(5, 3) = 5

Answer: 5  (buy day 1 at price 1, sell day 4 at price 6)

## The Picture

Think of a valley and a peak on a price chart.
You want the biggest gap: lowest valley on the left,
highest peak to its right.

```
Price
  7 |*
  6 |          *
  5 |    *
  4 |              *
  3 |      *
  2 |
  1 |  *          <-- min (buy here)
    +--+--+--+--+--+-->
     0  1  2  3  4  5   Day

           buy=1        sell=6
             |_____________|
                profit = 5
```

As you scan left to right:
- Keep a pin at the lowest point seen so far.
- At every point, measure the gap up to current price.
- Remember the biggest gap.

## When To Use This Pattern

- When you need to find the best **buy low, sell high** pair
  where buy comes before sell.
- When the problem asks for max difference where the
  smaller value appears first in the array.
- When a brute force O(n^2) loop over all pairs is too slow.
- When you only need to look backward (not forward) at
  each step — one pass is enough.

## The Approach

Walk through the prices once, left to right.
Keep track of the cheapest price seen so far.
At each day, ask: if I sell today, what is my profit?
Update the best profit if today beats it.
At the end, return the best profit found (or 0).

````python
from typing import List  # for type hints in the function signature
````

````python
def test_harness(func):
    """Run all test cases against func and report results."""

    tests = [
        # (prices, expected)
        ([7, 1, 5, 3, 6, 4],  5),   # normal case — buy 1 sell 6
        ([7, 6, 4, 3, 1],     0),   # always falling — no profit
        ([1],                 0),   # single element — no trade
        ([1, 2],              1),   # two elements — profit 1
        ([2, 1],              0),   # two elements — price falls
        ([3, 3, 3],           0),   # flat prices — no profit
        ([1, 10000],       9999),   # max spread
        ([10000, 1],          0),   # buy high sell low — skip
    ]

    passed = 0
    for i, (prices, expected) in enumerate(tests):
        result = func(prices)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        # Trim prices display to keep line under 80 chars
        prices_str = str(prices[:6])
        if len(prices) > 6:
            prices_str = prices_str[:-1] + ", ...]"
        print(
            f"Test {i+1}: {status} | "
            f"prices={prices_str} | "
            f"expected={expected} got={result}"
        )

    print(f"\n{passed}/{len(tests)} tests passed")
````

````python
from typing import List  # for type hints in the function signature
import math
def maxProfit(prices: List[int]) -> int:
    """
    Return the max profit from one buy-then-sell transaction.

    Approach:
    Walk the prices array once left to right.
    Keep min_price as the lowest price seen so far.
    At each day compute profit = price - min_price.
    Update max_profit if that profit is better.
    Return max_profit at the end (0 if never positive).

    Time:  O(n) — one pass through prices.
    Space: O(1) — only two variables stored.
    """
    profit, buy = 0, math.inf
    for p in prices:
        if p < buy:
            buy = p 
        profit = max(profit, p-buy)
    return profit
r'''
5
0
0
1
9999
Test 1: PASSED | prices=[7, 1, 5, 3, 6, 4] | expected=5 got=5
Test 2: PASSED | prices=[7, 6, 4, 3, 1] | expected=0 got=0
Test 3: PASSED | prices=[1] | expected=0 got=0
Test 4: PASSED | prices=[1, 2] | expected=1 got=1
Test 5: PASSED | prices=[2, 1] | expected=0 got=0
Test 6: PASSED | prices=[3, 3, 3] | expected=0 got=0
Test 7: PASSED | prices=[1, 10000] | expected=9999 got=9999
Test 8: PASSED | prices=[10000, 1] | expected=0 got=0

8/8 tests passed
'''

# Debug prints — run this cell while building your solution
print(maxProfit([7, 1, 5, 3, 6, 4]))  # expected: 5
print(maxProfit([7, 6, 4, 3, 1]))     # expected: 0
print(maxProfit([1]))                  # expected: 0
print(maxProfit([1, 2]))               # expected: 1
print(maxProfit([1, 10000]))           # expected: 9999
test_harness(maxProfit)
````

## Complexity

| Approach        | Time   | Space |
|-----------------|--------|-------|
| Brute Force     | O(n^2) | O(1)  |
| Single Pass     | O(n)   | O(1)  |

Brute force checks every buy/sell pair.
Single pass removes the inner loop by remembering
the minimum price seen so far at each step.
