## Official Problem Statement

Given a string `s`, find the length of the **longest substring**
without repeating characters.

**Constraints:**
- `0 <= s.length <= 5 * 10^4`
- `s` consists of English letters, digits, symbols, and spaces.


## What This Is Actually Asking

Imagine sliding a magnifying glass over the string.
The glass can only grow as long as every character inside it
is unique.
The moment a duplicate shows up on the right, you push the
left edge past the previous copy of that character.
You want to know the widest the glass ever got.


## Walk Through an Example by Hand

String: `"abcabcbb"`

```
Step 1: right=0 ('a')  window={a}         len=1  best=1
Step 2: right=1 ('b')  window={a,b}       len=2  best=2
Step 3: right=2 ('c')  window={a,b,c}     len=3  best=3
Step 4: right=3 ('a')  'a' in set!
        remove s[left]='a', left=1
        window={b,c}  add 'a' -> {b,c,a}  len=3  best=3
Step 5: right=4 ('b')  'b' in set!
        remove s[left]='b', left=2
        window={c,a}  add 'b' -> {c,a,b}  len=3  best=3
Step 6: right=5 ('c')  'c' in set!
        remove s[left]='c', left=3
        window={a,b}  add 'c' -> {a,b,c}  len=3  best=3
Step 7: right=6 ('b')  'b' in set!
        shrink left until 'b' gone -> left=5
        window={c}   add 'b' -> {c,b}     len=2  best=3
Step 8: right=7 ('b')  'b' in set!
        shrink -> left=7  window={b}      len=1  best=3
Answer: 3
```


## The Picture

Window condition: **all chars inside must be unique**

```
s = a  b  c  a  b  c  b  b
    0  1  2  3  4  5  6  7

Expanding (no duplicate yet):
    [  a  b  c  ]              window size = 3  ✓
    L              R

Duplicate 'a' detected at index 3:
    a  [  b  c  a  ]           shrink left past old 'a'
           L        R

Duplicate 'b' detected at index 4:
          [  c  a  b  ]        shrink left past old 'b'
             L        R

Window never exceeds size 3 → answer = 3
```

Key move: when `s[R]` is already in the set,
pop `s[L]` and advance `L` until the duplicate is gone.


## When To Use This Pattern

- When you see **"longest substring"** + a uniqueness constraint,
  think **sliding window with a set**.
- When the question asks for a **contiguous subarray/substring**
  with some property that breaks as soon as a condition is
  violated, think **shrink from left on violation**.
- When you need O(n) instead of O(n^2) for substring problems,
  think **two-pointer / sliding window**.
- When the window validity depends only on what's inside it
  (not absolute positions), think **set or frequency map**.


## The Approach

Use a set to track which characters are currently in the window.
Move the right pointer one step at a time, adding each character.
If the new character is already in the set, remove from the left
and advance the left pointer until the duplicate is gone.
After each right-pointer move, record the window size if it is
the largest seen so far.

## Real World Connection

At Citi, telemetry streams from ~6,000 endpoints emit event
records in real time through AWS Kinesis.
A common requirement is to find the longest run of records
where no event type repeats — essentially a uniqueness window
over a stream.
This sliding-window technique maps directly: the set becomes
the in-flight event-type tracker, and the left pointer is
advanced whenever a duplicate event type arrives.
In a DE pipeline you can implement this as a stateful Flink
or Kinesis Data Analytics operator with O(n) throughput.
