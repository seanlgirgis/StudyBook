## Official Problem Statement

Given an integer array `nums`, find a subarray that has the largest
product, and return the product.

The test cases are generated so that the answer will fit in a
**32-bit** integer.

**Constraints:**
- `1 <= nums.length <= 2 * 10^4`
- `-10 <= nums[i] <= 10`
- The product of any subarray of `nums` is guaranteed to fit in a
  32-bit integer.

## What This Is Actually Asking

Pick any contiguous slice of the array and multiply all its elements
together. Return the largest possible result of that multiplication.

The twist is zeroes reset the product to 1 and negatives flip the
sign, so the standard Kadane's max-sum idea needs an upgrade.

You need two trackers at each position: the best product ending here
and the *worst* (most negative) product ending here, because a future
negative could flip the worst into the best.

The global answer is the maximum value seen in `max_prod` across all
positions.

## Walk Through an Example by Hand

**Input:** `nums = [2, 3, -2, 4]`

```
Start: max_p=1, min_p=1, global=nums[0]=2

i=0, num=2:
  candidates = (2, 1*2, 1*2) = (2, 2, 2)
  max_p = max(2,2,2) = 2
  min_p = min(2,2,2) = 2
  global = max(2, 2) = 2

i=1, num=3:
  candidates = (3, 2*3, 2*3) = (3, 6, 6)
  max_p = 6
  min_p = 3
  global = max(2, 6) = 6

i=2, num=-2:
  candidates = (-2, 6*-2, 3*-2) = (-2, -12, -6)
  max_p = max(-2, -12, -6) = -2
  min_p = min(-2, -12, -6) = -12
  global = max(6, -2) = 6

i=3, num=4:
  candidates = (4, -2*4, -12*4) = (4, -8, -48)
  max_p = 4
  min_p = -48
  global = max(6, 4) = 6

Answer: 6  (subarray [2,3])
```

## The Picture

**Input:** `[-2, 3, -4]`  — expected answer: `24`

```
index :   0     1     2
num   :  -2     3    -4
       +-----+-----+-----+
max_p :  -2    3    24   <-- -4 flips min_p(-6) -> 24
min_p :  -2   -6   -12
global:  -2    3    24
       +-----+-----+-----+

At i=2:
  max_p_prev=-2  min_p_prev=-6  (saved before overwrite!)
  curr_max = max(-4, -2*-4, -6*-4) = max(-4, 8, 24) = 24
  curr_min = min(-4,  8,    24)           = -4

KEY: always save old max_p BEFORE computing new min_p
```

## When To Use This Pattern

- When the problem asks for an **optimal contiguous subarray** value,
  think Kadane-style DP.
- When **negatives can flip** the running value (product, not sum),
  think *track both max and min*.
- When a **zero resets** the streak, think: treat the element itself
  as a fresh start candidate.
- When you see "subarray" + "product" together, think LC 152 pattern.
- When O(n) time and O(1) space are needed, think running-state DP
  instead of a dp array.

## The Approach

Iterate once through the array keeping two variables: the largest
product ending at the current index (`max_p`) and the smallest
(most negative) product ending here (`min_p`).

At each element, the new max is the best of three choices: start
fresh at `num`, extend the previous max by `num`, or extend the
previous min by `num` (in case `num` is negative and flips it).
The new min is the worst of the same three choices.

Because we use the *old* `max_p` to compute the new `min_p`, we
must save the old `max_p` before overwriting it.

Track the running global maximum across all positions and return it.
