# LC 5 — Longest Palindromic Substring
**Difficulty:** Medium | **Category:** String DP / Palindromes
**Pattern:** Expand Around Center

## Official Problem Statement

Given a string `s`, return the longest palindromic substring in `s`.

A string is a palindrome if it reads the same forward and backward.

**Constraints:**
- `1 <= s.length <= 1000`
- `s` consists of only digits and English letters.


## What This Is Actually Asking

Find the longest stretch of the string that reads the same forwards
and backwards. "aba" is a palindrome. "abba" is a palindrome.
"abc" is not. You must return one of the valid longest substrings
(if there's a tie, any valid answer is accepted).
The substring must be contiguous — no skipping characters.


## Walk Through an Example by Hand

Input: `s = "babad"`

Try each center:
```
i=0 center='b': expand -> 'b' (len 1)
i=1 center='a': expand -> 'a', then b==b? No. -> 'a'
              odd: 'a'. even: s[0]s[1]='ba'? b!=a -> skip
i=1 center='a': left=0,right=2 -> s[0]='b',s[2]='b' match!
              expand to 'bab' (len 3). left=-1 stop.
i=2 center='b': left=1,right=3 -> 'a'=='a' match!
              expand to 'aba' (len 3). left=0,right=4 'b'!='d' stop.
i=3 center='a': expand -> 'a' (len 1)
i=4 center='d': expand -> 'd' (len 1)
```
Best: 'bab' or 'aba' — both length 3. Return either.


## The Picture

Expand-Around-Center visualization for `s = "racecar"`:

```
Index:  0   1   2   3   4   5   6
Char:   r   a   c   e   c   a   r

Center at i=3 ('e'), odd expansion:

Step 0:         [e]           len=1
Step 1:       c [e] c         match! len=3
Step 2:     a c [e] c a       match! len=5
Step 3:   r a c [e] c a r     match! len=7  <-- WINNER
Step 4:  out of bounds, stop

Odd center (single char):
  L <--- i ---> R
  while s[L] == s[R]: expand

Even center (between chars i and i+1):
  L <-- | --> R
  while s[L] == s[R]: expand
```

Each call to `expand(l, r)` walks outward, returning the
max palindrome string starting from that center.


## When To Use This Pattern

- When you see "longest palindromic substring", think expand-around-center.
- When the answer is a contiguous palindrome (not subsequence),
  think two-pointer expansion from every index.
- When O(n²) time and O(1) space is acceptable, prefer this
  over DP table (same time, less space).
- When you need to count or find palindromes, think center iteration.
- When you see a string problem asking for symmetry around a point,
  think expand-around-center.


## The Approach

For each index in the string, treat it as the center of a palindrome
and expand outward in both directions as long as characters match.
Handle both odd-length (single char center) and even-length
(two-char center) expansions separately. Track the start and end
indices of the longest palindrome seen so far, update whenever
a longer one is found. Return the substring at those indices.

````python
def test_harness(func):
    """Test LC 5 — checks length, not exact string (ties allowed)."""
    tests = [
        # (input_s, expected_length)
        ("babad",   3),   # 'bab' or 'aba'
        ("cbbd",    2),   # 'bb'
        ("a",       1),   # single char
        ("ac",      1),   # no palindrome longer than 1
        ("racecar", 7),   # whole string
        ("aacabdkacaa", 3),  # 'aca' or 'aaa'? -> 'aca' len 3
        ("aaaa",    4),   # all same
    ]
    passed = 0
    for s, exp_len in tests:
        result = func(s)
        # verify it's actually a palindrome
        is_palindrome = (result == result[::-1])
        # verify it appears in s
        in_string = (result in s)
        got_len = len(result)
        ok = is_palindrome and in_string and (got_len == exp_len)
        status = "PASSED" if ok else "FAILED"
        if ok:
            passed += 1
        else:
            print(
                f"{status} s={s!r} | "
                f"expected len={exp_len} | "
                f"got={result!r} len={got_len} "
                f"palindrome={is_palindrome}"
            )
    total = len(tests)
    print(f"\n{passed}/{total} tests passed.")
````

````python
def longestPalindrome(s: str) -> str:
    """
    LC 5 — Longest Palindromic Substring

    Restatement:
        Given string s, return the longest contiguous
        palindromic substring.

    Approach:
        Expand around center. For each index i, expand
        odd (center i) and even (center i,i+1). Track
        best start/end by comparing lengths.

    Time:  O(n^2) — n centers, each O(n) expansion
    Space: O(1)   — only index pointers
    """
    res = ""
    def traverseIt(ss, l,r, result):
        while l>=0 and r < len(ss) and ss[r]==ss[l]:
            if (r-l+1) > len(result):
                result=ss[l:r+1]
            l-=1
            r+=1
        return result

        
    for i in range(len(s)):
        #odd Length 1 center
        l,r=i,i
        res = traverseIt(s,l,r, res)
        #Even length
        l,r = i, i+1
        res = traverseIt(s,l,r, res)
    return res
r'''
bab
bb
racecar
a
aaaa

7/7 tests passed.

'''
# Quick debug prints (expected in comments)
print(longestPalindrome("babad"))    # 'bab' or 'aba' (len 3)
print(longestPalindrome("cbbd"))     # 'bb'
print(longestPalindrome("racecar"))  # 'racecar'
print(longestPalindrome("a"))        # 'a'
print(longestPalindrome("aaaa"))     # 'aaaa'
test_harness(longestPalindrome)

````

````python
def longestPalindrome(s: str) -> str:
    """
    LC 5 — Longest Palindromic Substring

    Restatement:
        Given string s, return the longest contiguous
        palindromic substring.

    Approach:
        Expand around center. For each index i, expand
        odd (center i) and even (center i,i+1). Track
        best start/end by comparing lengths.

    Time:  O(n^2) — n centers, each O(n) expansion
    Space: O(1)   — only index pointers
    """
    res = ""
    for i in range(len(s)):
        #odd Length 1 center
        l,r=i,i
        while l>=0 and r < len(s) and s[r]==s[l]:
            if (r-l+1) > len(res):
                res=s[l:r+1]
            l-=1
            r+=1
        #Even length
        l,r = i, i+1
        while l>=0 and r < len(s) and s[r]==s[l]:
            if (r-l+1) > len(res):
                res=s[l:r+1]
            l-=1
            r+=1
    return res

# Quick debug prints (expected in comments)
print(longestPalindrome("babad"))    # 'bab' or 'aba' (len 3)
print(longestPalindrome("cbbd"))     # 'bb'
print(longestPalindrome("racecar"))  # 'racecar'
print(longestPalindrome("a"))        # 'a'
print(longestPalindrome("aaaa"))     # 'aaaa'
test_harness(longestPalindrome)
````

## Complexity

| Approach              | Time   | Space |

| DP table              | O(n²)  | O(n²) |
| Expand around center  | O(n²)  | O(1)  |
| Manacher's algorithm  | O(n)   | O(n)  |
