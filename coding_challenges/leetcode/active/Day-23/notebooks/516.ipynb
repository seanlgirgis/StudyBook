# LC 516 — Longest Palindromic Subsequence
**Difficulty:** Medium | **Category:** String DP / Palindromes
**Pattern:** Interval DP (Bottom-Up by Length)

<div style="border-left:4px solid purple; padding:10px 16px;
background:#f5f0ff; margin-top:12px;">
<strong>Core Insight:</strong> If the outer characters match,
they contribute +2 to the LPS of the inner substring.
If they don't match, take the best of shrinking from either
end. Fill the DP table diagonally from length 1 upward so
smaller subproblems are always solved first.
</div>


## Official Problem Statement

Given a string `s`, find the longest palindromic subsequence's
length in `s`.

A subsequence is a sequence that can be derived from another
sequence by deleting some or no elements without changing the
order of the remaining elements.

**Constraints:**
- `1 <= s.length <= 1000`
- `s` consists only of lowercase English letters.


## What This Is Actually Asking

You can delete characters from `s` (but keep their order),
and you want the longest result that's a palindrome.
Unlike substrings, subsequences don't have to be contiguous.
"bbbab" -> delete 'a' -> "bbbb" or keep 'b','b','b' in order
-> longest palindromic subsequence is "bbbb" (length 4).


## Walk Through an Example by Hand

s = "bbbab"

dp[i][j] = LPS length of s[i..j] (inclusive)

```
Length 1: dp[i][i] = 1 for all i
  dp[0][0]=1 'b'
  dp[1][1]=1 'b'
  dp[2][2]=1 'b'
  dp[3][3]=1 'a'
  dp[4][4]=1 'b'

Length 2:
  dp[0][1]: s[0]='b'==s[1]='b' -> dp[1][0]+2 = 0+2 = 2
  dp[1][2]: s[1]='b'==s[2]='b' -> 2
  dp[2][3]: s[2]='b'!=s[3]='a' -> max(dp[2][2],dp[3][3])=1
  dp[3][4]: s[3]='a'!=s[4]='b' -> 1

Length 3:
  dp[0][2]: s[0]='b'==s[2]='b' -> dp[1][1]+2=1+2=3
  dp[1][3]: s[1]='b'!=s[3]='a' -> max(dp[1][2],dp[2][3])=2
  dp[2][4]: s[2]='b'==s[4]='b' -> dp[3][3]+2=1+2=3

Length 4:
  dp[0][3]: s[0]='b'!=s[3]='a' -> max(dp[0][2],dp[1][3])=3
  dp[1][4]: s[1]='b'==s[4]='b' -> dp[2][3]+2=1+2=3

Length 5:
  dp[0][4]: s[0]='b'==s[4]='b' -> dp[1][3]+2=2+2=4
```
Answer: dp[0][4] = 4  ("bbbb")


## The Picture

DP table for `s = "bbbab"` (rows=i, cols=j):

```
     j=0  j=1  j=2  j=3  j=4
       b    b    b    a    b
i=0 b  1    2    3    3    4  <-- answer
i=1 b  .    1    2    2    3
i=2 b  .    .    1    1    3
i=3 a  .    .    .    1    1
i=4 b  .    .    .    .    1

Fill order: diagonal by (j - i) length = 0, 1, 2, 3, 4

Rule:
  s[i] == s[j]:  dp[i][j] = dp[i+1][j-1] + 2
       +---------+
       | i+1..j-1|  <- inner subproblem
       +---------+
  [i]            [j]  <- outer chars both match -> +2

  s[i] != s[j]:  dp[i][j] = max(dp[i+1][j], dp[i][j-1])
  shrink left end  OR  shrink right end
```


## When To Use This Pattern

- When you see "longest palindromic subsequence", think
  interval DP on s[i..j].
- When outer characters matching gives a +2 bonus over the
  inner subproblem, think this exact recurrence.
- When the problem requires filling a 2D table diagonally
  (shorter intervals before longer ones), think interval DP.
- When a problem can be reduced to LCS of s and reverse(s),
  that's equivalent to LPS.
- When you see subsequence + palindrome together, this is
  the go-to pattern.


## The Approach

Create a 2D DP table where dp[i][j] is the length of the
longest palindromic subsequence in s[i..j]. Initialize the
diagonal (single characters) to 1. Fill by increasing interval
length: if s[i] == s[j], dp[i][j] = dp[i+1][j-1] + 2; else
dp[i][j] = max(dp[i+1][j], dp[i][j-1]). Return dp[0][n-1].

## Complexity

| Approach               | Time   | Space  |
|------------------------|--------|--------|
| Brute force (all subseq)| O(2^n) | O(n)   |
| Memoized recursion     | O(n²)  | O(n²)  |
| Bottom-up DP (table)   | O(n²)  | O(n²)  |
| LCS of s + reverse(s)  | O(n²)  | O(n²)  |
