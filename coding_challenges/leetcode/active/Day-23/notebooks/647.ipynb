## Official Problem Statement

Given a string `s`, return the number of palindromic substrings
in it.

A string is a palindrome when it reads the same backward as
forward. A substring is a contiguous sequence of characters
within the string.

**Constraints:**
- `1 <= s.length <= 1000`
- `s` consists of lowercase English letters.


## What This Is Actually Asking

Count every substring that is a palindrome — including
single characters, which are always palindromes.
"abc" has 3 palindromes: 'a', 'b', 'c'.
"aaa" has 6: 'a'(x3), 'aa'(x2), 'aaa'(x1).
You are counting occurrences, not unique palindromes.


## Walk Through an Example by Hand

Input: `s = "aaa"`

All substrings and palindrome check:
```
"a" (i=0)  -> palindrome YES
"a" (i=1)  -> palindrome YES
"a" (i=2)  -> palindrome YES
"aa"(0..1) -> palindrome YES
"aa"(1..2) -> palindrome YES
"aaa"      -> palindrome YES
Total = 6
```

Expand around center trace:
```
i=0 odd:  'a'       count+1=1
i=0 even: s[0]s[1]='aa' match, count+1=2. expand-> oob stop.
i=1 odd:  'a'       count+1=3
          expand: s[0]s[2]='a'=='a' match, count+1=4
          expand: oob stop
i=1 even: s[1]s[2]='aa' match, count+1=5. expand-> oob stop.
i=2 odd:  'a'       count+1=6
i=2 even: s[2]s[3] oob skip
Total = 6  CORRECT
```


## The Picture

Expand-Around-Center for `s = "abc"` vs `s = "aba"`:

```
s = "abc"  (no matches beyond single chars)

  i=0  [a]  b  c      odd expansion: 'a' -> count 1
              <->     even: a!=b, stop
  i=1   a  [b] c      odd: 'b' -> count 2
           <->        even: b!=c, stop
  i=2   a   b [c]     odd: 'c' -> count 3
  Total = 3

s = "aba"

  i=0  [a]  b  a      odd: 'a' -> count 1
              <->     even: a!=b, stop
  i=1   a  [b] a      odd: 'b' -> count 2
        <--     -->   expand: s[0]=a, s[2]=a MATCH -> count 3
        oob, stop
              <->     even: b!=a, stop
  i=2   a   b [a]     odd: 'a' -> count 4
  Total = 4

Each arrow step that succeeds = +1 palindrome found.
```


## When To Use This Pattern

- When you need to count palindromic substrings, think
  expand-around-center with a counter.
- When every valid expansion step equals one new palindrome,
  increment inside the while loop.
- When the problem is about contiguous substrings (not subsequences),
  prefer expand-around-center over DP.
- When you see `count all` instead of `find longest`, the same
  center expansion applies — just count instead of tracking max.
- When constraints are n <= 1000, O(n²) is perfectly fine.


## The Approach

Start with a count of 0. For each index, treat it as the center
of a palindrome and expand outward while characters match, adding
1 to count for each expansion including the initial single-char.
Do this twice per index: once for odd-length (center at char)
and once for even-length (center between chars i and i+1).
Return the total count.
