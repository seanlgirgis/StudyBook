## Official Problem Statement

You are a professional robber planning to rob houses along a street.
Each house has a certain amount of money stashed. The only constraint
stopping you from robbing each of them is that adjacent houses have
security systems connected — if two adjacent houses are broken into
on the same night, the police will be alerted.

Given an integer array `nums` representing the amount of money of
each house, return the maximum amount of money you can rob tonight
**without alerting the police**.

**Constraints:**
- `1 <= nums.length <= 100`
- `0 <= nums[i] <= 400`

## What This Is Actually Asking

Pick a subset of indices from the array so that no two chosen
indices are adjacent, and maximise the sum of chosen values.
This is a classic 1-D DP problem where the future decision at
index `i` depends only on the best results at `i-1` and `i-2`.
The adjacency constraint means we can never take two values in
a row, so every step is a simple max of two previous states.
Because we only look back two steps, we don't need to store the
whole DP array — two variables suffice.

## Walk Through an Example by Hand

`nums = [2, 7, 9, 3, 1]`

| i | nums[i] | prev2 (dp[i-2]) | prev1 (dp[i-1]) | cur = max(prev1, prev2+val) |
|---|---------|-----------------|-----------------|-----------------------------|
| 0 |    2    |        0        |        0        | max(0, 0+2) = **2**         |
| 1 |    7    |        0        |        2        | max(2, 0+7) = **7**         |
| 2 |    9    |        2        |        7        | max(7, 2+9) = **11**        |
| 3 |    3    |        7        |       11        | max(11, 7+3) = **11**       |
| 4 |    1    |       11        |       11        | max(11, 11+1) = **12**      |

Answer: **12** (rob houses 0, 2, 4 → 2+9+1=12)

## The Picture

```
nums =  [ 2,  7,  9,  3,  1 ]
index     0   1   2   3   4

dp[i] = max amount robbing up to house i

        dp[0]=2
           |
           v
        dp[1]=7          (skip house 0, take 7)
               \
                v
            dp[2]=11     (skip house 1, take dp[0]+9)
                    \
                     v
                 dp[3]=11 (skip house 2, dp[1]+3=10 < 11)
                        \
                         v
                     dp[4]=12 (dp[2]+1=12 > dp[3]=11)

Rolling variables (prev2, prev1) slide one step at a time:

 step 0:  prev2=0  prev1=0  -> cur=2   ; prev2=0,  prev1=2
 step 1:  prev2=0  prev1=2  -> cur=7   ; prev2=2,  prev1=7
 step 2:  prev2=2  prev1=7  -> cur=11  ; prev2=7,  prev1=11
 step 3:  prev2=7  prev1=11 -> cur=11  ; prev2=11, prev1=11
 step 4:  prev2=11 prev1=11 -> cur=12  ; answer=12
```

## When To Use This Pattern

- When you must **choose a subset with an adjacency restriction**,
  think DP with a skip-or-take recurrence.
- When the **optimal substructure** relies on only the previous
  1–2 states, think rolling variables instead of a full array.
- When the problem says "no two adjacent" or "at least one gap",
  think House Robber pattern.
- When you see a linear sequence with a local binary decision at
  each step, think 1-D DP.
- When you need O(1) space from an O(n) DP, think rolling
  variables — store only what the recurrence references.

## The Approach

Initialise two variables `prev2 = 0` and `prev1 = 0` representing
the best loot up to two houses ago and one house ago respectively.
For each house, compute `cur = max(prev1, prev2 + nums[i])` — either
skip this house or rob it plus the best from two steps back.
Slide the window: set `prev2 = prev1`, `prev1 = cur`.
After the loop, `prev1` holds the answer.

````python
from typing import List
````

````python
def test_harness(func):
    cases = [
        # (input_nums, expected)
        ([1, 2, 3, 1],        4),   # rob 0,2
        ([2, 7, 9, 3, 1],    12),   # rob 0,2,4
        ([0],                 0),   # single zero
        ([5],                 5),   # single element
        ([1, 1],              1),   # two houses
        ([2, 1, 1, 2],        4),   # rob 0,3
        ([0, 0, 0, 0],        0),   # all zeros
        ([400]*100,       20000),   # max constraints, even indices
    ]
    passed = 0
    for nums, expected in cases:
        result = func(nums)
        status = "PASSED" if result == expected else "FAILED"
        if status == "PASSED":
            passed += 1
        else:
            print(
                f"  {status} | nums={nums[:6]}{'...' if len(nums)>6 else ''}"
                f" | expected={expected} | got={result}"
            )
    total = len(cases)
    print(f"\nResults: {passed}/{total} passed")
    if passed == total:
        print("All tests passed!")
````

````python
def rob(nums: List[int]) -> int:
    """
    Return the maximum amount you can rob from a row of houses
    without robbing two adjacent houses.

    Strategy: 1-D DP with two rolling variables.
      prev2 = best loot up to two houses ago
      prev1 = best loot up to one house ago
      cur   = max(prev1, prev2 + nums[i])

    Args:
        nums: List of non-negative integers (house values).

    Returns:
        Maximum money that can be robbed.

    Time:  O(n)
    Space: O(1)
    """
    if len(nums) == 1 :return (nums[0])
    if len(nums) == 2 :return (max(nums[0], nums[1]))
    cur , prev, pprev = 0, max(nums[0], nums[1]), nums[0]
    for i in range (2, len(nums)):
        cur = max(nums[i]+pprev , prev)
        pprev = prev
        prev = cur
    return cur
        




def test():

    # Quick debug — run this cell while building
    print(rob([1,2,3,1]))      # 4
    print(rob([2,7,9,3,1]))    # 12
    print(rob([5]))             # 5
    print(rob([2,1,1,2]))      # 4
    test_harness(rob)
test()
````

## Complexity

| Approach      | Time   | Space  | Notes                          |
|---------------|--------|--------|--------------------------------|
| Brute force   | O(2^n) | O(n)   | Try every subset (recursion)   |
| Memoised DFS  | O(n)   | O(n)   | Cache each index result        |
| DP array      | O(n)   | O(n)   | Store all dp[i] values         |
| **Optimal**   | **O(n)**| **O(1)**| Two rolling variables         |
