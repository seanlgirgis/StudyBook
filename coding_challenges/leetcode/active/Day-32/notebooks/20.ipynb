## Official Problem Statement

Given a string `s` containing just the characters `'('`, `')'`,
`'{'`, `'}'`, `'['` and `']'`, determine if the input string
is valid.

An input string is valid if:
1. Open brackets must be closed by the same type of brackets.
2. Open brackets must be closed in the correct order.
3. Every close bracket has a corresponding open bracket of the
   same type.

**Constraints:**
- `1 <= s.length <= 10^4`
- `s` consists of parentheses only `'()[]{}'`

## What This Is Actually Asking

You have a string of bracket characters and need to decide if they
are properly nested and matched.

Every opening bracket must eventually be closed by the same type,
and inner pairs must close before outer ones.

Think of it like checking HTML tags — the most recently opened
tag must be the next one closed.

There is no arithmetic or counting — just structural correctness
of the nesting order.

## Walk Through an Example by Hand

Input: `s = "({[]})"` — expected: True

```
char  action           stack
----  ---------------  -----------
'('   push             ['(']
'{'   push             ['(', '{']
'['   push             ['(', '{', '[']
']'   top='[', match!  ['(', '{']
'}'   top='{', match!  ['(']
')'   top='(', match!  []

Stack is empty → return True
```

Counter-example: `s = "([)]"` — expected: False

```
char  action           stack
----  ---------------  -----------
'('   push             ['(']
'['   push             ['(', '[']
')'   top='[', NO match → return False
```

## The Picture

Stack evolution for `"({[]})"` — reading left to right:

```
index: 0    1    2    3    4    5
char:  (    {    [    ]    }    )

After index 0:        After index 2:
  +-----+               +-----+
  |  (  |  <-- top      |  [  |  <-- top
  +-----+               +-----+
                         |  {  |
                         +-----+
                         |  (  |
                         +-----+

After index 3 (']' pops '['):
  +-----+

  +-----+

  (empty) → True!
```

## When To Use This Pattern

- When you see **nested structures** that must open and close in
  order, think **stack**.
- When you see **matching pairs** where the last opened must be
  first closed (LIFO), think **stack**.
- When a problem involves **validating syntax** (brackets, tags,
  expressions), think **stack**.
- When a close token must match the **most recent** open token,
  think **stack**.
- When the problem has a "balanced" keyword, think **stack**.

## The Approach

Keep a mapping from each closing bracket to its expected opening
bracket. Scan the string left to right — push every open bracket
onto a stack.

When you encounter a closing bracket, check the top of the stack.
If the top matches the expected opener, pop it and continue.
Otherwise return False immediately.

After processing every character, the string is valid only if the
stack is completely empty — meaning every opener was matched.

````python
from typing import List  # standard type hints
````

````python
def test_harness(func):
    """
    Runs test cases for LC 20 Valid Parentheses.
    Each case: (input_string, expected_output)
    """
    cases = [
        ("()",       True),   # simple pair
        ("()[]{}",   True),   # multiple pairs
        ("(]",       False),  # wrong close type
        ("([)]",     False),  # interleaved, invalid
        ("{[]}",     True),   # nested pairs
        ("",         True),   # edge: empty string
        ("(",        False),  # edge: only open
        (")",        False),  # edge: only close
        ("({}[])",   True),   # multiple nested
        (")(",       False),  # reversed order
    ]
    passed = 0
    for s, expected in cases:
        result = func(s)
        status = "PASSED" if result == expected else "FAILED"
        if status == "FAILED":
            print(
                f"  {status} | input={repr(s)!r:15} "
                f"expected={expected} got={result}"
            )
        else:
            passed += 1
    total = len(cases)
    print(f"\nResult: {passed}/{total} tests passed.")
````

````python
def is_valid(s: str) -> bool:
    """
    Determine if the bracket string s is valid.

    Strategy:
      - Use a stack to track unmatched open brackets.
      - Map each closer to its expected opener.
      - On closer: if stack empty or top != opener, False.
      - Return True iff stack is empty at end.

    Args:
        s: string of bracket characters
    Returns:
        True if valid, False otherwise
    """
    matcher = { '}':'{' , ']':'[' , ')':'(' }
    stack = []

    for c in s:
        if c not in matcher:
            stack.append(c)         # if it is an open push it
        else:                       # It is a close
            if (not stack) or (stack and stack[-1] != matcher[c]):
                return False
            else:
                stack.pop()
    return not stack
            
        

# --- Debug prints (remove before submitting) ---

# Basic case — expect True
print(is_valid("()"))          # True

# Nested valid — expect True
print(is_valid("{[]}"))        # True

# Interleaved invalid — expect False
print(is_valid("([)]"))        # False

# Only open bracket — expect False
print(is_valid("("))           # False

# Only open bracket — expect False
print(is_valid(")"))           # False

# Empty string — expect True
print(is_valid(""))            # True
test_harness(is_valid)
````

## Complexity

| Approach    | Time   | Space  |
|-------------|--------|--------|
| Brute force | O(n²)  | O(n)   |
| Optimal (stack) | O(n) | O(n) |

- **Time O(n):** each character is pushed and popped at most once.
- **Space O(n):** worst case the stack holds n/2 open brackets
  (e.g., `"((((("`).
