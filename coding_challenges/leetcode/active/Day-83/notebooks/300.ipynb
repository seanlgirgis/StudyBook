## Official Problem Statement

Given an integer array `nums`, return the length of the longest
strictly increasing subsequence.

**Example 1:**
- Input: `nums = [10,9,2,5,3,7,101,18]`
- Output: `4`
- Explanation: `[2,3,7,101]` is the longest increasing subsequence.

**Example 2:**
- Input: `nums = [0,1,0,3,2,3]`
- Output: `4`

**Example 3:**
- Input: `nums = [7,7,7,7,7,7,7]`
- Output: `1`

**Constraints:**
- `1 <= nums.length <= 2500`
- `-10^4 <= nums[i] <= 10^4`

## What This Is Actually Asking

You have a list of numbers. Pick some of them — in order, left to
right — so that each picked number is bigger than the last.
You do not have to pick neighbours. You can skip numbers.
Return how many you picked in the longest such chain.

## Walk Through an Example by Hand

Input: `[10, 9, 2, 5, 3, 7, 101, 18]`

Start: every dp[i] = 1 (each number is a subsequence of length 1).

```
i=0  nums[0]=10   no j before it       dp=[1,1,1,1,1,1,1,1]
i=1  nums[1]=9    10>9, skip           dp=[1,1,1,1,1,1,1,1]
i=2  nums[2]=2    10>2,9>2 skip        dp=[1,1,1,1,1,1,1,1]
i=3  nums[3]=5    2<5 -> dp[3]=dp[2]+1 dp=[1,1,1,2,1,1,1,1]
i=4  nums[4]=3    2<3 -> dp[4]=dp[2]+1 dp=[1,1,1,2,2,1,1,1]
i=5  nums[5]=7    2<7,5<7,3<7          dp=[1,1,1,2,2,3,1,1]
                  best from j=3: dp[3]+1=3
i=6  nums[6]=101  all < 101            dp=[1,1,1,2,2,3,4,1]
                  best from j=5: dp[5]+1=4
i=7  nums[7]=18   2,5,3,7 < 18         dp=[1,1,1,2,2,3,4,4]
                  best from j=5: dp[5]+1=4
```

Answer: max(dp) = 4

## The Picture

Think of each number as a stair step. You want the longest
staircase going upward — but you can skip steps.

```
nums:  10   9   2   5   3   7  101  18
        |   |   |   |   |   |   |   |
dp:     1   1   1   2   2   3   4   4

Winning staircase:

           101
          /
         7
        /
       5   (or 3)
      /
     2

Each step points back to the tallest valid step before it.
dp[i] = how many stairs tall is the staircase ending HERE.
```

## When To Use This Pattern

- When the problem asks for the *longest* or *shortest* chain
  that follows a rule about ordering.
- When you need to pick elements left-to-right with a
  constraint on consecutive picks.
- When "what is the best I can do ending at position i?"
  is a useful question.
- When overlapping sub-problems appear — the same sub-answer
  is reused many times.

## The Approach

Build a dp array the same length as nums, filled with 1s.
For each position i, scan every j before it.
If nums[j] is smaller than nums[i], we could extend the
subsequence ending at j — so dp[i] = max(dp[i], dp[j] + 1).
The answer is the largest value in the whole dp array.

## Complexity

| Approach           | Time       | Space |
|--------------------|------------|-------|
| Brute force (all subsequences) | O(2^n) | O(n) |
| 1D DP (this approach)          | O(n^2) | O(n) |
| Patience sorting (optimal)     | O(n log n) | O(n) |

The DP approach avoids re-exploring all subsets by reusing
previously computed sub-answers — each dp[i] is built once.
